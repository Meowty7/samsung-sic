## Objetivos de aprendizaje

## Contexto

En el análisis de datos real, los conjuntos de datos rara vez están listos para su análisis: contienen valores faltantes, formatos inconsistentes o están desorganizados. La ordenación de datos (Tidy Data) consume una gran parte del tiempo del científico de datos y sigue principios como «cada variable en su columna» y «cada observación en su fila». Con Pandas se detectan los NaN con isnull(), se eliminan con dropna() o se reemplazan con fillna(), usando valores como la media o la moda. Los datos de series temporales son secuencias de puntos indexados en el tiempo, como precios de acciones o producción anual: Pandas ofrece Timestamp para puntos específicos y Period para intervalos, y convertir una columna de fechas a datetime64 y usarla como índice (DatetimeIndex) permite cortes temporales intuitivos; rolling().mean() suaviza fluctuaciones y revela tendencias. La segunda clase aplica todo en dos mini proyectos: el análisis de datos financieros del mercado estadounidense (índices, bonos, materias primas y oro obtenidos de Yahoo Finance; cambio porcentual diario, rendimiento acumulado, volatilidad y correlación entre activos, incluido el periodo de la pandemia de COVID-19) y el análisis de datos públicos de la pandemia de coronavirus, con estadísticas de vacunación por país y un mapa coroplético interactivo de casos acumulados creado con Folium.

## Vocabulario clave

- **NaN:** Not a Number; valor utilizado en Pandas para representar datos faltantes u omitidos en un DataFrame.
- **Datos ordenados (Tidy Data):** forma estándar de estructurar datos donde cada variable es una columna, cada observación es una fila y cada valor tiene su propia celda.
- **Estadísticas descriptivas:** métricas como media, mediana, moda, varianza y desviación estándar que resumen y describen las características principales de un conjunto de datos.
- **Timestamp:** un punto específico en el tiempo (por ejemplo, una fecha y hora exactas), representado en Pandas con el tipo datetime64.
- **Media móvil:** cálculo estadístico de series temporales que suaviza fluctuaciones a corto plazo y revela tendencias a largo plazo, calculado como la media de una ventana deslizante de datos.
- **Ticker:** código único que identifica un valor negociado públicamente en una bolsa (por ejemplo, SPY para el ETF del S&P 500).
- **Serie temporal:** conjunto de datos recolectados en intervalos de tiempo sucesivos, como los precios de cierre diarios de una acción.
- **Cambio porcentual diario:** variación relativa de un valor de un día para otro: ((precio hoy - precio ayer) / precio ayer) * 100.
- **Coeficiente de correlación:** medida de la fuerza y dirección de la relación lineal entre dos variables; en Pandas se calcula con .corr().
- **Mapa coroplético:** mapa temático donde las áreas geográficas se colorean en proporción a un valor estadístico, como la tasa de infección por país.
