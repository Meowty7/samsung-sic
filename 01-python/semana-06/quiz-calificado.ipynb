## Q 34-01 (20 puntos)

### Consigna

Convierta los siguientes datos del diccionario en un dataframe usando Pandas. (Denomina el dataframe `df`)

```python
d = { 'col1':[1,2], 'col2':[3,4], 'col3':[5,6], 'col4':[7,8] }
```

**Dónde se ve:** [Unidad 34](../semana-05/clase-2-modulos-y-pandas/u34-modulos-de-python.ipynb) (uso de módulos de Python, semana 5). En resumen: `import pandas as pd` trae el módulo y `pd.DataFrame(d)` convierte un diccionario de listas en tabla: cada clave es una columna.

### Respuesta

````
import pandas as pd

d = {'col1': [1, 2], 'col2': [3, 4], 'col3': [5, 6], 'col4': [7, 8]}

df = pd.DataFrame(d)
print(df)
````

**Explicación:** **Respuesta: `df = pd.DataFrame(d)`.** Cada clave del diccionario pasa a ser el nombre de una columna y cada lista sus valores; el índice se crea solo (0 y 1). La respuesta incluye `import pandas as pd` y la definición de `d` para que el código corra solo en la plataforma.

## Q 36-01 (20 puntos)

### Consigna

El índice del dataframe creado en la pregunta Q 34-01 es 0 y 1. Escriba un comando para cambiar el nombre de estos índices a `primero` y `segundo`.

**Dónde se ve:** [Unidad 36](../semana-05/clase-2-modulos-y-pandas/u36-dataframes-de-pandas.ipynb) (Pandas DataFrame, semana 5). En resumen: el atributo `df.index` se puede reasignar con una lista de la misma longitud, y `df.rename(index={...})` cambia etiquetas concretas.

### Respuesta

````
import pandas as pd

d = {'col1': [1, 2], 'col2': [3, 4], 'col3': [5, 6], 'col4': [7, 8]}
df = pd.DataFrame(d)

df.index = ['primero', 'segundo']
print(df)

# Otra forma equivalente (devuelve un dataframe nuevo):
df2 = pd.DataFrame(d).rename(index={0: 'primero', 1: 'segundo'})
print(df2.equals(df))
````

**Explicación:** **Respuesta: `df.index = ['primero', 'segundo']`.** La lista debe tener tantos elementos como filas. También sirve `df.rename(index={0: 'primero', 1: 'segundo'})`, que devuelve un dataframe nuevo (hay que asignarlo o usar `inplace=True`). La última línea comprueba que ambas formas dan el mismo resultado.

**Nota sobre las preguntas siguientes:** las respuestas de Q 37-01 y Q 38-01 trabajan con el `df` **original** de la Q 34-01 (índice 0 y 1); como el cambio de índice no altera los datos, el resultado sería el mismo con las filas llamadas `primero` y `segundo`, solo cambiarían las etiquetas de fila.

## Q 37-01 (20 puntos)

### Consigna

Escriba un comando para buscar datos faltantes en el dataframe `df` creado en la pregunta Q 34-01 e imprima el resultado. (Sin embargo, los datos que faltan deben devolverse como verdaderos).

**Dónde se ve:** [Unidad 37](clase-1-ordenacion-y-series-temporales/u37-ordenacion-de-datos.ipynb) (y la [Unidad 36](../semana-05/clase-2-modulos-y-pandas/u36-dataframes-de-pandas.ipynb) de la semana 5). En resumen: `isnull()` (o `isna()`) devuelve un dataframe del mismo tamaño con `True` donde falta el dato.

### Respuesta

````
import pandas as pd

d = {'col1': [1, 2], 'col2': [3, 4], 'col3': [5, 6], 'col4': [7, 8]}
df = pd.DataFrame(d)

print(df.isnull())

# Para contar los faltantes por columna:
print(df.isnull().sum())
````

**Explicación:** **Respuesta: `print(df.isnull())`.** Devuelve `True` donde falta el dato y `False` donde existe (lo que pide el paréntesis de la consigna; `notnull()` haría lo contrario). En este dataframe no falta nada, así que todo es `False`. `df.isnull().sum()` cuenta los faltantes por columna; `df.isna()` es un sinónimo.

## Q 38-01 (20 puntos)

### Consigna

Escriba un comando para verificar el resumen de las estadísticas descriptivas (desviación estándar, valor mínimo, moda, etc.) del dataframe `df` creado en la pregunta Q 34-01 e imprima el resultado.

**Dónde se ve:** [Unidad 38](clase-1-ordenacion-y-series-temporales/u38-series-temporales.ipynb) (y la [Unidad 36](../semana-05/clase-2-modulos-y-pandas/u36-dataframes-de-pandas.ipynb) de la semana 5). En resumen: `describe()` resume cada columna numérica con cantidad, media, desviación estándar, mínimo, cuartiles y máximo.

### Respuesta

````
import pandas as pd

d = {'col1': [1, 2], 'col2': [3, 4], 'col3': [5, 6], 'col4': [7, 8]}
df = pd.DataFrame(d)

print(df.describe())

# describe() no incluye la moda; se obtiene aparte:
print(df.mode())
````

**Explicación:** **Respuesta: `print(df.describe())`.** Muestra `count`, `mean`, `std` (desviación estándar), `min`, `25%`, `50%`, `75%` y `max`.

**Nota sobre el enunciado:** menciona la moda, pero `describe()` **no la incluye**. Se agrega `print(df.mode())` como complemento: aquí todos los valores de cada columna aparecen una sola vez, así que la moda devuelve ambos valores (dos filas). Si la plataforma solo espera `describe()`, esa línea basta.
