## Contexto

La primera clase presenta técnicas avanzadas de diseño de algoritmos. La programación dinámica aborda problemas con subestructura óptima y subproblemas superpuestos, almacenando soluciones parciales para evitar cálculos redundantes; se aplica a la secuencia de Fibonacci y al triángulo de Pascal, comparando la recursión simple con la memorización y el cálculo ascendente. El backtracking es una búsqueda en profundidad que construye soluciones candidatas paso a paso y retrocede cuando una candidata no puede llevar a una solución válida; se aplica al problema de las N-Reinas, podando las ramas no prometedoras. La segunda clase extiende Python con módulos: los estándar (random, time, datetime) en la misión del juego de dados, y los externos pandas y matplotlib. Con pandas se trabaja con Series (estructura unidimensional que asocia índices a datos) y DataFrames (tablas bidimensionales), cargando datos desde CSV, Excel y JSON, filtrando, seleccionando y calculando estadísticas; con matplotlib se grafican, por ejemplo, la densidad de población de ciudades o la correlación entre variables musicales.

## Vocabulario clave

- **Programación dinámica:** técnica que resuelve problemas con subproblemas superpuestos guardando las soluciones parciales para no recalcularlas.
- **Memorización (memoization):** guardar el resultado de una llamada recursiva para reutilizarlo cuando se repiten los mismos argumentos.
- **Cálculo ascendente (bottom-up):** resolver primero los subproblemas más pequeños y construir con ellos la solución de los más grandes.
- **Backtracking (retroceso):** búsqueda en profundidad que construye candidatas paso a paso y retrocede cuando una no puede llevar a una solución válida.
- **Poda:** descartar ramas no prometedoras del espacio de búsqueda para no explorarlas.
- **Módulo:** archivo de código Python (.py) que agrupa funciones, clases y variables para su reutilización; puede ser estándar (incluido en Python) o externo (instalado por el usuario).
- **Serie (pandas):** estructura de datos unidimensional de pandas que almacena datos asociados a un índice.
- **DataFrame (pandas):** estructura de datos bidimensional de pandas, similar a una hoja de cálculo, compuesta por filas y columnas; cada columna es una Serie.
- **Índice:** etiqueta o posición que identifica cada elemento de una Serie o cada fila de un DataFrame.
- **API:** conjunto de reglas que permite a distintas aplicaciones comunicarse; en el curso se usa para obtener datos de servicios remotos.
