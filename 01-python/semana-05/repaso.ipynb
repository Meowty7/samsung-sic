## Unidad 32 — Programación dinámica

[Ir a la unidad](clase-1-programacion-dinamica-y-backtracking/u32-programacion-dinamica.ipynb)

```python
fib1(n)  # recursión pura: fib1(n-1) + fib1(n-2), recalcula subproblemas
F = {0: 0, 1: 1}  # tabla de fib2 con los casos base
if n in F: return F[n]  # fib2: devuelve el valor ya calculado
F[n] = fib2(n - 1) + fib2(n - 2)  # fib2: calcula, guarda y devuelve
F = [0, 1] + [-1] * (N - 1)  # fib3: lista de largo N + 1 con -1 = no calculado
if F[n] < 0: F[n] = fib3(F, n - 1) + fib3(F, n - 2)  # fib3: calcular solo si falta
a, b = b, a + b  # fib4: avance ascendente con dos variables
bin1(n - 1, k - 1) + bin1(n - 1, k)  # bin1: recurrencia del coeficiente binomial
B = [[0] * (i + 1) for i in range(n + 1)]  # bin2: triángulo con n + 1 filas
B[i][j] = B[i - 1][j - 1] + B[i - 1][j]  # bin2: celda = suma de las dos de arriba
if j == 0 or j == i: B[i][j] = 1  # bin2: bordes del triángulo
for j in range(n, -1, -1): B[j] = B[j] + B[j - 1]  # bin3: una sola lista, de derecha a izquierda
mejor_hasta = max(S[i], mejor_hasta + S[i])  # subarreglo máximo: recurrencia ascendente
```

**Errores comunes**

- Creer que la recursión es lenta por sí misma: lo que cuesta es recalcular subproblemas superpuestos.
- Olvidar que la tabla `F` de `fib2` es global y se conserva entre llamadas; hay que reiniciarla para medir tiempos con justicia.
- Crear la lista de `fib3` con menos de `n + 1` posiciones y provocar un `IndexError`.
- Recorrer `bin3` de izquierda a derecha: sobrescribe valores de la fila anterior que aún se necesitan.
- Llamar a `bin1(n, k)` con `k > n` o `k < 0`: nunca alcanza el caso base.
- Inicializar el máximo del subarreglo en 0 cuando todos los números pueden ser negativos.

## Unidad 33 — Retroceso (backtracking)

[Ir a la unidad](clase-1-programacion-dinamica-y-backtracking/u33-backtracking.ipynb)

```python
def depth_first_search(v): visit(v); [depth_first_search(u) for u in hijos(v)]  # DFS recursiva
if not promising(i, col): return  # poda: volver sin expandir
col[i] == col[k]  # misma columna
abs(col[i] - col[k]) == i - k  # misma diagonal
for k in range(i):  # reinas ya colocadas (filas 0 a i-1)
col[i + 1] = j  # colocar la reina de la fila siguiente en la columna j
n_queens(i + 1, col)  # continuar la DFS en la fila siguiente
n_queens(-1, [-1] * N)  # llamada inicial para N reinas
resultado.append(col.copy())  # guardar una solución (copiar la lista)
actual.pop()  # deshacer la elección al retroceder
visitado[f][c] = True  # marcar casilla en un laberinto para evitar ciclos
if suma > objetivo: return  # poda en suma de subconjuntos
isqrt(n) ** 2 == n  # comprobar cuadrado perfecto
```

**Errores comunes**

- No detener la recursión en la última fila y seguir expandiendo.
- Guardar `col` sin copiarla: la lista se modifica en el resto de la búsqueda.
- Omitir el valor absoluto al comparar diferencias de columnas en la diagonal.
- No deshacer la elección (pop o desmarcar) al volver de una rama cuando el estado se comparte.
- Confundir "prometedor" con "solución": un nodo prometedor aún puede no ser solución.
- En laberintos, no marcar casillas visitadas y caer en ciclos infinitos.
