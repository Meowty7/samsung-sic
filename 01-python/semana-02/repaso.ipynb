## Unidad 18 — Llamada a función de recursión

[Ir a la unidad](clase-1-funciones-recursividad-lambda/u18-recursividad.ipynb)

```python
def factorial(n): return 1 if n <= 1 else n * factorial(n - 1)  # factorial recursivo con caso base n <= 1
for i in range(1, n + 1): fact = fact * i  # factorial con bucle (fact empieza en 1)
def fibonacci(n): return n if n <= 1 else fibonacci(n-1) + fibonacci(n-2)  # Fibonacci recursivo
dic = {0: 0, 1: 1}  # diccionario de memorización con los casos base
if n in dic: return dic[n]  # devuelve el resultado ya calculado sin recursión
dic[n] = fibonacci_2(n-1) + fibonacci_2(n-2)  # calcula, guarda y evita recálculos
Timer("fibonacci_1(30)", globals=globals()).timeit(number=20)  # mide el tiempo de 20 ejecuciones
hanoi(n-1, de_, a_, via_)  # mueve n-1 discos de origen a auxiliar (paso 1 de Hanoi)
hanoi(n-1, via_, de_, a_)  # mueve n-1 discos de auxiliar a destino (paso 3 de Hanoi)
2 ** n - 1  # movimientos mínimos de la Torre de Hanoi con n discos
sorted("EABFD")  # devuelve lista ordenada ['A', 'B', 'D', 'E', 'F']
sorted(lista, reverse=True)  # ordena en forma descendente
abs(-100), min(200, 100), max(200, 100)  # valor absoluto, mínimo y máximo
eval("100+200+300")  # evalúa una cadena como expresión (600); usar con precaución
chr(65), ord('A')  # 'A' y 65: conversión entre carácter y Unicode
```

**Errores comunes**

- Olvidar el caso base: la función no se detiene y Python lanza RecursionError (profundidad máxima superada).
- Escribir un caso recursivo que no se acerca al caso base (por ejemplo llamar a factorial(n + 1)).
- Usar Fibonacci recursivo sin memorización con valores grandes (30 o más): repite los mismos cálculos y es muy lento.
- Confundir el orden de los parámetros de hanoi(n, de_, via_, a_): el tercero es la barra auxiliar y el cuarto el destino.
- Usar eval() con texto escrito por un usuario no confiable: ejecuta cualquier código.
- Fiarse de los nombres traducidos del PDF (filtro, mapa, clasificado, gama): en Python son filter, map, sorted y range.

## Unidad 19 — Lambda

[Ir a la unidad](clase-1-funciones-recursividad-lambda/u19-lambda-filter-map-reduce.ipynb)

```python
lambda x, y: x + y  # función anónima que suma dos valores
(lambda x, y: x + y)(100, 200)  # define y llama la lambda a la vez (300)
add = lambda x, y: x + y  # lambda asignada a un nombre
filter(lambda x: x >= 19, ages)  # iterador con los elementos que cumplen la condición
list(filter(lambda x: x % 2 == 0, n_list))  # lista con los números pares
list(map(lambda x: x**2, a))  # lista con el cuadrado de cada elemento
list(map(int, "10 20 30".split()))  # convierte cadenas a enteros
from functools import reduce  # reduce vive en functools
reduce(lambda x, y: x + y, [1, 2, 3, 4])  # combina todo en un valor (10)
it = iter([10, 20, 30]); next(it)  # convierte en iterador y pide el siguiente elemento (10)
[x**2 for x in range(1, 8) if x % 2 == 1]  # comprensión de listas con condición
all([1, 2, 3]), any([0, 0, 1])  # True si todos son verdaderos; True si alguno lo es
try: ... except ZeroDivisionError: ... else: ... finally: ...  # manejo de excepciones completo
valid_date('1915.5.33')  # proyecto Sherlock: imprime 'La fecha es incorrecta.' y devuelve False
```

**Errores comunes**

- Olvidar list(...) alrededor de filter o map: se obtiene un objeto <filter object ...> o <map object ...> y no los valores.
- Usar return o sentencias dentro de una lambda: solo admite una expresión; si la lógica crece, usar def.
- Olvidar from functools import reduce (en Python 3 no es una función incorporada) o llamar a reduce con una lista vacía sin valor inicial.
- Recorrer dos veces el mismo iterador: la segunda vez está vacío porque ya se agotó.
- Llamar a next() sobre una lista (no es un iterador, hay que usar iter()) o sobre un iterador agotado (StopIteration).
- Copiar del PDF palabras reservadas traducidas ([x para x en lista si ...]) o un punto sobrante tras print(...): provocan SyntaxError.

## Unidad 20 — Ámbito de variables y clausuras (closure)

[Ir a la unidad](clase-2-ambito-clausuras-y-clases/u20-ambito-y-clausuras.ipynb)

```python
x = 10; def f(): x = 20  # x local: no cambia la x global
def f(): global x; x = 20  # modifica la variable global x
def g(): nonlocal v  # enlaza con v de la función envolvente más cercana
L -> E -> G -> B  # orden de búsqueda de nombres (regla LEGB)
l_list = [plus, minus]; l_list[0](1, 2)  # funciones guardadas en una lista y llamadas por índice
def f(g, a, b): return g(a, b)  # función de orden superior: recibe otra función
def externa(): def interna(): ...; return interna  # función anidada devuelta (clausura)
c = externa(); c(3)  # c conserva las variables de externa
return lambda x: a * x + b  # clausura escrita con lambda
c.__closure__[0].cell_contents  # lee el valor que una clausura capturó
del abs  # elimina la global que tapaba a la función incorporada
re.search('Life', txt)  # primera coincidencia (Match) o None
re.search('^Life', txt)  # coincide solo al inicio; '$' al final
re.search('AB*', txt)  # A seguida de cero o más B ('+' una o más)
re.findall('[Mm]y', txt)  # lista con todas las coincidencias
```

**Errores comunes**

- Creer que asignar dentro de una función modifica la variable global: crea una local si no se declara `global`.
- Dar a una variable el nombre de una función incorporada (`abs`, `list`, `sum`) y bloquearla; se arregla con `del`.
- Leer y asignar una variable sin `global`/`nonlocal`: `UnboundLocalError`.
- Usar `global` para una variable de la función exterior (`NameError`) o `nonlocal` para una global (`SyntaxError`).
- Devolver `mul_add()` en vez de `mul_add` al crear la clausura, o pasar `f()` en vez de `f` como argumento.
- Llamar a `re.search(...).group()` sin comprobar si el resultado es `None`.

**Errores comunes**

- Olvidar `self` como primer parámetro del método o escribir `name` en vez de `self.name`.

- Pensar que `m = n` copia el objeto: solo copia la referencia (con listas, ambos nombres modifican la misma lista).
