## 1. La función como componente básico del programa

Una **función** es un conjunto de código que realiza una operación específica dentro de un programa y se ejecuta **cuando es llamada**. Es el componente básico de un programa: el PDF la compara con construir un coche a partir de bloques de Lego; igual que se arma una forma combinando piezas, un programa grande se arma combinando funciones, módulos y clases.

Estructura básica:

```python
def nombre_funcion(x1, x2, ...):
    codigo1
    codigo2
    ...
    return n1[, n2, ...]
```

- `def` es la palabra clave que define la función.
- `nombre_funcion` debe ser un nombre significativo.
- Entre paréntesis van los **parámetros** (pueden omitirse si la función no los necesita).
- Los dos puntos `:` indican que sigue un bloque de código, igual que en `if`, `for` y `while`; el bloque debe estar **sangrado**.
- `return` devuelve el resultado cuando la función termina su trabajo; también puede omitirse.

**Referencia:** PDF parte 1, págs. 20–21

### Definir y llamar una función

Definir una función **no la ejecuta**: solo la guarda con su nombre. La llamada `print_star()` es la que ejecuta el bloque. Cada llamada ejecuta el bloque completo otra vez.

````python
# Define una función para mostrar un asterisco (25 asteriscos en una línea)
def print_star():
    print('*' * 25)

# Llama a una función para mostrar un asterisco
print_star()
````

````python
# Cada llamada ejecuta de nuevo el bloque
print_star()  # Función de salida Asterisk llamada 1
print_star()  # Función de salida Asterisk llamada 2
print_star()  # Llamada a la función de salida Asterisk 3
print_star()  # Llamada a la función de salida de Asterisk 4
````

**Ventajas de las funciones** (PDF, pág. 24):

1. La programación estructurada es posible porque un programa grande puede dividirse en varias partes.
2. Es posible reutilizar funciones en otros programas.
3. Aumenta la legibilidad del código.
4. Incluso cuando se modifica un programa, el mantenimiento es fácil porque solo hay que modificar algunas funciones.
5. Si se utiliza una función ya desarrollada, se ahorra tiempo y dinero en el desarrollo del programa.

**Clave:** definir (`def`) y llamar son dos pasos distintos; el bloque sangrado solo corre al llamar.

**Error común:** olvidar los paréntesis al llamar: `print_star` (sin `()`) solo hace referencia al objeto función y no ejecuta nada.

**Para el quiz:** los dos puntos `:` del encabezado anuncian un bloque sangrado, igual que en `if`, `for` y `while`.

## 2. Parámetros y argumentos

- **Parámetro:** variable definida en la cabecera de la función. Recibe el valor real cuando se llama a la función. En `def foo(m, n):`, `m` y `n` son parámetros.
- **Argumento:** el valor real que se pasa al llamar a la función. En `foo(3, 4)`, `3` y `4` son argumentos.

En el ejemplo del PDF, `x = get_suma(1, 10)` pasa los argumentos `1` y `10` a los parámetros `inicio` y `fin`; la función calcula la suma de los enteros de 1 a 10 y devuelve el resultado con `return`.

**Referencia:** PDF parte 1, págs. 25–27

````python
# Función que devuelve la suma de los enteros de inicio a fin (PDF, pág. 27)
def get_sum(inicio, fin):
    s = 0
    for i in range(inicio, fin + 1):
        s += i
    return s

x = get_sum(1, 10)  # 1 y 10 son argumentos; inicio y fin son parámetros
print(x)
````

**Nota sobre el PDF:** en la diapositiva de la pág. 27 el texto traducido aparece como `get_suma`, `para i en rango(...)` y `devolver s`. Son traducciones del código; en Python real las palabras clave y funciones son `for`, `in`, `range` y `return`, y el nombre debe ser el mismo en la definición y en la llamada (`get_sum`), como se muestra arriba.

### Función de salida de asterisco con parámetros

Para repetir la salida `n` veces, `print_star` recibe el parámetro `n` y hace un bucle `for` de `n` iteraciones. Con el argumento `4` se imprimen cuatro líneas.

**Referencia:** PDF parte 1, pág. 28

````python
# Programa que repite la salida del asterisco para el parámetro n veces
def print_star(n):
    for _ in range(n):
        print('*' * 25)

print_star(4)  # Dar el valor de argumento de 4 para la salida de asterisco
````

### Caso C: `TypeError` por número de argumentos

Si al llamar a la función solo se da un argumento, aparece el mensaje `TypeError` y la función no se ejecuta, porque no existe el valor del parámetro requerido. **El número de argumentos debe coincidir con el de parámetros.**

**Referencia:** PDF parte 1, pág. 31

````python
def print_sum(a, b):
    result = a + b
    print('La suma de', a, 'y', b, 'es', result)

# Los errores se capturan con try/except para que el notebook continúe
try:
    print_sum(10)  # error: falta el argumento b
except TypeError as error:
    print("TypeError:", error)
````

**Clave:** parámetro = variable de la cabecera; argumento = valor pasado en la llamada. Con parámetros posicionales, la cantidad de argumentos debe coincidir.

**Error común:** llamar `print_sum(10)` cuando la función espera dos valores: produce `TypeError: print_sum() missing 1 required positional argument: 'b'`.

## 4. Parámetro arbitrario: `*args`

A veces no se sabe de antemano cuántos argumentos recibirá una función. Esto se denomina **argumento arbitrario**. Se declara poniendo un asterisco `*` delante del parámetro; la variable dentro de la función recibe **todos los argumentos en forma de tupla**. Se puede recorrer con `for ... in`.

**Referencia:** PDF parte 1, págs. 32–33

````python
def greet(*names):
    for name in names:
        print('Hello', name, '!')

greet('A', 'B', 'C')         # 3 argumentos
greet('James', 'Thomas')     # 2 argumentos
````

Con `len` se cuenta cuántos argumentos llegaron. Observando el resultado, los argumentos `10`, `20` y `30` salen en forma de **tupla**.

**Referencia:** PDF parte 1, pág. 34

````python
def foo(*args):
    print('El número de argumentos:', len(args))
    print('Argumentos :', args)

foo(10, 20, 30)
````

### Suma de números pasados como argumentos

Si no se conoce el número de argumentos de antemano, `*numbers` recoge todos en una tupla y un bucle `for` los suma.

**Referencia:** PDF parte 1, pág. 35

````python
def sum_nums(*numbers):
    result = 0
    for n in numbers:
        result += n
    return result

print(sum_nums(10, 20, 30))          # Imprime la suma de argumentos de 10, 20, 30
print(sum_nums(10, 20, 30, 40, 50))  # Imprime la suma de argumentos de 10, 20, 30, 40, 50
````

**Nota sobre el PDF:** el texto traducido llama a la función `suma_números` y al parámetro `números`; el código de la diapositiva usa `sum_nums` y `numbers`, como arriba.

**Ampliación (no está en el PDF):** `**kwargs` recoge los argumentos de palabra clave sobrantes en un **diccionario**. Así se combinan los tres tipos de parámetros (el orden en la definición es: obligatorios, por defecto, `*args`, `**kwargs`).

````python
# Ampliación: **kwargs recoge argumentos de palabra clave en un diccionario
def mostrar_datos(**datos):
    print(datos)
    for clave, valor in datos.items():
        print(clave, '=', valor)

mostrar_datos(nombre='David', edad=23)

# Orden: obligatorio, por defecto, *args, **kwargs
def todo(a, b=2, *args, **kwargs):
    print('a =', a, '| b =', b, '| args =', args, '| kwargs =', kwargs)

todo(1)
todo(1, 5, 7, 8, color='azul')
````

**Clave:** `*args` entrega una tupla; permite funciones que aceptan una cantidad variable de argumentos.

**Error común:** olvidar que `args` es una tupla (no se puede modificar) y no una lista.

**Para el quiz:** el tipo de dato que recibe un parámetro `*args` es una tupla.

## 5. Parámetro por defecto

### Programas robustos

Si una función con parámetro obligatorio se llama sin argumento, ocurre `TypeError`. Se puede crear un programa más robusto dando al parámetro un **valor por defecto**: `def print_star(n=1):`. Si no se pasa argumento, se usa el valor por defecto y no hay error.

**Referencia:** PDF parte 1, págs. 36–37

````python
def print_star(n):  # Un argumento requerido
    for _ in range(n):
        print('*' * 25)

try:
    print_star()  # Se produce un error porque no hay argumento
except TypeError as error:
    print("TypeError:", error)
````

````python
def print_star(n=1):  # El parámetro n tiene un valor por defecto de 1
    for _ in range(n):
        print('*' * 25)

print_star()  # Ejecutar sin error aunque no haya argumento
````

Si se pasa un valor como argumento, **no se toma el valor por defecto**. Con `print_star(2)` salen dos líneas y el valor por defecto 1 se ignora.

**Referencia:** PDF parte 1, pág. 38

````python
def print_star(n=1):  # El parámetro n tiene un valor por defecto de 1
    for _ in range(n):
        print('*' * 25)

print_star(2)  # Dado que el valor del argumento es 2, no se usa el valor por defecto n=1
````

### Práctica con el parámetro por defecto

`div(a, b=2)` devuelve `a / b`. `a` debe recibir un argumento, pero `b` ya tiene el valor por defecto 2.

**Referencia:** PDF parte 1, pág. 39

````python
def div(a, b=2):
    return a / b

print('div(4) =', div(4))
print('div(6, 3) =', div(6, 3))
````

**Regla (y `SyntaxError`):** los parámetros por defecto deben asignarse a todos los parámetros, o a partir del último en el orden de aparición. Poner un valor por defecto al primero y omitirlo en el segundo provoca `SyntaxError`.

**Referencia:** PDF parte 1, pág. 40

````python
# Un SyntaxError no se puede capturar en la propia celda; se compila el texto del código para mostrarlo
codigo_incorrecto = '''
def div(a=2, b):
    return a / b
'''
try:
    compile(codigo_incorrecto, '<ejemplo>', 'exec')
except SyntaxError as error:
    print("SyntaxError:", error.msg)
````

**Nota sobre el PDF:** el mensaje en español de la diapositiva ("argumento no predeterminado sigue a argumento predeterminado") es una traducción del mensaje original `non-default argument follows default argument`. Según la versión de Python el texto exacto puede variar (en versiones recientes dice `parameter without a default follows parameter with a default`); el significado es el mismo.

### Cómo se resuelven las llamadas

Con `def div(a=1, b=2)` las llamadas con argumentos vacíos se completan automáticamente con los valores por defecto:

| Llamada | Argumentos reales (azul = valor por defecto) | Valor devuelto |
|---|---|---|
| `div()` | `div(1, 2)` | 0.5 |
| `div(4)` | `div(4, 2)` | 2.0 |
| `div(6, 3)` | `div(6, 3)` | 2.0 |

**Referencia:** PDF parte 1, pág. 41

````python
def div(a=1, b=2):
    return a / b

print('div() =', div())
print('div(4) =', div(4))
print('div(6, 3) =', div(6, 3))
````

**Ampliación (no está en el PDF):** el valor por defecto se evalúa **una sola vez**, al definir la función. Por eso no se debe usar una lista o diccionario vacío como valor por defecto.

````python
# Ampliación: error clásico con un valor por defecto mutable
def agregar(elemento, lista=[]):
    lista.append(elemento)
    return lista

print(agregar(1))
print(agregar(2))  # sorprende: la lista se conserva entre llamadas

# Forma correcta: usar None como valor por defecto
def agregar_ok(elemento, lista=None):
    if lista is None:
        lista = []
    lista.append(elemento)
    return lista

print(agregar_ok(1))
print(agregar_ok(2))
````

**Clave:** un parámetro por defecto hace opcional el argumento. Debe ir **al final** de los parámetros obligatorios.

**Error común:** `def div(a=2, b):` produce `SyntaxError`.

**Para el quiz:** con `div(a, b=2)`, la llamada `div(4)` equivale a `div(4, 2)` y devuelve `2.0`.

## 6. Parámetros de palabra clave

### Argumentos posicionales

El método de paso por defecto en Python es el de **argumentos posicionales**: el primer argumento va al primer parámetro, el segundo al segundo, etc. En el ejemplo, `get_root(1, 2, -8)` calcula las dos raíces de la ecuación cuadrática `ax^2 + bx + c = 0`.

**Referencia:** PDF parte 1, pág. 42

````python
def get_root(a, b, c):
    r1 = (-b + (b ** 2 - 4 * a * c) ** 0.5) / (2 * a)
    r2 = (-b - (b ** 2 - 4 * a * c) ** 0.5) / (2 * a)
    return r1, r2

# Al llamar a una función, utilice un valor constante como argumento
# Devuelve el valor del resultado 1 y el resultado 2
result1, result2 = get_root(1, 2, -8)
print('El valor del resultado es', result1, 'o', result2)
````

### Necesidad del parámetro de palabra clave

En el método posicional **el orden importa**: pasar `-8, 2, 1` da un resultado distinto de `1, 2, -8`. Con **argumentos de palabra clave** se indica el nombre del parámetro (`a=1`, `b=2`, `c=-8`) y el orden deja de importar.

**Referencia:** PDF parte 1, págs. 43–44

````python
result1, result2 = get_root(-8, 2, 1)  # a=-8, b=2, c=1: otra ecuación
print('El valor del resultado es', result1, 'o', result2)
````

````python
# Los tres códigos de abajo son a=1, b=2, c=-8, por lo que el resultado es el mismo
result1, result2 = get_root(1, 2, -8)
print('El valor del resultado es', result1, 'o', result2)

result1, result2 = get_root(a=1, c=-8, b=2)
print('El valor del resultado es', result1, 'o', result2)

result1, result2 = get_root(c=-8, b=2, a=1)
print('El valor del resultado es', result1, 'o', result2)
````

**Nota sobre el PDF:** en la pág. 43 el resultado aparece con coma decimal (`-0,25 o 0,5`) porque el documento está traducido; Python siempre imprime el punto decimal (`-0.25 o 0.5`).

### Cómo mezclar argumentos posicionales y de palabra clave

Al mezclar ambos tipos, los argumentos **posicionales deben ir primero**. Si un argumento posicional sigue a uno de palabra clave se produce `SyntaxError: positional argument follows keyword argument`. En `get_root(1, 2, c=-8)`, `a` y `b` se pasan por posición y `c` por palabra clave.

**Referencia:** PDF parte 1, págs. 45–46

````python
# Se produce un error cuando un argumento posicional sigue a uno de palabra clave
codigo_incorrecto = "result1, result2 = get_root(c=-8, b=2, 1)"
try:
    compile(codigo_incorrecto, '<ejemplo>', 'exec')
except SyntaxError as error:
    print("SyntaxError:", error.msg)

# Mezcla correcta: primero los posicionales, después los de palabra clave
result1, result2 = get_root(1, 2, c=-8)
print('El valor del resultado es', result1, 'o', result2)
````

**Clave:** con palabras clave el orden no importa; al mezclar, primero posicionales y después palabra clave.

**Error común:** `get_root(c=-8, b=2, 1)` provoca `SyntaxError`.

**Para el quiz:** el método por defecto es el de argumentos posicionales y en él el orden es importante.

**Ampliación (no está en el PDF):** un mismo parámetro no puede recibir el valor dos veces: `get_root(1, 2, a=3)` produce `TypeError: get_root() got multiple values for argument 'a'`.

````python
# Ampliación: el mismo parámetro recibido por posición y por palabra clave
try:
    get_root(1, 2, a=3)
except TypeError as error:
    print("TypeError:", error)
````

## 7. Devolución: la sentencia `return`

Una función puede verse como una caja negra: recibe **entradas** (parámetros), realiza una tarea específica y **devuelve** una o más salidas con `return`. En el ejemplo, `get_sum(a, b)` pone la suma en `result` y la devuelve; así `n1` vale 30 y `n2` vale 300.

**Referencia:** PDF parte 1, págs. 47–48

````python
def get_sum(a, b):      # Una función que devuelve la suma de dos números
    result = a + b
    return result       # Devuelve el resultado utilizando la sentencia return

n1 = get_sum(10, 20)
print('La suma de 10 y 20 =', n1)

n2 = get_sum(100, 200)
print('La suma de 100 y 200 =', n2)
````

También es posible devolver el resultado de una **expresión** directamente (`return a + b`), con el mismo efecto. El PDF muestra que ambas versiones dan el mismo resultado.

**Referencia:** PDF parte 1, pág. 49

````python
def get_sum(a, b):
    return a + b

result = get_sum(100, 200)
print('La suma de dos números', result)
````

### Devolver una tupla

Otra forma es devolver **dos valores en forma de tupla**: `return r1, r2`. El valor devuelto se puede recibir en dos variables a la vez (`result1, result2 = ...`), que reciben los valores de `r1` y `r2`.

**Referencia:** PDF parte 1, pág. 50

````python
def get_root(a, b, c):
    r1 = (-b + (b ** 2 - 4 * a * c) ** 0.5) / (2 * a)  # raíz 1
    r2 = (-b - (b ** 2 - 4 * a * c) ** 0.5) / (2 * a)  # raíz 2
    return r1, r2  # devuelve las dos raíces como una tupla

resultado = get_root(1, 2, -8)
print('La tupla devuelta es', resultado)

result1, result2 = get_root(1, 2, -8)
print('El valor del resultado es', result1, 'o', result2)
````

**Ampliación (no está en el PDF):** `return` termina la función de inmediato, y una función sin `return` devuelve `None`.

````python
# Ampliación: una función sin return devuelve None; el código tras un return no se ejecuta
def solo_imprime(x):
    print('Recibí', x)

valor = solo_imprime(5)
print('Valor devuelto:', valor)

def corta(x):
    return x * 2
    print('Esta línea nunca se ejecuta')

print(corta(4))
````

**Clave:** `return` entrega el resultado al lugar de la llamada; con `return a, b` el resultado es una tupla.

**Error común:** usar `print` dentro de la función creyendo que "devuelve" el valor; `print` muestra, `return` entrega y permite reutilizar el resultado.

**Para el quiz:** `return r1, r2` devuelve una tupla; se puede desempaquetar en dos variables.

### Funciones incorporadas `max`, `min`, `ord` y `chr`

## 9. Soluciones de ecuaciones cuadráticas con una función

Para `ax^2 + bx + c = 0` con `a` distinto de 0, la solución es `x = (-b ± raíz(b^2 - 4ac)) / 2a`. Una función que recibe `a`, `b`, `c` como parámetros y calcula `r1` y `r2` hace el código mucho más conciso y fácil de reutilizar: basta llamarla con diferentes coeficientes.

**Referencia:** PDF parte 1, pág. 64

````python
def print_root(a, b, c):
    r1 = (-b + (b ** 2 - 4 * a * c) ** 0.5) / (2 * a)
    r2 = (-b - (b ** 2 - 4 * a * c) ** 0.5) / (2 * a)
    print('La solución es', r1, 'o', r2)

# Resolver ecuaciones cuadráticas con diferentes valores de coeficiente
print_root(1, 2, -8)
print_root(2, -6, -8)
````

## 10. Devolución múltiple: área y perímetro de un círculo

Las sentencias `return` que devuelven dos o más valores se denominan **sentencias de devolución múltiple**: los valores separados por coma se devuelven como una **tupla**. La función `circle_area_circum(radius)` recibe un radio y devuelve el área y la circunferencia; fuera de la función, el resultado se guarda en `area` y `circum`.

**Referencia:** PDF parte 1, págs. 65–66

````python
def circle_area_circum(radius):
    area = 3.14 * radius ** 2
    circum = 2 * 3.14 * radius
    return area, circum  # Devuelve una tupla: (área, circunferencia)

radius = 10
area, circum = circle_area_circum(radius)  # Devuelve el área y el perímetro de un círculo
print('El área de un círculo de radio {} es {:.1f}, y la circunferencia del círculo es {:.1f}'.format(radius, area, circum))
````

**Nota sobre el PDF:** en la pág. 65 el código aparece traducido y roto: `3,14` con coma (en Python eso crea una tupla, no un decimal), `área`/`radio`/`circunferencia` con tildes y `zona de retorno, circum` en lugar de `return area, circum`. La pág. 66 trae la versión correcta (la usada arriba). Además, la salida impresa en la diapositiva (`314,0` y `62,8`) usa coma decimal; Python imprime `314.0` y `62.8`.

**Clave:** `return a, b` devuelve una tupla que se desempaqueta en `area, circum`.

**Error común:** escribir los decimales con coma (`3,14`): Python lo interpreta como una tupla `(3, 14)`.

**Para el quiz:** un `return` con valores separados por comas devuelve una tupla.

## 11. Variables globales y locales

### Variable global

Una **variable global** se declara **fuera de una función** y está disponible para todo el código. En el ejemplo, `a` y `b` se crean en el archivo de script y la función `print_sum` los **lee** directamente: ambas salidas muestran 10 y 20.

**Referencia:** PDF parte 1, pág. 67

````python
def print_sum():
    result = a + b  # lee las variables globales a y b
    print('print_sum() inside : la suma de', a, 'y', b, 'es', result)

a = 10
b = 20
print_sum()
result = a + b
print('print_sum() fuera : la suma de', a, 'y', b, 'es', result)
````

### ¿Qué pasa si se modifica la variable dentro de la función?

Si dentro de la función se **asigna** `a = 100` y `b = 200`, Python crea **nuevas variables locales** `a` y `b` que apuntan a 100 y 200. Las variables globales `a` y `b` siguen apuntando a 10 y 20, por eso las dos salidas difieren.

**Referencia:** PDF parte 1, págs. 68–69

````python
def print_sum():
    a = 100          # Crea nuevas variables a y b referenciando 100, 200 (locales)
    b = 200
    result = a + b
    print('print_sum() inside : la suma de', a, 'y', b, 'es', result)

a = 10               # Crea variables a y b referenciando 10 y 20 (globales)
b = 20
print_sum()
result = a + b
print('print_sum() fuera : la suma de', a, 'y', b, 'es', result)
````

### Variables locales

La sentencia de asignación `a = 100, b = 200` dentro de una función **crea variables locales**. Las variables globales `a` y `b` se refieren a los valores 10 y 20; dentro de `print_sum`, `a` y `b` son otros objetos (100 y 200) que solo existen mientras la función se ejecuta.

**Referencia:** PDF parte 1, pág. 70

### La palabra clave `global`

Para que la función **cambie** las variables globales, hay que declararlas con la palabra clave `global` dentro de la función. Así `a` y `b` dentro de la función se refieren a las variables globales y, al asignar 100 y 200, el cambio también se ve fuera de la función.

**Referencia:** PDF parte 1, págs. 71 y 73

````python
def print_sum():
    global a, b      # a y b usan a y b declarados fuera de la función
    a = 100
    b = 200
    result = a + b
    print('print_sum() inside : la suma de', a, 'y', b, 'es', result)

a = 10
b = 20
print_sum()
result = a + b
print('print_sum() fuera : la suma de', a, 'y', b, 'es', result)
````

**Cuidado con `SyntaxError`:** la declaración `global` y el operador de asignación **no pueden utilizarse en la misma línea**. `global a = 0` provoca un error de sintaxis; hay que declarar primero (`global a`) y asignar después (`a = 0`).

**Referencia:** PDF parte 1, pág. 72

````python
codigo_incorrecto = '''
def print_sum():
    global a = 0
'''
try:
    compile(codigo_incorrecto, '<ejemplo>', 'exec')
except SyntaxError as error:
    print("SyntaxError:", error.msg)
````

**Ampliación (no está en el PDF):** `global` solo es necesario para **asignar** a una variable global desde la función. Para solo leerla, como en el primer ejemplo de esta sección, no hace falta. Además, si una función asigna a un nombre en alguna parte de su cuerpo, ese nombre es local en **toda** la función; leerlo antes de asignarlo produce `UnboundLocalError`.

````python
# Ampliación: UnboundLocalError al leer un nombre que más abajo se asigna sin global
a = 10

def mal():
    print(a)   # intenta leer la local a, aún sin valor
    a = 5      # esta asignación hace que a sea local en toda la función

try:
    mal()
except UnboundLocalError as error:
    print("UnboundLocalError:", error)
````

### Variables globales frente a constantes globales

Usar variables globales es una **mala práctica** en todos los lenguajes de programación: cuando el código se hace largo, se convierten en una de las principales causas de errores. Las **constantes globales**, en cambio, no son necesariamente malas. Una constante es un valor que no cambia; suele declararse fuera de las funciones y se escribe en **mayúsculas** para distinguirla fácilmente.

**Referencia:** PDF parte 1, pág. 74

````python
GLOBAL_VALUE = 1024   # Mayúsculas: constante global, más fácil de distinguir

def foo():
    global GLOBAL_VALUE
    a = GLOBAL_VALUE * 100
    print(a)

foo()
````

**Nota sobre el PDF:** el texto de la pág. 74 dice que se asigna 1024 a `GLOBAL_VALUE` y el código calcula `GLOBAL_VALUE * 100`, por lo que la salida es `102400` (no `1124`). Aquí `global GLOBAL_VALUE` no es estrictamente necesario, porque la función solo lee la constante.

**Clave:** asignar dentro de una función crea una variable local; para modificar la global hay que usar `global`.

**Error común:** `global a = 0` en una sola línea (`SyntaxError`); y esperar que `a = 100` dentro de la función cambie la `a` global sin declararla.

**Para el quiz:** sin `global`, el cambio es local y la variable global conserva su valor.

## 13. Conjunto de productos con funciones

El **conjunto producto** (producto cartesiano) `A x A x A` de un conjunto `A` se puede obtener repitiendo una función `product_set` que, con un doble bucle `for`, combina cada elemento del primer conjunto con cada uno del segundo y guarda las parejas en un `set`. La función `exp(input_set, exponent)` repite `product_set` `exponent - 1` veces: con `exp(A, 3)` se calcula la tercera potencia.

**Referencia:** PDF parte 1, págs. 81–82

````python
def product_set(set1, set2):
    res = set()
    for i in set1:
        for j in set2:
            res = res | {(i, j)}  # Product set using double for loop
    return res

def exp(input_set, exponent):     # Una función que realiza potencias en input_set
    res = input_set               # res inicialización
    for _ in range(exponent - 1): # Se repite para que (exponente-1) se convierta en una potencia
        res = product_set(res, input_set)
    return res

A = {1, 3}
A3 = exp(A, 3)  # Realiza 3 potencias en el conjunto A
print(A3)
print('Cantidad de elementos:', len(A3))
````

### Todos los resultados posibles al lanzar un dado dos veces

`cases` es el conjunto de 1 a 6. Al multiplicarlo consigo mismo con `product_set`, cada pareja de dos números queda como un elemento del conjunto; se devuelven todos los casos **sin duplicados**.

**Referencia:** PDF parte 1, pág. 83

````python
cases = {1, 2, 3, 4, 5, 6}                 # Enteros de 1 a 6 en el dado
cases_2times = product_set(cases, cases)   # Obtener product_set para el caso
print(cases_2times)
print('Total de casos:', len(cases_2times))
````

**Nota sobre el PDF:** el orden en que se imprimen los elementos de un conjunto no está garantizado, por lo que puede diferir del que muestra la diapositiva; el contenido (6 x 6 = 36 parejas) es el mismo.

**Clave:** una función puede llamar a otra (`exp` llama a `product_set`), y los conjuntos eliminan duplicados automáticamente.

## Proyecto de la unidad

### Problema

Crear un programa de **cifrado César** que reciba una frase como entrada y la cifre desplazando cada letra mayúscula **tres posiciones** más adelante en el alfabeto. Los caracteres que no son letras mayúsculas se dejan tal cual. Para resolverlo se importa el módulo `string` y se usa `string.ascii_uppercase`; se trabaja con cortes (troceado) de cadenas.

```
src_str = ABCDEFGHIJKLMNOPQRSTUVWXYZ
dst_str = DEFGHIJKLMNOPQRSTUVWXYZABC
```

**Referencia:** PDF parte 1, págs. 14–15

### Plan de programación (pseudocódigo del PDF)

1. Inicio
2. Introduzca una frase
3. Repetir para el número de caracteres de la cadena
4. Si el carácter introducido es una letra mayúscula, entonces
5. Obtener el tercer carácter siguiente del alfabeto introducido, mediante una función (`cipher`)
6. Imprimir el carácter obtenido
7. Si no, imprimir el carácter tal como se introdujo
8. Fin

El diagrama de flujo del PDF recorre `src` con `for ch in src`, pregunta `ch in string.ascii_uppercase` y, según el caso, imprime `cipher(ch)` o `ch` con `end=''`.

**Referencia:** PDF parte 1, págs. 16–17

### Código final

El PDF presenta el programa con `input`. Para que corra sin interrupciones, el bucle se encapsula en la función `procesar` y el programa interactivo en `programa_cesar`. La versión que se ejecuta aquí procesa una lista de frases de prueba.

**Referencia:** PDF parte 1, pág. 18

````python
import string

def cipher(a):
    idx = src_str.index(a)  # posición de la letra en el alfabeto original
    return dst_str[idx]     # letra en la misma posición del alfabeto desplazado

src_str = string.ascii_uppercase
dst_str = src_str[3:] + src_str[:3]  # alfabeto desplazado tres posiciones

def procesar(src):
    """Imprime el texto cifrado carácter por carácter, como en el PDF."""
    print('Texto cifrado : ', end='')
    for ch in src:
        if ch in src_str:
            print(cipher(ch), end='')
        else:
            print(ch, end='')
    print()

def programa_cesar():
    """Versión interactiva del PDF."""
    src = input('Introduzca una frase: ')
    procesar(src)

# Frases de prueba (sustituyen a la entrada del usuario)
for frase in ['ATTACK TONIGHT', '¡ATAQUE ESTA NOCHE!', 'COME TO ROME']:
    print('Introduzca una frase:', frase)
    procesar(frase)

# Para usarlo de forma interactiva, quita el comentario de la siguiente línea
# programa_cesar()
````

### Resultado esperado y verificación

**Nota sobre el PDF:** la pág. 16 muestra la entrada `¡ATAQUE ESTA NOCHE!` con la salida `DWWDFN WRQLJKW`, que en realidad corresponde a `ATTACK TONIGHT` (la frase del servicio web de la pág. 14). Con la entrada `¡ATAQUE ESTA NOCHE!` el programa produce `¡DWDTXH HVWD QRFKH!`. Además, la `¡` antes de `Texto cifrado` en la salida es un artefacto de la traducción.

Para comprobar el programa sin imprimir, se puede devolver la cadena. Esta función auxiliar reutiliza `cipher` y permite usar `verificar`.

````python
def cifrar_frase(src):
    resultado = ''
    for ch in src:
        resultado += cipher(ch) if ch in src_str else ch
    return resultado

verificar('ATTACK TONIGHT', cifrar_frase('ATTACK TONIGHT'), 'DWWDFN WRQLJKW')
verificar('COME TO ROME', cifrar_frase('COME TO ROME'), 'FRPH WR URPH')
verificar('¡ATAQUE ESTA NOCHE!', cifrar_frase('¡ATAQUE ESTA NOCHE!'), '¡DWDTXH HVWD QRFKH!')
verificar('XYZ vuelve al inicio del alfabeto', cifrar_frase('XYZ'), 'ABC')
````

**Ampliación (no está en el PDF):** el programa del PDF solo cifra mayúsculas; las minúsculas pasan sin cambios. Una mejora es generalizar el desplazamiento y descifrar con el desplazamiento opuesto, usando parámetros por defecto (`desplazamiento=3`). Esta idea se practica en el Ejercicio 8.

````python
# Ampliación: desplazamiento como parámetro por defecto, también para descifrar
def cesar(frase, desplazamiento=3):
    origen = string.ascii_uppercase
    destino = origen[desplazamiento:] + origen[:desplazamiento]
    return ''.join(destino[origen.index(c)] if c in origen else c for c in frase)

cifrado = cesar('ATTACK TONIGHT')
print(cifrado)
print(cesar(cifrado, -3))  # descifrar: desplazamiento opuesto
````

````python
import io
from contextlib import redirect_stdout

def capturar_salida(funcion, *args):
    """Ejecuta la función y devuelve lo que imprimió como una cadena."""
    buffer = io.StringIO()
    with redirect_stdout(buffer):
        funcion(*args)
    return buffer.getvalue()

def redondear(valor, cifras=2):
    """Redondea números; deja pasar otros valores (por ejemplo None) para que verificar informe sin lanzar excepciones."""
    return round(valor, cifras) if isinstance(valor, (int, float)) else valor
````

<details><summary>Ver solución</summary>

```python
import string


def cifrar(frase, desplazamiento=3):
    origen = string.ascii_uppercase
    destino = origen[desplazamiento:] + origen[:desplazamiento]
    resultado = ''
    for ch in frase:
        if ch in origen:
            resultado += destino[origen.index(ch)]
        else:
            resultado += ch
    return resultado
```

Se crea el alfabeto desplazado con `origen[desplazamiento:] + origen[:desplazamiento]` y cada letra se reemplaza por la que está en la misma posición. Los caracteres que no están en `origen` se copian sin cambios.
</details>

## Apuntes para estudiar

### Clave

- Una función se define con `def nombre(parametros):` y un bloque sangrado; se ejecuta solo cuando se llama.
- **Parámetro** (variable de la definición) y **argumento** (valor de la llamada) no son lo mismo.
- `return` entrega el resultado; `return a, b` devuelve una tupla que se puede desempaquetar.
- `*args` recibe una cantidad variable de argumentos como tupla.
- Un parámetro con valor por defecto es opcional y debe ir después de los obligatorios.
- Con argumentos de palabra clave el orden no importa; al mezclarlos, primero los posicionales.
- Una variable creada o asignada dentro de una función es local; para modificar una global se declara `global` dentro de la función.
- Las constantes globales se escriben en mayúsculas; abusar de las variables globales es mala práctica.
- El cifrado César desplaza cada letra de forma fija; se implementa con `string.ascii_uppercase`, cortes de cadena e `index`.

### Errores comunes

- Llamar con distinto número de argumentos que de parámetros: `TypeError`.
- `def div(a=2, b):` (obligatorio después de uno por defecto): `SyntaxError`.
- `get_root(c=-8, b=2, 1)` (posicional tras palabra clave): `SyntaxError`.
- `global a = 0` en una sola línea: `SyntaxError`.
- Asignar a una variable dentro de la función esperando cambiar la global: se crea una local.
- Usar `print` en lugar de `return` cuando se necesita reutilizar el valor.
- Escribir decimales con coma (`3,14`): Python crea una tupla.

### Para el quiz

- En `def foo(m, n)`, `m` y `n` son parámetros; en `foo(3, 4)`, `3` y `4` son argumentos.
- `*args` entrega una tupla (`foo(10, 20, 30)` imprime `(10, 20, 30)`).
- `div(a, b=2)`: `div(4)` devuelve `2.0`.
- El método por defecto de paso de argumentos es el posicional; en él el orden es importante.
- Al llamar `print_sum(10)` a una función de dos parámetros ocurre `TypeError`.
- `return r1, r2` devuelve una tupla.
- Con `global a, b`, los cambios dentro de la función también se ven fuera.
- `'I like {1} and {0}'.format('Python', 'Java')` da `I like Java and Python`.
