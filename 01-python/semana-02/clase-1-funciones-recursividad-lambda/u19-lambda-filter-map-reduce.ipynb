## 1. Expresión lambda

Para devolver la suma de dos parámetros `x` e `y` se puede escribir una función con `def`, ponerle nombre (`add`) y asignar el valor de retorno con `return`. Eso resulta engorroso para una función tan corta. En Python se puede crear una **función de un solo uso** que englobe todos esos pasos: se llama **función lambda** o **expresión lambda**. Es una función **sin nombre** (anónima).

Comparación entre una función regular y una lambda:

- La función regular consta de la palabra `def`, el nombre de la función, los parámetros, los dos puntos y el cuerpo.
- La función lambda solo tiene los parámetros y la expresión que devuelve.

Sintaxis:

```
lambda parámetro1, parámetro2, ... : expresión
(lambda parámetro1, parámetro2, ...: expresión)(argumento1, argumento2, ...)     # definir y llamar
```

**Nota sobre el PDF:** en la pág. 154 los `[ ]` de `lambda [parámetro 1, parámetro 2, ...] : [expresión]` son la notación del docente para "parte variable" y no se escriben en el código. La traducción dice "insertar argumentos entre corchetes", pero en Python los parámetros van sin corchetes: `lambda x, y: x + y`.

````python
# Función regular: adición mediante la función add()
def add(x, y):
    return x + y


print('suma de 100 y 200 :', add(100, 200))
````

````python
# La misma operación con una función lambda asignada a un nombre
add = lambda x, y: x + y
print('suma de 100 y 200 :', add(100, 200))

# Ni siquiera hace falta el nombre: la lambda se define y se llama en la misma línea
print('suma de 100 y 200 :', (lambda x, y: x + y)(100, 200))
````

En la línea de la llamada directa, la función lambda ejecuta `x + y` y devuelve el valor; se pasan los argumentos 100 y 200 a los parámetros `x` e `y`.

La función lambda también se usa junto con `map`, que veremos más abajo. Aquí un adelanto del PDF (pág. 158): elevar al cuadrado cada elemento de una lista.

````python
a = [1, 2, 3, 4, 5, 6, 7]
square_a = list(map(lambda x: x**2, a))
print(square_a)
````

**Ampliación (no está en el PDF):** otras características de `lambda`.

- Una lambda contiene **una sola expresión**; no admite sentencias como `return`, `if` en bloque o asignaciones.
- Puede no tener parámetros, tener valores por defecto o usar una expresión condicional (`valor_si_cierto if condición else valor_si_falso`).
- Se usa a menudo como argumento `key` de `sorted`, `min` o `max`.

````python
saludo = lambda: "hola"                                   # sin parámetros
potencia = lambda base, exp=2: base ** exp                # parámetro con valor por defecto
paridad = lambda n: "par" if n % 2 == 0 else "impar"      # expresión condicional
print(saludo(), potencia(5), potencia(2, 10), paridad(7))

palabras = ["pera", "kiwi", "manzana", "uva"]
print(sorted(palabras, key=lambda p: len(p)))             # ordena por longitud
print(max(palabras, key=lambda p: len(p)))                # la palabra más larga

# Una lambda no admite sentencias: esto es un error de sintaxis, capturado aquí con try/except
try:
    compile("lambda x: return x", "<ejemplo>", "eval")
except SyntaxError as error:
    print("SyntaxError:", error.msg)
````

**Clave:** `lambda parámetros: expresión` crea una función anónima de una sola línea; el valor de la expresión es el valor devuelto. Es apropiada cuando la funcionalidad es simple y no hace falta reutilizarla.

**Error común:** intentar usar `return` dentro de una lambda, o escribir una lambda larga y difícil de leer. Si la lógica crece, es mejor una función con `def`.

**Para el quiz:** `(lambda x, y: x + y)(100, 200)` devuelve 300.

## 2. La función `filter`

Python ofrece funciones incorporadas para procesar colecciones. La función `filter` **recibe un objeto iterable y devuelve solo los elementos para los que la función dada devuelve `True`**. Su sintaxis es:

```
filter(función_a_aplicar, objeto_iterable)
```

**Nota sobre el PDF:** la sintaxis de la pág. 159 aparece como `filter((Tobeapplied_function, {objeto iterable})` (con un paréntesis de más y nombres traducidos). La sintaxis correcta es `filter(función, iterable)`. Además, todo el PDF llama a la función "filtro", pero el nombre real es `filter`.

**Ejemplo del PDF (págs. 160–163).** Una lista guarda las edades de muchas personas y se quiere extraer las de los adultos (19 o más). Primero con una función regular `adult_func`, que devuelve `True` si la edad `n` es adulta y `False` si no:

````python
# Devuelve True para valores iguales o superiores a 19, y False para los que no lo son
def adult_func(n):
    if n >= 19:
        return True
    else:
        return False


ages = [34, 39, 20, 18, 13, 54]
print('lista de adultos :')
for a in filter(adult_func, ages):       # filtra las edades utilizando la función filter()
    print(a, end=' ')
print()
````

`filter` pasa los elementos de `ages` uno a uno a `adult_func` y devuelve solo los que devolvieron `True`. Las edades 18 y 13 no se imprimen.

**Nota sobre el PDF:** el enunciado dice "adultos mayores de 19 años", pero el código usa `n >= 19`, de modo que 19 sí cuenta como adulto. La salida mostrada (`34 39 20 54`) es correcta para ese código.

Características de `adult_func` (pág. 162): el algoritmo es relativamente sencillo, no necesita una variable interna adicional y no hace falta reutilizarla una vez que filtra la lista de adultos. En casos así, donde la funcionalidad es simple y no se reutiliza, la función **lambda** es apropiada. Se transforma la condición en una lambda de una línea:

````python
ages = [34, 39, 20, 18, 13, 54]
print('lista de adultos :')
for a in filter(lambda x: x >= 19, ages):    # filtrar edades usando la función filter() con una lambda
    print(a, end=' ')
print()
````

La condición de filtro es `lambda x: x >= 19`: devuelve `True` si `x` es 19 o más, y solo se imprimen esos valores. El PDF (págs. 164–165) lo resume en un diagrama: la lista `[34, 39, 20, 18, 13, 54]` entra en el filtro y salen `[34, 39, 20, 54]`.

`filter` **no devuelve una lista**, sino un objeto iterable (un iterador) que entrega los elementos de a uno. Para obtener una lista hay que convertirlo con `list()`.

````python
from collections.abc import Iterator

resultado = filter(lambda x: x >= 19, ages)
print("¿Es un iterador?", isinstance(resultado, Iterator), "| tipo:", type(resultado).__name__)
print(list(resultado))          # list() consume el iterador y devuelve una lista
print(list(resultado))          # ya se consumió: ahora está vacío
````

**Clave:** `filter(función, iterable)` conserva los elementos para los que la función devuelve verdadero. Con `lambda` se evita definir una función auxiliar.

**Error común:** esperar que `filter` devuelva una lista y tratar de indexarla (`filter(...)[0]` falla). Además, el iterador solo se puede recorrer una vez.

**Para el quiz:** `list(filter(lambda x: x >= 19, [34, 39, 20, 18, 13, 54]))` devuelve `[34, 39, 20, 54]`.

## 3. La función `map`

`map` es otra función incorporada. **Aplica una función a cada elemento de un objeto iterable y devuelve un iterable con los resultados**. Sintaxis:

```
map(función_a_aplicar, objeto_iterable, ...)
```

`map` recibe dos o más argumentos: el primero es la función de mapeo y el segundo es un objeto iterable, como una lista.

**Ejemplo del PDF (págs. 166–168).** Dada la lista `a = [1, 2, 3, 4, 5, 6, 7]`, obtener `[1, 4, 9, 16, 25, 36, 49]`. Con un bucle `for` y `append`:

````python
a = [1, 2, 3, 4, 5, 6, 7]
square_a = []
for n in a:
    square_a.append(n ** 2)        # añade n cuadrado a la lista square_a
print(square_a)
````

Con `map` y una función `square` el código es más corto:

````python
def square(x):
    return x ** 2


a = [1, 2, 3, 4, 5, 6, 7]
# aplicar el valor de retorno de la función cuadrada a cada término de a
square_a = list(map(square, a))
print(square_a)
````

Combinado con `lambda` (págs. 169–170), la función auxiliar desaparece. También sirve para otras potencias:

````python
a = [1, 2, 3, 4, 5, 6, 7]
square_a = list(map(lambda x: x**2, a))
print(square_a)

cubic_a = list(map(lambda x: x**3, a))
print(cubic_a)
````

**Ampliación (no está en el PDF):** `map` admite varios iterables; la función lambda recibe un argumento de cada uno y el resultado termina cuando se acaba el más corto.

````python
precios = [10, 20, 30]
cantidades = [1, 2, 3]
totales = list(map(lambda p, c: p * c, precios, cantidades))
print(totales)

# Convertir cadenas a enteros con map (muy habitual)
texto = "10 20 30 40 50"
print(list(map(int, texto.split())))
````

**Clave:** `map(función, iterable)` transforma cada elemento; el resultado tiene tantos elementos como el original. `filter` selecciona; `map` transforma.

**Error común:** olvidar `list(...)` y tratar de imprimir el resultado: se muestra `<map object at ...>` en lugar de los valores.

**Para el quiz:** `list(map(lambda x: x**2, [1, 2, 3]))` devuelve `[1, 4, 9]`.

**Referencia:** PDF parte 1, págs. 165–170

## 4. Tipos de datos iterables y las funciones `filter` y `map`

`filter` y `map` aplican la función a cada elemento de un **tipo de dato iterable** (lista, tupla, rango, etc.) y **devuelven un objeto iterador**. Ese objeto puede convertirse en una lista con `list()`. Cada elemento del iterable se procesa con la función de asignación (`filter` o `map`) y se devuelve; después se puede volver a convertir en un tipo iterable como lista.

El PDF (pág. 173) cierra con la pregunta: ¿por qué filtrar y mapear los elementos con `filter` y `map` en lugar de hacerlo con llamadas a funciones normales? Su respuesta: hacer esa tarea con llamadas a funciones es posible, pero demasiadas llamadas tienen un defecto, porque cada llamada guarda el estado antes de la llamada y lo restaura después, y la ejecución frecuente de ese proceso es un problema que reduce la velocidad. Afirma que usar lambda en `filter` y `map` acelera la ejecución al reducir la sobrecarga de las llamadas a funciones.

**Nota sobre el PDF:** esa afirmación es discutible. Una función lambda **también es una función** y se llama igual que una `def`; su ventaja es que ahorra escribir y nombrar una función auxiliar, no que sea más rápida. Esta medición lo muestra (los tiempos dependen del equipo, pero deben ser parecidos):

````python
from timeit import timeit

datos = list(range(1000))


def cuadrado(x):
    return x ** 2


t_def = timeit(lambda: list(map(cuadrado, datos)), number=2000)
t_lambda = timeit(lambda: list(map(lambda x: x ** 2, datos)), number=2000)
print(f"map con def    : {t_def:.4f} s")
print(f"map con lambda : {t_lambda:.4f} s")
````

**Clave:** `filter` y `map` devuelven iteradores (objetos que entregan un elemento cada vez); con `list()` se materializan en una lista.

**Referencia:** PDF parte 1, págs. 171–173

## 5. La función `reduce`

`reduce` está incluida en un módulo llamado `functools`, por lo que hay que importarla. **Devuelve un único valor ejecutando una operación con una función dada sobre los elementos de un objeto iterable.**

Funcionamiento con una secuencia `seq = [s1, s2, s3, s4]` y una función `func`:

1. Los dos primeros elementos, `s1` y `s2`, se pasan a `func(s1, s2)` y se obtiene un valor.
2. Ese valor se pasa junto con el siguiente elemento: `func(func(s1, s2), s3)`.
3. El proceso se repite hasta el último elemento: `func(func(func(s1, s2), s3), s4)`.
4. Al final devuelve **un único valor**.

**Ejemplo del PDF (pág. 176):** sumar los elementos de la lista `[1, 2, 3, 4]` con una lambda. El cálculo es `(((1 + 2) + 3) + 4)`, que devuelve 10.

````python
from functools import reduce

a = [1, 2, 3, 4]
n = reduce(lambda x, y: x + y, a)
print(n)
````

El PDF explica el recorrido: en primer lugar se devuelve 3 de `lambda x, y: x + y` cuando entran los dos primeros valores, 1 y 2. Ese 3 entra junto con el tercer elemento y da 6; con el último elemento, 4, la función devuelve 10.

**Ampliación (no está en el PDF):** una versión con trazas para ver cada paso, y el valor inicial opcional.

````python
def sumar_con_traza(x, y):
    print(f"func({x}, {y}) = {x + y}")
    return x + y


print("resultado:", reduce(sumar_con_traza, [1, 2, 3, 4]))

# El tercer argumento es un valor inicial: sirve también con listas vacías
print(reduce(lambda x, y: x + y, [], 0))
print(reduce(lambda x, y: x * y, [1, 2, 3, 4], 1))          # producto
print(reduce(lambda x, y: x if x > y else y, [7, 3, 9, 2])) # máximo
````

**Clave:** `reduce(función, iterable)` combina todos los elementos en uno solo, de izquierda a derecha. Se importa con `from functools import reduce`.

**Error común:** usar `reduce` sin importarla (en Python 3 ya no es una función incorporada), o llamarla con una lista vacía sin valor inicial (lanza `TypeError`).

**Para el quiz:** `reduce(lambda x, y: x + y, [1, 2, 3, 4])` devuelve 10.

## 6. Gestión de excepciones (Codifiquemos)

La parte de práctica del PDF incluye un apartado sobre excepciones. Un programador debe codificar en previsión de excepciones: un código puede terminar de forma anormal por una entrada incorrecta. Para evitarlo se encierra el código con la sentencia `try ~ except`.

El programa del PDF (pág. 183) pide dos números, los divide y, si algo falla, imprime un aviso. Aquí se encapsula en una función que recibe el texto que escribiría el usuario:

````python
def dividir_texto(entrada):
    """Versión no interactiva del programa de la pág. 183: recibe el texto que escribiría el usuario."""
    try:
        a, b = entrada.split()
        result = int(a) / int(b)
        return '{}/{} = {}'.format(a, b, result)
    except:
        return 'Comprueba si los números son correctos.'


print(dividir_texto('10 2'))        # entrada correcta: se comporta como está previsto
print(dividir_texto('100 dos'))     # int('dos') falla: la sentencia except procesa el error
print(dividir_texto('100 0'))       # 100/0 falla: la sentencia except se encarga de la excepción
````

````python
# Para usarlo de forma interactiva, quita el comentario de la siguiente línea
# print(dividir_texto(input('Introduzca dos números : ')))
````

**Nota sobre el PDF:** en las salidas de la pág. 183, el tercer ejemplo aparece escrito como `100 dos`, pero el texto explica el caso "si el usuario introduce 100 y 0". La entrada que provoca el `ZeroDivisionError` es `100 0`, como se muestra arriba. Además, el PDF usa `except :` sin tipo, que captura **cualquier** error. Es más seguro nombrar los errores esperados.

**Ampliación (no está en el PDF):** la misma función capturando solo los errores previstos.

````python
def dividir_texto_seguro(entrada):
    try:
        a, b = entrada.split()
        return '{}/{} = {}'.format(a, b, int(a) / int(b))
    except ValueError:                 # int('dos') o una cantidad de valores distinta de dos
        return 'Escribe dos números enteros separados por un espacio.'
    except ZeroDivisionError:
        return 'No se puede dividir entre cero.'


for entrada in ('10 2', '100 dos', '100 0', '7'):
    print(repr(entrada), '->', dividir_texto_seguro(entrada))
````

### `try - except - else - finally` (págs. 186–187, "Un paso más")

Un manejo más completo usa los cuatro bloques:

| En el PDF | Palabra real | Qué hace |
|---|---|---|
| Inténtalo | `try` | Encierra el código que podría fallar. Si se produce una excepción, salta al `except` correspondiente. |
| excepto | `except` | Bloque que actúa en caso de excepción; describe cómo manejarla. |
| si no | `else` | Se ejecuta cuando **no** hay ninguna excepción. |
| finalmente | `finally` | Se ejecuta **siempre**, haya o no excepción. |

**Nota sobre el PDF:** la tabla de la pág. 187 traduce las palabras reservadas (`try` aparece como "Inténtalo", `except` como "excepto", `else` como "si no" y `finally` como "finalmente"). En el código hay que escribir las palabras en inglés.

````python
def divide(x, y):
    try:
        result = x / y
    except ZeroDivisionError:
        print('error por división por cero')
    else:
        print('resultado :', result)
    finally:
        print('ejecución finalizada')


print('llamada a la función divide(100, 2) :')
divide(100, 2)
print('llamada a la función divide(100, 0) :')
divide(100, 0)
````

**Clave:** `try` contiene el código de riesgo; `except` lo atiende; `else` corre solo si no hubo error; `finally` corre siempre (por ejemplo, para liberar recursos).

**Error común:** usar un `except:` sin tipo que oculta errores inesperados (incluidos errores de programación). Conviene nombrar el tipo (`ValueError`, `ZeroDivisionError`, ...).

**Para el quiz:** el bloque `finally` se ejecuta tanto si hubo excepción como si no.

## 7. Objeto iterador (Codifiquemos)

El PDF estudia ahora el **iterador**, una de las funciones de alto nivel de Python.

Un **objeto iterador** es un objeto que, de forma consecutiva, entrega los datos de una estructura de datos con uno o más elementos. Detrás de una sentencia `for` ocurre lo siguiente: el **objeto iterable** (lista, tupla, rango) se convierte en un iterador con la función `iter()`, y la función `next()` pide el siguiente elemento. Cuando ya no quedan elementos, el iterador lanza la excepción `StopIteration` y el `for` termina.

La lista `[10, 20, 30]` se recorre repetidamente en iteraciones como `for i in [10, 20, 30]:`. ¿Cómo sabe `for` que llegó al final? Porque `next()` lanza `StopIteration`, que notifica que no hay más elementos que devolver.

**Nota sobre el PDF:** la pág. 189 dice que el objeto `StopIteration` "se devuelve al final de [10, 20, 30, StopIteration]", como si fuera un elemento más de la lista. No lo es: es una **excepción** que lanza `next()` cuando el iterador se agota. La lista sigue teniendo solo tres elementos.

### `iter()` y `next()`

El tipo lista no es un iterador: no se le puede aplicar `next()` directamente. Hay que convertirla con `iter()` y luego pedir los elementos con `next()` uno a uno (págs. 191–194):

````python
a = [10, 20, 30]
try:
    next(a)                                  # una lista no es un iterador
except TypeError as error:
    print("TypeError:", error)

iter_a = iter([10, 20, 30])                  # iter() convierte la lista en un iterador
print(next(iter_a))
print(next(iter_a))                          # devuelve la siguiente vez del iterador
````

````python
lst = [10, 20, 30]
l_iter = iter(lst)                           # convierte un objeto de tipo lista en un iterador
print(type(l_iter).__name__)                 # list_iterator
print(next(l_iter))
print(next(l_iter))
print(next(l_iter))
````

Si se llama a `next()` sobre un iterador que ya devolvió todos sus elementos, se genera la excepción `StopIteration` (pág. 195). Y si se pasa un tipo que no es iterable, como un entero, a `iter()`, se produce un `TypeError` (pág. 193):

````python
try:
    next(l_iter)                             # el iterador ya está agotado
except StopIteration:
    print("StopIteration: no hay más elementos que devolver")

try:
    n_iter = iter(100)                       # un int no es iterable
except TypeError as error:
    print("TypeError:", error)
````

**Ampliación (no está en el PDF):** un bucle `for` equivale a un `while` que llama a `iter()` una vez y a `next()` repetidamente hasta que aparece `StopIteration`:

````python
def recorrer(iterable):
    iterador = iter(iterable)
    while True:
        try:
            elemento = next(iterador)
        except StopIteration:
            break
        print("elemento:", elemento)


recorrer([10, 20, 30])
````

### El método especial `__next__` (págs. 196–197)

Un método de una clase que realiza tareas específicas se denomina **método especial** o método mágico. Se escribe con doble barra baja delante y detrás (por ejemplo, `__next__()`) y sirve para redefinir un operador o una función ya definida en Python. La función `next(l_iter)` equivale a llamar al método `l_iter.__next__()`. De la misma forma, el operador suma `10 + 20` es el resultado del método especial `__add__` sobre el entero 10: `(10).__add__(20)`. Los métodos especiales se tratan en la Unidad 21 (clases).

````python
lst = [10, 20, 30]
l_iter = iter(lst)
print(l_iter.__next__())                     # llama al método especial __next__
print(l_iter.__next__())
print(l_iter.__next__())

print(10 + 20)
print((10).__add__(20))                      # lo mismo: el operador + usa el método especial __add__
````

### Funciones incorporadas para objetos iterables: `all` y `any` (págs. 190, 202–203)

Muchas funciones incorporadas se pueden aplicar a objetos iterables: `min` y `max` reciben un iterable y devuelven el mínimo y el máximo; también existen `all`, `any`, `ascii`, `bool`, `filter`, `iter`, entre otras.

- `all(iterable)` devuelve `True` **solo cuando todos** los elementos son verdaderos. Si hay un único valor `0`, `''` o `None`, devuelve `False`.
- `any(iterable)` devuelve `True` si **al menos un** elemento es verdadero.

````python
l1 = [1, 2, 3, 4]       # todos los elementos no son 0
l2 = [0, 2, 4, 8]       # un elemento es 0
l3 = [0, 0, 0, 0]       # todos los elementos son 0

print("all:", all(l1), all(l2), all(l3))     # True False False
print("any:", any(l1), any(l2), any(l3))     # True True False
````

**Clave:** iterable = se puede recorrer (lista, tupla, cadena, rango); iterador = entrega un elemento por vez con `next()` y lanza `StopIteration` al agotarse. `iter()` convierte un iterable en iterador.

**Error común:** llamar a `next()` directamente sobre una lista (`TypeError: 'list' object is not an iterator`) o seguir llamándolo cuando el iterador ya se agotó (`StopIteration`).

**Para el quiz:** `filter` y `map` devuelven iteradores; `next()` sobre un iterador agotado lanza `StopIteration`.

## 8. Comprensión de listas frente a la expresión lambda (Codifiquemos)

La **comprensión de listas** es otra forma de crear una lista a partir de un iterable. Su sintaxis es:

```
[ expresión for variable in iterador/secuencia if expresión_condicional ]
```

La parte `if expresión_condicional` es opcional:

```
[ expresión for variable in iterador/secuencia ]
```

**Nota sobre el PDF:** en las págs. 204 y 206 la traducción automática tradujo también las palabras reservadas dentro del código y el código no se ejecuta: `[x para x en edades si x >= 19]`. En Python las palabras son `for`, `in` e `if`. En la pág. 206 los nombres de variables cambian de una celda a otra (`ages`/`edades`, `adult_ages`/`edad_adultos`); aquí se usan `ages` y `adult_ages`.

Valores cuadrados con tres técnicas equivalentes (págs. 205–206):

````python
a = [1, 2, 3, 4, 5, 6, 7]                          # lista de valores consecutivos
a1 = list(map(lambda x: x**2, a))                  # map + lambda
a2 = [x**2 for x in a]                             # comprensión de listas
a3 = [x**2 for x in range(1, 8)]                   # comprensión con range
print(a1)
print(a2)
print(a3)
print("las tres son iguales:", a1 == a2 == a3)
````

Filtrado con `filter` + `lambda` frente a comprensión de listas:

````python
ages = [34, 39, 20, 18, 13, 54]
adult_ages = list(filter(lambda x: x >= 19, ages))
print('lista de adultos :', adult_ages)
print('lista de adultos :', [x for x in ages if x >= 19])
````

El PDF concluye que ambos códigos utilizan filtros condicionales similares y expresiones sencillas: devuelven los mismos resultados.

## Proyecto de la unidad: Sherlock Holmes (validación de fechas)

**Problema (págs. 145–148).** Recibir del usuario una fecha con la forma `año.mes.día` (por ejemplo `1887.11.20`) y verificar si es la fecha de publicación de un libro de la serie Sherlock Holmes. Requisitos:

- Construir una función que imprima "Este no es un año de publicación de la serie Sherlock Holmes" si el año no es uno de los años de publicación de la serie.
- Construir una función lambda llamada `is_sherlock_book_yr` que verifique si el año de entrada es un año de publicación.
- Comprobar que el mes y el día introducidos son correctos. Por ejemplo, `1915.5.3` es una entrada correcta, pero `1915.5.33` no lo es y el programa debe imprimir "La fecha es incorrecta".

**Plan de programación (pág. 150, función `valid_date`).**

1. Inicio.
2. Almacenar en una lista los años de publicación de la serie.
3. Recibir la fecha como entrada del usuario.
4. Si la fecha introducida no es un año de publicación, entonces: devolver `False` con un `print` (la función termina).
5. Si el mes es incorrecto (menor que 1 o mayor que 12), devolver `False` con un `print`.
6. Si el día es incorrecto (menor que 1 o mayor que los días de ese mes), devolver `False` con un `print`.
7. Si no se cumple ninguna condición de error, devolver `True`.
8. Fin.

**Nota sobre el PDF:** el código de la pág. 149 tiene un detalle que impide ejecutarlo tal cual. En la línea 11 aparece `print('Este no es un año de publicación de la serie Sherlock Holmes').` con un **punto final** después del paréntesis, que produce `SyntaxError`. Se corrige quitando ese punto. Además, el código pide la fecha con `input()` y se escribe fuera de una función; aquí se encapsula en `programa_sherlock(date)` para poder probarlo.

El código corregido, con comentarios:

````python
def valid_date(date):
    days = [31, 29, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31]          # días máximos de cada mes
    sherlock_series = [1887, 1890, 1892, 1894, 1901, 1905, 1915, 1917, 1927]   # años de publicación
    splitted = date.split('.')                                       # separa año, mes y día
    year = int(splitted[0])
    month = int(splitted[1])
    day = int(splitted[2])
    is_sherlock_book_yr = lambda x: x in sherlock_series             # lambda: ¿es un año de la serie?

    if not is_sherlock_book_yr(year):
        print('Este no es un año de publicación de la serie Sherlock Holmes')
        return False
    if month < 1 or month > 12:
        print('El mes es incorrecto.')
        return False
    if day < 1 or day > days[month - 1]:
        print('La fecha es incorrecta.')
        return False

    return True


def programa_sherlock(date):
    """Versión no interactiva del programa de la pág. 149: recibe la fecha como texto."""
    if valid_date(date) == True:
        print('¡Esta es la fecha de publicación de una serie de Sherlock Holmes!')
    else:
        print('Esta no es la fecha de publicación de una serie de Sherlock Holmes o la entrada tiene un error.')


# Un caso de la pág. 149 (1927.9.10) y otros de prueba
for fecha in ('1927.9.10', '1887.11.20', '1915.5.3', '1915.5.33', '1915.13.1', '1900.5.5'):
    print('Fecha:', fecha)
    programa_sherlock(fecha)
    print()
````

**Resultado esperado.** Con `1927.9.10` el programa imprime "¡Esta es la fecha de publicación de una serie de Sherlock Holmes!" (salida de la pág. 149). `1900.5.5` falla en el año, `1915.13.1` en el mes y `1915.5.33` en el día.

**Limitaciones del programa** (no son errores del PDF, pero conviene saberlas):

- La lista `days` usa 29 para febrero siempre; por eso `1915.2.29` se acepta aunque 1915 no fue bisiesto.
- Si el texto no tiene tres partes numéricas, `int()` o el índice fallan con `ValueError` o `IndexError`.

**Ampliación (no está en el PDF):** una versión que usa el manejo de excepciones de la sección 6 y la comprensión de listas para validar esas entradas mal formadas.

````python
def valid_date_segura(date):
    """Como valid_date, pero devuelve False en lugar de fallar con entradas mal formadas."""
    try:
        year, month, day = [int(parte) for parte in date.split('.')]
    except ValueError:
        print('El formato debe ser año.mes.día con números enteros.')
        return False
    return valid_date(f'{year}.{month}.{day}')


print(valid_date_segura('1917.6.10'))
print(valid_date_segura('1917-6-10'))
print(valid_date_segura('1917.6'))
print(valid_date_segura('1915.2.29'))     # se acepta: limitación de la lista days
````

**Clave:** `is_sherlock_book_yr = lambda x: x in sherlock_series` es una lambda de una sola expresión que devuelve `True` o `False`; las tres validaciones (año, mes, día) se aplican en orden y la función termina en la primera que falla.

**Error común:** dejar el punto final tras un `print(...)` (error de sintaxis) o comparar el año como texto: `'1917' in sherlock_series` es `False` porque la lista contiene enteros.

**Para el quiz:** `valid_date('1915.5.33')` imprime "La fecha es incorrecta." y devuelve `False`.

**Clave**

- `lambda parámetros: expresión` crea una función anónima de una sola expresión; sirve cuando la funcionalidad es simple y no se reutiliza. `(lambda x, y: x + y)(100, 200)` devuelve 300.
- `filter(función, iterable)` conserva los elementos para los que la función devuelve `True`.
- `map(función, iterable)` aplica la función a cada elemento y devuelve los resultados.
- `reduce(función, iterable)` combina todos los elementos en un único valor (`from functools import reduce`); con `[1, 2, 3, 4]` y suma da 10.
- `filter` y `map` devuelven iteradores; se convierten a lista con `list()`.
- Un iterable se convierte en iterador con `iter()`; `next()` entrega un elemento por vez y lanza `StopIteration` cuando se agota.
- `try - except - else - finally`: `else` solo si no hubo excepción, `finally` siempre.
- Comprensión de listas: `[expresión for x in iterable if condición]`, equivalente a `map` y `filter` con lambda.
- `all()` exige todos los elementos verdaderos; `any()` al menos uno.

**Errores comunes**

- Intentar usar `return` o sentencias dentro de una lambda.
- Olvidar `list(...)` y obtener un `<filter object ...>` o `<map object ...>` en lugar de los valores.
- Recorrer dos veces un iterador (la segunda vez está vacío).
- Olvidar `from functools import reduce`, o usar `reduce` con una lista vacía sin valor inicial.
- `next()` sobre una lista (no es un iterador) o sobre un iterador ya agotado.
- Escribir `except:` sin tipo, que oculta errores inesperados.
- Fiarse de los nombres traducidos del PDF (`filtro`, `mapa`) y de las palabras reservadas traducidas (`para`, `en`, `si`).

**Para el quiz**

- `list(filter(lambda x: x >= 19, [34, 39, 20, 18, 13, 54]))` devuelve `[34, 39, 20, 54]`.
- `list(map(lambda x: x**2, [1, 2, 3, 4, 5, 6, 7]))` devuelve `[1, 4, 9, 16, 25, 36, 49]`.
- `reduce(lambda x, y: x + y, [1, 2, 3, 4])` evalúa `(((1 + 2) + 3) + 4)` y devuelve 10.
- `next(iter([10, 20, 30]))` devuelve 10; al terminar los elementos se lanza `StopIteration`.
- `all([0, 2, 4, 8])` es `False` y `any([0, 2, 4, 8])` es `True`.
- En el proyecto, `valid_date` valida primero el año (con la lambda `is_sherlock_book_yr`), luego el mes (1 a 12) y luego el día.

## Autoevaluación

**1. ¿Qué es una expresión lambda y en qué se diferencia de una función definida con `def`?**

<details><summary>Ver respuesta</summary>

Es una función anónima (sin nombre) de una sola expresión, escrita como `lambda parámetros: expresión`. Una función `def` tiene nombre, puede tener varias sentencias y usa `return`; la lambda devuelve directamente el valor de su expresión y se usa para tareas simples de un solo uso.
</details>

**2. ¿Qué devuelve `list(filter(lambda x: x % 2 == 0, [1, 2, 3, 4, 5, 6]))` y qué papel cumple la lambda?**

<details><summary>Ver respuesta</summary>

Devuelve `[2, 4, 6]`. La lambda es la condición de filtro: devuelve `True` para los pares; `filter` conserva solo esos elementos y `list()` convierte el iterador en una lista.
</details>

**3. ¿En qué se diferencian `map` y `filter`?**

<details><summary>Ver respuesta</summary>

`filter` selecciona los elementos que cumplen una condición (el resultado puede tener menos elementos). `map` transforma cada elemento aplicándole una función (el resultado tiene tantos elementos como el original).
</details>

**4. Explica paso a paso `reduce(lambda x, y: x + y, [1, 2, 3, 4])` y de qué módulo se importa.**

<details><summary>Ver respuesta</summary>

Se importa con `from functools import reduce`. Primero suma 1 y 2 (3); luego 3 y 3 (6); luego 6 y 4 (10). Devuelve un único valor, 10, que equivale a `(((1 + 2) + 3) + 4)`.
</details>

**5. ¿Qué diferencia hay entre un iterable y un iterador, y qué ocurre al llamar a `next()` cuando el iterador se agotó?**

<details><summary>Ver respuesta</summary>

Un iterable (lista, tupla, rango) se puede recorrer; un iterador entrega un elemento por vez con `next()`. `iter()` convierte un iterable en iterador. Cuando ya no hay elementos, `next()` lanza la excepción `StopIteration` (así sabe el `for` que debe terminar).
</details>
