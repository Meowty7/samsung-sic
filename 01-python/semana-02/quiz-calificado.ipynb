## Q 17-01

### Consigna

Reciba las coordenadas (x1, y1), (x2, y2) de dos puntos del usuario e imprima la distancia entre los dos puntos. Para hacer esto, implemente la función `distancia(x1, y1, x2, y2)`.

Pauta: consulte la fórmula para encontrar la distancia entre dos puntos.

**Dónde se ve:** [Unidad 17](clase-1-funciones-recursividad-lambda/u17-funciones.ipynb)

### Respuesta

````
import math


def distancia(x1, y1, x2, y2):
    """Distancia euclidiana entre los puntos (x1, y1) y (x2, y2)."""
    return math.sqrt((x2 - x1) ** 2 + (y2 - y1) ** 2)


x1 = float(input("x1: "))
y1 = float(input("y1: "))
x2 = float(input("x2: "))
y2 = float(input("y2: "))

print("La distancia entre ({}, {}) y ({}, {}) es {}".format(x1, y1, x2, y2, distancia(x1, y1, x2, y2)))
````

**Explicación:** Fórmula: d = √((x2 − x1)² + (y2 − y1)²). Con los puntos (0, 0) y (3, 4) la distancia es 5.0. `float()` convierte el texto de `input()` en número y acepta decimales.

**Nota sobre el enunciado:** dice "de la siguiente manera" pero no muestra el formato; se imprime una frase con los dos puntos y la distancia.

Usa `input()`: la salida de arriba es una prueba respondiendo `0`, `0`, `3`, `4`. En la plataforma se abre una ventana para cada dato.

## Q 18-01

### Consigna

Un palíndromo es una oración, palabra o cadena que se lee igual en ambos sentidos (level, kayak, salas, somos, radar, reconocer). Defina una función recursiva `is_palindrome` y escriba un programa que reciba una cadena del usuario e imprima si es palíndromo o no.

Pauta: llame a la función `is_palindrome` dentro de la función `is_palindrome`.

**Dónde se ve:** [Unidad 18](clase-1-funciones-recursividad-lambda/u18-recursividad.ipynb)

### Respuesta

````
def is_palindrome(texto):
    """True si texto se lee igual al derecho y al revés (recursiva)."""
    if len(texto) <= 1:                     # caso base: 0 o 1 carácter siempre es palíndromo
        return True
    if texto[0] != texto[-1]:               # los extremos no coinciden: no es palíndromo
        return False
    return is_palindrome(texto[1:-1])       # llamada recursiva sin los extremos


cadena = input("Ingrese una cadena: ")
limpia = cadena.lower().replace(" ", "")      # ignora mayúsculas y espacios

if is_palindrome(limpia):
    print(cadena, "es un palíndromo.")
else:
    print(cadena, "no es un palíndromo.")

# Pruebas rápidas con los ejemplos del enunciado
for palabra in ["level", "kayak", "salas", "somos", "radar", "reconocer", "python"]:
    print(palabra, "->", is_palindrome(palabra))
````

**Explicación:** Cada llamada compara el primer y el último carácter y, si coinciden, se llama a sí misma con la cadena sin ellos (`texto[1:-1]`). Termina cuando quedan 0 o 1 caracteres (caso base) o cuando encuentra una diferencia. La entrada se pasa a minúsculas y sin espacios para que "Reconocer" también cuente.

Usa `input()`: la salida de arriba es una prueba respondiendo `Reconocer`. En la plataforma se abre una ventana para cada dato.

### Consigna

En el siguiente código, ¿qué valor se incluye finalmente en `num_list`? Calcule el resultado esperado a mano (el `for` se ejecuta 5 veces).

```python
def calc_digit(n):
    def findal(digit):
        return digit**n
    return final

num_list = []
for num in range(1,6):
    num_list.append( calc_digit(num) )

```

**Dónde se ve:** [Unidad 20](clase-2-ambito-clausuras-y-clases/u20-ambito-y-clausuras.ipynb)

### Respuesta

````
def calc_digit(n):
    def final(digit):          # función interna: recuerda n (clausura)
        return digit ** n
    return final


num_list = []
for num in range(1, 6):
    num_list.append(calc_digit(num))
    print(num_list[num - 1](num))

print(len(num_list), "funciones en num_list")   # 5 clausuras, una por cada n de 1 a 5\nprint(num_list[2](2))   # la tercera recuerda n = 3: 2 ** 3 = 8
````

**Explicación:** **Resultado esperado (a mano):** en cada vuelta se guarda la función que eleva a la potencia `num` y se llama con ese mismo `num`, así que se imprime num elevado a num:

| Vuelta | Cálculo | Imprime |
|---|---|---|
| 1 | 1¹ | 1 |
| 2 | 2² | 4 |
| 3 | 3³ | 27 |
| 4 | 4⁴ | 256 |
| 5 | 5⁵ | 3125 |

Al final, `num_list` contiene **5 funciones** (clausuras), no números: cada una recuerda su propio `n` (1, 2, 3, 4 y 5). Por ejemplo, `num_list[2](2)` devolvería 2³ = 8.

**Nota sobre el enunciado:** la función interna se llama `findal` pero se devuelve `final`; así daría `NameError`. Se corrige usando el mismo nombre en los dos lugares.

## Q 21-01

### Consigna

Implemente la multiplicación (`*`) y la división (`/`) de dos vectores usando los métodos `__mul__` y `__truediv__` de la unidad 21. Con v1 = (30, 40) y v2 = (10, 20), escriba la salida de la multiplicación y la división de los dos vectores.

Pauta: amplíe la funcionalidad de la clase `Vector2D` aprendida en este capítulo.

**Dónde se ve:** [Unidad 21](clase-2-ambito-clausuras-y-clases/u21-clases-y-objetos.ipynb)

### Respuesta

````
class Vector2D:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):        # v1 + v2
        return Vector2D(self.x + other.x, self.y + other.y)

    def __sub__(self, other):        # v1 - v2
        return Vector2D(self.x - other.x, self.y - other.y)

    def __mul__(self, other):        # v1 * v2: multiplica componente a componente
        return Vector2D(self.x * other.x, self.y * other.y)

    def __truediv__(self, other):    # v1 / v2: divide componente a componente
        return Vector2D(self.x / other.x, self.y / other.y)

    def __str__(self):
        return "({}, {})".format(self.x, self.y)


v1 = Vector2D(30, 40)
v2 = Vector2D(10, 20)

print('v1 + v2 =', v1 + v2)
print('v1 - v2 =', v1 - v2)
print('v1 * v2 =', v1 * v2)
print('v1 / v2 =', v1 / v2)
````

**Explicación:** Se parte de la `Vector2D` de la unidad 21 (`__init__`, `__add__`, `__sub__`, `__str__`) y se agregan `__mul__` y `__truediv__`. Python llama a `v1.__mul__(v2)` al escribir `v1 * v2` y a `v1.__truediv__(v2)` con `v1 / v2`. Resultado: `v1 * v2 = (300, 800)` y `v1 / v2 = (3.0, 2.0)` (la división con `/` devuelve float).

**Nota sobre el enunciado:** dice "el siguiente resultado" pero no lo muestra; se usa la operación componente a componente, que es la extensión natural de `__add__` y `__sub__` de la unidad.
