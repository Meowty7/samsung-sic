## 1. Objetos, atributos y métodos

## 2. Programación procedimental frente a programación orientada a objetos

El PDF compara los dos paradigmas.

**Programación procedimental (por procedimientos):**
- Resuelve el problema llamando a funciones y modelando los datos en el orden de resolución del problema.
- Es la adoptada por lenguajes tradicionales como C, Fortran, Basic, etc.
- Exige muchas "flechas" (llamadas a funciones y datos que viajan entre ellas) cuando los datos y funciones aumentan, y es difícil de gestionar en proyectos grandes.
- Es incapaz de tratar bien sistemas con muchos componentes (por ejemplo, una interfaz gráfica de usuario).

**Programación orientada a objetos (POO):**
- Es una técnica que modela un programa de forma parecida al mundo real: una tarea se expresa como **interacciones entre objetos**.
- Los objetos se generan a partir de clases; las clases se construyen con acciones (métodos) y atributos.
- Reduce los costes de mantenimiento al desarrollar o actualizar el software; hoy la mayoría de programadores la prefiere.
- La adoptan Java, Python, C++, C#, Swift, etc.

Ejemplos de objetos del PDF: en un juego (objeto automóvil, objeto jugador, objeto terreno, objeto de suelo, objeto de construcción, que interactúan entre sí) y en una interfaz gráfica (etiqueta, campo de texto, botón de radio, casilla de verificación, pestaña, botón).

**Limitaciones del diagrama de flujo:** cuando un programa se vuelve demasiado complejo, expresar su flujo con un diagrama de flujo resulta inconveniente. Para ese caso se puede usar un **diagrama de flujo de datos** (el PDF muestra uno del sistema bancario con procesos como abrir cuenta, comprobar transacción, depositar y retirar efectivo).

**Referencia:** PDF parte 2, págs. 92–95 y 97

**Ampliación (no está en el PDF):** el mismo problema (una cuenta con saldo) resuelto de las dos maneras. En el enfoque procedimental los datos viajan por separado de las funciones; en el orientado a objetos datos y funciones quedan juntos en el objeto.

````python
# Enfoque procedimental: los datos (un diccionario) y las funciones están separados
def depositar_proc(cuenta, importe):
    cuenta["saldo"] += importe

cuenta_p = {"titular": "David", "saldo": 0}
depositar_proc(cuenta_p, 2000)
print("Procedimental:", cuenta_p["saldo"])


# Enfoque orientado a objetos: datos y comportamiento viven juntos
class CuentaSimple:
    def __init__(self, titular):
        self.titular = titular
        self.saldo = 0

    def depositar(self, importe):
        self.saldo += importe


cuenta_o = CuentaSimple("David")
cuenta_o.depositar(2000)
print("Orientado a objetos:", cuenta_o.saldo)
````

## 3. Sintaxis para llamar objetos y métodos: el operador punto

Se usa el operador `.` (punto) para llamar a los métodos de un objeto. En `'tiger'.upper()` el texto `'tiger'` es un objeto cuyo tipo de datos es `str` y `upper()` es un método de la clase `str` que lo convierte en `'TIGER'`.

El PDF resume una idea fuerte: **todo en Python es un objeto**. Incluso los números que participan en operaciones aritméticas: `n = 100 + 200` equivale a `n = (100).__add__(200)`; es decir, el objeto entero `100` suma el objeto `200` usando su método `__add__`. Desde este punto de vista, llamar a `animals.append('conejo')` y ejecutar una suma son, en esencia, la misma cosa: enviar un mensaje a un objeto.

````python
print('tiger'.upper())

n = 100 + 200
m = (100).__add__(200)
print(n, m)
verificar("100 + 200 equivale a (100).__add__(200)", n, m)

# Continuación de la lista de animals de la sección 1
animals = ['rabbit', 'tiger', 'lion', 'dog', 'cat']
s = animals.pop()
print(s)
print(s.upper())
print(s.find('a'))
````

**Clave:** el punto invoca un método del objeto situado a su izquierda. Cadenas, listas y enteros son objetos de las clases `str`, `list` e `int`.

## 4. Clases ya conocidas: módulos, `type` e `id`

Las clases son **plantillas (tipos) para construir objetos**. Para programar algo real se necesitan objetos e interacciones entre objetos. Python trae ya muchas clases incorporadas (`list`, `str`, `int`, `tuple`, `dict`...) y otras dentro de módulos (`random`, `math`, `sys`, `datetime`...). Por ejemplo, el módulo `datetime` incluye la clase `date` (atributos año, mes, día; métodos `today`, `fromtimestamp`...) y la clase `time` (atributos hora, minuto, segundo; métodos `replace`, `isoformat`...). El código que escribe el usuario (`prog.py` en el PDF) crea objetos de esas clases y los hace interactuar.

````python
from datetime import date, time

fecha = date(2025, 1, 31)          # un objeto de la clase date
print(fecha.year, fecha.month, fecha.day)   # atributos
print(fecha.isoformat())                    # un método

hora = time(10, 30, 15)            # un objeto de la clase time
print(hora.hour, hora.minute, hora.second)
print(hora.replace(hour=12).isoformat())
````

### `type` e `id`

- `type(objeto)` devuelve el tipo (la clase) de un objeto.
- `id(objeto)` devuelve un identificador único para el objeto mientras existe (en CPython corresponde a su dirección de memoria).
- Si dos variables hacen referencia al **mismo objeto**, sus `id` coinciden. Un mismo objeto puede tener varios nombres de variable.

El PDF usa `animals` y `others = animals` para comprobarlo. Los valores concretos de `id` cambian de una ejecución a otra, por eso el código compara en lugar de imprimir números fijos.

````python
animals = ['lion', 'tiger', 'cat', 'dog']
print(type(animals))
print(id(animals))          # un número distinto en cada ejecución

s = 'tiger'
print(type(s))
print(id(s))

# Dos variables que referencian el mismo objeto de lista
others = animals
print(id(animals) == id(others))
verificar("animals y others son el mismo objeto", id(animals) == id(others), True)
````

### `__div__` frente a `__truediv__`

En Python 2 el operador `/` llamaba al método `__div__`; en Python 3 llama a `__truediv__`. Por eso `(200).__truediv__(100)` devuelve `2.0` (valor flotante) y `(200).__div__(100)` produce un error en Python 3.

````python
print((200).__truediv__(100))
try:
    print((200).__div__(100))
except AttributeError as error:
    print("AttributeError:", error)
````

**Error común:** buscar `__div__` en Python 3. El método correcto es `__truediv__` (y `__floordiv__` para `//`).

## 6. Clase, instancia y objeto

- **Clase:** concepto abstracto que denota un conjunto de atributos y acciones usados en los programas; es un plano, plantilla o esquema de objetos.
- **Instancia:** objeto individual creado a partir de una clase; distintas instancias pueden tener distintos valores de atributo.

El PDF lo ilustra con la clase `Gato` (concepto abstracto: atributos nombre y color; acciones maullar, correr, caminar) y dos instancias concretas: `nabi` (negro) y `nero` (blanco).

**Distinción precisa entre objeto e instancia.** Muchos textos mezclan ambos términos, pero para ser exactos: *objeto* es un material o cosa; *instancia* es un material o cosa **creada a partir de una clase**. `Gato` es una clase que actúa como marco; `nabi` creado con esa clase es un "objeto" y, al mismo tiempo, una "instancia de la clase `Gato`". Las siguientes descripciones son correctas:

1. Todos los objetos de Python tienen un tipo de datos.
2. Las instancias de Python se crean a partir de clases.
3. Los objetos de Python se crean a partir de clases.
4. La clase de Python es un objeto.
5. `nabi` es un objeto.
6. La instancia de la clase `Gato` es `nabi`.
7. `100` es un objeto de tipo entero.

````python
class Gato:
    pass

nabi = Gato()

verificar("1) todo objeto tiene tipo", type(100) is int, True)
verificar("2) nabi se crea a partir de una clase", type(nabi) is Gato, True)
verificar("4) la clase es un objeto (es instancia de type)", isinstance(Gato, type), True)
verificar("6) nabi es instancia de Gato", isinstance(nabi, Gato), True)
verificar("7) 100 es un objeto de tipo entero", isinstance(100, int), True)
````

**Clave:** la clase es el molde; la instancia es cada pieza fabricada con ese molde.

## 8. Métodos y el parámetro `self`

Un **método** (o función miembro) es una función definida dentro de una clase y utilizada por la clase o por sus instancias. El parámetro `self` del método es una variable que **hace referencia a sí misma** (a la instancia que lo llama); es obligatorio como **primer parámetro** de un método de instancia.

Se crea un objeto con `Cat()` y entonces se llama al método con `nabi.meow()`. Cada instancia (`nabi`, `nero`, `mimi`) puede ejecutar el método. El PDF imprime `miau miau~~~` una vez por llamada.

````python
class Cat:
    def meow(self):            # un método es una función dentro de la clase Cat
        print('miau miau~~~')

nabi = Cat()
nabi.meow()

nero = Cat()
nero.meow()

mimi = Cat()
mimi.meow()
````

Cuando se escribe `nabi.meow()`, Python hace internamente `Cat.meow(nabi)`: la instancia que está a la izquierda del punto es la que se pasa como `self`.

**Ampliación (no está en el PDF):** comprobar esa equivalencia, y qué pasa si se olvida `self` en la definición.

````python
class Cat:
    def meow(self):
        return 'miau miau~~~'

nabi = Cat()
verificar("nabi.meow() equivale a Cat.meow(nabi)", nabi.meow(), Cat.meow(nabi))


class CatSinSelf:
    def meow():                # error: falta el parámetro self
        return 'miau'

try:
    CatSinSelf().meow()
except TypeError as error:
    print("TypeError:", error)
````

**Error común:** olvidar `self` como primer parámetro. Python pasa la instancia automáticamente y el método recibe un argumento que no esperaba (`TypeError`).

## 9. Constructor `__init__` y representación en cadena `__str__`

- **Constructor `__init__`:** método que asigna valores iniciales a las variables de una instancia cuando se crea el objeto. Se ejecuta automáticamente al generar un objeto. (El PDF lo escribe `__int__` en una diapositiva: es una errata de la traducción; el nombre correcto es `__init__`.)
- **Método `__str__`:** devuelve la representación de un objeto como cadena. Es llamado automáticamente por la función `print`: `print(nabi)` ejecuta `print(nabi.__str__())`.

El primer parámetro `self` de `__init__` se refiere a la instancia que se está creando; los parámetros siguientes (`name`, `color`) son los valores con los que se inicializan los atributos de instancia. Con `Cat('nabi', 'negro')`, `self` es `nabi`, `name` es `'nabi'` y `color` es `'negro'`.

**Nota sobre el PDF:** en el código de la diapositiva el segundo gato se crea con `Cat(nero, white")`, con comillas rotas y un color inglés; aquí se corrige a `Cat("nero", "white")` para que el código corra y la salida sea la mostrada (`Cat(name=nero, color=white)`).

````python
class Cat:
    def __init__(self, name, color):
        # Inicializa los valores al generar instancias de Cat
        self.name = name
        self.color = color

    # Un formato de expresión de cadena de la clase Cat
    def __str__(self):
        return 'Cat(name=' + self.name + ', color=' + self.color + ')'


nabi = Cat('nabi', 'negro')    # generar instancia nabi
nero = Cat("nero", "white")    # generar instancia nero

print(nabi)
print(nero)
````

### ¿Cuándo se llama a `__str__`?

Normalmente se llama automáticamente: en `print(...)`, y también al formatear con `.format()` o al convertir con `str(...)`.

````python
print('información de nabi : {}'.format(nabi))
verificar("print llama a __str__", str(nabi), nabi.__str__())
verificar("format llama a __str__", '{}'.format(nabi), 'Cat(name=nabi, color=negro)')
````

**Ampliación (no está en el PDF):** sin `__str__`, `print` muestra la dirección del objeto; con `__str__` muestra lo que se defina. Existe también `__repr__`, que usa la consola interactiva y los contenedores (por ejemplo, al imprimir una lista de objetos); si solo se define `__str__`, una lista de instancias sigue mostrando la dirección.

````python
class GatoSinStr:
    def __init__(self, name):
        self.name = name

class GatoConRepr:
    def __init__(self, name):
        self.name = name
    def __repr__(self):
        return f"GatoConRepr(name={self.name!r})"

print(GatoSinStr("nabi").__str__().startswith("<__main__.GatoSinStr object at 0x"))
print(GatoConRepr("nabi"))          # print usa __repr__ si no hay __str__
print([GatoConRepr("nabi"), GatoConRepr("nero")])
````

**Clave:** `__init__` prepara cada instancia al crearla; `__str__` define cómo se ve como texto. Los dos son "métodos especiales" (doble guion bajo a cada lado).

**Error común:** que `__str__` use `print` en vez de `return`. Debe **devolver** la cadena; `print` la mostrará.

**Para el quiz:** `__init__` se llama automáticamente al crear una instancia; `__str__` se llama al usar `print` (o `str`) con el objeto.

## 10. Variables de instancia y ejemplo de llamada a métodos

Los términos **variable de instancia**, **variable miembro** y **campo** designan lo mismo: variables que almacenan los atributos propios de cada instancia. En `self.name = name` y `self.color = color`, cada instancia tiene su propia copia: `nabi.name` es `'Nabi'`, `nero.name` es `'Nero'`, `mimi.name` es `'Mimi'`.

Un método de la clase (aquí `meow`) puede ser llamado por todas las instancias con el operador punto, y cada una usa **sus propios** valores.

**Nota sobre el PDF:** la diapositiva escribe el nombre `'Mini'` en el diagrama y `'mimi'` en el código; el código y la salida usan `mimi`/`Mimi`. También escribe `'brown'` en el código frente a `marrón` en la salida; aquí se usa `'marrón'` para que ambos coincidan.

````python
class Cat:
    # Llamado como constructor o método de inicialización
    def __init__(self, name, color):
        self.name = name     # Genera una variable de instancia llamada name
        self.color = color   # Genera una variable de instancia llamada color

    # El método que imprime la información de un gato
    def meow(self):
        print("Mi nombre es {}, color es {}, miau miau~~".format(self.name, self.color))


nabi = Cat('Nabi', 'negro')      # Genera la instancia Nabi
nero = Cat('Nero', 'blanco')     # Genera la instancia Nero
mimi = Cat('Mimi', 'marrón')     # Genera la instancia Mimi

nabi.meow()
nero.meow()
mimi.meow()
````

**Clave:** `self.atributo = valor` dentro de `__init__` crea una variable de instancia. Cada instancia guarda sus propios valores.

## 11. Variables de instancia (clase `Circle`)

La clase `Circle` del PDF guarda tres variables de instancia: `__name`, `__radius` y `__PI`. Cada instancia puede tener valores distintos para ellas. El doble guion bajo (`__`) sirve para **ocultar** la variable de instancia del acceso externo.

El método `area` calcula el área multiplicando el radio al cuadrado por el `PI` de la instancia actual.

````python
class Circle:
    def __init__(self, name, radius, PI):
        self.__name = name       # variable de instancia
        self.__radius = radius   # variable de instancia
        self.__PI = PI           # variable de instancia

    # Calcula el área multiplicando el radio**2 por el PI de la instancia actual
    def area(self):
        return self.__PI * self.__radius ** 2


c1 = Circle("C1", 4, 3.14)
print("Área de c1:", c1.area())
c2 = Circle("C2", 6, 3.141)
print("Área de c2:", c2.area())
c3 = Circle("C3", 5, 3.1415)
print("Área de c3:", c3.area())
````

**Nota sobre el PDF:** la salida del PDF usa coma decimal (`50,24`, `113,076`) por la traducción; Python imprime siempre punto decimal (`50.24`).

El problema de este diseño: cada instancia repite el mismo `PI`. Si el valor de `PI` fuera constante, `c1`, `c2` y `c3` deberían **compartirlo**. Eso se resuelve con una variable de clase.

## 12. Variables de clase

Una **variable de clase** se define dentro de la clase pero **fuera de los métodos**; **todas las instancias la comparten**. Se reduce así la duplicación de datos y es más fácil detectar errores. Si se necesita más precisión, basta modificar `PI` en un solo lugar (el PDF sugiere `PI = 3.1415926`). Los atributos `__name` y `__radius` siguen siendo variables de instancia, propias de cada instancia.

Dentro de un método se accede a la variable de clase a través del nombre de la clase: `Circle.PI`.

````python
class Circle:
    PI = 3.1415   # variable de clase

    def __init__(self, name, radius):
        self.__name = name        # variable de instancia
        self.__radius = radius    # variable de instancia

    # Calcula el área usando la variable de clase PI de Circle
    def area(self):
        return Circle.PI * self.__radius ** 2


c1 = Circle("C1", 4)
print("Área de c1:", c1.area())
c2 = Circle("C2", 6)
print("Área de c2:", c2.area())
c3 = Circle("C3", 5)
print("Área de c3:", c3.area())

verificar("área de c1", round(c1.area(), 4), 50.264)
verificar("área de c2", round(c2.area(), 4), 113.094)
verificar("área de c3", round(c3.area(), 4), 78.5375)
````

**Nota sobre el PDF:** la salida del PDF escribe `Superficie de c2` en lugar de `Área de c2` (traducción automática) y usa coma decimal. Los valores como `113.09400000000001` aparecen porque los números de coma flotante no representan exactamente todos los decimales.

**Ampliación (no está en el PDF):** como la variable de clase es compartida, cambiarla en la clase afecta a todas las instancias; en cambio, asignar a `instancia.PI` crea una variable de instancia nueva que "tapa" a la de clase solo en esa instancia.

````python
class Circulo:
    PI = 3.14

    def __init__(self, radio):
        self.radio = radio

    def area(self):
        return self.PI * self.radio ** 2


a = Circulo(1)
b = Circulo(1)
Circulo.PI = 3.1415            # cambia para todas las instancias
verificar("a usa la variable de clase", a.area(), 3.1415)
verificar("b usa la variable de clase", b.area(), 3.1415)

a.PI = 3                       # crea una variable de instancia que tapa a la de clase, solo en a
verificar("a usa su propia PI", a.area(), 3)
verificar("b sigue usando la de clase", b.area(), 3.1415)
````

**Clave:** variable de instancia = propia de cada objeto (se crea con `self.x` en `__init__`); variable de clase = compartida por todas las instancias (se define en el cuerpo de la clase).

## 13. `__dict__`: ver los atributos de un objeto

El atributo `__dict__` es útil para construir programas: convierte las variables del objeto en un **diccionario** `{nombre: valor}`, y se pueden leer los valores con `obj.__dict__[clave]`. Solo muestra las variables de **instancia** (la variable de clase `PI` no aparece).

Si el atributo es **privado** (empieza con `__`), Python le cambia el nombre internamente a `_NombreClase__atributo` (*name mangling*) y esa es la clave del diccionario.

## 14. Herencia de clases: concepto y sintaxis

La **herencia** es una técnica que permite escribir una clase hija que hereda los métodos y atributos de una clase previamente construida. La razón de heredar es la **eficiencia del proceso de desarrollo** (reutilizar código), y es una de las principales características de la POO.

Terminología:
- Clase de nivel superior que transmite métodos y atributos: **clase padre**, **superclase** o **clase base**.
- Clase que hereda: **clase hija**, **subclase** o **clase derivada**.

En el PDF, `Persona` (clase padre) tiene características bastante generales (edad, sexo, nombre...). Sus hijas `Gestor`, `Empleado` y `Estudiante` heredan esas características y añaden otras más detalladas (nombre de la empresa, número de empleado, paga, número de identificación del estudiante, GPA...).

Sintaxis: la clase hija se define escribiendo el nombre de la clase padre entre paréntesis.

```python
class A:        # clase padre A
    ...
class B(A):     # clase hija B con A como padre
    ...
```

Una clase hija puede heredar los métodos y atributos de la clase padre y además definir los suyos propios. Enfoque del PDF: la instancia `b` de `B` puede acceder a `PI` aunque `B` no la tenga, porque la hereda de `A`.

````python
class A:                 # clase padre A
    PI = 3.14

class B(A):              # clase hija B con A como padre
    pass


a = A()                  # Genera la instancia a de la clase A
b = B()                  # Genera la instancia b de la clase B
print('a.PI =', a.PI)
print('b.PI =', b.PI)    # b puede acceder a los atributos de la clase padre

verificar("B hereda de A", issubclass(B, A), True)
verificar("b es instancia de B y de A", (isinstance(b, B), isinstance(b, A)), (True, True))
````

**Nota sobre el PDF:** el diagrama de la diapositiva dice `PI = 3,13` mientras que el código y la salida usan `3.14`.

**Ampliación (no está en el PDF):** una subclase puede **sobrescribir** un atributo o método del padre, y `B.__mro__` muestra el orden en que Python busca los nombres.

````python
class Animal:
    def sonido(self):
        return "..."

class Perro(Animal):
    def sonido(self):          # sobrescribe el método del padre
        return "guau"

class Pez(Animal):
    pass                       # hereda sonido() tal cual

print(Perro().sonido(), Pez().sonido())
print([clase.__name__ for clase in Perro.__mro__])
````

**Clave:** `class Hija(Padre):` hereda todo lo de `Padre`. Reutiliza código sin copiarlo.

**Para el quiz:** clase padre = superclase = base; clase hija = subclase = derivada.

## 15. Diseño de clases y encapsulación

**Por qué hace falta.** Si el atributo `age` de `Cat` es público, desde fuera se puede escribir `nabi.age = -5`. No es un error de sintaxis, sino un **problema lógico**: la edad de un gato no puede ser negativa, pero los valores no estaban protegidos.

````python
class Cat:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    # formato de expresión de cadena de objetos Cat
    def __str__(self):
        return 'Cat(name=' + self.name + ', age=' + str(self.age) + ')'


nabi = Cat('nabi', 3)    # genera la instancia nabi
print(nabi)
nabi.age = 4
nabi.age = -5            # un caso anormal de edad negativa
print(nabi)
````

**Encapsulación:** una forma de reducir los errores derivados del acceso externo a los atributos internos de una clase. Obliga a que el acceso externo pase por funciones designadas (métodos), evitando la manipulación aleatoria de los atributos. Conceptualmente, la clase envuelve método y atributo en una "cápsula": limita las modificaciones externas, protege los datos y evita cambios accidentales en los valores.

**Cómo se hace en Python:** el atributo se declara privado con doble guion bajo (`self.__age`) y se ofrecen métodos para leerlo y modificarlo:
- **setter** (`set_age`): métodos que empiezan por `set`; modifican el atributo, y aquí validan con un `if age > 0` para que no se asigne una edad negativa.
- **getter** (`get_age`): métodos que empiezan por `get`; devuelven el valor.

````python
class Cat:
    def __init__(self, name, age):
        self.__name = name
        self.__age = age

    # formato de expresión de cadena de objetos Cat
    def __str__(self):
        return 'Cat(name=' + self.__name + ', age=' + str(self.__age) + ')'

    # limita el acceso externo aleatorio a self.__age y evita que la edad sea negativa
    def set_age(self, age):
        if age > 0:
            self.__age = age

    def get_age(self):
        return self.__age


nabi = Cat('nabi', 3)    # genera la instancia nabi
print(nabi)
nabi.set_age(4)          # se aproxima a la edad mediante el método set_age()
nabi.set_age(-5)         # evita que la edad sea negativa mediante el método set_age()
print(nabi)

verificar("la edad negativa se ignora", nabi.get_age(), 4)
````

**Nota sobre el PDF:** en la diapositiva 140 el fragmento usa `si edad > 0:` (traducción de `if`) y los nombres `nombre`/`edad`; en el código real es `if age > 0:`.

**Ampliación (no está en el PDF):** el atributo privado no se puede leer desde fuera con su nombre, pero el nombre "mangleado" sigue siendo accesible: la privacidad en Python es una convención reforzada por el renombrado, no un bloqueo absoluto.

````python
try:
    print(nabi.__age)
except AttributeError as error:
    print("AttributeError:", error)

print(nabi._Cat__age)    # accesible, pero no se debe usar: rompe la encapsulación
````

**Clave:** privado (`__x`) + métodos `get`/`set` con validación = datos protegidos.

**Error común:** creer que `__x` hace el atributo totalmente inaccesible. Solo lo oculta con un nombre alterado.

**Para el quiz:** setter = método que modifica un atributo privado (con validación); getter = método que lo devuelve.

## 16. Operador de identidad: `is` e `is not`

El operador `is` devuelve `True` si dos operandos hacen referencia al **mismo objeto** (es decir, si tienen el mismo `id`); `is not` devuelve `True` si **no** son el mismo objeto. No compara valores, compara identidad.

Dos listas `[10, 20, 30]` creadas por separado son **objetos distintos** (distintos `id`) aunque tengan los mismos valores, por eso `list_a is list_b` es `False`.

````python
list_a = [10, 20, 30]     # list_a hace referencia a un objeto lista
list_b = [10, 20, 30]     # list_b hace referencia a otro objeto lista
if list_a is list_b:      # verifica si los dos objetos de lista son idénticos
    print('list_a is list_b')
else:
    print('list_a is not list_b')

print('id(list_a) =', id(list_a), ', id(list_b) =', id(list_b))
verificar("los ids de dos listas distintas difieren", id(list_a) != id(list_b), True)
verificar("sus valores sí son iguales (==)", list_a == list_b, True)
````

Con **cadenas** ocurre algo distinto: el PDF muestra que dos variables con la misma cadena `'ABC'` se refieren al mismo objeto. Como las cadenas son **inmutables**, ese enfoque no presenta ningún problema y ahorra memoria (se almacena un solo objeto cadena en una tabla de almacenamiento).

````python
a = 'ABC'                 # variable a que hace referencia a un objeto cadena
b = 'ABC'                 # variable b que hace referencia a un objeto cadena
if a is b:                # compara si el objeto de cadena a y b son idénticos
    print('a is b')       # el objeto de cadena a y b se refieren al mismo objeto
else:
    print('a is not b')
````

**Ampliación (no está en el PDF):** que dos cadenas iguales compartan objeto es una optimización de CPython, no una garantía del lenguaje. Para comparar **valores** se usa siempre `==`; `is` se reserva para comparar identidad (típicamente con `None`).

````python
valor = None
print(valor is None)        # uso correcto e idiomático de is

c = "".join(["AB", "C"])    # una cadena 'ABC' construida en tiempo de ejecución
print(c == a, c is a)       # == es True; is puede ser False, no se debe depender de él
````

**Error común:** usar `is` para comparar valores (`x is 5`, `texto is 'ABC'`). Puede funcionar por casualidad de la implementación y fallar en otro caso; usa `==`.

**Para el quiz:** `==` compara valores; `is` compara identidad (`id`).

## 17. Clase y métodos especiales

**Necesidad.** Se construye una clase `Vector2D` que expresa vectores bidimensionales. ¿Por qué `v1 + v2` produce un error? Porque el código de la clase no indica cómo debe funcionar la suma de sus instancias.

````python
class Vector2D:
    def __init__(self, x, y):
        self.x = x
        self.y = y


v1 = Vector2D(30, 40)
v2 = Vector2D(10, 20)
try:
    v3 = v1 + v2      # el operador + no ha sido definido en Vector2D: imprime error
except TypeError as error:
    print("TypeError:", error)
````

Una primera solución es definir un método llamado `add`, que suma la componente `x` y la componente `y` de los dos vectores y devuelve un nuevo `Vector2D`. Funciona, pero usar operadores como `+` o `-` es más conveniente que llamar a `add`.

````python
class Vector2D:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __str__(self):
        return "({}, {})".format(self.x, self.y)

    def add(self, other):
        return Vector2D(self.x + other.x, self.y + other.y)


v1 = Vector2D(30, 40)
v2 = Vector2D(10, 20)
v3 = v1.add(v2)
print('v1.add(v2) =', v3)
````

### Definir métodos especiales

Python permite **darle significado a los operadores** definiendo métodos especiales. `__add__` devuelve la suma de dos vectores y `__sub__` la diferencia. `v1 + v2` hace lo mismo que `v1.__add__(v2)`: el comportamiento interno de `v1 + v2` es idéntico al de llamar a `v1.__add__(v2)`.

````python
class Vector2D:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        return Vector2D(self.x + other.x, self.y + other.y)

    def __sub__(self, other):
        return Vector2D(self.x - other.x, self.y - other.y)

    def __str__(self):
        return "({}, {})".format(self.x, self.y)


v1 = Vector2D(30, 40)
v2 = Vector2D(10, 20)
v3 = v1 + v2
print('v1 + v2 =', v3)
v4 = v1 - v2
print('v1 - v2 =', v4)

verificar("v1 + v2 equivale a v1.__add__(v2)", str(v1 + v2), str(v1.__add__(v2)))
verificar("suma", str(v3), "(40, 60)")
verificar("resta", str(v4), "(20, 20)")
````

## 18. Objeto, referencia y operador de asignación

Una **variable** es solo un **nombre** dado a un objeto en memoria. En `n = 100`, la variable `n` hace referencia al objeto numérico `100`; `id(100)` es igual a `id(n)`. Con `m = n` (operador de asignación `=`) la variable `m` también hace referencia al mismo objeto: las dos variables se refieren al mismo objeto.

Los enteros, números reales, cadenas, booleanos y tuplas son **inmutables**. Si `n = 100` y `m = 100`, ambas se refieren al mismo objeto `100`, y este mecanismo permite un uso eficiente de la memoria.

````python
n = 100
print(id(100) == id(n))

m = n
print(id(n), id(m))
verificar("n y m son el mismo objeto", id(n) == id(m), True)

x = 100
y = 100
verificar("x e y (ambos 100) comparten objeto", id(x) == id(y), True)
````

### La asignación apunta a otro objeto

Si luego se ejecuta `n = 200`, `n` pasa a hacer referencia a **otro objeto** (el entero `200`), mientras `m` sigue haciendo referencia al `100`. Lo mismo con `n = n + 1`: el nuevo valor `101` es un objeto distinto con otro `id`. El objeto entero `100`, si ninguna variable lo referencia, es un "objeto basura" y la memoria que ocupa se libera (el PDF lo llama **recolección de basura**: procedimiento periódico de gestión de memoria que elimina los objetos sin referencias).

````python
n = 100
m = n
id_100 = id(n)

n = 200
print("id(n) cambió:", id(n) != id_100)
print("id(m) sigue igual:", id(m) == id_100)
verificar("m sigue apuntando al 100", m, 100)

n = 100
antes = id(n)
n = n + 1
print("n + 1 es otro objeto:", id(n) != antes)
verificar("n vale 101", n, 101)
````

**Nota sobre el PDF:** los números concretos de `id` que muestra (por ejemplo `140719510270864`) cambian en cada máquina y ejecución; lo importante es la relación entre ellos (iguales o distintos).

**Ampliación (no está en el PDF):** con objetos **mutables** (por ejemplo una lista) la asignación copia solo la referencia: dos nombres, un solo objeto. Modificar a través de uno se ve en el otro; para tener un objeto independiente hay que copiar.

````python
original = [1, 2, 3]
alias = original              # misma lista, dos nombres
copia = original.copy()       # otra lista con los mismos valores
alias.append(4)
print(original, copia)
verificar("original cambió a través de alias", original, [1, 2, 3, 4])
verificar("la copia no cambió", copia, [1, 2, 3])
verificar("alias is original", alias is original, True)
verificar("copia is not original", copia is not original, True)
````

**Clave:** asignar nunca copia el objeto; hace que el nombre apunte a un objeto. Reasignar mueve la referencia, no modifica el objeto anterior.

**Error común:** pensar que `m = n` crea una copia independiente (cierto en efecto con números, porque son inmutables, pero falso con listas y objetos propios).

## 19. Clases con estructura jerárquica: `Persona`, `Empleado`, `Gestor`

El PDF construye una jerarquía: la clase padre `Person` tiene el atributo `name` y el método `get_name()`. Las clases hijas `Manager` (gerente) y `Employee` (empleado) heredan de `Person`, de modo que contienen `name` y `get_name()` **sin ninguna codificación extra**; cada una añade lo propio: `Employee` el atributo `staff_id` y un método `info()`; `Manager` el atributo `position` y su propio `info()`.

En el constructor de cada clase hija se llama al constructor de la clase padre con `Person.__init__(self, name)` para inicializar el atributo heredado `name`.

````python
class Person:
    def __init__(self, name):
        self.name = name

    def get_name(self):
        return self.name


class Employee(Person):                        # clase hija de la clase Person
    def __init__(self, name, staff_id):
        Person.__init__(self, name)
        self.staff_id = staff_id

    def info(self):
        return 'Empleado : ' + self.get_name() + ', staffID : ' + str(self.staff_id)


class Manager(Person):                         # clase hija de la clase Person
    def __init__(self, name, position):
        Person.__init__(self, name)
        self.position = position

    def info(self):
        return 'Gerente : ' + self.get_name() + ', cargo : ' + str(self.position)


# Genera objetos con diferentes atributos
worker1 = Employee('David Doe', 1111)
worker2 = Employee('Paul Carter', 2222)
cfo = Manager('Anna Smith', 'CFO')

print(worker1.info())
print(worker2.info())
print(cfo.info())

verificar("Employee usa get_name heredado", worker1.get_name(), 'David Doe')
verificar("Manager es subclase de Person", issubclass(Manager, Person), True)
````

**Código final comentado.** Notas de diseño: la condición de retiro usa `>= 0` (el PDF muestra `> 0`, lo que rechazaría retirar justo todo el saldo); el mensaje de cada operación queda como en la salida del PDF.

````python
class BankAccount:
    """Una clase de Python que modela una cuenta bancaria."""

    def __init__(self, name, account_num, balance=0):
        # Constructor: crea las variables de instancia de la cuenta
        self.name = name
        self.account_num = account_num
        self.balance = balance

    def get_name(self):
        # Devuelve el nombre del titular de la cuenta
        return self.name

    def get_account_num(self):
        # Devuelve el número de cuenta
        return self.account_num

    def get_balance(self):
        # Devuelve el saldo
        return self.balance

    def deposit(self, amount):
        # Suma el importe al saldo, informa y devuelve el saldo
        self.balance += amount
        print('Se han depositado {} won. El saldo es de {} won.'.format(amount, self.balance))
        return self.balance

    def withdraw(self, amount):
        # Retira solo si el saldo alcanza; si no, informa del retiro denegado
        if self.balance - amount >= 0:
            self.balance -= amount
        else:
            print('El saldo de la cuenta es de {} won. Esto es menor que la cantidad solicitada de retiro {} won.'
                  .format(self.balance, amount))

    def __str__(self):
        # Representación en cadena: nombre, cuenta y saldo
        return 'El saldo de la cuenta {} de {} es de {} won.'.format(self.account_num, self.name, self.balance)


# Programa principal (sigue el plan de programación)
account1 = BankAccount("David", "1234-0001")
print(account1)           # [2] imprime la cuenta y el saldo
account1.deposit(2000)    # [3] deposita 2000 won
print(account1)
account1.withdraw(500)    # [4] retira 500 won
print(account1)
account1.withdraw(5000)   # [5] retiro denegado: supera el saldo
````

````python
class Estudiante:
    """Registra el nombre, el ID y las puntuaciones de tres pruebas de un estudiante."""

    def __init__(self, name, student_id):
        self.name = name
        self.student_id = student_id
        self.eng_quiz = 0
        self.math_quiz = 0
        self.science_quiz = 0

    def set_eng_quiz(self, puntuacion):
        self.eng_quiz = puntuacion

    def set_math_quiz(self, puntuacion):
        self.math_quiz = puntuacion

    def set_science_quiz(self, puntuacion):
        self.science_quiz = puntuacion

    def get_name(self):
        return self.name

    def get_student_id(self):
        return self.student_id

    def get_eng_quiz(self):
        return self.eng_quiz

    def get_math_quiz(self):
        return self.math_quiz

    def get_science_quiz(self):
        return self.science_quiz

    def get_total_score(self):
        return self.eng_quiz + self.math_quiz + self.science_quiz

    def get_avg_score(self):
        return self.get_total_score() / 3

    def __str__(self):
        return ("Nombre : {}, ID : {}\n"
                "Puntuación del examen de inglés : {}, Puntuación del examen de matemáticas : {}\n"
                "Puntuación del examen de ciencias : {},\n"
                "Total : {}, Media : {}").format(
                    self.name, self.student_id, self.eng_quiz, self.math_quiz,
                    self.science_quiz, self.get_total_score(), self.get_avg_score())


def leer_estudiante(leer=input):
    """Pide los datos con la función `leer` (input por defecto) y devuelve el Estudiante."""
    nombre = leer("Introduzca el nombre del estudiante : ")
    id_estudiante = int(leer("Introduzca el ID del estudiante : "))
    estudiante = Estudiante(nombre, id_estudiante)
    estudiante.set_eng_quiz(int(leer("Introduzca la puntuación obtenida por el alumno en el examen de inglés : ")))
    estudiante.set_math_quiz(int(leer("Introduzca la puntuación obtenida por el alumno en el cuestionario de matemáticas : ")))
    estudiante.set_science_quiz(int(leer("Introduzca la puntuación del alumno en el cuestionario de ciencias : ")))
    return estudiante


# Versión no interactiva: respuestas de prueba que reproducen el ejemplo de salida del PDF
respuestas = iter(["David Doe", "20213093", "90", "95", "100"])

def leer_de_prueba(pregunta):
    respuesta = next(respuestas)
    print(pregunta + respuesta)
    return respuesta

alumno = leer_estudiante(leer_de_prueba)
print(alumno)

verificar("total", alumno.get_total_score(), 285)
verificar("media", alumno.get_avg_score(), 95.0)

# Para usarlo de forma interactiva, quita el comentario de la siguiente línea
# print(leer_estudiante())
````

def simular_cuenta(operaciones):
    class BankAccount:
        def __init__(self):
            self.balance = 0

        def deposit(self, amount):

        def withdraw(self, amount):
            if self.balance - amount >= 0:
                self.balance -= amount
                return True
            return False

    cuenta = BankAccount()

## Apuntes para estudiar

### Clave

- **POO** = modelar el programa como objetos que interactúan; el objeto reúne **atributos** (datos) y **métodos** (funciones).
- Todo en Python es un objeto: `100`, `'tiger'`, las listas, las funciones e incluso las clases. `type(x)` da su clase; `id(x)` su identidad.
- **Clase** = plantilla abstracta (`class Cat:`); **instancia** = objeto concreto creado con ella (`nabi = Cat('nabi', 'negro')`). Los paréntesis al instanciar son obligatorios.
- **`__init__`** (constructor) se ejecuta al crear la instancia y define las variables de instancia con `self.x = x`.
- **`self`** es la propia instancia y debe ser el primer parámetro de los métodos de instancia; `nabi.meow()` equivale a `Cat.meow(nabi)`.
- **`__str__`** devuelve la representación en cadena (usada por `print`, `str` y `format`).
- **Variable de instancia** (`self.x`, propia de cada objeto) frente a **variable de clase** (en el cuerpo de la clase, compartida, `Clase.x`).
- **`__dict__`** muestra las variables de instancia como diccionario; los atributos privados aparecen como `_Clase__atributo`.
- **Herencia**: `class Hija(Padre):` reutiliza atributos y métodos; la hija llama a `Padre.__init__(self, ...)` o `super().__init__(...)`.
- **Encapsulación**: atributo privado `__x` + getters/setters con validación.
- **`is`** compara identidad (mismo `id`); `==` compara valores.
- **Métodos especiales**: `+` llama a `__add__`, `-` a `__sub__`, `<` a `__lt__`, `==` a `__eq__`, `len` a `__len__`...
- Una **variable** es un nombre que apunta a un objeto; asignar no copia: mueve la referencia. Sin referencias, el objeto es "basura" y se recolecta.

### Errores comunes

- Olvidar `self` como primer parámetro de un método (`TypeError`) o al leer un atributo (`name` en vez de `self.name`).
- Escribir `__int__` en lugar de `__init__` (o dejar `__init__` sin doble guion bajo a ambos lados).
- Instanciar sin paréntesis (`nabi = Cat`): se asigna la clase, no una instancia.
- Hacer `print` dentro de `__str__` en lugar de `return`.
- Olvidar llamar al constructor del padre desde el constructor de la hija: falta el atributo heredado.
- Creer que `__x` es inaccesible; solo cambia de nombre a `_Clase__x`.
- Usar `is` para comparar valores en lugar de `==`.
- Buscar `__div__` en Python 3 (es `__truediv__`).
- Creer que `m = n` copia un objeto mutable (la lista sigue siendo una sola).
- Olvidar `return` en `__add__`/`__sub__`: el resultado de la operación es `None`.
