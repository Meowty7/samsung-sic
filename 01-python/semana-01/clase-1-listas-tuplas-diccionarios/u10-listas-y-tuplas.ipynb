## 1. Por qué existen las listas

Cuando hay que guardar varios datos del mismo tipo, la solución ingenua es crear una variable por dato. El PDF lo ilustra con 7 puntuaciones de exámenes (págs. 18–19) y con las alturas de 5 estudiantes (pág. 64):

````python
# Una variable por cada dato: poco práctico
score0 = 87
score1 = 84
score2 = 95
score3 = 67
score4 = 88
score5 = 94
score6 = 63
print(score0, score3)
````

Con 100 personas habría 100 variables distintas, y con 1.000 personas, 1.000. Duplicar u operar sobre ellas exigiría mucho código y sería propenso a errores. La alternativa es una **lista**: una estructura de datos que guarda varios valores (llamados *elementos*) bajo un único nombre, separados por comas y entre corchetes `[]`.

````python
score_list = [87, 84, 95, 67, 88, 94, 63]
print(score_list)
print(score_list[0], score_list[3])   # la referencia usa el nombre de la lista y el índice
````

**Ampliación (no está en el PDF):** como la lista es un único objeto, se puede operar sobre todos sus elementos de una vez, por ejemplo con `sum()` y `len()`:

````python
promedio = sum(score_list) / len(score_list)
print(f"Promedio de {len(score_list)} puntuaciones: {promedio:.2f}")
````

**Clave:** una lista reemplaza muchas variables sueltas por una sola variable con varios elementos, que se distinguen por su posición (índice).

**Referencia:** PDF parte 1, págs. 18–19 y 64–65

## 2. Crear listas

Una lista se declara escribiendo sus elementos entre corchetes, separados por comas. No hay restricción sobre el tipo de dato de cada elemento: pueden mezclarse enteros, reales y cadenas (pág. 23). También se puede crear una lista vacía con `[]` y llenarla después (págs. 66–67), o convertir un rango con `list(range(...))` (págs. 24 y 69).

````python
fruits = ['banana', 'apple', 'orange', 'kiwi']   # lista de cadenas
print(fruits)
mixed_list = [100, 200, 'apple', 400]            # tipos distintos en una misma lista
print(mixed_list)
````

````python
list4 = list(range(1, 10))    # range(1, 10) genera 1, 2, ..., 9 (el 10 no se incluye)
list4
````

````python
print(list(range(1, 11)))     # ejemplo de la pág. 69
````

Una lista vacía sirve cuando no se sabe de antemano cuántos elementos habrá. Se le agregan elementos con `append` o con el operador `+` (pág. 67):

````python
bts = ['V', 'Jungkook', 'Jimin']   # lista con 3 elementos (pág. 66)
print(bts)

bts = []                           # lista vacía
bts.append("V")
print(bts)

bts = []
bts = bts + ["V"]                  # el operador + también agrega (concatena listas)
print(bts)
````

**Clave:** `[]` crea una lista vacía; `list(range(a, b))` crea una lista de enteros consecutivos desde `a` hasta `b - 1`.

**Error común:** olvidar que `range(1, 10)` termina en 9 y no en 10.

**Para el quiz:** las listas de Python admiten elementos de tipos distintos.

## 3. Índices, longitud e indexación

El **índice** es el número que indica la posición de un elemento. En una lista de `n` elementos el índice crece de `0` a `n - 1`. La **indexación** es acceder al valor mediante su índice de posición. La función `len` devuelve el número de elementos (la longitud).

````python
n_list = [11, 22, 33, 44, 55, 66]
print(n_list)
print(len(n_list))   # 6 elementos, índices 0 a 5
````

````python
print(n_list[0])   # el índice del primer elemento es 0
print(n_list[1])   # el del segundo elemento es 1
print(n_list[5])   # el último índice válido es len(n_list) - 1
````

### Índices negativos

Python permite contar desde el final: `-1` es el último elemento, `-2` el penúltimo, y así sucesivamente. El primer elemento es `-len(n_list)`.

````python
print(n_list[-1])   # último
print(n_list[-2])   # penúltimo
print(n_list[-3])
print(n_list[-6])   # equivale al primer elemento
````

### Precaución: no salirse del rango

Usar un índice mayor que `len(n_list) - 1` produce `IndexError: list index out of range`.

````python
try:
    n_list[6]   # la lista tiene 6 elementos: el índice 6 no existe
except IndexError as error:
    print("IndexError:", error)
````

**Nota sobre el PDF:** en la pág. 29 la lista de equivalencias indica `n_list[-5] = 23`, pero el valor correcto es `22` (la lista es `[11, 22, 33, 44, 55, 66]`). Es una errata del PDF.

**Clave:** índice positivo desde `0` al inicio; índice negativo desde `-1` al final. El índice máximo válido es `len(lista) - 1`.

**Error común:** usar `lista[len(lista)]` para el último elemento; debe ser `lista[len(lista) - 1]` o simplemente `lista[-1]`.

**Para el quiz:** `IndexError` aparece al usar un índice fuera de rango.

**Referencia:** PDF parte 1, págs. 25–29 y 70–71

## 4. Funciones incorporadas aplicables a listas

Con listas numéricas se pueden usar funciones incorporadas (*built-in*) de Python:

- `len(lista)`: longitud.
- `max(lista)` y `min(lista)`: valor máximo y mínimo.
- `sum(lista)`: suma de todos los elementos (si son numéricos).
- `any(lista)`: devuelve `True` si hay al menos un elemento "verdadero" (por ejemplo, distinto de 0); con `[0, '']` devuelve `False`.

````python
n_list = [200, 700, 500, 300, 400]
print(len(n_list))
print(max(n_list))
print(min(n_list))
print(sum(n_list))
````

````python
a_list = [0, '']
print(any(n_list))   # hay elementos distintos de 0
print(any(a_list))   # 0 y '' (cadena vacía) se consideran falsos
````

**Nota sobre el PDF:** la pág. 69 dice que `any` devuelve `True` si hay al menos un elemento "que no sea 0" y menciona "espacios en blanco". Con precisión: `any` evalúa la veracidad (*truthiness*) de cada elemento; `0`, `''` (cadena vacía), `None` y `False` son falsos. Una cadena con un espacio `' '` sí es verdadera.

**Ampliación (no está en el PDF):** `all(lista)` devuelve `True` solo si todos los elementos son verdaderos.

````python
print(any([0, ' ']))        # ' ' (un espacio) es una cadena no vacía: verdadera
print(all([1, 2, 3]), all([1, 0, 3]))
````

**Clave:** `max`, `min` y `sum` resuelven en una línea cálculos que de otro modo requerirían un bucle.

**Para el quiz:** `sum([200, 700, 500, 300, 400])` es `2100`.

**Referencia:** PDF parte 1, págs. 68–69

## 5. Agregar y ampliar elementos

El PDF presenta cuatro formas de agregar elementos a una lista:

| Operación | Efecto |
|---|---|
| `lista1 + lista2` | crea una lista nueva con los elementos de ambas |
| `lista.append(x)` | agrega `x` al final de la lista |
| `lista.extend(otra)` | agrega cada elemento de `otra` al final |
| `lista.insert(i, x)` | inserta `x` en la posición `i` |

Un **método** es una función incorporada de un objeto y se invoca con el operador punto: `lista.append(...)`.

````python
person1 = ['David Doe', 20, 1, 180.0, 100.0]
person2 = ['John Smith', 25, 1, 170.0, 70.0]

person_list = person1 + person2   # une los elementos de dos listas en una sola
print(person_list)
````

````python
a_list = ['a', 'b', 'c', 'd', 'e']
a_list.append('f')       # agrega 'f' al final
print(a_list)

n_list = [10, 20, 30, 40]
n_list.append(50)        # agrega 50
print(n_list)
````

### append contra extend

`append` agrega **un** elemento (aunque ese elemento sea otra lista). `extend` agrega **cada elemento** de otra secuencia.

````python
list1 = ['a', 'b', 'c']
list2 = [1, 2, 3]
list1.extend(list2)
print(list1)

list1.extend('d')        # una cadena de un solo carácter agrega ese carácter
print(list1)
````

````python
list1 = [11, 22, 33, 44]
list1.append([55, 66])   # agrega la lista completa como UN elemento
print(list1)

list1 = [11, 22, 33, 44]
list1.extend([55, 66])   # agrega 55 y 66 como elementos separados
print(list1)
````

**Ampliación (no está en el PDF):** `extend` recorre su argumento. Con una cadena de varios caracteres agrega cada carácter por separado.

````python
letras = ['a']
letras.extend('hola')
print(letras)
````

### insert

`insert(índice, elemento)` coloca el elemento en la posición indicada y desplaza los siguientes (pág. 72).

````python
slist = ['David', 178.9, 'John', 173.5, 'Jane', 176.1]
print(slist)
slist.insert(4, "Petter")   # entra en la posición 4
slist.insert(5, 168.1)      # entra en la posición 5
print(slist)
````

**Clave:** `append` agrega al final un único elemento; `extend` incorpora varios; `insert` elige la posición.

**Error común:** usar `append([55, 66])` esperando que agregue dos elementos: agrega una lista anidada.

**Para el quiz:** resultado de `list1.append([55, 66])` sobre `[11, 22, 33, 44]` es `[11, 22, 33, 44, [55, 66]]`.

**Referencia:** PDF parte 1, págs. 30–33 y 72

## 6. Borrar elementos: `del`, `pop` y `remove`

Hay tres formas de eliminar elementos:

- `del lista[i]`: instrucción de Python que elimina el elemento del índice `i`. **No es un método de lista**.
- `lista.pop()`: elimina y **devuelve** el último elemento; `lista.pop(i)` hace lo mismo con el índice `i`.
- `lista.remove(valor)`: elimina el primer elemento cuyo valor coincide con el indicado.

````python
n_list = [11, 22, 33, 44, 55, 66]
print(n_list)
del n_list[3]            # borra el cuarto elemento (44)
print(n_list)
````

````python
n_list = [10, 20, 30]
print(n_list)
n = n_list.pop()         # quita el último y lo devuelve
print('n =', n)
print('n_list =', n_list)
````

````python
n_list = [11, 22, 33, 44, 55, 66]
n_list.remove(44)        # elimina el elemento que vale 44
print(n_list)
````

### Precauciones

- `remove` con un valor que no está en la lista produce `ValueError`.
- `del` es una palabra clave, no un método: escribir `bts[0].del` produce `SyntaxError`.

````python
n_list = [11, 22, 33, 44, 55, 66]
try:
    n_list.remove(88)    # 88 no está en la lista
except ValueError as error:
    print("ValueError:", error)
````

````python
# La sintaxis correcta de del usa un índice
bts = ["V", "J-Hope", "Suga", "Jungkook"]
del bts[0]
print(bts)

# Usar del como si fuera un método es un error de sintaxis (se compila para capturarlo)
try:
    compile("bts[0].del", "<ejemplo>", "eval")
except SyntaxError as error:
    print("SyntaxError:", error.msg)
````

`pop` devuelve el elemento eliminado, lo que permite guardarlo (pág. 74):

````python
bts = ["V", "J-Hope", "Suga", "Jungkook"]
last_member = bts.pop()   # borra y devuelve el último elemento 'Jungkook'
print(last_member)
print(bts)
````

**Nota sobre el PDF:** la pág. 74 dice "a diferencia del `remove`, borra el último elemento". Es una simplificación: sin argumento `pop()` borra el último, pero con índice `pop(i)` borra el de esa posición, y la verdadera diferencia con `remove` es que `pop` trabaja por posición y devuelve el elemento, mientras que `remove` trabaja por valor y no devuelve nada.

**Clave:** `del` y `pop` actúan por posición; `remove` actúa por valor. Solo `pop` devuelve el elemento eliminado.

**Error común:** llamar a `remove` con un valor ausente. Se evita comprobando antes con `in` (sección 8).

**Para el quiz:** `del` no es un método de lista.

**Referencia:** PDF parte 1, págs. 34–38 y 73–75

## 7. Rebanado (slicing)

El **rebanado** crea una **nueva lista** a partir de una sección de otra. La sintaxis es `lista[inicio:fin]`: incluye `inicio` y llega hasta `fin - 1` (el índice final **no** se incluye). El rebanado no modifica la lista original, devuelve una copia parcial.

```
indice    [0]  [1]  [2]  [3]  [4]  [5]  [6]  [7]
a_list     10   20   30   40   50   60   70   80
a_list[1:5]     20   30   40   50
```

````python
a_list = [10, 20, 30, 40, 50, 60, 70, 80]
print(a_list[1:5])   # del índice 1 al 4
print(a_list[0:5])   # del índice 0 al 4
print(a_list[1:])    # del índice 1 hasta el final (se omite el segundo índice)
print(a_list[:5])    # desde el inicio hasta el índice 4 (se omite el primero)
print(a_list[:])     # toda la lista
````

### Resumen de sintaxis (pág. 45)

| Sintaxis | Función |
|---|---|
| `a_list[inicio:fin]` | desde `inicio` hasta `fin - 1` |
| `a_list[inicio:]` | desde `inicio` hasta el final |
| `a_list[:fin]` | desde el inicio hasta `fin - 1` |
| `a_list[:]` | toda la lista |
| `a_list[inicio:fin:paso]` | saltando de `paso` en `paso` |
| `a_list[-2:]` | los dos últimos elementos |
| `a_list[:-2]` | todo excepto los dos últimos |
| `a_list[::-1]` | todos los elementos en orden inverso |
| `a_list[1::-1]` | los dos primeros elementos en orden inverso |

**Nota sobre el PDF:** en la tabla de la pág. 45 aparecen dos filas con `a_list[-2:]`. La segunda ("todas las partidas desde el principio excepto las dos últimas") corresponde en realidad a `a_list[:-2]`, como confirman las págs. 50–52.

````python
a_list = [10, 20, 30, 40, 50, 60, 70, 80]
print(a_list[-2:])     # dos últimos
print(a_list[:-2])     # todos menos los dos últimos
print(a_list[::-1])    # orden inverso
print(a_list[1::-1])   # los dos primeros, invertidos
print(a_list[::2])     # de dos en dos
````

### Rebanado con índices negativos (págs. 49–52)

````python
a_list = [10, 20, 30, 40, 50, 60, 70, 80]
# índices negativos:  -8  -7  -6  -5  -4  -3  -2  -1
print(a_list[-7:-2])   # desde -7 hasta -3 (el -2 no se incluye)
print(a_list[-7:])     # desde -7 hasta el final
print(a_list[:-2])     # desde el inicio hasta -3
````

### Ejemplos con letras (págs. 76–78)

````python
letters = ['A', 'B', 'C', 'D', 'E', 'F']
print(letters[2:5])    # del tercer al quinto elemento
print(letters[:3])
print(letters[3:])
print(letters[:])      # "los dos puntos" se refieren de principio a fin
print(letters[::2])    # lee de principio a fin saltando con el paso
print(letters[::-1])   # lee de atrás hacia adelante
````

**Clave:** `lista[inicio:fin]` incluye `inicio` y excluye `fin`. El rebanado devuelve una lista nueva y deja intacta la original.

**Error común:** creer que `a_list[1:5]` incluye el índice 5. Llega hasta el 4.

**Para el quiz:** `a_list[1:5]` con `[10, ..., 80]` da `[20, 30, 40, 50]`; `a_list[::-1]` invierte la lista.

**Referencia:** PDF parte 1, págs. 39–52 y 76–78

## 8. Operadores `in` y `not in`

`in` devuelve `True` si el elemento está en la secuencia y `False` si no; `not in` devuelve lo contrario. Funcionan con cadenas, listas y tuplas. Son útiles para evitar errores: antes de `remove(10)`, se comprueba con `in` que el elemento exista.

````python
a_list = [10, 20, 30, 40]
print(10 in a_list)
print(50 in a_list)
print(10 not in a_list)
print(50 not in a_list)
````

````python
for n in a_list:          # in también se usa para recorrer una lista con for
    print(n, end=' ')
print()
````

````python
n_list = [11, 22, 33, 44, 55, 66]
print(88 in n_list)   # 88 no está en la lista
print(55 in n_list)   # 55 sí está
````

Uso defensivo antes de borrar (pág. 55):

````python
n_list = [11, 22, 33, 44, 55, 66]
if 55 in n_list:         # si 55 es un elemento de la lista
    n_list.remove(55)    # se borra
if 88 in n_list:         # 88 no está: no se ejecuta remove y no hay error
    n_list.remove(88)
print(n_list)
````

````python
bts = ["V", "J-Hope", "Suga", "Jungkook"]
if 'Suga' in bts:
    bts.remove('Suga')
print(bts)
````

**Clave:** `in` convierte `remove` en una operación segura.

**Error común:** llamar a `remove` sin comprobar antes que el elemento existe.

**Para el quiz:** `50 not in [10, 20, 30, 40]` es `True`.

**Referencia:** PDF parte 1, págs. 53–55 y 73

## 9. Ordenar y otros métodos de lista

`sort()` ordena la lista **en el mismo lugar** (modifica la lista) en orden ascendente; con la palabra clave `reverse=True` ordena en orden descendente (págs. 79–80).

````python
list1 = [20, 10, 40, 50, 30]
list1.sort()                   # orden ascendente
print(list1)

list1.sort(reverse=True)       # orden descendente
print(list1)
````

### Métodos que ofrece la lista (pág. 81)

| Método | Función |
|---|---|
| `index(x)` | busca la posición del elemento `x` |
| `append(x)` | agrega `x` al final |
| `count(x)` | cuenta cuántas veces aparece `x` |
| `extend([x1, x2])` | inserta los elementos de la lista dada |
| `insert(índice, x)` | agrega `x` en la posición indicada |
| `remove(x)` | elimina el elemento `x` |
| `pop(índice)` | elimina y devuelve el elemento del índice (sin índice, el último) |
| `sort()` | ordena (con `reverse=True`, en descendente) |
| `reverse()` | invierte el orden de los elementos |

**Nota sobre el PDF:** la tabla está traducida de forma literal y aparecen los nombres `insertar`, `eliminar`, `ordenar()` e `invertir()`. Los métodos reales de Python son `insert`, `remove`, `sort()` y `reverse()`.

````python
letras = ['b', 'a', 'c', 'a']
print(letras.index('c'))    # posición de 'c'
print(letras.count('a'))    # veces que aparece 'a'
letras.reverse()            # invierte la lista en el lugar
print(letras)
````

**Ampliación (no está en el PDF):** `sorted(lista)` devuelve una lista ordenada **nueva** sin modificar la original, a diferencia de `lista.sort()`. Además, `sort(key=len)` ordena por longitud (se usa en la programación en pareja, pág. 87).

````python
numeros = [3, 1, 2]
nuevos = sorted(numeros)
print(numeros, nuevos)

palabras = ['bcdefg', 'abc', 'abba']
palabras.sort(key=len)      # key=len ordena por longitud de cada cadena
print(palabras)
````

**Clave:** `sort()` modifica la lista y devuelve `None`.

**Error común:** escribir `lista = lista.sort()`; deja `lista` con el valor `None`.

**Para el quiz:** `reverse=True` produce orden descendente.

**Referencia:** PDF parte 1, págs. 79–81

## 10. Tuplas: datos inmutables

Los tipos de datos de Python se clasifican en **inmutables** (no cambian una vez creados) y **mutables** (se pueden modificar). Según el esquema de la pág. 56:

| Inmutables | Mutables |
|---|---|
| `int`, `float`, `str`, `tuple` (tupla) | `list` (lista), `set` (conjunto), `dict` (diccionario) |

**Nota sobre el PDF:** en el esquema de la pág. 56 el conjunto aparece con el texto roto "configur e" (traducción de *set*). Es el tipo `set`, que se ve más adelante en el curso.

Una **tupla** se parece a una lista, pero sus valores no se pueden cambiar una vez asignados. Se crea con paréntesis. Esa simplicidad le da ventajas: estructura sencilla y acceso más rápido que la lista. Hay que elegir el tipo adecuado según la finalidad.

````python
colors = ("red", "green", "blue")   # tupla de colores
print(colors)
numbers = (1, 2, 3, 4, 5)           # tupla de enteros
print(numbers)
````

Intentar cambiar un valor de la tupla produce `TypeError`; también falla `del t1[0]` (pág. 58).

````python
t1 = (1, 2, 3, 4, 5)
try:
    t1[0] = 100
except TypeError as error:
    print("TypeError:", error)

try:
    del t1[0]
except TypeError as error:
    print("TypeError:", error)
````

### Contraste con la lista

### Problema: base de datos personal y edad media

Se guardan los datos de cuatro personas, cada una en una lista con el formato `[nombre, edad, sexo (0 = mujer, 1 = hombre), altura, peso]`. Se unen en una sola lista `lista_personas` y el programa debe calcular la edad media usando **slicing**. Resultado esperado según el PDF: *La edad media es de 26,75 años.* (pág. 13).

La clave es que cada persona ocupa 5 posiciones consecutivas y la edad está siempre en la posición 1 de cada grupo, de modo que `lista_personas[1::5]` selecciona solo las edades (índices 1, 6, 11 y 16).

**Referencia:** PDF parte 1, págs. 14–15

### Código del PDF, corregido

**Nota sobre el PDF:** el código mostrado en las págs. 14 y 16 no funciona tal como está impreso:

- La última línea usa `edad_media`, pero la variable calculada se llama `average_age`, y además tiene un paréntesis de cierre extra y falta el cierre de la cadena (`print('la edad media es' + str(edad_media).)`). Por eso produce un `NameError` o un `SyntaxError`.
- El PDF mezcla identificadores en inglés (`person1`, `person_list`, `n_persons`) con los del plan en español (`persona1`, `lista_personas`, `n_personas`).
- El nombre de la segunda persona es `'John Smith'` en el código y `'Juan Pérez'` en la pág. 13.
- El resultado se muestra con coma decimal ("26,75") por la traducción; Python imprime `26.75`.

A continuación, la versión corregida con los identificadores de la pág. 13 y comentarios paso a paso:

````python
persona1 = ['David Doe', 20, 1, 180.0, 100.0]
persona2 = ['Juan Pérez', 25, 1, 170.0, 70.0]
persona3 = ['Jane Carter', 22, 0, 169.0, 60.0]
persona4 = ['Peter Kelly', 40, 1, 150.0, 50.0]

# Una sola lista con todos los datos: 4 personas x 5 columnas = 20 elementos
lista_personas = persona1 + persona2 + persona3 + persona4

# Cada persona ocupa 5 posiciones, así que el número de personas es len / 5
n_personas = int(len(lista_personas) / 5)

suma_edad = 0.0
for edad in lista_personas[1::5]:   # empieza en el índice 1 y salta de 5 en 5: solo las edades
    suma_edad += edad

edad_media = float(suma_edad) / n_personas
print('La edad media es de ' + str(edad_media) + ' años.')
````

Resultado esperado: `La edad media es de 26.75 años.` (20 + 25 + 22 + 40 = 107; 107 / 4 = 26.75).

**Ampliación (no está en el PDF):** el mismo cálculo con las funciones de la sección 4, sin bucle, y encapsulado en una función reutilizable.

````python
def edad_media_de(lista_plana, columnas=5):
    """Calcula la edad media de una lista plana donde la edad está en la posición 1 de cada grupo."""
    edades = lista_plana[1::columnas]
    return sum(edades) / len(edades)

print(f"La edad media es de {edad_media_de(lista_personas)} años.")
````

## Apuntes para estudiar

### Clave

- Lista: varios valores en una variable, entre `[]`, con índices de `0` a `n - 1`; admite tipos distintos.
- Índices negativos: `-1` es el último elemento; el índice máximo positivo es `len(lista) - 1`.
- `len`, `max`, `min`, `sum` y `any` operan sobre listas.
- Agregar: `+`, `append` (uno al final), `extend` (varios), `insert(i, x)` (en una posición).
- Borrar: `del lista[i]` (por posición), `pop(i)` (por posición, devuelve el elemento), `remove(x)` (por valor).
- Rebanado `lista[inicio:fin:paso]`: excluye `fin`, admite negativos y devuelve una copia parcial.
- `in` / `not in` verifican pertenencia; se usan antes de `remove`.
- `sort()` ordena en el lugar; `reverse=True` ordena en descendente.
- Tupla: como la lista pero inmutable, se crea con paréntesis.
- Inmutables: `int`, `float`, `str`, `tuple`. Mutables: `list`, `set`, `dict`.

### Errores comunes

- `IndexError` al usar un índice mayor que `len(lista) - 1`.
- `ValueError` al usar `remove` con un valor que no está.
- Confundir `append([a, b])` (agrega una lista anidada) con `extend([a, b])`.
- Creer que `a_list[1:5]` incluye el índice 5.
- `TypeError` al intentar modificar un elemento de una tupla.
- Escribir `lista = lista.sort()`, que deja `None`.
- Usar `del` como método (`lista[0].del`): es una instrucción, produce `SyntaxError`.
