## Tipos de datos de secuencia

La lista, la tupla, el rango y la cadena tienen algo en común: guardan **varios valores ordenados en una secuencia**. En Python se llaman *tipos de datos de secuencia*. Un objeto creado a partir de uno de ellos es un *objeto secuencia*, y cada valor que contiene se llama *elemento*.

Como comparten estructura, comparten también operaciones: los operadores `in` y `not in`, `+`, `*`, los métodos de recuento, la indexación, el corte y la función `len`. Esta unidad recorre esas operaciones una por una.

El PDF ilustra los cuatro tipos: la cadena `'Hello World!'` (cada carácter es un elemento), una lista de notas (`87, 84, 95, ...`) y un `range`.

````python
cadena = "Hello World!"
lista = [87, 84, 95, 67, 88, 94, 63]
tupla = (11, 22, 33)
rango = range(5)

for nombre, objeto in [("cadena", cadena), ("lista", lista), ("tupla", tupla), ("rango", rango)]:
    print(f"{nombre:7} tipo={type(objeto).__name__:6} elementos={len(objeto)}")
````

**Ampliación (no está en el PDF):** se puede comprobar que un objeto es una secuencia con `collections.abc.Sequence`. Los conjuntos y los diccionarios no lo son, porque no tienen un orden por posición.

````python
from collections.abc import Sequence

for objeto in ["texto", [1, 2], (1, 2), range(3), {1, 2}, {"a": 1}]:
    print(f"{str(objeto):12} -> es secuencia: {isinstance(objeto, Sequence)}")
````

**Clave:** cadena, lista, tupla y `range` son secuencias. Lo que se aprende para uno sirve, con pocas excepciones, para los demás.

## Búsqueda de un valor específico: `in` y `not in`

Los **operadores de pertenencia** devuelven `True` o `False`. Sirven para comprobar si un elemento está incluido en una cadena, una lista, una tupla o un rango.

- `x in secuencia` devuelve `True` si `x` está presente y `False` en caso contrario.
- `x not in secuencia` devuelve `False` si `x` está presente y `True` en caso contrario.

Ejemplos del PDF:

````python
list1 = [10, 20, 30, 40]
print(10 in list1)       # True: 10 es un elemento de list1
print(10 not in list1)   # False
````

````python
tup = (1, 2, 3, 4)
print(3 in tup)          # buscar en una tupla
print(11 in range(10))   # buscar en un rango: range(10) llega hasta 9
print('a' in 'abcd')     # buscar en una cadena: 'a' está incluida en 'abcd'
````

**Ampliación (no está en el PDF):** con cadenas, `in` también busca **subcadenas**, no solo caracteres sueltos. En listas y tuplas, en cambio, solo compara elementos completos.

````python
print('bc' in 'abcd')      # True: subcadena
print('ac' in 'abcd')      # False: no es contigua
print(2 in [[1, 2], [3]])  # False: 2 no es elemento de la lista externa
print([1, 2] in [[1, 2], [3]])  # True
````

**Clave:** `in` responde "¿está?" con un booleano; no indica dónde.

**Error común:** buscar un elemento anidado con `in` en la secuencia externa. `2 in [[1, 2], [3]]` es `False` porque los elementos de la lista externa son `[1, 2]` y `[3]`.

**Para el quiz:** `11 in range(10)` es `False`, porque `range(10)` produce de 0 a 9.

## Vinculación de objetos de secuencia: operador `+`

El operador `+` **enlaza** (concatena) dos objetos secuencia del mismo tipo y produce uno nuevo, con los elementos del primero seguidos de los del segundo. Funciona con listas, tuplas y cadenas. **No funciona con `range`.**

````python
# Enlazar listas
list1 = [11, 22, 33, 44]
list2 = [55, 66]
print(list1)
print(list1 + list2)

# Enlazar tuplas
tup1 = (1, 2, 3)
tup2 = (4, 5, 6)
print(tup1 + tup2)

# Enlazar cadenas
str1 = 'hello '
str2 = 'world'
print(str1 + str2)
````

### Posible error al enlazar objetos de secuencia

Entre los tipos de secuencia, el operador `+` no puede enlazar objetos `range`: produce `TypeError`. La solución del PDF es convertir el rango a lista o a tupla antes de enlazar.

````python
try:
    range(10) + range(10, 20)
except TypeError as error:
    print("TypeError:", error)

print(list(range(10)) + list(range(10, 20)))
print(tuple(range(10)) + tuple(range(10, 20)))
````

**Ampliación (no está en el PDF):** tampoco se pueden mezclar tipos distintos, por ejemplo una lista con una tupla. Hay que convertir uno de los dos.

````python
try:
    [1, 2] + (3, 4)
except TypeError as error:
    print("TypeError:", error)

print([1, 2] + list((3, 4)))
````

**Clave:** `+` crea un objeto nuevo; no modifica los originales.

**Error común:** intentar `range(a) + range(b)`. Se convierte primero con `list(...)` o `tuple(...)`.

**Para el quiz:** `(1, 2, 3) + (4, 5, 6)` da `(1, 2, 3, 4, 5, 6)`.

**Referencia:** PDF parte 1, págs. 163–165

## Iteración de objetos de secuencia: operador `*`

Los tipos de secuencia aceptan el operador de multiplicación, que **repite** los elementos del objeto tantas veces como indique un entero. Se escribe `[tipo de dato de secuencia] * entero`.

````python
list1 = [11, 22, 33, 44]
print(list1 * 2)

tup1 = (1, 2, 3)
print(tup1 * 2)

str2 = 'hello'
print(str2 * 3)
````

### Precauciones al iterar con `*`

1. El tipo `range` no admite `*`: produce `TypeError`. Se convierte a lista o a tupla antes de repetir.
2. No se puede multiplicar una secuencia por otra secuencia (por ejemplo `list1 * list2`). El segundo operando debe ser un **entero**.

````python
# range no se puede repetir con *
try:
    range(10) * 3
except TypeError as error:
    print("TypeError:", error)

# Solución: convertir el rango antes de repetir
ran = list(range(5)) * 3
print(ran)
ran = tuple(range(5)) * 3
print(ran)

# Dos secuencias no se pueden multiplicar entre sí
list1 = [11, 22, 33, 44]
list2 = [55, 66]
try:
    print(list1 * list2)
except TypeError as error:
    print("TypeError:", error)
````

**Ampliación (no está en el PDF):** `*` es la forma habitual de inicializar una lista con un valor repetido. Cuidado cuando el elemento repetido es mutable: se repite la **misma referencia**, no copias. Esto se retoma en la Unidad 13 al hablar de listas bidimensionales.

````python
ceros = [0] * 5
print(ceros)

filas = [[0]] * 3     # tres referencias a la MISMA lista interna
filas[0].append(1)
print(filas)          # el cambio aparece en las tres posiciones
````

**Clave:** la multiplicación solo acepta `secuencia * entero`.

**Error común:** `list1 * list2` y `range(n) * k` fallan con `TypeError`.

**Para el quiz:** `'hello' * 3` da `'hellohellohello'`.

**Referencia:** PDF parte 1, págs. 166–170

## Cálculo del número de elementos específicos: método `count`

`secuencia.count(valor)` devuelve cuántas veces aparece `valor` en la secuencia. No confundir con `len`, que cuenta **todos** los elementos.

````python
list1 = [11, 11, 11, 22, 33, 44]
print(list1.count(11))     # Número de elementos específicos: 3

tup1 = (11, 11, 11, 22, 33, 44)
print(tup1.count(11))      # tres 11 en tup1

str1 = 'hello world'
print(str1.count('l'))     # tres 'l' en str1
````

Ejemplo del PDF que compara `len` y `count` con un objeto `range`:

````python
ran = range(0, 5, 1)
print(len(ran))        # 5: ran tiene cinco elementos (0, 1, 2, 3, 4)
print(ran.count(2))    # 1: solo hay un 2
````

**Ampliación (no está en el PDF):** para saber **dónde** está un elemento se usa `index`, que devuelve la posición de la primera aparición y lanza `ValueError` si no existe.

````python
letras = ['a', 'b', 'c', 'b']
print(letras.index('b'))     # 1: primera aparición
try:
    letras.index('z')
except ValueError as error:
    print("ValueError:", error)
````

**Clave:** `count` cuenta apariciones de un valor; `len` cuenta todos los elementos.

**Para el quiz:** `ran = range(0, 5, 1)` tiene `len(ran) == 5` y `ran.count(2) == 1`.

**Referencia:** PDF parte 1, págs. 171–173

## Utilización del índice de objetos secuenciales

Un **índice** es un número que indica la posición de un elemento dentro de una lista o secuencia. En una secuencia de `n` elementos, los índices van de **0 a n-1**. El primer elemento siempre tiene el índice 0.

Se accede con corchetes: `secuencia[indice]`.

````python
# Lista
list1 = [11, 11, 11, 22, 33, 44]
print(list1[0])    # primer elemento
print(list1[2])

# Cadena
str1 = 'hello world'
print(str1[2])
print(str1[7])

# Tupla
tup1 = (1, 1, 1, 2, 3, 3, 4)
print(tup1[3])

# Rango: ran[4] es el quinto elemento
ran = range(0, 5, 1)
print(ran[4])
````

**Nota sobre el PDF:** en el diagrama de la pág. 174, el objeto `rango(5)` se dibuja con seis casillas (0 a 5). En realidad `range(5)` tiene cinco elementos, de 0 a 4, como muestra el ejemplo `range(0, 5, 1)` de la pág. 176.

**Ampliación (no está en el PDF):** usar un índice fuera de rango lanza `IndexError`.

````python
try:
    [10, 20, 30][3]
except IndexError as error:
    print("IndexError:", error)
````

**Clave:** el primer elemento es el índice 0; el último de una secuencia de `n` elementos es `n-1`.

**Error común:** pensar que el índice empieza en 1.

**Referencia:** PDF parte 1, págs. 174–176

## Corte (slicing) de objetos de secuencia

El corte extrae una porción de la secuencia indicando un índice de inicio y uno de fin: `secuencia[inicio:fin]`. Trae los elementos desde el índice `inicio` hasta el índice `fin - 1`; el índice `fin` **no se incluye**. Todos los tipos de secuencia usan la misma regla.

- `[1:5]` trae los elementos de los índices 1 a 4.
- `[1:]` omite el último índice y llega hasta el final.
- `[:]` omite ambos y trae todos los elementos.

````python
datos = [10, 20, 30, 40, 50, 60, 70, 80]
print(datos[1:5])   # [20, 30, 40, 50]
print(datos[1:])    # [20, 30, 40, 50, 60, 70, 80]
print(datos[:])     # [10, 20, 30, 40, 50, 60, 70, 80]
````

**Ampliación (no está en el PDF):** el corte admite un tercer valor, el *paso*: `[inicio:fin:paso]`. Además, a diferencia de la indexación, un corte fuera de rango **no** lanza error: devuelve lo que exista.

````python
datos = [10, 20, 30, 40, 50, 60, 70, 80]
print(datos[::2])      # de dos en dos
print(datos[::-1])     # invertida
print(datos[5:100])    # no falla: [60, 70, 80]
print("abcdef"[1:4])   # el corte funciona igual con cadenas
````

**Clave:** el índice final del corte es exclusivo. `datos[a:b]` tiene `b - a` elementos.

**Error común:** esperar que `[1:5]` incluya el elemento del índice 5.

**Para el quiz:** `[10, 20, 30, 40, 50, 60, 70, 80][1:5]` es `[20, 30, 40, 50]`.

**Referencia:** PDF parte 1, págs. 177–179

## Indexación negativa y corte con índices negativos

Los índices negativos cuentan **desde el final**: el último elemento es `-1`, el penúltimo `-2`, `-3`, y así hacia adelante. Para una secuencia de `n` elementos, el índice `-k` equivale al índice `n - k`.

Para la lista `[10, 20, 30, 40, 50, 60, 70, 80]`:

| Elemento | 10 | 20 | 30 | 40 | 50 | 60 | 70 | 80 |
|---|---|---|---|---|---|---|---|---|
| Índice | 0 | 1 | 2 | 3 | 4 | 5 | 6 | 7 |
| Índice negativo | -8 | -7 | -6 | -5 | -4 | -3 | -2 | -1 |

La regla funciona igual con cadenas, tuplas y rangos.

````python
# Cadena
str1 = 'abcdef'
print(str1[-1])    # f
print(str1[-6])    # a

# Tupla
tup1 = (11, 22, 33, 44, 55, 66, 77)
print(tup1[-1])    # 77
print(tup1[-6])    # 22

# Rango
ran = range(1, 7)
print(ran[-6])     # 1
print(ran[-1])     # 6
````

### Corte con índices negativos

Los cortes también aceptan índices negativos y se pueden omitir el inicio o el final.

- `[-7:-2]` trae `[20, 30, 40, 50, 60]` (del índice -7 al -3).
- `[-7:]` llega hasta el último elemento: `[20, 30, 40, 50, 60, 70, 80]`.
- `[:-2]` importa desde el primero hasta el índice -3: `[10, 20, 30, 40, 50, 60]`.

````python
datos = [10, 20, 30, 40, 50, 60, 70, 80]
print(datos[-7:-2])   # [20, 30, 40, 50, 60]
print(datos[-7:])     # [20, 30, 40, 50, 60, 70, 80]
print(datos[:-2])     # [10, 20, 30, 40, 50, 60]
````

**Clave:** `secuencia[-1]` es el último elemento sin necesidad de conocer la longitud.

**Error común:** creer que `[-7:-2]` incluye el elemento del índice -2. El final es exclusivo también con índices negativos.

**Para el quiz:** `[:-2]` de `[10, ..., 80]` termina en `60`.

**Referencia:** PDF parte 1, págs. 180–186

## Aplicaciones de `range`, `len` y el código ASCII

### `range` convertido en lista

`range(inicio, fin, paso)` genera una secuencia de números; `fin` no se incluye y el `paso` es opcional. Con `list(...)` se obtiene una lista con esos números.

Ejemplo del PDF: crear `even_list` con los números pares entre 1 y 10 (incluido el 10).

````python
even_list = list(range(2, 11, 2))
print('even_list =', even_list)   # el último argumento es el paso: 2, 4, 6, 8, 10

print(list(range(10)))
print(list(range(2, 10)))
print(list(range(2, 10, 3)))
````

### `len` y el último elemento

`len(secuencia)` devuelve el número de elementos. El último índice es `len(secuencia) - 1`, de modo que `secuencia[len(secuencia) - 1]` es el último elemento. Ejemplo del PDF con la lista `naciones`:

````python
naciones = ['Korea', 'China', 'Rusia', 'Malasia']
print('último elemento de naciones :', naciones[len(naciones) - 1])
print('equivale a naciones[-1]     :', naciones[-1])
````

### Código ASCII y comparación de tuplas

Una cadena se compara por el **código ASCII** de sus caracteres. `ord(c)` devuelve el código de un carácter. Con tuplas, la comparación se hace elemento por elemento: se comparan los primeros; si son iguales se pasa a los segundos, y así hasta encontrar una diferencia. En el ejemplo, `a = ('A', 'B', 'C')` y `b = ('A', 'B', 'D')` coinciden en los dos primeros elementos; el tercero decide: `ord('C')` es 67 y `ord('D')` es 68, así que `a < b`.

````python
a = ('A', 'B', 'C')
b = ('A', 'B', 'D')
print(ord('C'))
print(ord('D'))
print(a > b)    # False
print(a < b)    # True
````

**Un paso más (según el PDF):** ASCII es la abreviatura de *American Standard Code for Information Interchange*. Es un código de siete bits que asigna 128 números a alfabetos, números, letras especiales y caracteres de control, para que distintos programas y ordenadores se comuniquen con la misma representación numérica de los caracteres.

**Ampliación (no está en el PDF):** `chr(n)` hace la operación inversa de `ord`.

````python
print(ord('A'), ord('a'))   # las mayúsculas van antes que las minúsculas
print(chr(67))
````

**Clave:** `range` + `list` es la forma rápida de generar listas de números; `len(x) - 1` es el último índice.

**Error común:** `naciones[len(naciones)]` lanza `IndexError`; el último índice es `len - 1`.

**Para el quiz:** `list(range(2, 10, 3))` es `[2, 5, 8]`.

## La tupla: un objeto inmutable entre los objetos de secuencia

Una **tupla** es un tipo de colección con múltiples elementos. A diferencia de la lista, **no se puede cambiar** la secuencia de elementos una vez creada: es *inmutable*. No se pueden modificar ni eliminar sus elementos.

### Cómo crear una tupla

| Qué se crea | Código |
|---|---|
| Tupla vacía | `tuple0 = tuple()` (se deben usar paréntesis vacíos) |
| Tupla con un elemento | `tuple1 = (1,)` (debe llevar coma) |
| Tupla básica con paréntesis | `tupla2 = (1, 2, 3, 4)` |
| Tupla simple (sin paréntesis) | `tupla3 = 1, 2, 3, 4` |
| Tupla a partir de una lista | `n_list = [1, 2, 3, 4]` y `tupla4 = tuple(n_list)` |

````python
tuple0 = tuple()
tuple1 = (1,)
tupla2 = (1, 2, 3, 4)
tupla3 = 1, 2, 3, 4
n_list = [1, 2, 3, 4]
tupla4 = tuple(n_list)

for t in (tuple0, tuple1, tupla2, tupla3, tupla4):
    print(t, type(t).__name__)
````

**Nota sobre el PDF:** en la tabla de la pág. 196 la traducción automática cambió el código a `tupla4 = tupla(n_lista)`. El código válido es `tupla4 = tuple(n_list)`: `tuple` es el nombre de la función en Python y no se traduce.

### Tupla de un elemento: `(100)` frente a `(100,)`

Si se escribe `tup = (100)`, Python lo interpreta como el entero `100` entre paréntesis y `tup` es un `int`. Para crear una tupla de un elemento hay que añadir la **coma**: `tup = (100,)`.

````python
tup = (100)
print(tup)
print(type(tup))

tup = (100,)
print(tup)
print(type(tup))
````

### Diferencias entre tupla y lista

Los valores internos de una tupla no se pueden modificar. Asignar un elemento individual produce `TypeError`. Esta propiedad se llama **inmutabilidad**.

````python
t = (0, 1, 2, 3, 4)
try:
    t[0] = 100
except TypeError as error:
    print("TypeError:", error)

# La lista, en cambio, sí admite la asignación
lista = [0, 1, 2, 3, 4]
lista[0] = 100
print(lista)
````

### Empaquetado y desempaquetado

- **Empaquetar** (*packing*) es juntar varios valores en una variable: `a = (1, 2)`.
- **Desempaquetar** (*unpacking*) es tomar varios valores de una variable empaquetada y asignarlos a varias variables: `x, y = c`.

````python
# Empaquetado
a = (1, 2)
print(a[0])    # referencia a un elemento de la tupla
print(a[1])

# Desempaquetado
c = (3, 4)
x, y = c       # asigna a 2 variables la tupla c
print(x)
print(y)
````

### Intercambiar valores (swap)

En lenguajes como C o Java se necesita una variable temporal y tres asignaciones. En Python el empaquetado y desempaquetado permiten hacerlo en una sola línea: `a, b = b, a`.

````python
# Con variable temporal (estilo C o Java)
a = 100
b = 200
print('antes del intercambio : a =', a, 'b =', b)
temp = a
a = b
b = temp
print('después del intercambio : a =', a, 'b =', b)

# Con una tupla: una sola línea
a = 100
b = 200
print('antes del intercambio : a =', a, 'b =', b)
a, b = b, a    # intercambio muy sencillo
print('intercambio usando tupla: a =', a, 'b =', b)
````

### Cómo ordenar tuplas

Como los elementos de una tupla no se pueden modificar, **no existe** el método `sort` en tuplas. Para ordenar, se convierte la tupla en lista, se ordena la lista y, si hace falta, se vuelve a convertir.

````python
tup = (1, 2, 5, 4, 3, 2, 9, 3, 7, 3, 9)
temp = list(tup)
temp.sort()
print(temp)
````

**Ampliación (no está en el PDF):** la función `sorted` acepta cualquier secuencia y devuelve una lista ordenada, sin pasos intermedios.

````python
tup = (1, 2, 5, 4, 3, 2, 9, 3, 7, 3, 9)
print(sorted(tup))
print(tuple(sorted(tup)))
````

**Clave:** la tupla es inmutable; `(100,)` con coma es tupla de un elemento; `a, b = b, a` intercambia valores.

**Error común:** escribir `(100)` creyendo que es una tupla, o intentar `tupla.sort()` o `tupla[0] = x`.

**Para el quiz:** el tipo de `(100)` es `int`; el de `(100,)` es `tuple`.

### Calculadora del porcentaje de envejecimiento

**Problema.** Existen dos tuplas que describen la composición de la población de las ciudades A y B. Cada una tiene 11 elementos: habitantes de 0 a 9 años, de 10 a 19, y así hasta el último, de 100 años o más.

```python
población_a = (100, 150, 230, 120, 180, 100, 140, 95, 81, 21, 4)
población_b = (300, 420, 530, 420, 400, 300, 40, 5, 1, 1, 1)
```

Se define el **grado de envejecimiento** como la proporción de personas de 70 años o más (índice 7 en adelante) sobre el total. Hay que escribir un programa que lo calcule para cada ciudad usando el **corte**.

**Plan de programación (pseudocódigo del PDF).**

1. Inicio.
2. Introducir los datos de población de la ciudad.
3. Usar el corte para tomar solo a los ancianos de 70 años o más (`población[7:]`).
4. Sumar la población total de la ciudad.
5. Dividir la población anciana entre la población total.
6. Imprimir el porcentaje de envejecimiento.
7. Fin.

El diagrama de flujo del PDF es lineal: `población` -> `oldA = suma(población_a[7:])` -> `sumA = suma(población_a)` -> `oldRateA = oldA / sumA` -> imprimir.

**Código final.** Los nombres de variable son los del PDF, con los identificadores sin tilde porque Python los admite pero es más cómodo evitarlos.

````python
population_a = (100, 150, 230, 120, 180, 100, 140, 95, 81, 21, 4)
population_b = (300, 420, 530, 420, 400, 300, 40, 5, 1, 1, 1)

# [3] Cortar de 70 años en adelante (índice 7) y sumar ese grupo
oldA = sum(population_a[7:])
oldB = sum(population_b[7:])

# [4] Sumar la población total de cada ciudad
sumA, sumB = sum(population_a), sum(population_b)

# [5] Dividir la población anciana entre el total
oldRateA, oldRateB = oldA / sumA, oldB / sumB

# [6] Imprimir el resultado
print('Los grados de envejecimiento en la ciudad A y B son {:5.3f} y {:5.3f} respectivamente.'.format(oldRateA, oldRateB))
````

**Resultado esperado:** el PDF indica que los grados de envejecimiento de las ciudades A y B son 0,165 y 0,003 respectivamente. La salida de la celda anterior coincide.

**Nota sobre el PDF:** en el código de la pág. 156 (y 158) la sentencia `print(...)` usa `.format(gradoantiguoA, gradoantiguoB)`, nombres que no existen: las variables calculadas son `oldRateA` y `oldRateB`. Ejecutado tal cual, daría `NameError`. Además, la traducción del texto cambió algunos nombres (`población_a` frente a `population_a`). Arriba se muestra el código corregido.

**Ampliación (no está en el PDF):** la misma lógica empaquetada en una función reutilizable, con el índice de corte como parámetro:

````python
def grado_envejecimiento(poblacion, desde=7):
    """Proporción de la población a partir del grupo de edad con índice 'desde'."""
    return sum(poblacion[desde:]) / sum(poblacion)


print(f"Ciudad A: {grado_envejecimiento(population_a):.3f}")
print(f"Ciudad B: {grado_envejecimiento(population_b):.3f}")
print(f"Ciudad A, 80 años o más: {grado_envejecimiento(population_a, 8):.3f}")
````

## Apuntes para estudiar

**Clave**

- Lista, tupla, rango y cadena son secuencias: comparten `in`, `+`, `*`, `count`, índices, corte y `len`.
- `x in s` y `x not in s` devuelven booleanos. `+` enlaza del mismo tipo; `*` repite con un entero.
- Índices: de 0 a n-1 hacia adelante y de -1 a -n hacia atrás. El corte `[a:b]` excluye `b`.
- `range(inicio, fin, paso)` no incluye `fin`; `list(range(...))` lo convierte en lista.
- `len(s) - 1` es el último índice; `s[-1]` es el último elemento.
- La tupla es inmutable. Con un solo elemento necesita coma: `(100,)`.
- Empaquetado `a = (1, 2)`, desempaquetado `x, y = c`, intercambio `a, b = b, a`.
- Para ordenar una tupla se convierte a lista, se aplica `sort` y se vuelve a tupla.

**Errores comunes**

- `range(a) + range(b)` y `range(n) * k`: `TypeError`. Convertir antes con `list` o `tuple`.
- `list1 * list2`: `TypeError`; el multiplicador debe ser un entero.
- `(100)` es un entero, no una tupla.
- `tupla[0] = valor`: `TypeError`; la tupla es inmutable.
- Creer que el corte incluye el índice final o que el primer índice es 1.
- `secuencia[len(secuencia)]`: `IndexError`.

**Para el quiz**

- `11 in range(10)` es `False`.
- `[11, 22, 33, 44] * 2` repite la lista dos veces; `'hello' * 3` da `hellohellohello`.
- `[10, 20, 30, 40, 50, 60, 70, 80][1:5]` es `[20, 30, 40, 50]`; `[-7:-2]` es `[20, 30, 40, 50, 60]`.
- `list(range(2, 11, 2))` es `[2, 4, 6, 8, 10]`.
- `range(0, 5, 1)`: `len` es 5 y `count(2)` es 1.
- `ord('C')` es 67 y `ord('D')` es 68; por eso `('A','B','C') < ('A','B','D')`.
