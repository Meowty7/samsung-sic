## 1. Qué es un diccionario

Un **diccionario** (`dict`) es un tipo de datos formado por pares **clave-valor**. Se usa mucho porque permite referirse a un valor mediante su clave. Puntos que destaca el PDF (págs. 103–105):

- Cada elemento del diccionario se define con dos puntos: `clave : valor`.
- Un diccionario consta de uno o más elementos, separados por comas y entre llaves `{}`.
- El valor se puede buscar usando la clave, pero **no** se puede usar el valor para buscar la clave.
- Como el diccionario busca el valor mediante una clave, **las claves no pueden duplicarse**.
- Diferencia con la lista: la lista usa índices (`0` a `n - 1`) para referirse a cada elemento; el diccionario usa la clave.

```
Lista                      Diccionario
índice -> elemento         clave -> valor
  0    -> valor 1          clave 1 -> valor 1
  1    -> valor 2          clave 2 -> valor 2
 n-1   -> valor n          clave n -> valor n
```

Una lista se declara con corchetes y un diccionario con llaves: `[]` crea una lista vacía y `{}` un diccionario vacío (pág. 132).

**Referencia:** PDF parte 1, págs. 103–106 y 131–132

## 2. Crear un diccionario y leer valores con la clave

Sintaxis: `nombre = {clave1: valor1, clave2: valor2, ...}`. Para leer un valor se escribe el nombre del diccionario con la clave entre corchetes: `nombre[clave]`.

Ejemplo de la pág. 107: una persona con nombre, edad y peso.

````python
person = {'Name': 'David Doe', 'Age': 26, 'Weight': 82}   # claves de texto
print(person['Name'])
print(person['Age'])
print(person['Weight'])
````

Las claves no tienen que ser texto. En la pág. 108 se usa el número de ID como clave y el nombre como valor:

````python
students = {2019001: 'John Smith', 2019002: 'Jane Carter', 2019003: 'Peter Kelly'}
print(students[2019001])
print(students[2019002])
print(students[2019003])
````

Otra forma de construirlo (págs. 131–133): crear un diccionario vacío y agregar pares uno a uno, o inicializarlo directamente.

````python
phone_book = {}                          # diccionario vacío
phone_book["David"] = "010-1234-5678"
print(phone_book)

phone_book = {"David": "010-1234-5678"}  # creación e inicialización a la vez
phone_book["John"] = "010-1234-5679"
phone_book["Daniel"] = "010-1234-5680"
print(phone_book)
````

**Nota sobre el PDF:** en la pág. 108 el valor de la primera fila de la tabla aparece escrito como `'John Smith` (con una comilla suelta). Es una errata de maquetación: en el código es `'John Smith'`.

**Ampliación (no está en el PDF):** como las claves no pueden duplicarse, si se repite una clave Python conserva solo el último valor. Y acceder a una clave inexistente produce `KeyError`.

````python
duplicado = {'a': 1, 'a': 2}     # la clave 'a' aparece dos veces
print(duplicado)                  # solo queda el último valor

try:
    person['Job']                 # 'Job' no es una clave de person
except KeyError as error:
    print("KeyError:", error)
````

**Clave:** se consulta por clave, no por posición. Las claves son únicas.

**Error común:** intentar buscar una clave a partir de su valor, o repetir claves esperando conservar ambos valores.

**Para el quiz:** `{}` crea un diccionario vacío; `[]`, una lista vacía.

**Referencia:** PDF parte 1, págs. 106–108 y 131–133

## 3. Insertar, modificar y borrar elementos (enfoque clave y asignación)

El PDF llama "enfoque clave y asignación de valor" a usar la clave para leer y también para asignar:

- **Insertar:** `diccionario[clave] = valor` con una clave **nueva**.
- **Modificar:** la misma sintaxis con una clave **existente** reemplaza su valor.
- **Borrar:** `del diccionario[clave]`. Si la clave no existe se produce `KeyError`.

````python
person = {'Name': 'David Doe', 'Age': 26, 'Weight': 82}
person['Job'] = 'Data Scientist'    # clave nueva: se inserta el elemento
print(person)
````

````python
person = {'Name': 'David Doe', 'Age': 26, 'Weight': 82}
person['Age'] = 27                  # clave existente: se modifica el valor
print(person)
````

````python
person = {'Name': 'David Doe', 'Age': 26, 'Weight': 82}
del person['Age']                   # borra el elemento con la clave 'Age'
print(person)
````

````python
person = {'Name': 'David Doe', 'Age': 26, 'Weight': 82}
try:
    del person['Hometown']          # clave no válida
except KeyError as error:
    print("KeyError:", error)
````

### El método `pop` en el diccionario

Los objetivos de la unidad piden aprender a devolver el valor de un elemento con `pop`. El PDF lo compara con la lista (págs. 135–137); la descripción del método como tal no aparece en las diapositivas.

**Ampliación (no está en el PDF):** `diccionario.pop(clave)` elimina el elemento de esa clave y **devuelve su valor**. Si la clave no existe produce `KeyError`, salvo que se indique un valor por defecto como segundo argumento.

````python
person = {'Name': 'David Doe', 'Age': 26, 'Weight': 82}
edad = person.pop('Age')            # elimina el par y devuelve el valor 26
print(edad)
print(person)

print(person.pop('Job', 'sin dato'))   # clave ausente con valor por defecto: no hay error
````

**Clave:** la misma sintaxis `d[clave] = valor` inserta o modifica según exista o no la clave.

**Error común:** usar `del` o leer con una clave que no existe: produce `KeyError`.

**Para el quiz:** `KeyError` es el error al borrar o leer una clave no válida.

**Referencia:** PDF parte 1, págs. 109–112 y 135–137

## 4. Funciones y operadores del diccionario

Funciones y operadores similares a los de la lista (págs. 113–117):

- `len(d)`: número de elementos (pares) del diccionario.
- `clave in d`: `True` si la **clave** está en el diccionario; `not in` devuelve lo contrario.
- `d1 == d2` y `d1 != d2`: comparan si dos diccionarios tienen los mismos elementos.
- Los operadores `>`, `>=`, `<`, `<=` **no** son aplicables: producen `TypeError`.

````python
person = {'Name': 'David Doe', 'Age': 26, 'Weight': 82}
print(len(person))                  # número de elementos
print('Name' in person)             # 'Name' es una clave
print('Job' in person)              # 'Job' no es una clave
````

Atención: `in` solo mira las **claves**. `'David Doe'` es un valor, no una clave, así que `in` devuelve `False` (pág. 115).

````python
print('David Doe' in person)       # es un valor, no una clave
print('David Doe' not in person)
````

La comparación `==` no depende del orden en que se escribieron los pares:

````python
d1 = {'Name': 'David Doe', 'Age': 26}
d2 = {'Age': 26, 'Name': 'David Doe'}
print(d1 == d2)
print(d1 != d2)
````

````python
try:
    d1 > d2                         # los diccionarios no admiten comparación de tamaño
except TypeError as error:
    print("TypeError:", error)
````

**Ampliación (no está en el PDF):** el diccionario ofrece las vistas `keys()` (claves), `values()` (valores) e `items()` (pares). El PDF usa `keys()` y `items()` en los ejemplos del menú y del `pop`, y anuncia que los verá en detalle más adelante. Para buscar entre los **valores** se usa `values()`.

````python
print(person.keys())
print(person.values())
print(person.items())
print('David Doe' in person.values())   # ahora sí se busca entre los valores
````

**Clave:** `in` busca claves, no valores.

**Error común:** usar `valor in diccionario` esperando que busque en los valores.

**Para el quiz:** `d1 == d2` es `True` aunque los pares estén en distinto orden; `d1 > d2` produce `TypeError`.

**Referencia:** PDF parte 1, págs. 113–117

## 5. Lista contra diccionario

Comparación de las págs. 134–139:

| Aspecto | Lista | Diccionario |
|---|---|---|
| Cómo se refiere un valor | índice (empieza en 0, automático) | clave (la define quien programa) |
| Crear vacío | `lst = []` | `dic = {}` |
| Crear con elementos | `lst = [1, 2, 3]` | `dic = {'uno': 1, 'dos': 2}` |
| Leer un valor | `lst[0]` | `dic['uno']` |
| Cantidad de elementos | `len(lst)` | `len(dic)` |
| Borrar un elemento | `del lst[0]` | `del dic['uno']` |
| Borrar todo | `lst.clear()` | `dic.clear()` |
| Comprobar pertenencia | `2 in lst` (valores) | `'uno' in dic` o `'uno' in dic.keys()` (claves) |

**Nota sobre el PDF:** en la pág. 139 la tabla escribe `lst.clear`, `dic.clear` y `dic.keys` sin paréntesis, y las traduce `2 en lst`. En Python los métodos se llaman con paréntesis (`clear()`, `keys()`) y el operador es `in`. Sin paréntesis, `lst.clear` solo hace referencia al método y no lo ejecuta.

````python
lst = [11, 22, 33, 44, 55]
dic = {0: 11, 1: 22, 2: 33, 3: 44, 4: 55}    # mismas posiciones, pero ahora son claves
print(lst[0], dic[0])
print(len(lst), len(dic))
````

````python
lst = [1, 2, 3]
dic = {'uno': 1, 'dos': 2}
print(2 in lst)               # valor en la lista
print('uno' in dic.keys())    # clave en el diccionario
print('uno' in dic)           # equivalente más corto

lst.clear()                   # vacía la lista
dic.clear()                   # vacía el diccionario
print(lst, dic)
````

### Diferencia clave al usar `pop(0)` (págs. 135–137)

En la lista, al borrar un elemento los demás **se desplazan** y sus índices cambian. En el diccionario, cada valor conserva su clave: no se desplaza nada.

````python
lst = [11, 22, 33, 44, 55]
print('antes de pop(0) :', lst)
print('antes de pop(0) lst[1] =', lst[1])
lst.pop(0)                    # el índice 0 es el primer elemento
print('después de pop(0) :', lst)
print('después de pop(0) lst[1] =', lst[1])   # el valor referido por lst[1] cambia
````

````python
dic = {0: 11, 1: 22, 2: 33, 3: 44, 4: 55}
print('antes de pop(0) :', dic.items())     # items() devuelve los pares (clave, valor)
print('antes de pop(0) dic[1] =', dic[1])
dic.pop(0)                    # aquí 0 es una CLAVE: se borra el par (0, 11)
print('después de pop(0) :', dic.items())
print('después de pop(0) dic[1] =', dic[1])  # el valor referido por dic[1] no cambia
````

**Nota sobre el PDF:** en la pág. 136 el resultado impreso de la lista muestra `después de pop(0) list[1] = 22`, pero ejecutando el código el valor es `33` (tras borrar el 11, la lista queda `[22, 33, 44, 55]`). El propio comentario del PDF dice que "el valor referido por `lst[1]` cambia", así que `22` es una errata. En la pág. 135 también hay un error tipográfico en el diccionario resultante (`4:55. 5:66`): debe ser `{1:22, 2:33, 3:44, 4:55}`.

**Clave:** lista = acceso por posición (los índices se reacomodan); diccionario = acceso por clave (cada valor conserva la suya).

**Error común:** creer que `dic.pop(0)` elimina el primer elemento por posición: elimina la clave `0`.

**Para el quiz:** tras `lst.pop(0)` el valor de `lst[1]` cambia; tras `dic.pop(0)` el de `dic[1]` no.

**Referencia:** PDF parte 1, págs. 134–139

## 6. Formato de impresión con `format`

Hasta ahora las impresiones no tenían una separación uniforme. El método de cadenas `format` coloca valores en marcadores de posición `{}` y permite controlar el ancho y los decimales. Se usa en el proyecto de esta unidad.

**Referencia:** PDF parte 1, págs. 140–143

### Marcadores de posición e índices (págs. 140–141)

Cada `{}` se reemplaza por un argumento de `format`, en orden. Con `{0}`, `{1}`, `{2}` se elige qué argumento se usa y se puede repetir.

````python
print('{} Python!'.format('Hello'))
print('I like {} and {}'.format('Python', 'Java'))
````

````python
print('I like {0} and {1}'.format('Python', 'Java'))
print('I like {1} and {0}'.format('Python', 'Java'))
print('I like {0}, {0}, {0}'.format('Python', 'Java'))
````

### Ancho de campo en enteros: `{0:4d}`

`d` es "entero". El número antes de `d` es el ancho total del campo; el número se alinea a la derecha. Con 4, 5 y 6 posiciones:

````python
print('12345678901234567890')       # regla para contar las posiciones
print('{0:4d},{0:5d},{0:6d}'.format(123))
````

### Decimales y ancho en reales: `{0:.2f}` y `{0:10.3f}`

`f` es "real". Con `.2f` se muestran 2 decimales; con `10.3f` el campo mide 10 posiciones en total y lleva 3 decimales.

````python
print('Precisión de dos decimales: {0:.2f}, precisión de tres decimales: {0:.3f}'.format(1/3))
print('Números reales para 10 espacios: {0:10.3f}, {0:10.4f}'.format(3.1415926))
````

**Nota sobre el PDF:** en la pág. 142 los resultados se muestran con coma decimal (`0,33`, `3,142`) por la traducción. Python siempre usa punto decimal.

````python
# Tabla alineada (pág. 143): número, cuadrado y cubo en campos de 3, 4 y 5 posiciones
for i in range(2, 11, 2):
    print('{0:3d} {1:4d} {2:5d}'.format(i, i*i, i*i*i))
````

**Ampliación (no está en el PDF):** el proyecto usa `{:,}` para mostrar el separador de miles, y `{:16s}` para reservar 16 posiciones a una cadena (alineada a la izquierda). Las f-strings (`f"{valor:>5}"`) ofrecen una sintaxis equivalente.

````python
print('{:,}'.format(4500))                 # separador de miles
print('[{:16s}]'.format('Americano'))        # cadena en un campo de 16 posiciones
print(f"[{'Americano':16s}] {4500:,}")       # misma idea con f-string
````

**Clave:** `{}` marca dónde va cada valor; `d` es entero, `f` real, `s` cadena; el número antes de la letra es el ancho del campo.

**Error común:** contar mal el ancho; si el número no cabe en el ancho indicado, el campo se amplía en lugar de truncarse.

**Para el quiz:** `'{0:10.3f}'.format(3.1415926)` produce `'     3.142'` (10 posiciones, 3 decimales).

**Referencia:** PDF parte 1, págs. 140–143

## 7. JSON y conversión a diccionario

### Qué es JSON (págs. 118–120)

- El documento estándar de los servicios web es HTML (HyperText Markup Language), un subconjunto de XML (eXtensible Markup Language), un formato de texto que expresa datos estructurados.
- **JSON** (*JavaScript Object Notation*) es un formato de texto, más rápido de leer y escribir que XML, desarrollado para entregar objetos de datos formados por pares **clave-valor**.
- Es un formato estándar abierto, legible, muy usado para el intercambio de datos entre navegador y servidor.
- Surgió de JavaScript, pero es independiente del lenguaje: se puede analizar y generar en muchos lenguajes (C, C++, C#, Java, JavaScript, Perl, Python, etc.).

### Ejemplo de datos JSON (pág. 121)

Un JSON admite cadenas, números, listas, otros objetos y booleanos (`true` / `false`, en minúscula):

```
{
  "Nombre": "David Doe",                                   <- clave: cadena; valor: cadena
  "Edad": 25,                                              <- valor numérico (entero)
  "Hobby": ["baloncesto"],                                 <- valor: lista
  "Familia": {"padre": "Juan Pérez", "madre": "Marry Doe"}, <- valor: diccionario
  "Casado": true                                           <- valor booleano
}
```

(Las flechas son explicaciones de la diapositiva, no forman parte del JSON.)

**Referencia:** PDF parte 1, págs. 118–121

### De JSON a diccionario: `json.loads` (págs. 122–123)

El módulo `json` de la biblioteca estándar se carga con `import json`. La función `loads` (*load string*) convierte una **cadena** con formato JSON en un diccionario de Python.

````python
import json

data = '{"Name": "David Doe", "Age": 25, "Hobby": ["basketball"], "Family": {"father": "John Doe", "mother": "Marry Doe"}, "Married": true}'

json_data = json.loads(data)    # convierte la cadena JSON en un diccionario

print(type(json_data))
print(json_data['Name'])
print(json_data['Family'])
print(json_data['Married'])     # true de JSON pasa a ser True de Python
````

````python
# Segundo ejemplo del PDF: la novela El viejo y el mar
data = '{"title": "The Old man and The Sea", "ISBN": "12345", "Author" :"Ernest Hemingway"}'
json_data = json.loads(data)

print(type(json_data))
print(json_data['title'])
print(json_data['ISBN'])
print(json_data['Author'])
````

**Ampliación (no está en el PDF):** JSON es más estricto que un diccionario de Python: exige **comillas dobles** y escribe `true`, `false` y `null` en minúscula (equivalen a `True`, `False` y `None`).

````python
try:
    json.loads("{'title': 'x'}")     # comillas simples: JSON no válido
except json.JSONDecodeError as error:
    print("JSONDecodeError:", error)

print(json.loads('{"activo": true, "nota": null}'))
````

### De diccionario a JSON: `json.dump` y la sintaxis `with ... as` (págs. 124–126)

`json.dump(objeto, archivo, indent=...)` escribe el objeto en un archivo con formato JSON. El parámetro `indent='\t'` sangra con tabulaciones para facilitar la lectura. El archivo se abre con `with open(...) as f`, sintaxis que cierra el archivo automáticamente al terminar el bloque y que se estudia en detalle en la Unidad 13.

Código del PDF (crea `book.json` en la carpeta del notebook):

```python
import json

data = '{"title": "The Old man and The Sea", "ISBN": "12345", "Author" :"Ernest Hemingway"}'
json_data = json.loads(data)

# Código para crear json_data como archivo book.json
with open('book.json', 'w') as f:
    json.dump(json_data, f, indent='\t')
```

**Nota sobre el PDF:** en la pág. 126 el cuadro dice "Lee el archivo book.json con la función open en el modo de escritura (w)". Es una traducción confusa: el modo `'w'` (*write*) **escribe** el archivo; no lo lee. Para leer se usa `'r'`.

Para no dejar archivos en la carpeta del curso, en este notebook se ejecuta el mismo código sobre una carpeta temporal:

````python
import json
import os
import tempfile

data = '{"title": "The Old man and The Sea", "ISBN": "12345", "Author" :"Ernest Hemingway"}'
json_data = json.loads(data)

carpeta = tempfile.mkdtemp()                 # carpeta temporal
ruta = os.path.join(carpeta, 'book.json')    # en el PDF es simplemente 'book.json'

with open(ruta, 'w') as f:                   # 'w' = modo de escritura
    json.dump(json_data, f, indent='\t')     # guarda el diccionario como JSON, sangrando con tabulador

with open(ruta, 'r') as f:                   # 'r' = lectura (ampliación: el PDF no la muestra)
    print(f.read())

os.remove(ruta)                              # limpieza del archivo temporal
os.rmdir(carpeta)
````

**Ampliación (no está en el PDF):** `json.load(archivo)` lee un archivo JSON y devuelve un diccionario; `json.dumps(objeto)` convierte un diccionario en una **cadena** JSON (la "s" de *string*), y `loads`/`dumps` son los equivalentes de `load`/`dump` para cadenas.

````python
texto = json.dumps({'Name': 'David Doe', 'Age': 25, 'Married': True}, indent=2)
print(texto)
print(type(texto))
````

**Clave:** `json.loads` convierte cadena JSON en diccionario; `json.dump` escribe un diccionario como JSON en un archivo.

**Error común:** confundir `loads`/`dumps` (trabajan con cadenas) con `load`/`dump` (trabajan con archivos); usar comillas simples en una cadena JSON.

**Para el quiz:** JSON es un formato de texto de pares clave-valor, independiente del lenguaje; `loads` devuelve un `dict`.

## Proyecto de la unidad

### Problema: menú de una cafetería

Mostrar el menú y los precios de una cafetería con un diccionario y, cuando el usuario escriba una bebida, indicar su precio o avisar que no está en el menú. Menús y precios: Americano 3.000; Ice Americano 3.500; Cappuccino 4.000; Cafe Latte 4.500; Espresso 3.600 (KRW).

**Referencia:** PDF parte 1, págs. 93–95

### Plan de programación n.º 1: imprimir el menú (pág. 98)

1. Inicio.
2. Introducir el menú y el precio en un diccionario (`menu = {'Americano': 3000, ...}`).
3. Para cada clave del diccionario: imprimir el menú (clave y precio).
4. Fin.

### Código final n.º 1 (págs. 96 y 100)

El código del PDF funciona. El recorrido `for key in menu` itera sobre las **claves** del diccionario.

````python
menu = {'Americano': 3000, 'Ice Americano': 3500, 'Cappuccino': 4000, 'Cafe Latte': 4500, 'Espresso': 3600}

for key in menu:    # recorre las claves
    print("{:16s} precio : {:,}KRW".format(key, menu[key]))
````

**Nota sobre el PDF:** las salidas de las págs. 96–97 muestran `Precio : 3.000 KRW` (con punto de miles y un espacio antes de KRW) por la traducción y el formato de la diapositiva. Python con `{:,}` imprime `3,000` (coma como separador de miles) y el código del PDF no deja espacio antes de `KRW`.

### Plan de programación n.º 2: consultar un pedido (pág. 99)

1. Inicio.
2. Imprimir todo el menú.
3. Pedir al usuario que elija una bebida (`elección = entrada`).
4. Si la bebida elegida está en el menú (`elección in menu.keys()`), imprimir su precio.
5. En caso contrario, imprimir que no está en el menú.
6. Fin.

### Código final n.º 2, corregido (págs. 97 y 101)

**Nota sobre el PDF:** el código impreso no se puede ejecutar tal cual, por fallos de traducción:

- Las cadenas de las líneas 1 a 5 empiezan con `'` pero terminan con `"` (por ejemplo `print('Americano          Precio : 3.000 KRW")`): produce `SyntaxError`.
- La variable se define como `choice = input(...)`, pero la línea 8 usa `elección` (traducida) en `.format(elección, menú[elección])`: produce `NameError`. Además `menú` lleva tilde y no es el nombre de la variable (`menu`).
- Los textos de la línea 8 y 10 aparecen sin comillas de apertura: `print(Lo sentimos. {} no está en el menú.'.format(choice))`.
- El ejemplo de salida muestra "Café Latte cuesta 4,500KRW. Por favor, haga un pago.", pero el código imprime "`{}` es `{:,}`KRW. Por favor, realice el pago." y la clave del diccionario es `'Cafe Latte'` sin tilde: escribir "Café Latte" con tilde daría "no está en el menú".

El programa corregido se encapsula en funciones. La lógica del pedido (`procesar_pedido`) no usa `input()`, por lo que se puede probar con una lista de pedidos; la versión interactiva queda comentada.

````python
menu = {'Americano': 3000, 'Ice Americano': 3500, 'Cappuccino': 4000, 'Cafe Latte': 4500, 'Espresso': 3600}


def mostrar_menu(menu):
    """Imprime todos los menús con su precio (código final n.º 1)."""
    for key in menu:
        print("{:16s} precio : {:,}KRW".format(key, menu[key]))


def procesar_pedido(menu, choice):
    """Devuelve el mensaje que corresponde a la bebida elegida (código final n.º 2)."""
    if choice in menu.keys():      # la clave está en el menú
        return '{} es {:,}KRW. Por favor, realice el pago.'.format(choice, menu[choice])
    else:                          # la clave no está en el menú
        return 'Lo sentimos. {} no está en el menú.'.format(choice)


def menu_interactivo(menu):
    """Muestra el menú, pide una bebida con input() e imprime el resultado."""
    mostrar_menu(menu)
    choice = input('Seleccione un menú de la lista: ')
    print(procesar_pedido(menu, choice))
````

Versión **no interactiva**: se procesa una lista de pedidos de prueba (uno válido, uno inexistente y uno con tilde, que no coincide con la clave):

````python
mostrar_menu(menu)
print()

pedidos_de_prueba = ['Cafe Latte', 'Mocha', 'Café Latte']
for pedido in pedidos_de_prueba:
    print('Seleccione un menú de la lista:', pedido)
    print(procesar_pedido(menu, pedido))
    print()
````

## Apuntes para estudiar

### Clave

- Diccionario: pares `clave: valor` entre llaves; se accede con `d[clave]`. Las claves son únicas y no se busca por valor.
- `{}` crea un diccionario vacío; `[]` una lista vacía.
- `d[clave] = valor` inserta (clave nueva) o modifica (clave existente). `del d[clave]` elimina.
- `len(d)` cuenta pares. `clave in d` y `clave not in d` miran solo las claves.
- `==` y `!=` comparan contenido sin importar el orden; `>`, `<`, etc. producen `TypeError`.
- Lista: acceso por índice posicional; al borrar, los elementos se desplazan. Diccionario: acceso por clave; cada valor conserva su clave.
- `'{}'.format(...)`: `{0:4d}` entero con ancho 4, `{0:.2f}` real con 2 decimales, `{0:10.3f}` ancho 10 y 3 decimales.
- JSON: formato de texto de pares clave-valor, independiente del lenguaje. `json.loads(cadena)` produce un `dict`; `json.dump(obj, f, indent='\t')` escribe un archivo.
- `with open(ruta, 'w') as f:` abre el archivo para escribir y lo cierra solo.

### Errores comunes

- `KeyError` al leer o borrar con una clave inexistente.
- Creer que `valor in diccionario` busca en los valores.
- Repetir una clave: solo se conserva el último valor.
- Esperar que `dic.pop(0)` elimine el primer elemento por posición: elimina la clave `0`.
- Escribir `lst.clear` o `dic.keys` sin paréntesis: no ejecuta el método.
- Comparar diccionarios con `>` o `<`: `TypeError`.
- Usar comillas simples en una cadena JSON: `JSONDecodeError`.
- Confundir el modo `'w'` (escribir) con leer un archivo.
