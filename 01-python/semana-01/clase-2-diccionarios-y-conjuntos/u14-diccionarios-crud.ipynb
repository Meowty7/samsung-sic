## 1. Añadir y editar clave-valor

Un diccionario permite añadir y eliminar información según se necesite. La forma básica de añadir o editar un par es asignar con el mismo formato que ya se conoce: `dic[clave] = valor`.

- Si la clave **no existe**, se añade un par nuevo.
- Si la clave **ya existe**, se reemplaza su valor.

Hay casos en los que hace falta añadir o actualizar **por defecto**. Para ello el PDF presenta dos métodos avanzados:

- `setdefault`: añade el par clave-valor por defecto.
- `update`: modifica el valor de una clave y añade el par si la clave no existe.

**Referencia:** PDF parte 2, pág. 18

````python
dic = {'a': 10, 'b': 20}
dic['c'] = 30   # la clave 'c' no existe: se añade
dic['a'] = 90   # la clave 'a' existe: se edita su valor
print(dic)
````

### 1.1 Contar letras sin `setdefault`

El PDF plantea un código que cuenta cuántas veces aparece cada letra de una lista y lo guarda en un diccionario. Para construirlo conviene dar un valor inicial a cada clave, pero el diccionario de partida está vacío y no hay valor por defecto. Por eso se comprueba con `if` si la clave ya está en el diccionario y, si no, se inicializa en cero antes de sumar 1.

Líneas 4, 5 y 6 del PDF:

- Se examina si el valor de `ch` está entre las claves de `dic`.
- Si no está, se añade la clave con valor cero.
- Se suma 1 al valor de esa clave.

El resultado es la frecuencia de aparición de cada letra: `'a'` aparece 2 veces, `'b'` 3 veces y `'c'` 2 veces.

**Referencia:** PDF parte 2, págs. 19–20

````python
str1 = ['a', 'b', 'c', 'b', 'a', 'b', 'c']
dic = {}
for ch in str1:
    if ch not in dic.keys():
        dic[ch] = 0
    dic[ch] += 1

print('alphabet counting :', dic)
````

### 1.2 Contar letras con `setdefault`

`setdefault(clave, valor_por_defecto)` evita la cláusula `if` del código anterior. Si la clave ya existe, **no cambia nada** y devuelve el valor actual; si no existe, añade la clave con el valor por defecto indicado.

En el ejemplo del PDF, `dic.setdefault(ch, 0)` inicializa a cero cada letra la primera vez que aparece, y `dic[ch] += 1` suma 1. El `print(list(dic.items()))` dentro del bucle permite ver cómo el valor de cada clave cambia con cada letra procesada.

**Nota sobre el PDF:** la diapositiva dice que `setdefault` "devuelve la clave si tiene clave". En realidad devuelve el **valor** asociado a la clave (el existente, o el valor por defecto si la clave no estaba). Se muestra más abajo.

**Referencia:** PDF parte 2, págs. 21–23

````python
str1 = ['a', 'b', 'c', 'b', 'a', 'b', 'c']
dic = {}
for ch in str1:
    dic.setdefault(ch, 0)
    dic[ch] += 1
    print(list(dic.items()))

print('alphabet counting :', dic)
````

Ejemplo del PDF con una clave nueva: como no existe la clave `'f'`, se añade con el valor por defecto 50.

**Referencia:** PDF parte 2, pág. 24

````python
dic = {'a': 10, 'b': 20, 'c': 30, 'd': 40}
dic.setdefault('f', 50)
print(dic)
````

**Ampliación (no está en el PDF):** `setdefault` devuelve el valor de la clave. Eso permite usarlo en una sola línea y confirma la aclaración anterior.

````python
dic = {'a': 10, 'b': 20}
print(dic.setdefault('a', 999))   # la clave existe: devuelve su valor y no cambia nada
print(dic.setdefault('z', 999))   # la clave no existe: la crea y devuelve el valor por defecto
print(dic)
````

### 1.3 Modificar con `update`

`update` modifica el valor de una clave con la forma `clave=valor`. Los nombres de las claves se escriben sin comillas. Si la clave no existe, el método añade el par clave-valor.

- `dic.update(a=90)` cambia el valor de `'a'` a 90 (pág. 25).
- `dic.update(e=50)` como no hay clave `'e'`, la añade con valor 50 (pág. 26).
- Se pueden modificar varios elementos a la vez, separados por comas: `dic.update(a=900, f=60)` (pág. 27).

**Referencia:** PDF parte 2, págs. 25–27

````python
dic = {'a': 10, 'b': 20, 'c': 30, 'd': 40}

dic.update(a=90)          # cambia el valor de 'a'
print(dic)

dic.update(e=50)          # 'e' no existe: se añade
print(dic)

dic.update(a=900, f=60)   # varios a la vez: cambia 'a' y añade 'f'
print(dic)
````

### 1.4 Precauciones de `update`: claves que no son cadenas

La forma `update(clave=valor)` solo puede utilizarse cuando la clave es una **cadena**, porque los nombres se escriben sin comillas. Si la clave es un número, el valor se modifica insertando un diccionario: `update(diccionario)`.

Otra forma de modificar con claves no cadena es usar una **lista** o una **tupla** con pares: `update([[clave1, valor1], [clave2, valor2]])` o `update(((clave1, valor1), (clave2, valor2)))`. Ambas tienen el mismo efecto.

**Referencia:** PDF parte 2, págs. 28–29

````python
dic2 = {1: 'one', 2: 'two'}
dic2.update({1: 'ONE', 3: 'THREE'})   # claves numéricas: se pasa un diccionario
print(dic2)

# Con una lista de pares
dic2.update([[2, 'TWO'], [4, 'FOUR']])
print(dic2)

# Con una tupla de pares
dic2.update(((2, 'Two'), (4, 'Four')))
print(dic2)
````

**Ampliación (no está en el PDF):** la forma `update(clave=valor)` en realidad exige que el nombre sea un identificador válido de Python, no solo una cadena. Una clave como `'precio total'` (con espacio) no se puede escribir así y debe pasarse en un diccionario.

````python
dic = {}
dic.update({'precio total': 100})   # clave con espacio: va dentro de un diccionario
print(dic)
````

### 1.5 Diferencia entre `setdefault` y `update`

- `setdefault` solo puede **añadir** pares; el valor de las claves que ya existen no se modifica. Aunque la clave `'a'` ya esté, `setdefault('a', 90)` no cambia su valor.
- `update` puede tanto **añadir** como **modificar** pares clave-valor.

**Clave:** `setdefault` = "añade solo si falta"; `update` = "añade o sobrescribe".

**Referencia:** PDF parte 2, pág. 30

````python
dic = {'a': 10, 'b': 20, 'c': 30, 'd': 40}
print(dic)
dic.setdefault('a', 90)   # 'a' ya existe: no cambia
print(dic)

dic.update(a=90)          # update sí sobrescribe
print(dic)
````

**Error común:** esperar que `setdefault` sobrescriba un valor existente. Si se necesita cambiar un valor ya existente hay que usar `dic[clave] = valor` o `update`.

**Para el quiz:** `setdefault(k, v)` no modifica claves existentes; `update(...)` sí. `update(a=1)` solo funciona con claves que sean nombres de identificador.

**Referencia:** PDF parte 2, págs. 25–30

## 2. Borrar un clave-valor específico o aleatorio

### 2.1 `pop(clave)`

`pop` borra el par clave-valor de una clave específica y **devuelve el valor** borrado. Si se especifica un valor por defecto, `pop(clave, valor_por_defecto)`, borra el par y devuelve el valor eliminado; cuando la clave no existe, devuelve el valor por defecto en lugar de fallar.

En el ejemplo del PDF, `dic.pop('a')` borra la clave `'a'` y devuelve 10; y `dic.pop('f', 8)` devuelve 8 porque no existe la clave `'f'`.

**Referencia:** PDF parte 2, págs. 31–32

````python
dic = {'a': 10, 'b': 20, 'c': 30, 'd': 40}
print(dic.pop('a'))   # borra 'a' y devuelve su valor
print(dic)

print(dic.pop('f', 8))   # 'f' no existe: devuelve el valor por defecto
````

**Error común:** usar `pop` con una clave inexistente y sin valor por defecto lanza `KeyError` (así lo indica también la tabla de métodos del PDF, pág. 39).

**Referencia:** PDF parte 2, págs. 31–32 y 39

````python
dic = {'a': 10, 'b': 20}
try:
    dic.pop('z')
except KeyError as error:
    print('KeyError:', error)
````

### 2.2 `popitem()`

`popitem` borra un par clave-valor del diccionario y lo **devuelve como tupla** `(clave, valor)`. El comportamiento depende de la versión de Python: en Python 3.6 y superiores borra el **último** par insertado y en 3.5 e inferiores borra uno cualquiera. El material se basa en Python 3.6 o superior, por lo que se devuelve el último elemento.

**Nota sobre el PDF:** la tabla de la pág. 39 dice que `popitem()` "devuelve el elemento seleccionado aleatoriamente". Con las versiones actuales de Python no es aleatorio: siempre extrae el último par insertado (orden LIFO).

**Referencia:** PDF parte 2, pág. 33

````python
dic = {'a': 10, 'b': 20, 'c': 30, 'd': 40}
print(dic.popitem())   # devuelve el último par como tupla
print(dic)
````

### 2.3 `clear()`

`clear` borra todos los pares clave-valor. Después de borrar, el diccionario queda vacío: `{}`.

**Referencia:** PDF parte 2, pág. 34

````python
dic = {'a': 10, 'b': 20, 'c': 30, 'd': 40}
print(dic)
dic.clear()
print(dic)
````

**Clave:** `pop(k)` borra una clave concreta y devuelve su valor; `popitem()` borra el último par y devuelve una tupla; `clear()` borra todo y no devuelve nada.

**Para el quiz:** `pop` con clave ausente y sin valor por defecto lanza `KeyError`; `pop(k, d)` devuelve `d`.

**Referencia:** PDF parte 2, págs. 31–34

## 3. Traer un valor de una clave específica y guardarlo: `get`

`get(clave)` toma el valor de una clave específica del diccionario. Si se especifica un valor por defecto, `get(clave, valor_por_defecto)`, devuelve el valor de la clave cuando existe, pero devuelve el valor por defecto cuando la clave **no existe**. Si no existe y no se pasa valor por defecto, devuelve `None` (así lo indica la tabla del PDF, pág. 39).

A diferencia de `pop`, `get` **no borra** el par.

**Referencia:** PDF parte 2, pág. 35

````python
dic = {'a': 10, 'b': 20, 'c': 30, 'd': 40}
a = dic.get('a')
print(a)

b = dic.get('z', 0)   # 'z' no existe: devuelve el valor por defecto 0
print(b)

print(dic.get('z'))   # sin valor por defecto: None
print(dic)            # get no borra nada
````

**Clave:** `dic[clave]` lanza `KeyError` si falta; `dic.get(clave)` devuelve `None` y `dic.get(clave, d)` devuelve `d`.

**Para el quiz:** `get` es la forma segura de leer una clave que puede no existir.

**Referencia:** PDF parte 2, págs. 35 y 39

## 4. Elementos, claves y valores: `keys`, `values`, `items`

La operación más importante de un diccionario es encontrar el valor asociado a una clave: `dic['a']`.

- `keys()` devuelve todas las claves (`dict_keys`).
- `values()` devuelve todos los valores (`dict_values`).
- `items()` devuelve todos los pares en forma de tuplas `(clave, valor)` (`dict_items`).

Con `for clave, valor in dic.items():` se asigna la clave y el valor de cada par a dos variables y se repite el código por cada par.

**Referencia:** PDF parte 2, págs. 36–38

````python
dic = {'a': 10, 'b': 20, 'c': 30, 'd': 40}
print(dic['a'])
print(dic.keys())
print(dic.values())
print(dic.items())

for str1, num in dic.items():
    print(str1, ':', num)
````

### 5.1 `split`: dividir una cadena

`split` corta una cadena en una lista de subcadenas. Sin argumentos separa por espacios en blanco. Si se pasa un separador, la cadena se divide en cada aparición de ese separador.

**Referencia:** PDF parte 2, págs. 46–47

````python
s = 'Welcome to Python'
print(s.split())

s = '2021.8.15'
print(s.split('.'))      # separar por puntos

s = 'Hello, World!'
print(s.split(','))      # separar por coma: queda un espacio delante de World
````

### 5.2 `strip`, `lstrip` y `rstrip`: quitar espacios

El resultado de `s.split(',')` anterior tiene un espacio delante de `World!`. `strip` elimina los espacios antes y después de la cadena; `lstrip` solo los de la izquierda y `rstrip` solo los de la derecha. El PDF muestra además cómo limpiar cada elemento tras dividir con una lista por comprensión, que se estudia en la Unidad 19.

El PDF también advierte que un separador incorrecto no divide la cadena: el código `s.split(...)` con un separador que no aparece devuelve la cadena completa dentro de una lista.

**Referencia:** PDF parte 2, págs. 48–50

````python
s = 'Hello, World!'
print(s.split(';'))    # separador incorrecto: no divide

s = 'Welcome, to,  Python, and ,  bla, bla   '
print(s.split(','))                       # quedan espacios sobrantes
print([x.strip() for x in s.split(',')])  # strip a cada elemento

s = '   Hello, World!   '
print(repr(s.strip()))
print(repr(s.lstrip()))
print(repr(s.rstrip()))
````

### 5.3 `strip` con un carácter indicado

Si se pasa un carácter a `strip`, elimina ese carácter antes y después de la cadena. `lstrip` y `rstrip` también aceptan caracteres.

**Referencia:** PDF parte 2, pág. 51

````python
s = '########this is an example#####'
print(s.strip('#'))
print(s.lstrip('#'))
print(s.rstrip('#'))
````

### 5.4 `join`: unir cadenas

`split` separa una cadena en subcadenas; `join` hace lo contrario: recoge una lista de cadenas y las une en una sola, con el separador que se indique entre ellas (`separador.join(lista)`). Si no se especifica separador (cadena vacía), las cadenas se unen sin espacios. `join` también sirve para volver a unir una lista de caracteres.

`join` pone el carácter de unión **solo entre** las cadenas, no antes del primero ni después del último.

**Referencia:** PDF parte 2, págs. 52–54

````python
print(','.join(['apple', 'grape', 'banana']))   # une tres palabras con una coma
print('-'.join('010.1234.5678'.split('.')))     # cambiar el número de teléfono de . a -
print(''.join(['apple', 'grape', 'banana']))     # sin separador

s = 'hello world'
clist = list(s)
print(clist)
print(''.join(clist))
````

### 5.5 `split` y `join` juntos

Combinando `split` y `join` se eliminan los espacios innecesarios (incluidos tabuladores y saltos de línea) de una cadena: al dividir con `split()` desaparecen los espacios en blanco sobrantes y al unir con `' '.join(...)` la cadena queda con un único espacio entre palabras.

**Referencia:** PDF parte 2, pág. 55

````python
a_string = 'Actions \n\t speak louder than words'
print(a_string)

word_list = a_string.split()
print(word_list)

refined_string = ' '.join(word_list)
print(refined_string)
````

### 5.6 `lower`, `upper` y `capitalize`

`lower()` pasa a minúsculas, `upper()` pasa a mayúsculas y `capitalize()` convierte solo la primera letra en mayúscula.

**Referencia:** PDF parte 2, pág. 56

````python
s = 'Hello, World!'
print(s.lower())
print(s.upper())
print('hello'.capitalize())
````

### 5.7 `startswith` y `endswith`

`startswith` indica si una cadena empieza por un carácter dado; `endswith` indica si termina con él. Ambos aceptan un segundo y un tercer argumento que delimitan el tramo de la cadena donde se busca:

- `startswith(texto, inicio)` busca a partir de la posición `inicio`.
- `endswith(texto, inicio, fin)` busca solo en el tramo `[inicio:fin]`.

En el ejemplo del PDF, `s[1:3]` es `'BC'`, por lo que `endswith('C', 1, 3)` es `True` y `endswith('C', 1, 10)` es `False` (el tramo `[1:10]` termina en `'J'`).

**Referencia:** PDF parte 2, págs. 57–59

````python
s = 'ABCDEFGHIJKLMNOPQRSTUVWXYZ'
print(s.startswith('A'))       # True: empieza por A
print(s.startswith('B', 1))    # True: en la posición 1 hay una B
print(s.startswith('B', 5))    # False
print(s.endswith('Z'))         # True: termina en Z
print(s.endswith('C', 1, 3))   # True: s[1:3] = 'BC'
print(s.endswith('C', 1, 10))  # False
````

### 5.8 `replace`: sustituir

`replace(old, new, [count])` sustituye `old` por `new` en la cadena; el tercer argumento es opcional e indica cuántas veces como máximo se sustituye. El PDF lo usa para quitar las comas de un número con separadores de miles.

**Nota sobre el PDF:** la diapositiva titula esta sección "1.12. Método endwith" por error; el método que se explica es `replace`.

**Referencia:** PDF parte 2, págs. 60

````python
text = '123,456,789,999'

replaceAll = text.replace(',', '')      # todas las comas
replace_t1 = text.replace(',', '', 1)   # solo la primera
replace_t2 = text.replace(',', '', 2)   # las dos primeras
replace_t3 = text.replace(',', '', 3)   # las tres

print('Resultado :')
print(replaceAll)
print(replace_t1)
print(replace_t2)
print(replace_t3)
````

**Error común:** las cadenas son inmutables. `strip`, `replace`, `lower`, etc. **devuelven una cadena nueva**; si no se guarda el resultado, la cadena original no cambia.

````python
s = '  hola  '
s.strip()                  # el resultado se descarta
print(repr(s))
s = s.strip()              # hay que reasignar
print(repr(s))
````

**Clave:** `split` y `join` son operaciones inversas; `strip` limpia los bordes; `replace` sustituye. **Para el quiz:** `' '.join('a b c'.split())` normaliza espacios.

**Referencia:** PDF parte 2, págs. 46–60

## Proyecto de la unidad

### Problema

Crear un programa de menú de cafetería que guarde cada menú con su precio en un diccionario. El programa responde con el precio cuando se consulta un menú y permite añadir menús y precios nuevos. El PDF explica que los comandos se escriben en la ventana de ejecución: aparece el símbolo del sistema `$`, el comando de entrada (añadir) se expresa como `<`, el de búsqueda como `>` y para salir se introduce `q`.

**Referencia:** PDF parte 2, págs. 12–14

### Plan de programación (pseudocódigo del PDF)

1. Inicio.
2. Restablecer el diccionario del menú de cafetería.
3. `while True`:
4. Introducir el comando y el menú de cafetería.
5. Si el primer carácter del comando es `<` (entrada):
6. Obtener el menú y el precio de la cadena de entrada y añadirlos al diccionario.
7. En caso contrario, si el primer carácter es `>` (salida):
8. Imprimir el menú.
9. En caso contrario, si el primer carácter es `q`:
10. Salir del programa.
11. En caso contrario, imprimir error.
12. Salir del programa.
13. Fin.

### Código final

Se encapsula el código del PDF en una función `programa_menu_cafeteria`. El parámetro `leer` permite sustituir `input` por una función de prueba y así ejecutar el notebook sin interacción. El resto del código es el del PDF.

**Nota sobre el PDF:** en la diapositiva el primer `print` aparece cortado en el borde del recuadro; aquí se escribe completo: `"Iniciar programa de menú de cafetería... Pulsar q para salir"`. Además, la rama `elif command == 'q'` del PDF nunca se ejecuta, porque `str.startswith('q')` ya hizo `break` antes (se explica más abajo en *Errores comunes del programa*).

**Referencia:** PDF parte 2, págs. 13 y 16

````python
def programa_menu_cafeteria(leer=input):
    print("Iniciar programa de menú de cafetería... Pulsar q para salir")
    cafe_menu = {'Café helado': 3000}

    while True:
        str = leer('$ ')            # comando de entrada (el PDF nombra la variable str)
        if str.startswith('q'):     # q: salir
            break
        command = str[0]            # primer carácter: '<', '>' u otro

        if command == '<':          # añadir: "< menú : precio"
            str = str.replace('<', "")
            inputStr = str.split(':')
            if len(inputStr) < 2:
                print('error de entrada.')
                continue
            else:
                cafe_menu[inputStr[0].strip()] = inputStr[1].strip()

        elif command == '>':        # consultar: "> menú"
            str = str.replace('>', "")
            inputStr = str.strip()
            if inputStr in cafe_menu:
                print(cafe_menu[inputStr])
            else:
                print('{} no está en el menú.'.format(inputStr))
        elif command == 'q':        # (inalcanzable: ya se salió arriba)
            break
        else:
            print('error de entrada.')
    print("saliendo del programa cafe menu.")
    return cafe_menu                # el PDF no devuelve nada; se devuelve para poder verificar
````

### Ejecución con entradas de prueba

La función `simular_entradas` devuelve un reemplazo de `input` que lee de una lista y muestra cada comando como si se hubiera escrito. La sesión replica el ejemplo de la pág. 14 del PDF.

````python
def simular_entradas(comandos):
    """Devuelve una función que imita input() leyendo de una lista de comandos."""
    iterador = iter(comandos)
    def leer(mensaje=''):
        texto = next(iterador)
        print(f"{mensaje}{texto}")
        return texto
    return leer


comandos_prueba = [
    '> Café helado',
    '> té verde',
    '< espresso : 5000',
    'espresso',
    '> espresso',
    'q',
]
menu_final = programa_menu_cafeteria(simular_entradas(comandos_prueba))
print(menu_final)
````

El ejercicio de la pág. 64 propone actualizar el programa de inventario de una tienda de conveniencia con el diccionario:

```
items = {"Café": 7, "Bolígrafo": 3, "Vaso de papel": 2, "Leche": 1, "Coca-Cola": 4, "Libro": 5}
```

con un menú `1) comprobar stock 2) almacenaje 3) liberar 4) salir`: consultar existencias, añadir (almacenaje) o restar (liberar) cantidades. A continuación se resuelve con una función que procesa una lista de operaciones de prueba.

**Nota sobre el PDF:** el diccionario del enunciado tiene un error de sintaxis (`"Coca-Cola" 4`, faltan los dos puntos). La salida de ejemplo introduce `leche`, `coque` y `coca cola` en minúsculas o abreviados y obtiene resultados como si coincidieran con `Leche` y `Coca-Cola`; aquí se normaliza el nombre con una búsqueda que ignora mayúsculas.

**Referencia:** PDF parte 2, págs. 62–64

````python
def gestionar_inventario(items, operaciones):
    """Procesa operaciones (opcion, nombre, cantidad) sobre el diccionario items.
    1 = comprobar stock, 2 = almacenar, 3 = liberar."""
    por_minuscula = {nombre.lower(): nombre for nombre in items}   # búsqueda sin distinguir mayúsculas
    for opcion, nombre, cantidad in operaciones:
        clave = por_minuscula.get(nombre.lower())
        if clave is None:
            print(f"{nombre}: no existe en el inventario")
            continue
        if opcion == 1:
            print(f"Stock de {clave}: {items[clave]}")
        elif opcion == 2:
            items[clave] += cantidad
            print(f"Almacenado {cantidad} de {clave}. Stock: {items[clave]}")
        elif opcion == 3:
            items[clave] -= cantidad
            print(f"Liberado {cantidad} de {clave}. Stock: {items[clave]}")
    return items


items = {"Café": 7, "Bolígrafo": 3, "Vaso de papel": 2, "Leche": 1, "Coca-Cola": 4, "Libro": 5}
gestionar_inventario(items, [(1, 'leche', 0), (3, 'coca-cola', 1), (1, 'Coca-Cola', 0), (2, 'Libro', 2)])
````

## Apuntes para estudiar

**Clave**

- `dic[clave] = valor` añade o edita; `setdefault(k, v)` añade solo si falta (y devuelve el valor); `update(...)` añade o sobrescribe.
- `update(clave=valor)` solo sirve con claves que sean identificadores; con claves numéricas se pasa un diccionario, una lista o una tupla de pares.
- `pop(k)` borra y devuelve el valor; `pop(k, d)` evita el `KeyError`; `popitem()` borra y devuelve el último par como tupla; `clear()` vacía el diccionario.
- `get(k)` devuelve `None` si falta; `get(k, d)` devuelve `d`; no borra nada.
- `keys()`, `values()` e `items()` devuelven vistas; `for k, v in dic.items()` recorre pares.
- Cadenas: `split`, `strip`/`lstrip`/`rstrip`, `join`, `lower`/`upper`/`capitalize`, `startswith`, `endswith`, `replace`; devuelven cadenas nuevas.
- Proyecto: bucle `while True` con comandos `<` (añadir), `>` (consultar) y `q` (salir).

**Errores comunes**

- Esperar que `setdefault` sobrescriba un valor ya existente.
- `pop(clave)` con una clave inexistente y sin valor por defecto: `KeyError`.
- Usar `update(1=...)` con claves numéricas (no es válido): hay que pasar un diccionario.
- No guardar el resultado de `strip`/`replace`/`lower`: las cadenas no se modifican.
- Creer que `popitem` es aleatorio: en Python actual extrae el último par insertado.
- En el programa del menú: `startswith('q')` sale con cualquier palabra que empiece por `q` y una entrada vacía provoca `IndexError`.

**Para el quiz**

- `setdefault` devuelve el valor de la clave (no la clave).
- `get` y `pop` con valor por defecto no lanzan error si la clave falta.
- `clear()` deja `{}`; `popitem()` devuelve una tupla `(clave, valor)`.
- `','.join(['a', 'b'])` da `'a,b'` y `'a,b'.split(',')` da `['a', 'b']`.
- Los nombres traducidos en las diapositivas (llaves, valores, artículos, claro, banda/tira) corresponden a `keys`, `values`, `items`, `clear`, `strip`.
