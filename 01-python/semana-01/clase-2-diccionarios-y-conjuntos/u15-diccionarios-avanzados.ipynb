## 1. Convertir listas y tuplas en diccionarios

### 1.1 Varias formas de crear un diccionario

Se puede hacer un diccionario a partir de una lista o una tupla ya existentes. Una forma es usar el método `fromkeys` de la clase `dict`: los elementos de la lista o tupla pasan a ser las **claves**. Por defecto todos los valores se inicializan a `None`, y existe una forma de indicar un valor inicial distinto.

**Referencia:** PDF parte 2, pág. 78

### 1.2 `dict.fromkeys(claves[, valor])`

- `dict.fromkeys(keys)` crea un diccionario cuyas claves son los elementos de `keys` y cuyos valores son `None`.
- `dict.fromkeys(keys, 100)` hace que 100 sea el valor por defecto de todas las claves.

Funciona igual con listas y con tuplas.

**Referencia:** PDF parte 2, págs. 79–80

````python
keys = ['a', 'b', 'c', 'd']          # lista
x = dict.fromkeys(keys)
print(x)

keys = ('a', 'b', 'c', 'd')          # tupla
x = dict.fromkeys(keys)
print(x)

y = dict.fromkeys(keys, 100)         # valor inicial 100
print(y)
````

**Ampliación (no está en el PDF):** otras formas habituales de construir un diccionario a partir de secuencias. `dict(lista_de_pares)` convierte pares `(clave, valor)`, y `dict(zip(claves, valores))` junta dos secuencias.

````python
pares = [('a', 1), ('b', 2)]
print(dict(pares))

claves = ['x', 'y', 'z']
valores = [10, 20, 30]
print(dict(zip(claves, valores)))
````

**Error común (ampliación, no está en el PDF):** `fromkeys` usa **el mismo objeto** como valor para todas las claves. Con un valor mutable (por ejemplo una lista vacía) todas las claves comparten esa lista.

````python
d = dict.fromkeys(['a', 'b'], [])
d['a'].append(1)
print(d)                 # 'b' también cambió: es la misma lista
print(d['a'] is d['b'])
````

**Clave:** `dict.fromkeys(keys, valor)` crea claves con un valor inicial común; sin segundo argumento el valor es `None`.

**Para el quiz:** `dict.fromkeys(['a', 'b'])` da `{'a': None, 'b': None}`.

**Referencia:** PDF parte 2, págs. 78–80

## 2. `defaultdict`

### 2.1 Por qué se necesita

Acceder a una clave que no existe en un diccionario produce `KeyError`. Para evitar este error se utiliza `defaultdict`:

- `defaultdict` no produce error al intentar alcanzar una clave inexistente: devuelve un **valor por defecto**.
- `defaultdict` está incluido en el módulo `collections`.

**Referencia:** PDF parte 2, pág. 81

````python
x = {'a': 0, 'b': 0, 'c': 0, 'd': 0}
try:
    x['z']
except KeyError as error:
    print('KeyError:', error)
````

### 2.2 Cómo usarlo

Se importa `defaultdict` del módulo `collections` y se crea con una **función generadora** del valor por defecto, por ejemplo `int`. El valor por defecto de `int()` es 0, por eso `y['z']` devuelve 0 aunque la clave `'z'` no exista. Al consultar la clave `'z'`, el `defaultdict` la crea con el valor por defecto.

**Nota sobre el PDF:** el ejemplo escribe `y = dict.fromkeys(keys, 100)` y justo después `y = defaultdict(int)`. La segunda línea **reemplaza** a `y`, de modo que el diccionario resultante solo contiene `'z'`; el diagrama de la pág. 83 (con `'a'` y `'b'` en 100 y `'z'` en 0) muestra la idea, pero no es lo que produce ese código. Para obtener ese resultado habría que escribir `defaultdict(int, y)`.

**Referencia:** PDF parte 2, págs. 82–84

````python
from collections import defaultdict   # traer defaultdict del módulo collections

keys = ('a', 'b', 'c', 'd')
y = dict.fromkeys(keys, 100)
y = defaultdict(int)                 # int como valor por defecto (reemplaza a y)
print(y['z'])
print(y)
print(int())                         # el valor por defecto de int() es 0
````

````python
# Para conservar los datos anteriores hay que pasarlos al crear el defaultdict
y = defaultdict(int, dict.fromkeys(keys, 100))
print(y['z'])
print(dict(y))
````

### 2.3 Interpretación: contar con `defaultdict`

La clase `defaultdict` llama automáticamente a la función (`int`) que se le pasó al crearla para establecer el valor de salida cuando no hay valor para una clave. El ejemplo del PDF cuenta la frecuencia de aparición de cada letra de la palabra `'abcbaabca'`, igual que en la unidad anterior pero **sin** `if` ni `setdefault`.

**Referencia:** PDF parte 2, págs. 84–85

````python
from collections import defaultdict

word = 'abcbaabca'
counter = defaultdict(int)
for letter in word:
    counter[letter] += 1

print(counter)
````

### 2.4 Precauciones: otros valores por defecto

`defaultdict(int)` genera valores por defecto con la función `int`, cuyo resultado sin argumentos es 0. Para un valor por defecto distinto de cero se crea con otra función generadora: por ejemplo, `defaultdict(list)` usa una lista vacía como valor por defecto.

**Referencia:** PDF parte 2, pág. 86

````python
list_dict = defaultdict(list)
print(list_dict)
print(list_dict['list1'])      # lista vacía como valor por defecto
print(list_dict)
````

**Ampliación (no está en el PDF):** `defaultdict(list)` sirve para agrupar elementos; `defaultdict(lambda: 100)` usa cualquier función sin argumentos como valor por defecto.

````python
palabras = ['uva', 'higo', 'manzana', 'uña']
por_inicial = defaultdict(list)
for p in palabras:
    por_inicial[p[0]].append(p)
print(dict(por_inicial))

cien = defaultdict(lambda: 100)
print(cien['x'])
````

**Clave:** con `defaultdict(f)`, una clave ausente se crea automáticamente con el valor `f()`. **Error común:** olvidar `from collections import defaultdict`; y consultar con `y['z']` **inserta** la clave `'z'` (a diferencia de `dic.get('z')`).

**Para el quiz:** `defaultdict(int)` es ideal para contar; el valor por defecto de `int()` es 0.

**Referencia:** PDF parte 2, págs. 81–86

## 3. Imprimir un diccionario con sentencias de repetición

### 3.1 Recorrer con `for`

Al recorrer un diccionario con `for i in x:` solo se obtiene la **clave**. Para tener clave y valor hay que usar `items()`: `for clave, valor in diccionario.items():`. En cada vuelta se toma un par clave-valor, se guarda la clave en `clave` y el valor en `valor`, y se repite el código.

**Referencia:** PDF parte 2, págs. 87–88

````python
x = {'a': 10, 'b': 20, 'c': 30, 'd': 40}

for i in x:
    print(i, end=' ')      # solo las claves
print()

for key, value in x.items():
    print(key, value)
````

### 3.2 Solo claves o solo valores

La clave del diccionario `x` se puede obtener con `x.keys()` y el valor con `x.values()`.

**Referencia:** PDF parte 2, págs. 89–90

````python
for key in x.keys():
    print(key, end=' ')
print()

for value in x.values():
    print(value, end=' ')
print()
````

### 3.3 Repetir con los métodos de diccionario (sección "Codifiquemos")

El PDF vuelve al tema con un ejemplo de agenda: `phone_book.items()` devuelve elementos que son **tuplas** `(clave, valor)` separadas por comas entre paréntesis, como `('David Doe', '010-1234-5678')`. `phone_book.keys()` devuelve las claves y `phone_book.values()` los valores.

**Nota sobre el PDF:** los encabezados de esta parte dicen "El formulario de devolución de artículos", "método de las llaves" y "método de los valores"; se trata de `items()`, `keys()` y `values()` aplicados a `phone_book`. También el texto "método mes" se refiere a `keys`.

**Referencia:** PDF parte 2, págs. 103–105

````python
phone_book = {"David Doe": "010-1234-5678", "John Smith": "010-1234-5679", "Jane Carter": "010-1234-5680"}
print(phone_book.items())
print(phone_book.keys())
print(phone_book.values())

for name, phone_num in phone_book.items():
    print(name, ':', phone_num)
````

**Clave:** `for k in d` entrega claves; `for k, v in d.items()` entrega claves y valores. **Error común:** esperar valores al hacer `for x in diccionario`.

**Para el quiz:** `items()` devuelve tuplas `(clave, valor)`.

**Referencia:** PDF parte 2, págs. 87–90 y 103–105

## 4. Diccionario dentro de un diccionario (diccionario doble)

Un diccionario puede contener otro diccionario como valor: `diccionario = {clave1: {claveA: valorA}, clave2: {claveB: valorB}}`. El PDF lo ilustra con un diccionario `terrestrial_planet` cuyas claves son los planetas terrestres (Mercurio, Venus, Tierra y Marte), y cada uno tiene a su vez un diccionario con radio medio, masa y periodo orbital. Es útil para almacenar **datos jerárquicos**.

Para acceder a un valor interior se usan dos pares de corchetes, uno por cada nivel:

```
Diccionario[clave][clave]
Diccionario[clave][clave] = valor
```

El diccionario consta de dos pasos, por eso se usan dos corchetes. Para obtener el radio medio de Venus se busca primero Venus y luego `'mean_radius'`.

**Referencia:** PDF parte 2, págs. 91–92

````python
terrestrial_planet = {
    'Mercury': {
        'mean_radius': 2439.7,
        'mass': 3.3022E+23,
        'orbital_period': 87.969
    },
    'Venus': {
        'mean_radius': 6051.8,
        'mass': 4.8676E+24,
        'orbital_period': 224.70069,
    },
    'Earth': {
        'mean_radius': 6371.0,
        'mass': 5.97219E+24,
        'orbital_period': 365.25641,
    },
    'Mars': {
        'mean_radius': 3389.5,
        'mass': 6.4185E+23,
        'orbital_period': 686.9600,
    }
}

print(terrestrial_planet['Venus']['mean_radius'])
````

**Ampliación (no está en el PDF):** con `Diccionario[clave][clave] = valor` también se modifica o añade un dato interior, y se puede recorrer el diccionario doble con dos bucles anidados.

````python
terrestrial_planet['Mars']['mean_radius'] = 3390.0     # modificar un dato interior
terrestrial_planet['Earth']['moons'] = 1               # añadir un dato interior
print(terrestrial_planet['Earth'])

for planeta, datos in terrestrial_planet.items():
    for dato, valor in datos.items():
        print(planeta, dato, valor)
    print('-')
````

**Clave:** `d[k1][k2]` lee y escribe en dos niveles. **Error común:** `d[k1][k2]` con `k1` ausente lanza `KeyError` y con `k1` presente pero `k2` ausente también.

**Referencia:** PDF parte 2, págs. 91–92

## 5. Asignación y copia de diccionarios

### 5.1 Asignación: `y = x`

Si se crea un diccionario y se asigna a otra variable (`y = x`), parece que habrá dos diccionarios, pero en realidad hay **uno solo**: los nombres de variable son distintos pero apuntan al mismo objeto. La comparación con `is` devuelve `True`. Por eso, cambiar el valor desde una variable se refleja en la otra: si `y['a'] = 99`, también `x['a']` vale 99.

**Referencia:** PDF parte 2, págs. 93–94

````python
x = {'a': 0, 'b': 0, 'c': 0, 'd': 0}
y = x
print(x is y)

y['a'] = 99
print(y)
print(x)
````

### 5.2 `copy`: dos objetos distintos

Para que las dos variables sean objetos distintos hay que **copiar** el diccionario con `copy()`. Tras `y = x.copy()`:

- `x is y` devuelve `False` (objetos diferentes);
- `x == y` devuelve `True` mientras el contenido sea igual;
- al cambiar `y['a'] = 99`, `x` no cambia y `x == y` pasa a ser `False`.

**Referencia:** PDF parte 2, págs. 95–96

````python
x = {'a': 0, 'b': 0, 'c': 0, 'd': 0}
y = x.copy()
print(x is y)
print(x == y)

y['a'] = 99
print(x is y)
print(x == y)
print(x)
print(y)
````

### 5.3 `copy` no basta con diccionarios dobles

Con un diccionario doble, `copy()` copia solo el **primer nivel**: los diccionarios interiores se siguen compartiendo. Cambiar `y['a']['python']` se refleja tanto en `x` como en `y`.

**Referencia:** PDF parte 2, pág. 97

````python
x = {'a': {'python': '2.7'}, 'b': {'python': '3.6'}}
y = x.copy()
y['a']['python'] = '2.7.15'
print(x)
print(y)
````

### 5.4 `deepcopy`

Para copiar completamente un diccionario doble hay que usar la función `deepcopy` del módulo `copy`. La **copia profunda** no solo copia la dirección de los datos sino que crea y almacena una copia de los datos en una memoria independiente. Tras `y = copy.deepcopy(x)`, cambiar `y` no afecta a `x`.

**Referencia:** PDF parte 2, pág. 98

````python
import copy                            # llamar al módulo copy

x = {'a': {'python': '2.7'}, 'b': {'python': '3.6'}}
y = copy.deepcopy(x)                   # copia profunda
y['a']['python'] = '2.7.15'
print(x)
print(y)
````

**Clave:** `y = x` no copia; `x.copy()` copia un nivel (copia superficial); `copy.deepcopy(x)` copia todos los niveles.

**Error común:** usar `copy()` con diccionarios dobles y pensar que son independientes.

**Para el quiz:** `x is y` compara identidad de objeto; `x == y` compara contenido.

**Nota sobre el PDF:** la numeración de la sección salta de 4.1 a 4.3 (no existe un 4.2).

**Referencia:** PDF parte 2, págs. 93–98

### 6.5 Módulo `time`

El módulo `time` ofrece información relacionada con el tiempo. El UTC que marca el inicio del sistema Unix, 0:0:0 del 1 de enero de 1970, se denomina **epoch** (época), y el tiempo medido desde allí en segundos es el tiempo Unix o tiempo epoch. `time.time()` devuelve los segundos transcurridos desde el epoch.

`time.localtime(timestamp)` devuelve la hora local en forma de estructura de fecha y hora, y `time.strftime(formato, tiempo)` imprime ese tiempo con el formato que se desee: `%Y` es el año, `%m` el mes, `%d` el día, `%H:%M:%S` las horas (24 horas), los minutos (00 a 59) y los segundos (00 a 59).

**Referencia:** PDF parte 2, págs. 115–117

````python
import time

seconds = time.time()
print('Tiempo tras epoch = ', seconds)

unix_timestamp = 1630282234          # valor fijo para un resultado reproducible
local_time = time.localtime(unix_timestamp)   # depende de la zona horaria del equipo
print(local_time)
print(time.strftime('%Y-%m-%d %H:%M:%S', local_time))

# Ampliación (no está en el PDF): gmtime siempre usa UTC
print(time.strftime('%Y-%m-%d %H:%M:%S', time.gmtime(unix_timestamp)))
````

### 6.7 Funciones trigonométricas con radianes

Las funciones trigonométricas del módulo `math` usan el ángulo en **radianes**. Por eso `sin(90.0)` no da 1: para usar 90 grados hay que convertirlo a radianes (`math.pi / 2`, o `math.radians(90)`).

**Referencia:** PDF parte 2, pág. 121

````python
import math as m

print(m.sin(0.0))             # valor de sen 0
print(m.sin(90.0))            # no es el ángulo general de 90: son 90 radianes
print(m.sin(m.pi / 2))        # notación en radianes
print(m.sin(m.radians(90)))   # ampliación: radians convierte grados a radianes
````

### Código final

Se encapsula el código del PDF en la función `guia_telefonica`; el parámetro `leer` sustituye a `input` para ejecutar el notebook sin interacción. El diccionario inicial guarda cada teléfono como cadena (las comillas entre paréntesis `('010-1234-4500')` no crean una tupla, solo una cadena); después de la actualización cada valor es una **tupla** `(teléfono, dirección, reseña)`.

**Nota sobre el PDF:** el ejemplo del PDF escribe `Northern Busan Estaba bien` como entrada (cuatro palabras). Con `adr, review = input(...).split()` eso produce un `ValueError`, porque hay más de dos palabras. La salida de ejemplo del PDF (`'Northern Busan'`, `'It was ok'`) tampoco coincide con lo que produce el código. Por la restricción del enunciado, aquí se escriben dirección y reseña **sin espacios internos** (por ejemplo `Busan-norte bien`). Además, el mensaje de entrada del PDF concatena `key + 'Por favor...'` sin espacio; aquí se añade uno.

**Referencia:** PDF parte 2, págs. 74 y 76

````python
def guia_telefonica(leer=input):
    tele_dir_dic = {'Restaurante chino': ('010-1234-4500'),
                    'Restaurante japonés': ('010-2230-6540'),
                    'Restaurante italiano': ('010-3232-7788')}
    print("número anterior")
    for key, value in tele_dir_dic.items():
        print(key + " :", value)
    for key, value in tele_dir_dic.items():
        adr, review = leer(key + ' Por favor, escriba la dirección y la reseña de: ').split()
        update = (value, adr, review)
        tele_dir_dic[key] = update
    print("Número actualizado")
    for key, value in tele_dir_dic.items():
        print(key + " :", value)
    return tele_dir_dic        # el PDF no devuelve nada; se devuelve para poder verificar
````

## Apuntes para estudiar

**Clave**

- `dict.fromkeys(claves, valor)` crea un diccionario a partir de una lista o tupla; el valor por defecto es `None`.
- `defaultdict(f)` (módulo `collections`) evita el `KeyError`: una clave ausente se crea con el valor `f()`; `defaultdict(int)` empieza en 0 y `defaultdict(list)` en `[]`.
- `for k in d` recorre claves; `d.keys()`, `d.values()` y `d.items()` recorren claves, valores y pares.
- Diccionario doble: `d[k1][k2]` para leer y escribir.
- `y = x` no copia (mismo objeto, `x is y` es `True`); `x.copy()` copia un nivel; `copy.deepcopy(x)` copia todos los niveles.
- Módulos: `import m`, `import m as alias`, `from m import nombre`, `from m import *`; `dir(m)` lista su contenido.
- `datetime` (fecha y hora), `time` (epoch, `localtime`, `strftime`) y `math` (funciones matemáticas, trigonometría en radianes).
- Proyecto: guía telefónica que amplía cada valor a una tupla `(teléfono, dirección, reseña)`.

**Errores comunes**

- Reasignar la variable (`y = defaultdict(int)`) y perder el diccionario anterior.
- Olvidar `from collections import defaultdict`.
- Creer que consultar `y['z']` en un `defaultdict` no cambia nada: inserta la clave.
- Pensar que `y = x` crea un diccionario nuevo.
- Usar `copy()` con diccionarios dobles y esperar independencia total.
- Pasar grados a `math.sin`, `math.cos` o `math.tan` (esperan radianes).
- Dar entradas con espacios cuando el código hace `adr, review = input().split()` (`ValueError`).
- `dict.fromkeys(claves, [])` comparte una única lista entre todas las claves.

**Para el quiz**

- `dict.fromkeys(['a', 'b'])` da `{'a': None, 'b': None}`.
- El valor por defecto de `int()` es 0, por eso `defaultdict(int)` sirve para contar.
- `x is y` compara identidad; `x == y` compara contenido.
- `deepcopy` está en el módulo `copy`; `defaultdict` en `collections`.
- `from datetime import *` trae todo el módulo; `import datetime as dt` crea un alias.
- Los nombres traducidos (suelo, sen, grados, radianes, artículos, llaves) son `floor`, `sin`, `degrees`, `radians`, `items`, `keys`.
