## 1. Qué es un conjunto y cómo se declara

El tipo `set` es un tipo de datos básico de Python (no lo es en lenguajes como C o Java). Se parece al conjunto matemático: una colección de objetos que cumplen un criterio claro. Sus propiedades según el PDF:

- Es un tipo **desordenado** y se escribe con llaves `{}`, como el diccionario.
- **No permite elementos duplicados**: si un valor se repite, se guarda una sola vez.
- Permite operaciones de conjuntos: intersección, unión y diferencia.

Formas de declarar un conjunto (tabla del PDF):

| Qué se quiere | Código |
|---|---|
| Conjunto vacío | `set0 = set()` |
| Conjunto básico | `set1 = {1, 2, 3, 4}` |
| Desde una tupla | `set2 = set(n_tupla)` |
| Desde una lista | `set3 = set(n_list)` |

**Referencia:** PDF parte 2, págs. 142 y 172

````python
set0 = set()                 # conjunto vacío
set1 = {1, 2, 3, 4}          # conjunto básico
n_tupla = (1, 2, 3, 4)
set2 = set(n_tupla)          # desde una tupla
n_list = [1, 2, 3, 4]
set3 = set(n_list)           # desde una lista
print(set0, set1, set2, set3)

numbers = {2, 1, 3}          # ejemplo de la pág. 172: conjunto de 3 números
print(numbers)
````

**Clave:** un `set` es una colección sin orden y sin elementos repetidos; por eso no tiene índices.

**Error común:** escribir `x = {}` para un conjunto vacío. Eso crea un diccionario vacío; el conjunto vacío se declara con `set()` (el PDF lo muestra en la pág. 149).

**Para el quiz:** `set()` crea un conjunto vacío; `{1, 2, 3}` crea un conjunto básico.

Lo mismo ocurre en la lista de palabras clave de la pág. 129 y en la salida "1 está en el juego" de la pág. 145, que debería decir "1 está en el conjunto".

**Referencia:** PDF parte 2, págs. 142 y 172

## 2. Crear conjuntos desde otros tipos y conjuntos vacíos

Se puede aplicar `set()` a cualquier objeto iterable (lista, tupla o cadena). Al convertir, los elementos repetidos desaparecen. Con una cadena, cada carácter se convierte en un elemento. Como el conjunto no tiene orden, no se puede usar índice ni troceado.

**Referencia:** PDF parte 2, págs. 143–144, 148–149

````python
days_list = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']   # lista
days_set = set(days_list)                                       # conjunto desde lista
print(days_set)   # el orden mostrado puede variar entre ejecuciones

fruits_tuple = ('apple', 'orange', 'water melon')                # tupla
fruits_set = set(fruits_tuple)                                  # conjunto desde tupla
print(fruits_set)

h_str = 'hello'                                                 # cadena
h_set = set(h_str)
print(h_set)   # el conjunto no permite duplicar la 'l'
````

````python
print(set([1, 2, 3, 1, 2]))   # los elementos repetidos se eliminan
print(set("abcdefa"))         # cada carácter es un elemento; la 'a' repetida desaparece

numbers = set()               # conjunto vacío
print(numbers)
````

**Clave:** `set(iterable)` sirve para eliminar duplicados rápidamente.

**Error común:** esperar que el conjunto conserve el orden de la lista original; el orden no está garantizado.

**Para el quiz:** `set("hello")` produce 4 elementos (`'h'`, `'e'`, `'l'`, `'o'`), no 5.

**Nota sobre el PDF:** el texto de la pág. 144 habla de la cadena "hola" pero el código usa `'hello'` (resultado `{'e', 'h', 'l', 'o'}`); el texto es una traducción y el código es el que cuenta. En la pág. 149 la palabra "sting" de un comentario es "string".

## 3. Comprobar valores, añadir y eliminar

Para saber si un valor pertenece a un conjunto se usa el operador `in`, igual que con listas. `in` no es exclusivo de los conjuntos: sirve con cualquier dato que contenga varios elementos.

Como el conjunto no tiene índice, las operaciones de indexación o troceado no tienen sentido. Para modificarlo se usan métodos: `add(x)` añade un elemento y `remove(x)` lo elimina.

**Referencia:** PDF parte 2, págs. 145–147

````python
numbers = {2, 1, 3}
if 1 in numbers:                       # comprueba si 1 está en el conjunto
    print("1 está en el conjunto.")

a_list = ['hello', 'world', 'welcome', 'to', 'python']
print('python' in a_list)              # 'in' también sirve con listas

numbers = {1, 2, 3}
numbers.add(4)                         # añadir un elemento
print(numbers)
numbers.remove(4)                      # eliminar un elemento
print(numbers)
````

````python
# El conjunto no admite índices: el intento produce un error (ejemplo didáctico)
try:
    numbers[0]
except TypeError as error:
    print("TypeError:", error)
````

**Clave:** `in` comprueba pertenencia; `add` y `remove` modifican el conjunto en el sitio.

**Error común:** usar `numbers[0]` o troceado `numbers[1:]`; lanzan `TypeError`.

**Para el quiz:** los conjuntos no admiten indexación ni troceado porque no tienen orden.

**Ampliación (no está en el PDF):** `remove(x)` lanza `KeyError` si `x` no está en el conjunto; `discard(x)` hace lo mismo sin error. La tabla de la pág. 158 presenta `discard` (como "descartar") y `clear` (como "borrar"), que vacía todo el conjunto.

````python
# Ampliación (no está en el PDF): remove vs discard
s = {1, 2, 3}
s.discard(99)                  # no hace nada, sin error
try:
    s.remove(99)               # lanza KeyError
except KeyError as error:
    print("KeyError:", error)
s.clear()                      # elimina todos los elementos
print(s)
````

## 4. Cálculo de conjuntos: unión, intersección, diferencia y diferencia simétrica

Dados dos conjuntos `s1` y `s2`, el PDF los representa con un diagrama de Venn: `s1 = {1, 2, 3, 4, 5, 6}` y `s2 = {4, 5, 6, 7, 8, 9}`. Los operadores y métodos son:

| Cálculo | Operador | Método | Resultado con `s1` y `s2` |
|---|---|---|---|
| Unión | `s1 \| s2` | `s1.union(s2)` | `{1, 2, 3, 4, 5, 6, 7, 8, 9}` |
| Intersección | `s1 & s2` | `s1.intersection(s2)` | `{4, 5, 6}` |
| Diferencia | `s1 - s2` | `s1.difference(s2)` | `{1, 2, 3}` |
| Diferencia simétrica | `s1 ^ s2` | `s1.symmetric_difference(s2)` | `{1, 2, 3, 7, 8, 9}` |

La diferencia simétrica es la unión menos la intersección: `(s1 ∪ s2) - (s1 ∩ s2)`.

**Referencia:** PDF parte 2, págs. 150–156

````python
s1 = {1, 2, 3, 4, 5, 6}
s2 = {4, 5, 6, 7, 8, 9}

# Con operadores (pág. 155)
print(s1 | s2)    # unión
print(s1 & s2)    # intersección
print(s1 - s2)    # diferencia
print(s1 ^ s2)    # diferencia simétrica
````

````python
# Con métodos (pág. 156): mismos resultados
print(s1.union(s2))
print(s1.intersection(s2))
print(s1.difference(s2))
print(s1.symmetric_difference(s2))
````

**Clave:** los operadores `|`, `&`, `-`, `^` dan el mismo resultado que los métodos `union`, `intersection`, `difference`, `symmetric_difference`.

**Error común:** que `s1 - s2` es lo mismo que `s2 - s1`. La diferencia no es conmutativa; la simétrica sí lo es.

**Para el quiz:** `s1 & s2` es la intersección y `s1 ^ s2` es la diferencia simétrica.

**Nota sobre el PDF:** en las págs. 153, 154 y 158 los nombres de los métodos aparecen traducidos al español (`diferencia`, `diferencia_simétrica`, `sindicato`, `intersección`, `añadir`, `descartar`, `borrar`). Eso no funciona en Python: los nombres reales son `difference`, `symmetric_difference`, `union`, `intersection`, `add`, `discard` y `clear`. Las págs. 156 y 157 sí muestran los nombres correctos en inglés.

**Referencia:** PDF parte 2, págs. 150–156, 158

### Cálculos encadenados

Se pueden encadenar varios operadores con tres o más conjuntos. Con `s1 = {1, 2, 3, 4, 5, 6}`, `s2 = {4, 5, 6, 7, 8, 9}` y `s3 = {5, 6, 9, 10, 11}`: `s1 & s2 & s3` es `{5, 6}` y `s1 - s2 - s3` resta de `s1` los elementos de `s2` y de `s3`, dando `{1, 2, 3}`.

**Nota sobre el PDF:** en la pág. 157 el comentario de `s1 & s2 & s3` dice "Encuentra la unión de s1, s2, s3", pero la operación es una intersección (el propio diagrama muestra `s1 & s2 & s3 = {5, 6}`). El comentario del código de abajo está corregido.

**Referencia:** PDF parte 2, pág. 157

````python
s1 = {1, 2, 3, 4, 5, 6}
s2 = {4, 5, 6, 7, 8, 9}
s3 = {5, 6, 9, 10, 11}

print(s1 & s2 & s3)   # intersección de s1, s2 y s3
print(s1 - s2 - s3)   # resta de s1 los elementos de s2 y de s3
````

### Subconjuntos, superconjuntos y conjuntos disjuntos

Si todos los elementos de A pertenecen a B, A es **subconjunto** de B y B es **superconjunto** de A. Dos conjuntos son **disjuntos** si no comparten ningún elemento.

**Referencia:** PDF parte 2, págs. 159–160

````python
s1 = {1, 2, 3, 4, 5}
s2 = {1, 2, 3}
s3 = {1, 2, 6}

print(s2.issubset(s1))      # True: s2 es subconjunto de s1
print(s3.issubset(s1))      # False: el 6 no está en s1
print(s1.issuperset(s2))    # True: s1 es superconjunto de s2
print(s1.issuperset(s3))    # False

a = {1, 2, 3}
b = {10, 20, 30}
print(a.isdisjoint(b))      # True: no comparten elementos
````

**Clave:** `issubset` e `issuperset` devuelven booleanos; `isdisjoint` es `True` cuando la intersección está vacía.

**Error común:** confundir el sentido: `s2.issubset(s1)` pregunta si `s2` está contenido en `s1`, no al revés.

**Para el quiz:** `{1, 2, 3}.isdisjoint({10, 20, 30})` es `True`.

**Referencia:** PDF parte 2, págs. 159–160

### Producto cartesiano (Un paso más)

El producto cartesiano `A × B` es el conjunto de todas las tuplas `(a, b)` con `a` en `A` y `b` en `B`. Con `A = {1, 3}` y `B = {2, 4}` resulta `{(1, 2), (1, 4), (3, 2), (3, 4)}`. El PDF lo construye con un doble bucle `for` que va uniendo (`|`) tuplas al conjunto `res`. Como el conjunto no admite duplicados, no quedan tuplas repetidas. En el nombre `AxB`, la `x` no es el operador de multiplicación.

**Referencia:** PDF parte 2, págs. 161–162

````python
A = {1, 3}            # elementos del conjunto A
B = {2, 4}            # elementos del conjunto B
res = set()
for i in A:
    for j in B:
        res = res | {(i, j)}   # une la tupla (i, j) al conjunto de productos
AxB = res             # producto cartesiano de A y B (no A*B)
print('A =', A)
print('B =', B)
print('AxB =', AxB)
````

**Nota sobre el PDF:** en la pág. 162 el código aparece parcialmente tapado por un comentario; el bloque anterior es la reconstrucción coherente con las líneas visibles y con la explicación "Línea 4-7" de la misma página.

**Ampliación (no está en el PDF):** la biblioteca estándar ofrece `itertools.product(A, B)`, que genera las mismas parejas.

````python
# Ampliación (no está en el PDF): el mismo resultado con itertools.product
from itertools import product
print(set(product(A, B)) == AxB)
````

## 5. Comparación de conjuntos

Para comparar conjuntos se usan operadores:

- `==` indica si dos conjuntos tienen exactamente los mismos elementos (y `!=` si no).
- `<=` indica si un conjunto es subconjunto de otro.
- `<` indica si es **subconjunto propio**: está contenido en el otro y no es igual a él.

El PDF explica el subconjunto propio así: `B` es subconjunto de `A` y `B` no es igual a `A`.

**Referencia:** PDF parte 2, pág. 163

````python
A = {1, 2, 3}
B = {1, 2, 3}
print(A == B)         # True: mismos elementos

A = {1, 2, 3, 4, 5}
B = {1, 2, 3}
print(B < A)          # True: B es subconjunto propio de A
print(B <= A)         # True: B es subconjunto de A
print(A < A, A <= A)  # False True: un conjunto no es subconjunto propio de sí mismo
````

**Clave:** `<` exige que haya al menos un elemento de más en el conjunto mayor; `<=` admite igualdad.

**Error común:** usar `<` esperando que `A < A` sea verdadero; es falso porque `A` es igual a `A`.

**Para el quiz:** `B < A` significa "B es subconjunto propio de A".

**Ampliación (no está en el PDF):** también existen `>` y `>=` para superconjuntos propios y superconjuntos.

**Referencia:** PDF parte 2, pág. 163

## 6. Sentencias de iteración y orden de los conjuntos

El conjunto es un objeto iterable, de modo que se puede recorrer con `for`. Como no tiene orden, la secuencia de impresión puede diferir del orden de entrada. Para imprimirlo ordenado se usa la función `sorted()`, que devuelve una lista ordenada.

**Referencia:** PDF parte 2, págs. 164–165

````python
numbers = {2, 1, 3}
for x in numbers:
    print(x, end=" ")
print()
for x in sorted(numbers):      # impresión ordenada
    print(x, end=" ")
print()
````

**Clave:** `sorted(conjunto)` no modifica el conjunto; devuelve una lista nueva ordenada.

**Error común:** suponer que el orden de recorrido de un conjunto es el de inserción.

**Para el quiz:** para imprimir los elementos de un conjunto en orden se usa `sorted`.

**Referencia:** PDF parte 2, págs. 164–165

## 7. Declaración condicional en conjuntos: compresión de conjuntos

Un conjunto se puede crear con la sintaxis abreviada `{expresión for variable in iterable}`. Es el equivalente en conjunto de la lista por comprensión: se recorre el iterable de entrada y cada elemento resultante pasa a formar parte del conjunto, sin duplicados. Para recibir sólo algunos elementos se agrega `if` después del bucle `for`:

`{expresión for variable in iterable if condición}`

Ejemplo del PDF: extraer los caracteres de `'pineapple'` que no son `'a'`, `'p'` ni `'l'`.

**Referencia:** PDF parte 2, págs. 166–167

````python
a = {i for i in 'apple'}
print(sorted(a))                         # ['a', 'e', 'l', 'p']: la 'p' repetida desaparece

a = {i for i in 'pineapple' if i not in 'apl'}
print(sorted(a))                         # ['e', 'i', 'n']
````

**Clave:** el `if` va después del `for` y filtra qué elementos entran en el conjunto.

**Error común:** poner el `if` antes del `for`; en esa forma abreviada sólo se admite después del bucle.

**Para el quiz:** `{i for i in 'pineapple' if i not in 'apl'}` produce `{'e', 'i', 'n'}`.

**Nota sobre el PDF:** en la pág. 167 el texto dice "alfabetos 'a', 'p' y 'l'" y "los alfabetos 'e', 'I' y 'n'"; "alfabetos" es una traducción de "letras" y la `'I'` mayúscula debe ser `'i'`. El texto también menciona 'piña' cuando el código usa `'pineapple'`. En la pág. 166 la salida `{'a', 'e', 'l', 'p'}` está dada sin orden fijo.

**Referencia:** PDF parte 2, págs. 166–167

## 8. Funciones aplicables a conjuntos

Muchas funciones de Python procesan los elementos de un conjunto: `len`, `max`, `min`, `sorted`, `sum`. También `all` y `any`, que evalúan cada elemento como booleano: `all` devuelve `True` si todos son verdaderos y `any` devuelve `True` si al menos uno lo es.

**Referencia:** PDF parte 2, págs. 172–174

````python
a_set = {1, 5, 4, 3, 7, 4}     # conjunto de seis valores escritos, uno repetido
print(len(a_set))      # 5: no cuenta el duplicado
print(max(a_set))      # 7: el mayor
print(min(a_set))      # 1: el menor
print(sorted(a_set))   # lista ordenada y sin duplicados
print(sum(a_set))      # 20: el duplicado sólo cuenta una vez
````

````python
a_set = {1, 0, 2, 3, 3}
print(all(a_set), any(a_set))   # all: False porque hay un 0; any: True porque hay algún verdadero
````

**Clave:** `len` cuenta elementos únicos; `sum` suma elementos únicos.

**Error común:** esperar que `sum` repita los valores duplicados; ya se eliminaron al crear el conjunto.

**Para el quiz:** `all({1, 0, 2})` es `False` porque el `0` es falso.

**Nota sobre el PDF:** en la pág. 173 el comentario de `min` dice "el número más pequeño ... es 7", pero el mínimo es 1 (la salida mostrada también es 1). El texto menciona `stored`, que es `sorted`. En la pág. 174, el comentario "¿Hay - en a_set? check" no tiene sentido; se trata de comprobar si todos los elementos son verdaderos y si alguno lo es.

**Referencia:** PDF parte 2, págs. 172–174

## 9. Divisores y máximo común divisor con conjuntos

El PDF usa los conjuntos para pensar de forma programática: obtener los divisores propios de un número y luego el mayor divisor común de dos números.

1. Divisores de un número `num`: recorrer `i` desde 2 hasta `num - 1` y guardar los `i` con `num % i == 0`. Se empieza en 2 para excluir el 1 y se termina antes de `num` para excluir al propio número. Con `num = 10` salen `[2, 5]`.
2. Para 48 y 60 se declara un conjunto vacío por número (`divisors1`, `divisors2`) y se añaden los divisores con `add`.
3. El **máximo común divisor** es el mayor elemento de la intersección: `max(divisors1.intersection(divisors2))`. Para 48 y 60 resulta 12.

El PDF aclara que el método de Euclides es más rápido, pero este camino enseña a resolver problemas paso a paso.

**Referencia:** PDF parte 2, págs. 175–177

````python
num = 10
divisors = []

for i in range(2, num):
    if num % i == 0:
        divisors.append(i)

print(num, 'divisores propios:', divisors)
````

````python
num1 = 48
divisors1 = set()
for i in range(2, num1):
    if num1 % i == 0:
        divisors1.add(i)
print(num1, 'divisores propios:', divisors1)

num2 = 60
divisors2 = set()
for i in range(2, num2):
    if num2 % i == 0:
        divisors2.add(i)
print(num2, 'divisores propios:', divisors2)

print(divisors1.intersection(divisors2))
print(num1, num2, 'máximo divisor común:', max(divisors1.intersection(divisors2)))
````

**Clave:** la intersección de los conjuntos de divisores contiene los divisores comunes; `max` elige el mayor.

**Error común:** (ampliación) este método omite el 1 y el propio número. Con `12` y `24` daría 6 en lugar de 12, porque 12 no está en `range(2, 12)`. Para un cálculo exacto hay que incluir ambos extremos, como en el ejercicio 5.

**Para el quiz:** el máximo común divisor de 48 y 60 es 12.

## 10. Agregación con `zip`

Las listas, los diccionarios, los conjuntos y las tuplas se llaman tipos de datos **iterables** (en el PDF, "repetibles"). La función `zip(*iterables)` recibe varios iterables y devuelve un iterador de tuplas: la tupla `i` agrupa los elementos `i` de cada iterable. Esto se denomina agregación.

- Sin argumentos, `zip()` devuelve un iterador vacío.
- Si los iterables tienen distinta longitud, se descartan los elementos sobrantes de los más largos.
- Con una lista de cadenas, cada elemento aparece como una tupla de un elemento: `('a',)`, no `'a'`.
- Para desempaquetar, se llama a `zip(*lista_de_tuplas)`; el operador `*` delante del argumento recupera los datos originales.

**Referencia:** PDF parte 2, págs. 178–183

````python
empty_iterator = zip()
result = set(empty_iterator)
print(result)      # set(): zip sin argumentos devuelve un iterador vacío
````

````python
a = [10, 20, 30]                       # lista a
b = ('ten', 'twenty', 'thirty')        # tupla b
for val in zip(a, b):
    print(val)                         # tuplas creadas agregando la lista a y la tupla b
````

````python
lst = ['a', 'b', 'hello', 'this']
my_iterator = zip(lst)
result = set(my_iterator)
print(sorted(result))   # cada cadena queda dentro de una tupla de un elemento
````

````python
str_list = ['hello', 'world', 'python', 'rocks']
int_tuple = (1, 2, 3)
int_list = [400, 500, 600, 700, 800]
my_iterator = zip(str_list, int_tuple, int_list)
print(list(my_iterator))   # se descartan los elementos que sobran: 'rocks', 700 y 800
````

````python
zip_lst = list(zip(str_list, int_tuple, int_list))
a, b, c = zip(*zip_lst)    # desempaquetar las tuplas
print(a, b, c)
````

**Clave:** `zip` se detiene en el iterable más corto; `zip(*...)` hace la operación inversa.

**Error común:** olvidar que `zip` devuelve un iterador: para verlo hay que convertirlo con `list(...)`, `set(...)` o recorrerlo con `for`. Además, un iterador sólo se recorre una vez.

**Para el quiz:** `list(zip([1, 2, 3], ['a', 'b']))` es `[(1, 'a'), (2, 'b')]`.

### Problema: probabilidad en el Monopoly con tres dados

Se lanzan tres dados de seis caras. Se debe (1) imprimir el número total de casos posibles y (2) calcular, para cada valor `i` entre 3 y 18, la probabilidad de que la suma de los tres dados sea mayor o igual que `i`. Con esos porcentajes se decide qué casillas conviene comprar.

Reglas del PDF: se usan tres dados (en el juego típico son dos); cada caso se guarda como una tupla `(i, j, k)` y se almacena en un `set`, de modo que los casos repetidos no se duplican.

### Plan de programación (pseudocódigo del PDF)

Programa 1, casos posibles:
1. Inicio.
2. Declarar el conjunto que contiene todos los casos de los dados (`cases_3times = set()`).
3. Tres bucles anidados `for` con valores de 1 a 6 (`i`, `j`, `k`).
4. Cada resultado se guarda como tupla `(i, j, k)` con `add`.
5. Guardar el número de casos en `total_cases` con `len`.
6. Imprimir el número de casos. Fin.

Programa 2, probabilidades:
1. Partir de `cases_3times` y de `total_cases`.
2. Para `i` de 3 a 18 (`range(3, 19)`): poner `n_cases = 0`.
3. Para cada caso `c` en `cases_3times`: si `sum(c) >= i`, sumar 1 a `n_cases`.
4. `prob = n_cases * 100 / total_cases` e imprimir la probabilidad.

**Referencia:** PDF parte 2, págs. 131–138

### Código final n.º 1: todos los casos posibles

````python
cases_3times = set()
for i in range(1, 7):
    for j in range(1, 7):
        for k in range(1, 7):
            cases_3times.add((i, j, k))     # cada caso es una tupla (i, j, k)

total_cases = len(cases_3times)
print('Los sucesos que pueden ocurrir al tirar los dados 3 veces son', total_cases, 'casos.')
````

Resultado esperado según el PDF: 216 casos (6 × 6 × 6).

**Referencia:** PDF parte 2, págs. 135 y 139

### Código final n.º 2: probabilidad de obtener una suma mayor o igual que `i`

````python
for i in range(3, 19):          # encuentra la probabilidad de obtener más de 3 a más de 18
    total_cases = len(cases_3times)
    n_cases = 0
    for c in cases_3times:
        if sum(c) >= i:
            n_cases += 1
    prob = n_cases * 100 / total_cases
    print('la probabilidad de obtener más de {:2d} es {:6.2f}%'.format(i, prob))
````

Resultado esperado según el PDF: 100,00 % para 3; 99,54 % para 4; 98,15 % para 5; ...; 50,00 % para 11; ...; 0,46 % para 18. La ejecución anterior reproduce esos valores (con punto decimal).

**Referencia:** PDF parte 2, págs. 184–187

````python
mylist = [(1, 2), (4, 5), (4, 2), (3, 1), (9, 4)]

def consultar_par(mylist, a, b):
    """Devuelve el mensaje que corresponde a la búsqueda del par (a, b) en mylist."""
    if (a, b) in mylist:
        return f"Hay ({a},{b}) en la primera."
    elif (b, a) in mylist:
        return f"No hay ({a},{b}) pero hay ({b},{a}) en el segundo."
    else:
        return f"No hay ({a},{b}) ni ({b},{a})"

def ejecutar(mylist):
    """Versión interactiva: pide dos números enteros y muestra el resultado."""
    a, b = map(int, input("Introduzca dos números enteros: ").split())
    print(consultar_par(mylist, a, b))

# Versión no interactiva: procesa una lista de comandos de prueba (las del PDF)
for comando in ["1 2", "5 4", "3 9"]:
    a, b = map(int, comando.split())
    print("Introduzca dos números enteros:", comando)
    print(consultar_par(mylist, a, b))
    print()

# Para usarlo de forma interactiva, quita el comentario de la siguiente línea
# ejecutar(mylist)
````

Genera el conjunto `{'e', 'i', 'n'}`: recorre las letras y conserva las que no están en `'apl'`. En el Monopoly se usa un conjunto de tuplas `(i, j, k)` porque no admite duplicados y permite contar los casos con `len`; hay 216 casos posibles con tres dados.
</details>
