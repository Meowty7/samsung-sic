## Lista, tupla, diccionario y conjunto

| | Lista | Tupla | Diccionario | Conjunto |
|---|---|---|---|---|
| Sintaxis | `[1, 2]` | `(1, 2)` | `{'a': 1}` | `{1, 2}` |
| Vacío | `[]` | `()` | `{}` | `set()` |
| Mutable | Sí | No | Sí | Sí |
| Ordenado | Sí | Sí | Sí, por inserción | No |
| Acceso | Índice `l[0]` | Índice `t[0]` | Clave `d['a']` | Solo `in` |
| Duplicados | Sí | Sí | Claves no | No |
| Uso típico | Colección que cambia | Datos fijos, registros | Buscar por nombre | Quitar duplicados, operaciones de conjuntos |

**Clave:** si necesitas buscar por posición, lista o tupla; si buscas por nombre, diccionario; si solo te importa si algo está o no, conjunto.

## Unidad 10 — Tipos de datos de lista y tupla

[Ir a la unidad](clase-1-listas-tuplas-diccionarios/u10-listas-y-tuplas.ipynb)

```python
lista = [87, 84, 95]  # crea una lista
lista[0], lista[-1]  # primer y último elemento
len(lista), max(lista), min(lista), sum(lista)  # longitud, máximo, mínimo y suma
lista.append(x)  # agrega un elemento al final
lista.extend([x1, x2])  # agrega cada elemento de otra lista
lista.insert(i, x)  # inserta x en la posición i
del lista[i]  # borra por posición (instrucción, no método)
lista.pop(i)  # borra por posición y devuelve el elemento
lista.remove(x)  # borra el primer elemento igual a x
x in lista, x not in lista  # comprueba pertenencia
lista[inicio:fin:paso]  # rebanado; fin no se incluye
lista[::-1]  # lista en orden inverso
lista.sort(reverse=True)  # ordena en el lugar, descendente
colores = ("red", "green")  # tupla inmutable
a, b = (1, 2)  # asignación múltiple con una tupla
```

**Errores comunes**

- Usar un índice igual o mayor que len(lista): produce IndexError.
- Llamar a remove con un valor ausente (ValueError) sin comprobar antes con `in`.
- Confundir append([a, b]) (agrega una lista anidada) con extend([a, b]).
- Creer que lista[1:5] incluye el índice 5: el índice final no se incluye.
- Intentar modificar una tupla con tupla[0] = x: produce TypeError.
- Escribir lista = lista.sort(): sort modifica la lista y devuelve None.

## Unidad 12 — Tipos de datos de secuencia

[Ir a la unidad](clase-1-listas-tuplas-diccionarios/u12-iteracion-y-busqueda.ipynb)

```python
x in secuencia  # True si x es un elemento (en cadenas busca subcadenas)
x not in secuencia  # True si x no está
a + b  # enlaza secuencias del mismo tipo (no sirve con range)
secuencia * 3  # repite la secuencia 3 veces (el multiplicador es un entero)
secuencia.count(x)  # número de apariciones de x
len(secuencia)  # número de elementos
secuencia[0]  # primer elemento (índices de 0 a n-1)
secuencia[-1]  # último elemento
secuencia[1:5]  # elementos de los índices 1 a 4
secuencia[:-2]  # todo salvo los dos últimos
list(range(2, 11, 2))  # [2, 4, 6, 8, 10]
ord('C')  # 67, código ASCII del carácter
t = (100,)  # tupla de un elemento (la coma es obligatoria)
a, b = b, a  # intercambia dos valores
tuple(sorted(t))  # tupla ordenada (la tupla no tiene sort)
```

**Errores comunes**

- Escribir `(100)` creyendo que es una tupla: es un entero. Hace falta la coma: `(100,)`.
- Usar `+` o `*` con `range` (`TypeError`): convertir antes con `list(...)` o `tuple(...)`.
- Multiplicar dos secuencias (`list1 * list2`): el multiplicador siempre debe ser un entero.
- Suponer que el corte incluye el índice final, o que el primer índice es 1.
- Intentar `tupla[0] = valor` o `tupla.sort()`: la tupla es inmutable.
- Acceder a `secuencia[len(secuencia)]`: el último índice es `len(secuencia) - 1`.

## Unidad 13 — Listas bidimensionales

[Ir a la unidad](clase-1-listas-tuplas-diccionarios/u13-listas-bidimensionales.ipynb)

```python
m = [[1, 2, 3], [4, 5, 6]]  # lista bidimensional de 2 filas y 3 columnas
m[0][2]  # fila 0, columna 2 (devuelve 3)
m[1][1] = 300  # asigna a la fila 1, columna 1
len(m), len(m[0])  # número de filas y de columnas de la fila 0
for x in fila:  # dentro de 'for fila in m:' recorre los elementos de cada fila (doble bucle)
[[0 for j in range(2)] for i in range(3)]  # matriz 3 x 2 de ceros
[[0] * i for i in [1, 2, 3]]  # lista irregular de filas de 1, 2 y 3 elementos
copia = copy.deepcopy(m)  # copia independiente (import copy)
f = open('datos.txt', 'w')  # abre para escribir (sobrescribe)
f.write('texto\n')  # escribe y devuelve el número de caracteres
f.close()  # cierra el archivo y vacía el búfer
with open('datos.txt') as f:  # abre y cierra solo al salir del bloque
f.read()  # todo el contenido; f.readline() lee una línea
open('datos.txt', 'a+')  # añade al final y crea si no existe
```

**Errores comunes**

- Usar `for i, j, k in m` con filas de distinto tamaño: `ValueError` por no tener valores suficientes.
- Creer que `list1 = list_array` copia la lista: ambas variables comparten el mismo objeto.
- Crear la matriz con `[[0] * c] * f`: las filas son la misma lista y un cambio aparece en todas.
- Invertir los índices y escribir `m[columna][fila]` en lugar de `m[fila][columna]`.
- Abrir con `'w'` un archivo que se quería conservar, o no cerrarlo con `close` o `with`.
- Sumar valores leídos de un archivo sin convertirlos con `int`: son cadenas.

## Unidad 14 — Diccionario, método 1: añadir, editar, borrar y guardar

[Ir a la unidad](clase-2-diccionarios-y-conjuntos/u14-diccionarios-crud.ipynb)

```python
dic[clave] = valor  # añade o edita un par
dic.setdefault(k, v)  # añade k solo si falta y devuelve su valor
dic.update(a=1, b=2)  # añade o sobrescribe (claves tipo cadena)
dic.update({1: 'x'})  # update con claves que no son cadenas
dic.pop(k, None)  # borra k y devuelve su valor (o None si falta)
dic.popitem()  # borra y devuelve el último par como tupla
dic.clear()  # vacía el diccionario
dic.get(k, 0)  # lee k sin KeyError; 0 si no existe
for k, v in dic.items():  # recorre pares clave-valor
s.split(',')  # divide una cadena en una lista
s.strip()  # quita espacios de ambos extremos
','.join(lista)  # une una lista de cadenas con coma
s.replace(',', '', 1)  # sustituye la primera coma por nada
s.startswith('B', 1)  # True si desde la posición 1 empieza por 'B'
```

**Errores comunes**

- Esperar que `setdefault` sobrescriba un valor existente: solo añade claves que faltan; para sobrescribir se usa `update` o `dic[k] = v`.
- `dic.pop(k)` con una clave inexistente y sin valor por defecto lanza `KeyError`.
- `update(clave=valor)` solo sirve con nombres de identificador; las claves numéricas requieren pasar un diccionario, lista o tupla de pares.
- Olvidar reasignar el resultado de `strip`, `replace` o `lower`: las cadenas son inmutables.
- En el programa de menú, `startswith('q')` sale con cualquier entrada que empiece por q y una entrada vacía provoca `IndexError`.

## Unidad 16 — Establecer tipos de datos: conjuntos (set)

[Ir a la unidad](clase-2-diccionarios-y-conjuntos/u16-conjuntos.ipynb)

```python
set()  # conjunto vacío (no usar {})
{1, 2, 3}  # conjunto básico
set([1, 2, 2, 3])  # elimina duplicados: {1, 2, 3}
1 in s  # comprueba pertenencia
s.add(4)  # añade un elemento
s.remove(4)  # elimina (KeyError si no existe); discard no da error
s1 | s2  # unión (union)
s1 & s2  # intersección (intersection)
s1 - s2  # diferencia (difference)
s1 ^ s2  # diferencia simétrica (symmetric_difference)
s2.issubset(s1)  # True si s2 está contenido en s1
s1.isdisjoint(s2)  # True si no comparten elementos
B < A  # subconjunto propio; B <= A subconjunto
sorted(s)  # lista ordenada del conjunto
{i for i in 'pineapple' if i not in 'apl'}  # conjunto por comprensión con filtro
zip(a, b)  # agrupa en tuplas; zip(*pares) separa
```

**Errores comunes**

- Usar `{}` esperando un conjunto vacío: crea un diccionario; debe usarse `set()`.
- Indexar o trocear un conjunto (`s[0]`): no tiene orden y lanza `TypeError`.
- Suponer que el conjunto conserva el orden de inserción; para imprimir ordenado se usa `sorted`.
- Copiar los nombres de método traducidos del PDF (`diferencia`, `sindicato`, `añadir`): los reales son `difference`, `union`, `add`.
- Olvidar que `zip` devuelve un iterador y corta en el iterable más corto.
- Creer que `A < A` es verdadero: un conjunto no es subconjunto propio de sí mismo.
