## Q 04-01

### Consigna

Escriba un programa que realice las siguientes funciones utilizando una expresión condicional compuesta de una sentencia if. Guía: el código se ejecuta de forma diferente según la respuesta a la primera pregunta sea 0 o 1.

Ejemplo:

```text
¿Eres un adulto? (1 si es mayor de edad, 0 si es menor de edad): 1
¿Está casado? (1 si es casado, 0 si es soltero): 1
Usted es un adulto que está casado.
```

Las tres salidas posibles son: Usted es un adulto que está casado. / Usted es un adulto que está soltero. / Usted es menor de edad.

**Dónde se ve:** [Unidad 7](../semana-00/clase-2-control-de-flujo/u07-if-else-elif.ipynb)

### Respuesta

````
adulto = int(input("¿Eres un adulto? (1 si es mayor de edad, 0 si es menor de edad): "))

if adulto == 1:
    casado = int(input("¿Está casado? (1 si es casado, 0 si es soltero): "))
    if casado == 1:
        print("Usted es un adulto que está casado.")
    else:
        print("Usted es un adulto que está soltero.")
else:
    print("Usted es menor de edad.")
````

**Explicación:** Es un `if` anidado: solo si la primera respuesta es 1 se hace la segunda pregunta. Si es 0, se imprime directamente "Usted es menor de edad." sin preguntar el estado civil. La salida mostrada corresponde a las respuestas 1 y 1.

## Q 05-01

### Consigna

Entre los números naturales positivos distintos del 1, un número que no es primo se llama número compuesto. Escribe los números primos y compuestos del 2 al 12 (`2 : Primo`, `3 : Primo`, `4 : Compuesto`, ...).

Pauta: use la instrucción `for`. Con un `for` anidado, el lazo interno debe determinar si el número es primo.

**Dónde se ve:** [Unidad 8](../semana-00/clase-2-control-de-flujo/u08-bucle-for.ipynb), [Unidad 9](../semana-00/clase-2-control-de-flujo/u09-bucle-while.ipynb)

### Respuesta

````
for n in range(2, 13):
    es_primo = True
    for divisor in range(2, n):      # lazo interno: busca un divisor
        if n % divisor == 0:
            es_primo = False
            break
    if es_primo:
        print(n, ": Primo")
    else:
        print(n, ": Compuesto")
````

**Explicación:** El lazo externo recorre del 2 al 12. El interno prueba los divisores de 2 a n-1: si alguno divide exacto, el número es compuesto y `break` corta la búsqueda. Si ninguno divide, sigue siendo primo. `print(n, ": Primo")` separa con un espacio, igual que el formato pedido.

## Q 06-01

### Consigna

Un número de Armstrong es un número entero de tres dígitos que es igual a la suma de los cubos de cada dígito. Encuentre todos los números de Armstrong de tres dígitos e imprímalos.

Pauta: todos los números del 100 al 999 deben buscarse utilizando la instrucción `for`.

**Dónde se ve:** [Unidad 3](../semana-00/clase-1-fundamentos/u03-operadores-y-tipos-de-datos.ipynb), [Unidad 8](../semana-00/clase-2-control-de-flujo/u08-bucle-for.ipynb)

### Respuesta

````
for numero in range(100, 1000):
    centenas = numero // 100
    decenas = (numero // 10) % 10
    unidades = numero % 10
    if centenas ** 3 + decenas ** 3 + unidades ** 3 == numero:
        print(numero)
````

**Explicación:** Cada dígito se obtiene con división entera (`//`) y resto (`%`). Los números de Armstrong de tres dígitos son **153, 370, 371 y 407** (por ejemplo, 1³ + 5³ + 3³ = 1 + 125 + 27 = 153).

**Nota sobre el enunciado:** dice "imprímalos de la siguiente manera" pero no muestra el formato; se imprime uno por línea.

## Q 07-01

### Consigna

Si hay dos listas `l1` y `l2` con las siguientes cadenas, use su combinación para imprimir:

```python
l1 = ['I like','I love']
l1 = ['pancake.','kiwi juice.','espresso.']
```

```text
I like pancake.
I like kiwi juice.
I like espresso.
I love pancake.
I love kiwi juice.
I love espresso.
```

Pauta: use bucles anidados.

**Dónde se ve:** [Unidad 8](../semana-00/clase-2-control-de-flujo/u08-bucle-for.ipynb), [Unidad 13](clase-1-listas-tuplas-diccionarios/u13-listas-bidimensionales.ipynb)

### Respuesta

````
l1 = ['I like', 'I love']
l2 = ['pancake.', 'kiwi juice.', 'espresso.']

for inicio in l1:          # bucle externo: 'I like', luego 'I love'
    for final in l2:       # bucle interno: cada comida o bebida
        print(inicio, final)
````

**Explicación:** Por cada elemento de `l1` se recorre toda `l2`: 2 × 3 = 6 líneas.

## Q 09-01

### Consigna

La matriz bidimensional `a` contiene `[[1, 2], [3, 4], [5, 6]]`. Cámbiela a una matriz unidimensional como `[1, 2, 3, 4, 5, 6]` e imprímala.

Pauta: use un bucle `for`. Defina una nueva matriz y coloque los elementos de `a` en ella.

**Dónde se ve:** [Unidad 13](clase-1-listas-tuplas-diccionarios/u13-listas-bidimensionales.ipynb)

### Respuesta

````
a = [[1, 2], [3, 4], [5, 6]]
b = []                    # nueva matriz unidimensional

for fila in a:
    for valor in fila:
        b.append(valor)   # se agrega cada elemento en orden

print(b)
````

**Explicación:** El bucle externo toma cada fila y el interno cada valor de la fila; `append` los va colocando en `b`.

## Q 11-01

### Consigna

Declare una variable `school` con un diccionario anidado:

```python
school = {
    'kim':{ 'age':16, 'hei':170, grade:3 },
    'lee':{ 'age':15, 'hei':165, grade:2 },
    'cho':{ 'age':14, 'hei':167, grade:1 },
}
```

Luego, use `deepcopy()` del módulo `copy` para copiarlo a otra variable, `school2`, y verifique con el operador `is` que son variables diferentes.

Pauta: imprima el resultado de `school is school2`, que debe ser falso.

**Dónde se ve:** [Unidad 15](clase-2-diccionarios-y-conjuntos/u15-diccionarios-avanzados.ipynb)

### Respuesta

````
import copy

school = {
    'kim': {'age': 16, 'hei': 170, 'grade': 3},
    'lee': {'age': 15, 'hei': 165, 'grade': 2},
    'cho': {'age': 14, 'hei': 167, 'grade': 1},
}

school2 = copy.deepcopy(school)   # copia profunda: también copia los diccionarios internos

print(school is school2)              # False: son objetos distintos
print(school == school2)              # True: tienen el mismo contenido
print(school['kim'] is school2['kim'])  # False: los internos también son copias
````

**Explicación:** `is` compara identidad (si es el mismo objeto en memoria), no contenido. `deepcopy` crea objetos nuevos en todos los niveles, por eso incluso `school['kim'] is school2['kim']` es `False`.

**Nota sobre el enunciado:** `grade` aparece sin comillas; así daría `NameError`, porque Python lo toma como una variable. Se escribe `'grade'`.
