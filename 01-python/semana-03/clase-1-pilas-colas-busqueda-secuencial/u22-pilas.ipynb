## Qué es una pila (LIFO, push y pop)

La **pila** (stack) es un tipo de datos abstracto en el que un elemento solo se puede añadir o eliminar en **un lado**, la parte superior (la **cima**, `top`). Funciona por el método **LIFO** (Last-In-First-Out): el elemento que se introdujo al final es el primero que se saca. El otro extremo es el **fondo**.

Una pila debe proporcionar dos interfaces:

- `push`: añade un nuevo elemento en la parte superior.
- `pop`: elimina el elemento de la parte superior (y lo devuelve).

En el PDF, el diagrama muestra la pila `A, B, C` (fondo `A`, cima `C`); tras `push X` la cima es `X`, y tras `pop` sale `X` y la cima vuelve a ser `C`.

````python
pila = ["A", "B", "C"]   # el fondo es "A" y la cima es "C"
print("cima:", pila[-1])
pila.append("X")          # push X
print("tras push X:", pila)
sacado = pila.pop()       # pop
print("pop devolvió", sacado, "->", pila)
````

**Clave:** LIFO = el último en entrar es el primero en salir; solo se opera por la cima.

**Error común:** pensar que `pop` puede sacar un elemento del medio o del fondo: en una pila solo sale la cima.

**Para el quiz:** `push` añade en la cima; `pop` quita y devuelve la cima.

### Aplicación: el aparcamiento

El PDF plantea un problema con 5 coches aparcados en un callejón sin salida, y el carril del fondo también es un callejón sin salida. Los conductores llegaron en el orden 2, 3, 1, 4, 5 y los coches tienen que salir por la derecha en orden (1, 2, 3, 4, 5, según los diagramas de las págs. 23–28). La solución del PDF, paso a paso: 5 y 4 entran en el callejón del fondo, el 1 sale directamente, el 3 entra encima del 4, el 2 sale, y después salen 3, 4 y 5 en ese orden. El callejón del fondo se comporta como una **pila**: el coche que entra último es el primero en salir. El PDF también muestra un tren: si los vagones 1 a 5 entran por la izquierda y salen por la derecha pasando por una vía así, el orden se invierte.

````python
# Ampliación (no está en el PDF): simulación del aparcamiento con una lista usada como pila
def aparcamiento(callejon, mostrar=False):
    # callejon va de izquierda a derecha; el último coche es el más cercano al cruce
    sin_salida = []          # la pila: el callejón del fondo
    salida = []              # coches que ya salieron, en orden
    siguiente = 1            # próximo coche que debe salir
    for coche in reversed(callejon):
        if coche == siguiente:
            salida.append(coche)
            siguiente += 1
        else:
            sin_salida.append(coche)           # push
        while sin_salida and sin_salida[-1] == siguiente:
            salida.append(sin_salida.pop())    # pop
            siguiente += 1
        if mostrar:
            print(f"coche {coche}: pila={sin_salida} salida={salida}")
    return salida, sin_salida

print(aparcamiento([2, 3, 1, 4, 5], mostrar=True))

# El tren: apilar 1..5 y desapilar invierte el orden
vagones = []
for vagon in [1, 2, 3, 4, 5]:
    vagones.append(vagon)
print([vagones.pop() for _ in range(5)])
````

**Clave:** una pila sirve para invertir un orden y para guardar lo pendiente, recuperándolo en orden inverso al de llegada.

## Implementar una pila con una lista

En Python una pila se puede definir con el tipo de datos **lista**. Como una lista permite añadir o eliminar elementos en cualquier posición, hay que ofrecer una interfaz adicional (`push` y `pop`) que solo actúe en la cima. Por comodidad, se supone que el **último** elemento de la lista es la cima.

- `push(stack, item)` recibe la pila (lista) y el elemento; usa `append()` para añadirlo a la cima.
- `pop(stack)` recibe la pila; usa el `pop()` de la lista, que borra y devuelve el último elemento.

````python
stack = []

def push(stack, item):
    stack.append(item)

def pop(stack):
    return stack.pop()
````

El PDF prueba la estructura con este código; la salida es la que muestra la pág. 32.

````python
stack = []
print(stack)
push(stack, "A")
print(stack)
push(stack, "B")
print(stack)
pop(stack)
print(stack)
push(stack, "C")
print(stack)
pop(stack)
print(stack)
pop(stack)
print(stack)
````

## Implementar una pila como clase

En Python un tipo de datos abstracto se define como una **clase**. El PDF define la estructura `Stack` con el campo `self.stack` (una lista vacía creada en el constructor `__init__`) y los métodos `is_empty()`, `push()` y `pop()`:

- `__init__(self)` inicializa el campo `self.stack` con una lista vacía.
- `is_empty(self)` devuelve `True` si la longitud de `self.stack` es 0, y `False` si no.
- `push(self, item)` añade el elemento a la cima (con `append`).
- `pop(self)` devuelve `None` si `self.is_empty()` es `True`; si no, borra y devuelve el elemento superior.

````python
class Stack:

    def __init__(self):
        self.stack = []

    def is_empty(self):
        return True if len(self.stack) == 0 else False

    def push(self, item):
        self.stack.append(item)

    def pop(self):
        return None if self.is_empty() else self.stack.pop()
````

El PDF prueba la clase con este código. `stack` guarda la instancia creada con el constructor `Stack()`, y los métodos se llaman con el operador punto. La salida de la pág. 40 es `B`, `C`, `A` y `None`.

````python
stack = Stack()
stack.push("A")
stack.push("B")
print(stack.pop())
stack.push("C")
print(stack.pop())
print(stack.pop())
print(stack.pop())
````

**Nota sobre el PDF:** en la pág. 37 se dice que `push` "añade el elemento al principio de la lista `self.stack`". En realidad `append()` lo añade al **final** de la lista, y ese final es la cima de la pila.

**Clave:** la clase encapsula la lista (`self.stack`) y expone solo `push`, `pop` e `is_empty`.

**Error común:** olvidar `self` en la definición de los métodos o llamar a `Stack` sin paréntesis (`stack = Stack`): se guardaría la clase, no una instancia.

**Referencia:** PDF parte 1, págs. 36–40

### Un paso más: `peek()`

El `pop()` de la lista borra el elemento superior y lo devuelve. El PDF propone una función `peek()` que permite **ver** el elemento superior **sin borrarlo**. Como la cima es el último elemento, se usa el índice `-1`. También se puede definir como método de la clase `Stack`. Para no alterar la clase anterior, aquí el método se añade en una subclase (el PDF lo muestra como método de `Stack`).

````python
def peek(stack):
    return None if is_empty(stack) else stack[-1]

lista_pila = ["A", "B", "C"]
print(peek(lista_pila), lista_pila)          # C ['A', 'B', 'C']: no se borró nada
print(peek([]))                              # None

class StackConPeek(Stack):
    def peek(self):
        return None if self.is_empty() else self.stack[-1]

s = StackConPeek()
s.push(1)
s.push(2)
print(s.peek(), s.stack)
````

**Clave:** con el índice `-1` se llega a la cima: `pila[-1] == pila[len(pila) - 1]`.

**Ampliación (no está en el PDF):** el PDF define `peek` solo como función y como método suelto; `StackConPeek` es una forma de añadirlo sin modificar la clase original.

**Para el quiz:** `peek` lee la cima sin borrarla; `pop` la lee y la borra.

## Proyecto de la unidad

### Problema: verificador de paréntesis equilibrados

Escribir un algoritmo que compruebe si los paréntesis son válidos cuando se da una cadena formada por paréntesis. Se imprime "Valido" si la cadena es válida y, si no, "Invalido". Hay tres tipos de paréntesis: corchete cuadrado `[]`, llave `{}` y paréntesis redondo `()`. Los paréntesis solo son válidos si cumplen estas condiciones:

1. El tipo de paréntesis de cierre debe emparejar con el de apertura.
2. Los paréntesis deben tener un orden correcto.
3. Una cadena sin paréntesis se considera válida.

Ejemplos del PDF:

| Válido | No válido |
|---|---|
| `[]{}()` | `(` |
| `((()))` | `(]` |
| `({[]})` | `)(` |
| `[]{}()({[]})((()))` | `([()})` |
| `([]{()[]{}}())[{}]` | `([]{()[]){}}()[{}]` |

### Plan de programación (el algoritmo con una pila)

Se recorre la cadena de izquierda a derecha:

1. Si el carácter es un paréntesis **abierto**, se empuja (`push`) a la pila.
2. Si es un paréntesis **cerrado**, se saca (`pop`) el último abierto de la pila y se comprueba que sean del mismo tipo.
3. Si al terminar de recorrer toda la cadena la pila está **vacía**, los paréntesis están equilibrados.

Casos en que **no** están equilibrados:

- **Caso 1:** el paréntesis de cierre no coincide con el que sale de la pila.
- **Caso 2:** hay un paréntesis de cierre pero la pila está vacía.
- **Caso 3:** la pila no está vacía después de comprobar toda la cadena.

### Código final

La función `check_parentheses(expr)` devuelve `True` si los paréntesis coinciden y `False` si no. `opening` y `closing` están escritos de modo que los paréntesis que se corresponden ocupan la **misma posición**, y por eso `index()` comprueba la coincidencia. Se usa la clase `Stack` definida antes (el PDF escribe "chek_parentheses" en una de sus diapositivas; el nombre correcto es `check_parentheses`).

````python
def check_parentheses(expr):
    opening = "[{("            # paréntesis de apertura
    closing = "]})"            # paréntesis de cierre, en la misma posición que su apertura
    stack = Stack()            # pila donde se guardan los paréntesis abiertos
    for char in expr:
        if char in opening:
            stack.push(char)                      # abierto: se empuja a la pila
        elif char in closing:
            if stack.is_empty():
                return False                      # Caso 2: cierre con la pila vacía
            if opening.index(stack.pop()) != closing.index(char):
                return False                      # Caso 1: el cierre no coincide
    return stack.is_empty()                       # Caso 3: sobran abiertos si no está vacía
````

Se prueba con los ejemplos de las págs. 44–45 del PDF (los espacios entre los símbolos que aparecen en las salidas del PDF se ignoran, porque solo se examinan los paréntesis).

````python
validos = ["[]{}()", "((()))", "({[]})", "[]{}()({[]})((()))", "([]{()[]{}}())[{}]",
           "([ ] { ( ) [ ] { } } ())[ {} ]", ""]
no_validos = ["(", "(]", ")(", "([()})", "([]{()[]){}}()[{}]", "([]{()[]{}()[{]"]

for cadena in validos:
    verificar(f"valida {cadena!r}", check_parentheses(cadena), True)
for cadena in no_validos:
    verificar(f"no valida {cadena!r}", check_parentheses(cadena), False)
````

**Ampliación (no está en el PDF):** la misma función sirve para un código fuente completo, porque ignora los caracteres que no son paréntesis. Se prueba con el programa en C de la pág. 9 y con una versión que le falta una llave. (Limitación: no distingue paréntesis que estén dentro de cadenas de texto o comentarios.)

````python
programa_c = '''#include <stdio.h>
char *s[] = {"Hello", "World"};
int main() {
    printf("%s, %s!\\n", s[0], s[1]);
}'''
print(check_parentheses(programa_c))
print(check_parentheses(programa_c[:-1]))    # se quita la última llave
````

### Seguimiento paso a paso (Ampliación)

**Ampliación (no está en el PDF):** esta función muestra el estado de la pila en cada paso y señala cuál de los tres casos ocurre. Sirve para entender los diagramas de las págs. 47–52.

````python
def trazar(expr):
    opening, closing = "[{(", "]})"
    stack = Stack()
    for posicion, char in enumerate(expr, start=1):
        if char in opening:
            stack.push(char)
            print(f"{posicion}: push {char}  pila={stack.stack}")
        elif char in closing:
            if stack.is_empty():
                print(f"{posicion}: {char} con la pila vacía -> Caso 2")
                return False
            abierto = stack.pop()
            if opening.index(abierto) != closing.index(char):
                print(f"{posicion}: {char} no coincide con {abierto} -> Caso 1")
                return False
            print(f"{posicion}: pop  {abierto} coincide con {char}  pila={stack.stack}")
    if not stack.is_empty():
        print(f"fin: la pila no está vacía {stack.stack} -> Caso 3")
        return False
    print("fin: pila vacía -> equilibrado")
    return True

for ejemplo in ["[{()}]", "[{(}])", "[{()}])", "[[["]:
    print("Cadena:", ejemplo)
    print("Resultado:", trazar(ejemplo))
    print()
````

### Programación en pareja: etiquetas HTML coincidentes

**Pregunta Q1 del PDF:** un documento HTML consta de muchas etiquetas (`<html>`, `<body>`, `<h1>`, `<p>`, `<ul>`, `<li>`, ...). Escribir un programa que **empareje las etiquetas** del documento HTML. Es el mismo problema de los paréntesis, con etiquetas de apertura (`<p>`) y de cierre (`</p>`).

Consejo del PDF: el método `find()` localiza la posición inicial de una subcadena; se buscan `'<'` y `'>'` para distinguir las etiquetas del documento. Si la etiqueta empieza por `/`, es una etiqueta de **cierre**; si no, es de apertura. La coincidencia se comprueba con una pila, igual que con los paréntesis.

**Nota sobre el PDF:** el consejo de la pág. 67 dice "si la etiqueta incluye '/', entonces es una etiqueta abierta, y si no, es una etiqueta cerrada". Está invertido: `</p>` (con `/`) es una etiqueta de **cierre**. Además, el HTML de la pág. 65 trae un punto suelto tras `</li>.` y la entrada de ejemplo de la pág. 67 está en inglés ("Input the string of HTML document").

Primero, el bucle de extracción de etiquetas del PDF (pág. 67):

````python
text = ("<html> <body> <h1>¡Hola, mundo!</h1> <p> Estamos aprendiendo el arte de la codificación "
        "con el lenguaje de programación Python. Aquí estamos aprendiendo ... </p> <ul> "
        "<li> Estructuras de datos, </li> <li> Algoritmos, </li> "
        "<li> Desarrollo del Pensamiento Computacional. </li> </ul> </body> </html>")

start = text.find('<')
while start != -1:
    end = text.find('>', start + 1)
    tag = text[start:end + 1]
    print(tag, end=' ')
    start = text.find('<', end + 1)
print()
````

Ahora la solución completa con una pila: se apila el nombre de cada etiqueta de apertura y, al leer una de cierre, se saca de la pila y se compara.

````python
def etiquetas_html_coinciden(texto):
    stack = Stack()
    inicio = texto.find("<")
    while inicio != -1:
        fin = texto.find(">", inicio + 1)
        if fin == -1:
            return False                              # etiqueta sin cerrar el signo >
        etiqueta = texto[inicio + 1:fin]              # contenido entre < y >
        if etiqueta.startswith("/"):                  # etiqueta de cierre
            abierta = stack.pop()
            if abierta is None or abierta != etiqueta[1:].strip():
                return False                          # cierre sin apertura o distinto tipo
        else:                                         # etiqueta de apertura
            stack.push(etiqueta.strip())
        inicio = texto.find("<", fin + 1)
    return stack.is_empty()

print(etiquetas_html_coinciden(text))
print(etiquetas_html_coinciden("<html><body><p>Hola</body></p></html>"))
print(etiquetas_html_coinciden("<ul><li>uno</li>"))
````

def primer_error(expr):

    pila = []                      # guarda las posiciones de los abiertos

def html_balanceado(html):
    vacias = {"br", "img", "hr", "input", "meta"}
    pila = []
    inicio = html.find("<")
    while inicio != -1:
        fin = html.find(">", inicio + 1)
        if fin == -1:
            return False
        etiqueta = html[inicio + 1:fin].strip()
        partes = etiqueta.lstrip("/").split()

## Apuntes para estudiar

**Clave**

- Pila = tipo de datos abstracto LIFO: el último en entrar es el primero en salir; solo se opera por la cima.
- Interfaz básica: `push` (añade en la cima) y `pop` (quita y devuelve la cima); `peek` solo la lee.
- Con una lista de Python: `append()` hace de `push` y `pop()` hace de `pop`; la cima es el último elemento (`pila[-1]`).
- Como clase: `Stack` guarda `self.stack = []` en `__init__` y define `is_empty()`, `push()` y `pop()`.
- `pop()` seguro: si `is_empty()` devuelve `None` en lugar de lanzar `IndexError`.
- Paréntesis equilibrados: abierto -> `push`; cerrado -> `pop` y comparar con `opening.index(...) == closing.index(...)`; al final la pila debe estar vacía.
- Aplicaciones: aparcamiento, historial del navegador, deshacer, paréntesis y etiquetas HTML.

**Errores comunes**

- Hacer `pop` sobre una pila vacía sin comprobarlo (`IndexError: pop from empty list`).
- Olvidar el Caso 3 del verificador: devolver `True` aunque queden paréntesis abiertos en la pila.
- Comparar solo cuántos paréntesis hay de cada tipo sin respetar el orden.
- Creer que `append` añade "al principio" (añade al final, que es la cima).
- Escribir `Stack` sin paréntesis al crear la instancia, o llamar a un método sin `self`.
- Invertir la regla de HTML: la etiqueta con `/` es de cierre, no de apertura.

**Para el quiz**

- LIFO = Last-In-First-Out.
- `push` = añadir en la cima; `pop` = quitar la cima.
- Tipo de datos abstracto = tipo definido por el usuario (una clase).
- Programa informático = estructuras de datos + algoritmos.
- `is_empty()` devuelve `True` si la longitud es 0.
- Tres casos de paréntesis no equilibrados: cierre que no coincide, cierre con pila vacía, pila no vacía al final.
