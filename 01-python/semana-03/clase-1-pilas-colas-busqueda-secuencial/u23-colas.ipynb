## 1. ¿Qué es una cola?

Una **cola** (*queue*) es un tipo de dato abstracto en el que solo es posible **añadir por la parte trasera** y **eliminar por la parte delantera**. Como el primer elemento que se añade es también el primero que se elimina, funciona con el método **FIFO** (*First-In-First-Out*: el primero en entrar es el primero en salir).

Una cola debe proporcionar dos interfaces importantes:

- `enqueue`: añade un nuevo elemento al final (parte trasera) de la cola.
- `dequeue`: elimina un elemento de la parte delantera (frente) de la cola y lo devuelve.

Ejemplo del PDF: con una cola `C B A` (trasera a la izquierda, frente a la derecha), `enqueue(X)` deja `X C B A`; después `dequeue()` saca `A`, que fue el primero en entrar.

Ejemplos en la vida diaria (PDF): la gente que espera para reservar un vuelo en un aeropuerto, la fila de un restaurante o de un banco; la primera persona que llegó es la primera que recibe el servicio.

**Clave:** en una cola se inserta por un extremo y se extrae por el otro. Eso la diferencia de la pila (Unidad 22), donde ambas operaciones ocurren en el mismo extremo (LIFO).

**Ampliación (no está en el PDF):** otras aplicaciones habituales de las colas son la cola de impresión, los turnos de procesos en un sistema operativo y el recorrido de grafos en anchura. La comparación con la pila es solo un resumen mío para el estudio.

**Referencia:** PDF parte 1, págs. 71, 87–88

Primero una cola mínima con una lista de Python (idea del PDF: la lista se puede usar para definir la cola). El frente de la cola es el índice 0 y la parte trasera es el final de la lista:

````python
fila_banco = []                  # cola vacía
fila_banco.append("Ana")         # enqueue: llega Ana
fila_banco.append("Luis")        # enqueue: llega Luis
fila_banco.append("Eva")         # enqueue: llega Eva
print("Fila:", fila_banco)

atendido = fila_banco.pop(0)     # dequeue: sale el primero en haber llegado
print("Se atiende a:", atendido)
print("Fila:", fila_banco)
````

## 2. Definir una cola como un tipo de clase

Igual que con la pila, la clase `Queue` guarda una lista en un campo miembro. El PDF explica el código línea por línea:

- Líneas 1–4: la estructura de datos se define como la clase `Queue`; el constructor `__init__(self)` inicia el campo `self.queue` con una lista vacía.
- Líneas 6–7: `is_empty()` devuelve `True` si la longitud de `self.queue` es 0 y `False` en caso contrario.
- Líneas 9–10: `enqueue(self, item)` recibe el elemento y lo añade al final de la lista `self.queue`.
- Líneas 12–13: `dequeue(self)` devuelve `None` si la cola está vacía; en caso contrario devuelve el elemento eliminado de la lista con `pop(0)`.

````python
class Queue:

    def __init__(self):
        self.queue = []

    def is_empty(self):
        return True if len(self.queue) == 0 else False

    def enqueue(self, item):
        self.queue.append(item)

    def dequeue(self):
        return None if self.is_empty() else self.queue.pop(0)
````

**Nota sobre el PDF:** hay descuidos de traducción en estas diapositivas. El título dice "Implementar una pila" y la diapositiva de `dequeue` dice "clase stack", cuando se trata de la clase `Queue`. Además el texto afirma que `dequeue` borra "el último elemento", pero el código usa `pop(0)`, que elimina el **primero** (el frente). El código es el correcto para una cola: sacar el último elemento sería el comportamiento de una pila.

Prueba del PDF: se crea la instancia con `Queue()` y se usa el operador `.` para llamar a `enqueue()` y `dequeue()`. Cada `print` muestra el valor devuelto por `dequeue()`, y la última llamada devuelve `None` porque la cola ya está vacía.

````python
queue = Queue()
queue.enqueue("A")
queue.enqueue("B")
print(queue.dequeue())
queue.enqueue("C")
print(queue.dequeue())
print(queue.dequeue())
print(queue.dequeue())
````

Salida esperada según el PDF (pág. 94): `A`, `B`, `C`, `None`. Al ejecutarlo, la salida coincide. Observa el orden: salen en el mismo orden en que entraron (FIFO).

**Clave:** `dequeue()` sobre una cola vacía no lanza error: devuelve `None`, por eso conviene comprobar `is_empty()` antes de usar el resultado.

**Error común:** usar `pop()` (sin índice) en lugar de `pop(0)`. `pop()` saca el último elemento y convierte la cola en una pila.

**Para el quiz:** en la clase `Queue` del PDF, `enqueue` usa `append` y `dequeue` usa `pop(0)`; el atributo es `self.queue`.

### Un paso más: el método `size()`

El PDF propone conocer cuántos elementos hay en la cola con un método `size()` que devuelve la longitud de la lista interna. Este método se necesita después en el problema de Josefo (`queue.size() > 1`). Desde aquí se usa esta versión completa de la clase:

````python
class Queue:

    def __init__(self):
        self.queue = []

    def is_empty(self):
        return True if len(self.queue) == 0 else False

    def enqueue(self, item):
        self.queue.append(item)

    def dequeue(self):
        return None if self.is_empty() else self.queue.pop(0)

    def size(self):
        return len(self.queue)


cola = Queue()
for fruta in ["Banana", "Manzana", "Tomate"]:
    cola.enqueue(fruta)
print("Elementos:", cola.size())
cola.dequeue()
print("Después de un dequeue:", cola.size(), cola.queue)
````

## 3. Problema del mundo real: el relato de Josefo

Josefo fue un historiador judío del siglo I. Después de ser derrotados por Roma, él y otras 40 personas (41 en total) escaparon a una cueva. Allí acordaron colocarse en círculo y que **cada tercera persona** se suicidara hasta que no sobreviviera nadie. Josefo pensó que eso no tenía sentido, así que hizo cálculos y ocupó la posición que le permitiría sobrevivir hasta el final; la historia dice que finalmente sobrevivió convenciendo a las otras personas restantes.

**Enunciado como problema de codificación (PDF):**

3. Se excluye a la K-ésima persona en el sentido de las agujas del reloj hasta eliminar `N-1` personas.
4. Se calcula la persona que sobrevivirá hasta el último cuando se dan dos enteros positivos `N` y `K` (`K <= N`).

### Primera solución (con una lista)

El PDF da dos versiones del código. La primera, `josephus_sequence(n, k)`, usa una lista de Python como cola (`append` para encolar y `pop(0)` para desencolar). Cada vez que se saca un elemento, el contador `j` aumenta en 1: si `j % k == 0` es el K-ésimo y se añade a la secuencia de eliminados; si no, se vuelve a encolar al final.

````python
def josephus_sequence(n, k):
    sequence = []
    queue = []
    for i in range(1, n + 1):
        queue.append(i)
    j = 1
    while len(queue) > 1:
        item = queue.pop(0)
        if j % k == 0:
            sequence.append(item)
        else:
            queue.append(item)
        j += 1
    sequence.append(queue.pop(0))
    return sequence


print(josephus_sequence(7, 3))
````

**Clave:** la lista `queue` actúa como una cola porque se agrega al final y se saca por el principio; el último que queda se añade al final de la secuencia.

**Referencia:** PDF parte 1, págs. 84–85

## Proyecto de la unidad: generador de secuencias de Josephus

**Problema:** dados el número de personas `N` y el salto `K`, devolver el orden en que son eliminadas, de modo que el último elemento de la lista es el sobreviviente.

**Plan de programación (pasos del PDF):**

1. Colocar los enteros de 1 a `N` en una cola, por orden. Como todavía no se ha eliminado a nadie, la secuencia se define como una lista vacía.
2. Sacar el primer elemento de la cola (`dequeue`) y contar. Si es el K-ésimo (`j % k == 0`), se añade al final de la secuencia; si no, se vuelve a poner en la cola (`enqueue`) por el final.
3. Repetir hasta que solo quede un elemento en la cola.
4. Sacar ese último elemento y añadirlo al final de la secuencia.

Las diapositivas de las págs. 98–112 muestran este proceso con `N = 7` y `K = 3`: la cola `7 6 5 4 3 2 1` (trasera a la izquierda, frente a la derecha), el `dequeue` del 1 con `recuento = 1`, la reinserción por la parte trasera y así sucesivamente. En el tercer `dequeue` (recuento = 3) sale el 3 y la secuencia pasa a `[3]`; después `[3, 6]`, `[3, 6, 2]`, `[3, 6, 2, 7]`, `[3, 6, 2, 7, 5]`, `[3, 6, 2, 7, 5, 1]` y, al final, el 4 queda solo y se añade: `[3, 6, 2, 7, 5, 1, 4]`.

**Referencia:** PDF parte 1, págs. 97–112

### Código final con la clase `Queue`

Explicación del PDF por líneas:

- Líneas 1–3: `josephus(n, k)` devuelve la permutación de Josephus en una lista; `sequence` se inicia como lista vacía y `queue` como objeto de la clase `Queue`.
- Líneas 4–5: coloca los enteros de 1 a `N` en la cola.
- Líneas 6–7 y 13: `j` es un contador de los elementos sacados de la cola (empieza en 1) y aumenta en 1 en cada vuelta hasta que solo queda un elemento.
- Líneas 8–12: se saca un elemento; si `j % k == 0` se añade a la secuencia; si no, se vuelve a encolar.
- Líneas 14–15: al terminar el bucle queda un elemento, que se saca y se añade al final de la secuencia.

````python
def josephus(n, k):
    sequence = []                       # eliminados, en orden
    queue = Queue()                     # cola con la clase definida antes
    for i in range(1, n + 1):           # personas del 1 al N
        queue.enqueue(i)
    j = 1                               # contador de elementos sacados
    while queue.size() > 1:
        item = queue.dequeue()
        if j % k == 0:                  # es el K-ésimo: queda eliminado
            sequence.append(item)
        else:                           # no es el K-ésimo: vuelve al final
            queue.enqueue(item)
        j += 1
    sequence.append(queue.dequeue())    # el último que queda
    return sequence


print(josephus(7, 3))
print(josephus(41, 3))
````

Resultado esperado según el PDF (pág. 118): para `N = 7` y `K = 3` la salida es `[3, 6, 2, 7, 5, 1, 4]`, y para `N = 41` y `K = 3` es `[3, 6, 9, 12, 15, 18, 21, 24, 27, 30, 33, 36, 39, 1, 5, 10, 14, 19, 23, 28, 32, 37, 41, 7, 13, 20, 26, 34, 40, 8, 17, 29, 38, 11, 25, 2, 22, 4, 35, 16, 31]`. Los dos últimos, `16` y `31`, son las posiciones que ocuparon Josefo y la otra persona que sobrevivió en el relato.

**Nota sobre el PDF:** los nombres y los mensajes de `input()` varían entre diapositivas: la pág. 85 usa `josephus_sequence` con una lista, la pág. 113 usa `josephus` con la clase `Queue` y la pág. 118 muestra mensajes de `input` en inglés (`Input the number of people(N)`) mientras la pág. 84 los muestra en español. Son la misma lógica; aquí se mantienen los identificadores del PDF y los mensajes se dan en español.

### Versión interactiva y versión no interactiva

El PDF pide `N` y `K` con `input()` (págs. 84 y 118). Para que el código pueda probarse sin escribir, la entrada se encapsula en una función y la lógica de decisión (`josephus`) queda separada. En la plataforma del curso `input()` abre una ventana emergente y `print()` se ve al final; si el usuario cancela, `input` devuelve vacío o `None` y la función debe terminar sin error.

````python
def leer_entero(mensaje):
    """Pide un entero con input(). Devuelve None si el usuario cancela o no escribe nada."""
    texto = input(mensaje)
    if texto is None or texto.strip() == "":
        return None
    try:
        return int(texto)
    except ValueError:
        return None


def josefo_interactivo():
    """Pide N y K al usuario e imprime la secuencia de Josephus. Termina si se cancela."""
    n = leer_entero("Ingrese el número de personas (N): ")
    if n is None:
        print("Entrada cancelada o no válida. Fin.")
        return
    k = leer_entero("Ingrese el número de posiciones que se desean saltar (K): ")
    if k is None:
        print("Entrada cancelada o no válida. Fin.")
        return
    if n < 1 or k < 1:
        print("N y K deben ser enteros positivos.")
        return
    print(josephus(n, k))


def josefo_no_interactivo(casos):
    """Misma salida que la versión interactiva, pero con una lista de pares (N, K) de prueba."""
    for n, k in casos:
        print(f"N = {n}, K = {k} -> {josephus(n, k)}")


josefo_no_interactivo([(7, 3), (5, 2), (10, 1), (6, 6)])

# Para usarlo de forma interactiva, quita el comentario de la siguiente línea
# josefo_interactivo()
````

**Para el quiz:** el problema de Josefo se resuelve con una cola porque "saltar" `K-1` personas equivale a sacar y volver a poner al final; la condición `j % k == 0` identifica al eliminado. El generador devuelve el orden de eliminación completo.

**Ampliación (no está en el PDF):** el sobreviviente se puede calcular sin simular, con la recurrencia `J(1) = 0`, `J(i) = (J(i-1) + k) % i` (posición contando desde 0). Se comprueba abajo contra el último elemento de la secuencia.

````python
def sobreviviente_formula(n, k):
    """Ampliación: sobreviviente de Josephus con la recurrencia (sin simular la cola)."""
    r = 0
    for i in range(2, n + 1):
        r = (r + k) % i
    return r + 1


for n, k in [(7, 3), (41, 3), (10, 2), (12, 5)]:
    verificar(f"fórmula = último de la secuencia para N={n}, K={k}",
              sobreviviente_formula(n, k), josephus(n, k)[-1])
````

### Programación en pareja: la clase `Deque` (problema adicional)

Las págs. 123–124 son diapositivas generales sobre cómo practicar la programación en pareja (resumen): preparar pautas y mecanismos de emparejamiento y turnarse para manejar el mouse; emparejar estudiantes de habilidades similares, no necesariamente iguales; ofrecer incentivos; evaluar resultados individuales para evitar trampas; y recordar que el aprendizaje colaborativo no es "divide y vencerás" sino un esfuerzo conjunto en todo el proyecto.

La pág. 125 (Q1) pide completar y probar la clase `Deque`, una cola de doble extremo (*double-ended queue*), y la pág. 126 explica el esquema: en una cola solo se añade por un lado y se elimina por el otro, mientras que en el `Deque` la adición y la eliminación son posibles **por ambos lados** (`añadir_primero`, `añadir_último`, `eliminar_primero`, `eliminar_último`).

**Nota sobre el PDF:** la plantilla de la pág. 125 está rota por la traducción: los cuatro métodos se llaman `add_first`, `remove_first`, `add_last` y `remove_first` (el último repite el nombre, debería ser `remove_last`) y los comentarios están cruzados (por ejemplo, `add_last` dice "remover y retornar un elemento a la parte final"). Abajo se muestra la versión corregida y completada con los nombres `add_first`, `remove_first`, `add_last` y `remove_last` y el comportamiento que indica el esquema.

````python
class Deque:
    """Cola de doble extremo. Frente = índice 0; parte trasera = final de la lista."""

    def __init__(self):
        self.queue = []

    def is_empty(self):
        return len(self.queue) == 0

    def add_first(self, item):
        """Agrega un elemento a la parte inicial (frente) de la cola."""
        self.queue.insert(0, item)

    def remove_first(self):
        """Remueve y retorna el elemento de la parte inicial; None si está vacía."""
        return None if self.is_empty() else self.queue.pop(0)

    def add_last(self, item):
        """Agrega un elemento a la parte final (trasera) de la cola."""
        self.queue.append(item)

    def remove_last(self):
        """Remueve y retorna el elemento de la parte final; None si está vacía."""
        return None if self.is_empty() else self.queue.pop()


d = Deque()
d.add_last("B")
d.add_last("C")
d.add_first("A")
print(d.queue)                 # ['A', 'B', 'C']
print(d.remove_first())        # A
print(d.remove_last())         # C
print(d.queue)                 # ['B']
````

````python
verificar("add_first y add_last colocan en los extremos correctos", d.queue, ["B"])
d2 = Deque()
verificar("remove_first en vacía devuelve None", d2.remove_first(), None)
verificar("remove_last en vacía devuelve None", d2.remove_last(), None)
d2.add_last(1); d2.add_last(2); d2.add_first(0)
verificar("orden después de tres inserciones", d2.queue, [0, 1, 2])
````

**Clave:** un `Deque` generaliza la cola y la pila: con `add_last` + `remove_first` se comporta como cola; con `add_last` + `remove_last` se comporta como pila.

def rotar(elementos, k):

    for e in elementos:
        cola.enqueue(e)

def banco(eventos):
    cola = Queue()
    atendidos = []
    for evento in eventos:
        if evento == "atender":

def ultimo_sobreviviente(n, k):
    cola = Queue()
    for i in range(1, n + 1):

## Apuntes para estudiar

**Clave**

- Cola = tipo de dato abstracto FIFO: se añade por la parte trasera (`enqueue`) y se elimina por la parte delantera (`dequeue`).
- Con una lista de Python: `enqueue` = `append(item)`, `dequeue` = `pop(0)`, `is_empty` = `len(self.queue) == 0`, `size` = `len(self.queue)`.
- `dequeue` sobre una cola vacía devuelve `None` en la clase del PDF.
- Problema de Josefo: se cuentan elementos con `j`; si `j % k == 0` se elimina (va a la secuencia), si no se vuelve a encolar; el último que queda se añade al final.
- Con `N = 7` y `K = 3` el orden de eliminación es `[3, 6, 2, 7, 5, 1, 4]`; con `N = 41` y `K = 3` el sobreviviente es el 31.
- Un `Deque` (cola de doble extremo) permite añadir y eliminar por ambos lados.

**Errores comunes**

- Usar `pop()` en lugar de `pop(0)` en `dequeue`: convierte la cola en una pila.
- Olvidar la condición `is_empty()` antes de usar el resultado de `dequeue()`, que puede ser `None`.
- No incrementar `j` en cada vuelta del bucle de Josefo.
- Llamar a `queue.size()` sin haber definido el método `size()` en la clase.
- Confundir "pila" con "cola" por las traducciones del PDF (títulos que dicen "pila" y comentarios que dicen "clase stack").

**Para el quiz**

- FIFO = First-In-First-Out; la pila es LIFO.
- `enqueue` añade al final; `dequeue` quita del frente.
- La clase `Queue` guarda su lista en `self.queue` y usa `pop(0)`.
- La condición clave del problema de Josefo es `j % k == 0`.
- `josephus(7, 3)` devuelve `[3, 6, 2, 7, 5, 1, 4]`.
- Los cuestionarios #1 (`['Tomato', 'Strawberry', 'Grapes']`) y #2 (`[60, 70, 80, 90, 100]`) del PDF.
