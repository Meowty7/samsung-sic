## Problema de búsqueda: dos tipos de búsqueda

Según el PDF, hay dos formas de buscar, y la elección depende de si los datos siguen alguna regla:

- **Datos sin ninguna regla (desordenados):** hay que comprobarlos todos. Si el valor `x` no existe en la lista `S`, no se puede saberlo hasta haber comprobado todos los elementos. Esto es la **búsqueda secuencial**, que el PDF llama también "encontrar una aguja en un pajar".
- **Datos con una regla (por ejemplo, una secuencia ordenada):** el problema se resuelve de forma más eficaz con la **búsqueda binaria**, que se estudia en la unidad siguiente (unidad 25).

La búsqueda secuencial consiste en revisar los elementos uno a uno, en orden.

**Ampliación (no está en el PDF):** Python ya trae operaciones de búsqueda secuencial integradas. El operador `in` indica si un valor está en la lista y el método `index()` devuelve su posición. Ambas recorren la lista elemento por elemento, igual que el algoritmo que vamos a escribir.

````python
S = [11, 37, 45, 26, 59, 28, 17, 53]

print(45 in S)        # True: el valor está en la lista
print(88 in S)        # False: no está
print(S.index(45))    # 2: posición de la primera coincidencia
try:
    S.index(88)       # index() lanza ValueError si no encuentra el valor
except ValueError as error:
    print("ValueError:", error)
````

**Clave:** sin información sobre el orden de los datos, la única manera de estar seguro de que un valor no existe es revisarlos todos.

**Referencia:** PDF parte 1, págs. 140–141

## Búsqueda secuencial: el algoritmo

Planteamiento del PDF para encontrar un entero concreto en una lista de enteros:

- **Problema:** ¿existe un entero aleatorio `x` en `S`, el conjunto de enteros dados?
- **Entrada:** `S` como lista de enteros y `x` como el entero que se quiere encontrar.
- **Salida:** la posición (índice) de `x` si existe en `S`, y `-1` si no existe.

Funcionamiento paso a paso con `S = [11, 37, 45, 26, 59, 28, 17, 53]`:

1. Se compara `x` con el primer elemento `S[0]`. Si son iguales, se devuelve `0`. Si no, se pasa al siguiente.
2. Con `x = 45`: `S[0]` es 11 (distinto), `S[1]` es 37 (distinto) y `S[2]` es 45 (igual), por lo que se devuelve `2` y se termina.
3. **Mejor caso (caso base):** si `x` es el primer elemento (por ejemplo `x = 11`), basta una sola comparación y se devuelve `0`.
4. **Peor caso:** si `x` no existe (por ejemplo `x = 88`), se compara con los `n` elementos y se devuelve `-1`; son `n` comparaciones en total.

**Referencia:** PDF parte 1, págs. 142–147

## Búsqueda secuencial: implementación

El PDF recibe `nums` y `x` y recorre los índices con un bucle `for`:

- El bucle `for` recorre los índices de `nums`.
- Si `x == nums[i]`, se devuelve `i`.
- Si el bucle termina sin coincidencias, se devuelve `-1` porque `nums` no contiene `x`.

````python
def seq_search(nums, x):
    for i in range(len(nums)):
        if x == nums[i]:
            return i
    return -1

# Ejemplos del PDF
S = [11, 37, 45, 26, 59, 28, 17, 53]
for x in (53, 11, 77):
    pos = seq_search(S, x)
    print(f"In S, {x} is at position {pos}.")
````

**Nota sobre el PDF:** el PDF imprime los mensajes en inglés (`In S, 53 is at position 7.`) aunque el resto está traducido al español; se mantienen tal cual para que coincidan con la salida de las diapositivas. En el PDF los valores buscados de los ejemplos son 53 (posición 7), 11 (posición 0) y 77 (posición -1).

**Clave:** `-1` es el valor centinela que indica "no encontrado", porque ningún índice válido de una lista es negativo en este algoritmo.

**Error común:** devolver `-1` dentro del bucle (con un `else` mal colocado). El `return -1` debe ir **después** del bucle, cuando ya se comprobaron todos los elementos.

**Referencia:** PDF parte 1, págs. 148–149

### Contando comparaciones: mejor, peor y caso medio

**Ampliación (no está en el PDF):** para ver el costo del algoritmo, esta versión cuenta cuántas comparaciones `==` realiza.

````python
def seq_search_contando(nums, x):
    """Devuelve (posición, número de comparaciones realizadas)."""
    comparaciones = 0
    for i in range(len(nums)):
        comparaciones += 1
        if x == nums[i]:
            return i, comparaciones
    return -1, comparaciones

S = [11, 37, 45, 26, 59, 28, 17, 53]
print("Mejor caso  (x = 11):", seq_search_contando(S, 11))   # 1 comparación
print("Caso x = 45         :", seq_search_contando(S, 45))   # 3 comparaciones
print("Peor caso   (x = 88):", seq_search_contando(S, 88))   # 8 comparaciones (n)
````

## Un paso más: complejidad temporal de la búsqueda secuencial

El PDF analiza la complejidad con la notación Big-O:

- **Mejor caso:** la búsqueda termina tras una única comparación, por lo que es O(1).
- **Peor caso:** hay que comparar `n` veces, por lo que es O(n).
- **Caso medio:** si se busca una sola vez cada elemento de `S` (que tiene `N` elementos), buscar el elemento en la posición 1 cuesta 1 comparación, el de la posición 2 cuesta 2, y así hasta `N`. El total es:

  1 + 2 + ... + N = N(N + 1) / 2

  El tiempo medio por búsqueda es T(N) = N(N + 1) / 2 ÷ N = (N + 1) / 2, que pertenece a O(n): **tiempo lineal**.

**Referencia:** PDF parte 1, pág. 150

````python
# Comprobación del caso medio con el ejemplo de 8 elementos
S = [11, 37, 45, 26, 59, 28, 17, 53]
total = sum(seq_search_contando(S, x)[1] for x in S)
print("Comparaciones totales:", total)                  # 36 = 8 * 9 / 2
print("Promedio por búsqueda:", total / len(S))         # 4.5 = (8 + 1) / 2

# La media crece de forma lineal con n (tiempo lineal)
for n in (10, 100, 1000):
    datos = list(range(n))
    media = sum(seq_search_contando(datos, x)[1] for x in datos) / n
    print(f"n = {n:>4}: media = {media} (formula (n + 1) / 2 = {(n + 1) / 2})")
````

**Clave:** mejor caso O(1), peor caso O(n), caso medio (n + 1) / 2, que también es O(n). Duplicar el tamaño de la lista duplica aproximadamente el trabajo.

**Para el quiz:** el caso medio asume que el elemento buscado existe y que cualquier posición es igual de probable.

## Encontrar el mayor

El PDF aplica la misma estrategia secuencial para encontrar la clave mayor de una lista:

- **Problema:** ¿dónde está la mayor clave de `S`, un conjunto de enteros dados?
- **Entrada:** `S` como lista de números enteros.
- **Salida:** el **índice** del mayor elemento de `S`.

Idea: se supone que la mayor clave es el primer elemento (`largest = 0`). Luego se compara `S[i]` con `S[largest]` desde `i = 1` hasta el último elemento; si `S[i]` es mayor, se actualiza `largest = i`. Al terminar de recorrer todos los elementos, `largest` es el índice del valor máximo.

Con `S = [11, 37, 45, 26, 59, 28, 17, 53]` el índice del mayor evoluciona así: 0, 1, 2, 2, 4, 4, 4, 4 (el resultado es el índice 4, que contiene 59).

**Referencia:** PDF parte 1, págs. 151–160

### Implementación de `find_largest`

- La línea 2 inicializa `largest` con el índice del primer elemento.
- El bucle compara `nums[largest]` con `nums[i]`, desde el segundo elemento hasta el último.
- Si `nums[i]` es mayor, se actualiza `largest = i`.
- Al terminar, se devuelve `largest`: el índice del elemento más grande.

````python
def find_largest(nums):
    largest = 0
    for i in range(1, len(nums)):
        if nums[largest] < nums[i]:
            largest = i
    return largest

nums = [11, 37, 45, 26, 59, 28, 17, 53]
pos = find_largest(nums)
print(f"The largest is {nums[pos]} at {pos}")
````

**Nota sobre el PDF:** en la diapositiva de la pág. 151 la salida muestra primero la lista `[11, 37, 45, 26, 59, 28, 17, 53]` y después `The largest is 59 at 4`, pero el código mostrado solo tiene tres líneas y no imprime la lista. Ejecutado tal cual, solo imprime `The largest is 59 at 4`.

### Traza paso a paso

**Ampliación (no está en el PDF):** esta versión imprime las mismas columnas que las diapositivas (`i`, `largest`, `S[largest]` y `S[i]`) para seguir el algoritmo.

````python
def find_largest_traza(nums):
    largest = 0
    print(f"{'i':>2} {'largest':>8} {'S[largest]':>11} {'S[i]':>5}")
    for i in range(1, len(nums)):
        print(f"{i:>2} {largest:>8} {nums[largest]:>11} {nums[i]:>5}")
        if nums[largest] < nums[i]:
            largest = i
    print("Resultado: largest =", largest)
    return largest

find_largest_traza([11, 37, 45, 26, 59, 28, 17, 53])
````

**Error común:** guardar el **valor** mayor en lugar del **índice**, y devolver uno cuando se pide el otro. El PDF devuelve el índice; el valor se obtiene con `nums[pos]`.

**Ampliación (no está en el PDF):** la función falla con una lista vacía, porque `nums[0]` no existe. Una lista sin elementos no tiene mayor, así que conviene decidir qué devolver antes de usarla.

````python
try:
    find_largest([])
except IndexError as error:
    print("IndexError con lista vacía:", error)

def find_largest_seguro(nums):
    """Devuelve el índice del mayor o -1 si la lista está vacía."""
    if not nums:
        return -1
    return find_largest(nums)

print(find_largest_seguro([]), find_largest_seguro([4, 9, 2]))
````

## Proyecto de la unidad: el problema de la caída de huevos

### Problema

Escribir un algoritmo que encuentre el **piso más alto** desde el cual un huevo no se rompe al dejarlo caer, dada la altura (número de pisos) del edificio. Condiciones (pág. 164):

1. La **rotura** (`breaking`), que es el piso más bajo donde se rompe el huevo, se selecciona aleatoriamente entre 1 y la altura.
2. El experimento se expresa como `do_experiment(floor)`: devuelve `True` si el huevo se rompe al caer desde ese piso y `False` si no.
3. Solo hay un huevo, así que el programa debe detenerse en cuanto `do_experiment()` devuelva `True`: no se puede continuar si el huevo se rompe.

Reglas del resultado (pág. 133): si el huevo no se rompe, se prueba el piso siguiente hasta el último; si se rompe en el piso `n`, la respuesta es `n - 1`. Si no se rompe ni siquiera desde el piso máximo, la respuesta es el piso máximo; si se rompe desde el piso 1, la respuesta es `0`.

Ejemplo con 10 pisos y rotura en el piso 6 (págs. 134–136): los experimentos en los pisos 1 a 5 tienen éxito, el del piso 6 rompe el huevo, y por tanto el piso 5 es el más alto donde el huevo no se rompe.

### Plan de programación (págs. 164–169)

1. Recibir del usuario la altura del edificio con `input()` y convertirla a `int`.
2. Crear la rotura como un número aleatorio entre 1 y la altura con `randint(1, height)`.
3. Implementar `do_experiment(floor, breaking)`, que devuelve `floor >= breaking`.
4. Implementar `find_highest_safe_floor(height, breaking)`: repetir el experimento desde el piso 1 hasta `height`; si el huevo se rompe en el piso `n`, devolver `n - 1`; si resiste todos, devolver `height`.
5. Imprimir el resultado con una f-string.

**Referencia:** PDF parte 1, págs. 133–138 y 164–169

````python
def do_experiment(floor, breaking):
    """True si el huevo se rompe al caer desde `floor`."""
    return floor >= breaking

def find_highest_safe_floor(height, breaking):
    """Búsqueda secuencial del piso más alto en el que el huevo no se rompe."""
    for n in range(1, height + 1):      # se prueba desde el piso 1 hasta la altura
        if do_experiment(n, breaking):  # el huevo se rompió en el piso n
            return n - 1                # el piso anterior es el más alto seguro
    return height                       # resistió todos los pisos

# Ejemplo de las págs. 134-136: 10 pisos, rotura en el 6
print(find_highest_safe_floor(10, 6))   # 5
# Otros casos límite
print(find_highest_safe_floor(10, 1))   # 0: se rompe desde el piso 1
print(find_highest_safe_floor(10, 11))  # 10: nunca se rompe
````

**Nota sobre el PDF:** en la pág. 164 el punto 3 dice "el programa debe cerrarse cuando la función `do_experiment()` llame a `True`" (traducción literal). Significa que la búsqueda se detiene cuando el experimento devuelve `True`. Además, en la pág. 168 el texto dice "si el valor del suelo es igual o mayor que el valor de ruptura devuelve `True`", lo que coincide con `floor >= breaking`.

### Versión interactiva y no interactiva

El PDF pide la altura con `input()`. Para que el programa se pueda probar sin teclado, se encapsula la lectura: `programa_caida_de_huevos` recibe la función que lee (por defecto `input`) y la que genera el azar (por defecto `randint`). Si el usuario cancela (entrada vacía o `None`), el programa termina.

````python
from random import randint

def leer_entero(mensaje, leer=input):
    """Pide un entero. Devuelve None si el usuario cancela (entrada vacía, None o fin de entrada)."""
    while True:
        try:
            texto = leer(mensaje)
        except (EOFError, KeyboardInterrupt):
            return None
        if texto is None or texto.strip() == "":
            return None
        try:
            return int(texto)
        except ValueError:
            print("Entrada no válida: escribe un número entero.")

def programa_caida_de_huevos(leer=input, azar=randint):
    height = leer_entero("Ingrese el número de pisos: ", leer)
    if height is None or height < 1:
        print("Programa terminado: no se recibió una altura válida.")
        return None
    breaking = azar(1, height)
    floor = find_highest_safe_floor(height, breaking)
    print(f"El huevo estará a salvo hasta el piso número {floor}.")
    return floor
````

### Resultado esperado y análisis

Con 100 pisos y rotura aleatoria, el programa imprime el piso más alto seguro, que siempre es `rotura - 1`. La búsqueda secuencial hace tantos experimentos como indique la rotura. En el mejor caso, el huevo se rompe en el piso 1: un único experimento, O(1). En el peor caso, el huevo no se rompe nunca o se rompe en el último piso: `n` experimentos, O(n).

````python
def find_highest_safe_floor_contando(height, breaking):
    """Devuelve (piso seguro, número de experimentos)."""
    experimentos = 0
    for n in range(1, height + 1):
        experimentos += 1
        if do_experiment(n, breaking):
            return n - 1, experimentos
    return height, experimentos

for rotura in (1, 50, 100):
    piso, exp = find_highest_safe_floor_contando(100, rotura)
    print(f"rotura = {rotura:>3}: piso seguro = {piso:>3}, experimentos = {exp}")
````

### Un paso más: complejidad del problema de la caída de huevos

El PDF analiza (pág. 170):

- Con **un solo huevo**, igual que en la búsqueda secuencial, el mejor caso es O(1) y el peor caso es O(n); el caso medio también es O(n).
- Si hay un **número infinito de huevos** (no importa que se rompan), se puede usar la **búsqueda binaria** (unidad 25), con complejidad media O(log n).
- Con **dos huevos**, el PDF indica que este problema se llama "problema de los dos huevos" y que en un edificio de 100 plantas se necesitan al menos 14 experimentos en el peor de los casos.

**Referencia:** PDF parte 1, pág. 170

**Ampliación (no está en el PDF):** el valor 14 del problema de los dos huevos se puede comprobar. La estrategia consiste en lanzar el primer huevo en saltos decrecientes (pisos 14, 27, 39, ...), y cuando se rompe, usar el segundo huevo piso a piso en el tramo anterior. Con `k` experimentos se cubren `k(k + 1) / 2` pisos, así que se busca el menor `k` que cubra 100 pisos.

````python
def minimo_experimentos_dos_huevos(pisos):
    k = 0
    while k * (k + 1) // 2 < pisos:
        k += 1
    return k

def dos_huevos(pisos, rotura, k):
    """Devuelve (piso seguro, número de experimentos) con la estrategia de saltos decrecientes."""
    experimentos = 0
    anterior, salto = 0, k
    while True:
        actual = min(anterior + salto, pisos)    # piso donde cae el primer huevo
        experimentos += 1
        if actual >= rotura:                     # el primer huevo se rompe
            break
        if actual == pisos:                      # resistió todo el edificio
            return pisos, experimentos
        anterior = actual
        salto -= 1
    seguro = anterior                            # el segundo huevo recorre piso a piso
    for piso in range(anterior + 1, actual):
        experimentos += 1
        if piso >= rotura:
            break
        seguro = piso
    return seguro, experimentos

k = minimo_experimentos_dos_huevos(100)
print("Experimentos mínimos en el peor caso:", k)   # 14
peor = max(dos_huevos(100, rotura, k)[1] for rotura in range(1, 101))
correcto = all(dos_huevos(100, r, k)[0] == r - 1 for r in range(1, 101))
print("Peor caso medido:", peor, "| siempre correcto:", correcto)
````

def piso_mas_alto_seguro(altura, rotura):
    experimentos = 0
    for piso in range(1, altura + 1):

        if piso >= rotura:
            return piso - 1, experimentos
    return altura, experimentos

## Apuntes para estudiar

**Clave**

- La búsqueda secuencial revisa los elementos de uno en uno y sirve con datos sin ningún orden.
- Devuelve el índice del elemento o `-1` si no existe.
- Mejor caso O(1) (el elemento es el primero); peor caso O(n) (no existe o es el último); caso medio (n + 1) / 2, que es O(n): tiempo lineal.
- Encontrar el mayor usa la misma idea: guardar el índice del mayor visto hasta ahora y actualizarlo en un solo recorrido (`find_largest`).
- Caída de huevos con un huevo: probar piso a piso desde el 1; si se rompe en el piso `n`, la respuesta es `n - 1`; si resiste todo, es la altura.
- Con infinitos huevos se puede usar búsqueda binaria (unidad 25), O(log n).

**Errores comunes**

- Colocar el `return -1` dentro del bucle en lugar de después.
- Confundir el índice del mayor con el valor del mayor.
- Olvidar que `find_largest` falla con listas vacías.
- Devolver `n` en lugar de `n - 1` en el problema de los huevos: el piso donde se rompe no es el piso seguro.
- Creer que la búsqueda secuencial necesita datos ordenados: no los necesita (ese es el caso de la binaria).

- Número de comparaciones medias para N elementos: (N + 1) / 2.
