## Q 22-01

### Consigna

La siguiente es la implementación de una pila. ¿Cuál será el resultado del siguiente código?

```python
stack = Stack()
stack.push("Banana")
stack.push("Apple")
stack.push("Tomato")
stack.pop()
stack.push("Strawberry")
stack.push("Grapes")
stack.pop()
print(stack.stack)
```

**Dónde se ve:** [Unidad 22](clase-1-pilas-colas-busqueda-secuencial/u22-pilas.ipynb)

### Respuesta

````
class Stack:

    def __init__(self):
        self.stack = []

    def is_empty(self):
        return True if len(self.stack) == 0 else False

    def push(self, item):
        self.stack.append(item)

    def pop(self):
        return None if self.is_empty() else self.stack.pop()


stack = Stack()
stack.push("Banana")
stack.push("Apple")
stack.push("Tomato")
stack.pop()
stack.push("Strawberry")
stack.push("Grapes")
stack.pop()
print(stack.stack)
````

**Explicación:** **Respuesta:** `['Banana', 'Apple', 'Strawberry']`.

La pila saca siempre el último que entró (LIFO): el primer `pop` quita `Tomato` y el segundo quita `Grapes`. Para ejecutarlo se incluye la clase `Stack` de la unidad 22 (la misma del curso).

```python
stack = Stack()

```

**Dónde se ve:** [Unidad 22](clase-1-pilas-colas-busqueda-secuencial/u22-pilas.ipynb)

### Respuesta

````
class Stack:

    def __init__(self):
        self.stack = []

    def is_empty(self):
        return True if len(self.stack) == 0 else False

    def push(self, item):
        self.stack.append(item)

    def pop(self):
        return None if self.is_empty() else self.stack.pop()


stack = Stack()
items = [10 * i for i in range(1, 10)]
for item in items:
    stack.push(item)
    if (item // 10) % 2 == 0:
        stack.pop()
print(stack.stack)
````

¿Cuál es el algoritmo de la siguiente función `find_two()`? Analice el código y escriba el resultado.

```python
def find_two():
    x = y = 0
    for i in range(1, len(nums)):
        if nums[x] < nums[i]:
            x = i
        elif nums[y] > nums[i]:
            y = i
    return x,y

nums = [11, 37, 45, 26, 59, 28, 17, 53]
i,j = find_two(nums)
print(nums[i], nums[j])
```

**Dónde se ve:** [Unidad 24](clase-1-pilas-colas-busqueda-secuencial/u24-busqueda-secuencial.ipynb)

### Respuesta

````
def find_two(nums):
    x = y = 0                          # x: índice del mayor, y: índice del menor
    for i in range(1, len(nums)):
        if nums[x] < nums[i]:
            x = i
        elif nums[y] > nums[i]:
            y = i
    return x, y


nums = [11, 37, 45, 26, 59, 28, 17, 53]
i, j = find_two(nums)
print(nums[i], nums[j])
````

**Explicación:** **Algoritmo:** búsqueda secuencial que encuentra **a la vez el máximo y el mínimo** en un solo recorrido; devuelve sus índices.

**Resultado:** `59 11` (59 es el mayor y 11 el menor).

**Nota sobre el enunciado:** la función se define sin parámetros, `def find_two():`, pero se llama con `find_two(nums)`; así daría `TypeError`. Se agrega el parámetro `nums`.

## Q 24-02

### Consigna

¿Cuántas comparaciones debe realizar la función `find_two()` de la pregunta anterior (24-01)?

**Dónde se ve:** [Unidad 24](clase-1-pilas-colas-busqueda-secuencial/u24-busqueda-secuencial.ipynb)

### Respuesta

````
def find_two_contando(nums):
    x = y = 0
    comparaciones = 0
    for i in range(1, len(nums)):
        comparaciones += 1                 # nums[x] < nums[i]
        if nums[x] < nums[i]:
            x = i
        else:
            comparaciones += 1             # nums[y] > nums[i] (solo si la primera fue falsa)
            if nums[y] > nums[i]:
                y = i
    return x, y, comparaciones


nums = [11, 37, 45, 26, 59, 28, 17, 53]
x, y, total = find_two_contando(nums)
print('Comparaciones con esta lista:', total)
print('Mínimo posible (n - 1):', len(nums) - 1)
print('Máximo posible 2(n - 1):', 2 * (len(nums) - 1))
````

**Explicación:** **Respuesta:** con esta lista hace **11 comparaciones**. En general, para n elementos hace entre **n − 1 = 7** (si cada número es un nuevo máximo, el `elif` nunca se evalúa) y **2(n − 1) = 14** (si la primera comparación siempre es falsa). Recorrido: i = 1, 2 y 4 encuentran un nuevo máximo (1 comparación cada uno); i = 3, 5, 6 y 7 hacen 2 comparaciones cada uno: 3 × 1 + 4 × 2 = 11.

Código del juego de adivinar el número. Si el máximo es 100 y el número es 51, ¿cuál es la salida de `count`?

```python
maximo = int(input("Ingrese el número maximo: "))

low, high = 1, maximo

```

**Dónde se ve:** [Unidad 25](clase-2-busqueda-binaria-tablas-hash/u25-busqueda-binaria.ipynb)

### Respuesta

````
maximo = int(input("Ingrese el número maximo: "))
numero = int(input("Ingrese tu intento de adivinar el numero: "))
low, high = 1, maximo
count = 0
while low < high:
    count += 1
    mid = (low + high) // 2
    print(f"Intento {count}: low={low}, high={high}, mid={mid}")   # traza para ver cada paso
    if mid == numero:
        print(f"El número es {numero}")
        break
    elif mid > numero:
        high = mid - 1
    else:
        low = mid + 1

print(f"Total {count} veces fue buscado.")
````

**Explicación:** **Respuesta:** `count` = **6** ("Total 6 veces fue buscado."). Los valores de `mid` son 50, 75, 62, 56, 53 y 51. La línea de traza (`print` de cada intento) se agregó solo para mostrar los pasos; el valor de `count` es el mismo sin ella.

Usa `input()`: la salida de arriba es una prueba respondiendo `100`, `51`. En la plataforma se abre una ventana para cada dato.

## Q 25-02

### Consigna

En el mismo código, si el máximo es 100 y el número es 25, ¿cuál es el resultado del conteo?

**Dónde se ve:** [Unidad 25](clase-2-busqueda-binaria-tablas-hash/u25-busqueda-binaria.ipynb)

### Respuesta

````
def contar_busqueda(maximo, numero):
    low, high = 1, maximo
    count = 0
    while low < high:
        count += 1
        mid = (low + high) // 2
        print(f"Intento {count}: mid={mid}")
        if mid == numero:
            break
        elif mid > numero:
            high = mid - 1
        else:
            low = mid + 1
    return count


print('Total', contar_busqueda(100, 25), 'veces fue buscado.')
````

**Explicación:** **Respuesta:** `count` = **2**. Primer intento: `mid` = (1 + 100) // 2 = 50, mayor que 25, entonces `high` = 49. Segundo intento: `mid` = (1 + 49) // 2 = 25, lo encuentra.

### Consigna

Usando la función hash de la clase `HashTable`, calcule la clave y la clave hash de "Alice in Wonderland" (el valor clave es 1763).

```python
table = HashTable(8)
book  = "Alice in Wonderland"
key = sum(map(ord, book))
print(key, table.hash(key))
```

**Dónde se ve:** [Unidad 26](clase-2-busqueda-binaria-tablas-hash/u26-tablas-hash.ipynb)

### Respuesta

````
class HashTable:

    def __init__(self, size):
        self.size = size
        self.table = {}
        for i in range(size):
            self.table[i] = []

    def hash(self, key):
        return key % self.size

    def get(self, key):
        return self.table[self.hash(key)]

    def put(self, key, value):
        bucket = self.table[self.hash(key)]
        if value not in bucket:
            bucket.append(value)


table = HashTable(8)
book  = "Alice in Wonderland"
key = sum(map(ord, book))
print(key, table.hash(key))

# Valor clave indicado en el enunciado
key = 1763
print(key, table.hash(key))
````

**Explicación:** **Respuesta:** `1763 3` (la clave que da el enunciado y su clave hash). El código copia la consigna tal cual y además muestra el cálculo con el valor indicado: 1763 % 8 = 3 (8 × 220 = 1760).

**Nota sobre el enunciado:** el texto "Alice in Wonderland" suma 1795, no 1763; el 1763 corresponde a "Alice in WonderLand" (con L mayúscula), como aparece en el PDF de la unidad 26. La clave hash es 3 en los dos casos. Por eso la respuesta muestra las dos líneas: la del código literal (`1795 3`) y la del valor indicado (`1763 3`). Para ejecutarlo se incluye la clase `HashTable` de la unidad 26 (la misma del curso).

Si la nueva estantería tiene 10 compartimentos, use el siguiente código para averiguar qué libro hay en cada compartimento.

```python

books = [
    "The Little Prince",
```

**Dónde se ve:** [Unidad 26](clase-2-busqueda-binaria-tablas-hash/u26-tablas-hash.ipynb)

### Respuesta

````
class HashTable:

    def __init__(self, size):
        self.size = size
        self.table = {}
        for i in range(size):
            self.table[i] = []

    def hash(self, key):
        return key % self.size

    def get(self, key):
        return self.table[self.hash(key)]

    def put(self, key, value):
        bucket = self.table[self.hash(key)]
        if value not in bucket:
            bucket.append(value)


table = HashTable(10)
books = [
    "The Little Prince",
    "The Old Man and the Sea",
    "The Little Memaid",
    "Beauty and the Beast",
    "The Last Leaf",
    "Alice in Wonderland",
]

for book in books:
    key = sum(map(ord, book))
    table.put(key, book)

for key in table.table.keys():
    print(key, table.table[key])
````
