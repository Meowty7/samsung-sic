## Mapa rápido de la semana

| Estructura o algoritmo | Idea | Operaciones clave | Costo típico |
|---|---|---|---|
| Pila | LIFO: sale primero lo último que entró | `push` (`append`), `pop` (`pop()`), `peek` (`[-1]`) | O(1) |
| Cola | FIFO: sale primero lo primero que entró | `enqueue` (`append`), `dequeue` (`pop(0)` o `deque.popleft()`) | O(1) con `deque` |
| Búsqueda secuencial | Revisar uno por uno | recorrer con `for` | O(n) |
| Búsqueda binaria | Partir el rango a la mitad (lista ordenada) | `bajo`, `alto`, `medio` | O(log n) |
| Tabla hash | Clave → función hash → posición | guardar y buscar por clave (`dict`) | O(1) promedio |

**Clave:** la búsqueda binaria solo funciona sobre datos **ordenados**. Con 1 000 000 de elementos, la secuencial puede necesitar 1 000 000 de comparaciones y la binaria unas 20.

## Unidad 22 — Pila

[Ir a la unidad](clase-1-pilas-colas-busqueda-secuencial/u22-pilas.ipynb)

```python
pila = []  # una pila con una lista (la cima es el último elemento)
pila.append(x)  # push: añade x en la cima
pila.pop()  # pop: quita y devuelve la cima (IndexError si está vacía)
pila[-1]  # peek: lee la cima sin borrarla
len(pila) == 0  # comprueba si la pila está vacía
def is_empty(stack): return len(stack) == 0  # versión funcional de la comprobación
return None if is_empty(stack) else stack.pop()  # pop seguro que evita IndexError
class Stack:  # define la pila como tipo de datos abstracto
self.stack = []  # campo de la clase creado en __init__
stack = Stack()  # crea una instancia (los paréntesis son obligatorios)
stack.push("A")  # llama al método push con el operador punto
opening.index(stack.pop()) != closing.index(char)  # Caso 1: el cierre no coincide
return stack.is_empty()  # Caso 3: equilibrada solo si la pila queda vacía
texto.find("<")  # posición de la siguiente etiqueta HTML
```

**Errores comunes**

- Hacer `pop()` sobre una pila vacía sin comprobar `is_empty()` y obtener `IndexError`.
- Olvidar el Caso 3 del verificador: devolver `True` aunque queden paréntesis abiertos en la pila.
- Pensar que `append` añade al principio: añade al final, que es la cima de la pila.
- Crear la pila sin paréntesis (`stack = Stack`) u olvidar `self` en los métodos.
- Comprobar solo el número de paréntesis de cada tipo sin respetar el orden (por ejemplo `([)]`).
- En HTML, creer que la etiqueta con `/` es de apertura: `</p>` es de cierre.

## Unidad 24 — Búsqueda secuencial

[Ir a la unidad](clase-1-pilas-colas-busqueda-secuencial/u24-busqueda-secuencial.ipynb)

```python
x in S                      # True si x está en la lista (búsqueda secuencial integrada)
S.index(x)                  # posición de la primera coincidencia; ValueError si no existe
for i in range(len(S)):     # recorre los índices de la lista
if x == S[i]: return i      # coincidencia: devuelve el índice y termina
return -1                   # tras el bucle: no se encontró x
largest = 0                 # el primer elemento se supone el mayor
for i in range(1, len(S)):  # compara desde el segundo elemento
if S[largest] < S[i]: largest = i   # actualiza el índice del mayor
S[largest]                  # valor del mayor elemento
randint(1, height)          # entero aleatorio entre 1 y height (módulo random)
floor >= breaking           # True si el huevo se rompe desde ese piso
return n - 1                # piso más alto seguro cuando se rompe en el piso n
frase.split()               # lista de palabras separadas por espacios
```

**Errores comunes**

- Colocar `return -1` dentro del bucle: debe ir después, cuando ya se revisaron todos los elementos.
- Confundir el índice del mayor con el valor del mayor: `find_largest` devuelve el índice.
- Llamar a `find_largest` con una lista vacía: `nums[0]` lanza `IndexError`.
- Devolver `n` en lugar de `n - 1` en el problema de los huevos: el piso donde se rompe no es el piso seguro.
- Creer que la búsqueda secuencial necesita datos ordenados; no los necesita.

## Unidad 25 — Búsqueda binaria

[Ir a la unidad](clase-2-busqueda-binaria-tablas-hash/u25-busqueda-binaria.ipynb)

```python
low, high = 0, len(nums) - 1      # espacio de búsqueda inicial: toda la lista
while low <= high:                # mientras queden candidatos
mid = (low + high) // 2           # índice del elemento medio
if nums[mid] == x: return mid     # encontrado: devuelve la posición
elif nums[mid] > x: high = mid - 1  # x está a la izquierda: descarta la mitad derecha
else: low = mid + 1               # x está a la derecha: descarta la mitad izquierda
return -1                         # se agotó el espacio: x no existe
print(low, high, mid)             # traza para contar las repeticiones del bucle
sorted(nums)                      # ordena la lista antes de buscar
math.log2(n)                      # pasos máximos aproximados: log2 de n
low, high = 1, height             # huevos: pisos entre 1 y la altura
if do_experiment(mid, breaking): high = mid   # se rompe: la solución está en mid o debajo
else: low = mid + 1               # no se rompe: la solución está por encima
return low - 1                    # piso más alto seguro
```

**Errores comunes**

- Aplicar la búsqueda binaria a una lista sin ordenar: puede descartar la mitad que contiene el valor.
- Usar `low < high` en lugar de `low <= high` al buscar un elemento: no se comprueba el último candidato (por ejemplo, adivinar el 100 en el cuestionario del PDF).
- Olvidar el `+ 1` o el `- 1` al mover `low` o `high`, lo que puede producir un bucle infinito.
- Devolver `low` en lugar de `low - 1` en el problema de los huevos.
- No contemplar que el huevo no se rompa en ningún piso (`breaking > height`): el código del PDF devuelve `height - 1` en ese caso.
