| Paso | Espacio de búsqueda | Piso medio | Resultado del experimento | Nuevo espacio |

| 1 | pisos 1 a 10 | 5 | no se rompe | pisos 6 a 10 |

| 3 | pisos 6 a 8 | 7 | se rompe | pisos 6 a 7 |

## Búsqueda binaria: el algoritmo

Planteamiento del PDF para encontrar un entero en una lista **ordenada** de forma ascendente:

- **Problema:** ¿existe un entero aleatorio `x` en `S`, un conjunto de enteros ordenados?
- **Entrada:** `S` como lista de enteros ordenada de forma ascendente y `x` como el entero que se quiere encontrar.
- **Salida:** la posición (índice) de `x` si existe en `S`, y `-1` si no existe.

**Espacio de búsqueda:** el rango de índices donde `x` todavía podría estar. Al principio es toda la lista: `bajo` (`low`) es el índice del primer elemento y `alto` (`high`) es el índice del último.

Con `S = [11, 17, 26, 28, 37, 45, 53, 59]` y `x = 37`:

1. `low = 0`, `high = 7`. El elemento medio es `mid = (low + high) // 2 = 3`, y `S[3] = 28`.
2. Como `S[mid] < x`, el valor `x` solo puede estar entre `mid + 1` y `high`: `low = mid + 1 = 4`. El espacio pasa de `[0, 7]` a `[4, 7]`, la mitad.
3. `mid = (4 + 7) // 2 = 5`, `S[5] = 45`. Como `S[mid] > x`, `high = mid - 1 = 4`. El espacio es ahora `[4, 4]`.
4. `mid = 4`, `S[4] = 37`: son iguales, se devuelve `4`.

Si el valor no existe (por ejemplo `x = 31`), `high` baja a `mid - 1` hasta quedar por debajo de `low`; entonces la condición del bucle deja de cumplirse y se devuelve `-1`.

**Mejor caso:** que `x` sea justo el elemento medio, de modo que baste una sola comparación (por ejemplo `x = 28`).

**Peor caso:** que `x` no exista. Para 77 en la lista de 8 elementos se realizan 4 comparaciones en total.

**Referencia:** PDF parte 2, págs. 20–29

## Búsqueda binaria: implementación

Código del PDF. Recibe `nums` y `x`; el espacio de búsqueda inicial es toda la lista:

- Línea 2: `low` y `high` se inicializan con el primer y el último índice. El espacio de búsqueda es `[0, len(nums) - 1]`.
- Líneas 3–10: se calcula `mid = (low + high) // 2` (el operador `//` devuelve el cociente entero). Si `nums[mid] == x`, se devuelve `mid`. Si `nums[mid] > x`, `high = mid - 1`. En caso contrario, `low = mid + 1`. En cada caso el espacio se reduce a la mitad.
- Línea 11: si `low` supera a `high`, el bucle termina y se devuelve `-1` porque `x` no existe en `nums`.

````python
def bin_search(nums, x):
    low, high = 0, len(nums) - 1
    while low <= high:
        mid = (low + high) // 2
        if nums[mid] == x:
            return mid
        elif nums[mid] > x:
            high = mid - 1
        else:
            low = mid + 1
    return -1

S = [11, 17, 26, 28, 37, 45, 53, 59]
for x in (53, 37, 28, 31, 77):
    pos = bin_search(S, x)
    print(f"In S, {x} is at position {pos}.")
````

**Nota sobre el PDF:** en la pág. 27 el texto dice que, para un valor que no existe como 31, "la condición `low < high` no se satisface"; sin embargo, el código de las págs. 30–32 usa `while low <= high`. La condición correcta es `low <= high` (con `<` se dejaría de comprobar el último candidato). Además, en la pág. 30 aparece "numbs entero" por "nums entero" (error de traducción) y en la pág. 25 la diapositiva marca el medio en 5 cuando el espacio ya es `[4, 4]` (el medio real es 4).

**Nota sobre el PDF:** los mensajes de salida siguen en inglés (`In S, 53 is at position 6.`), como en las diapositivas. El PDF obtiene `x` con `input()`; aquí se prueba con valores fijos y más abajo hay una versión que lee con `input()`.

**Clave:** la lista debe estar ordenada para que descartar una mitad sea correcto. `bin_search` devuelve `-1` si `x` no está.

**Error común:** usar `low < high` en lugar de `low <= high` (no comprueba el último elemento), o actualizar `low = mid` o `high = mid` sin sumar o restar 1, lo que puede producir un bucle infinito.

**Referencia:** PDF parte 2, págs. 20–32

### Versión que lee el dato con `input()`

El PDF (pág. 20) lee `x` con `input()` y muestra la posición. Esta versión recibe la función de lectura para poder probarla sin teclado. Si el usuario cancela (entrada vacía o `None`), termina sin error.

````python
def leer_entero(mensaje, leer=input):
    """Pide un entero. Devuelve None si el usuario cancela (entrada vacía, None o fin de entrada)."""
    while True:
        try:
            texto = leer(mensaje)
        except (EOFError, KeyboardInterrupt):
            return None
        if texto is None or texto.strip() == "":
            return None
        try:
            return int(texto)
        except ValueError:
            print("Entrada no válida: escribe un número entero.")

def programa_busqueda(S, leer=input):
    x = leer_entero("Input the number to search: ", leer)
    if x is None:
        print("Programa terminado: no se recibió ningún número.")
        return None
    pos = bin_search(S, x)
    print(f"In S, {x} is at position {pos}.")
    return pos

S = [11, 17, 26, 28, 37, 45, 53, 59]
respuestas = iter(["quince", "53"])            # primero una entrada no válida, luego 53
verificar("lee 53 y lo encuentra", programa_busqueda(S, lambda m: next(respuestas, None)), 6)
verificar("entrada vacía cancela", programa_busqueda(S, lambda m: ""), None)

# Para usarlo de forma interactiva, quita el comentario de la siguiente línea
# programa_busqueda(S)
````

## Un paso más: complejidad temporal

Análisis del PDF:

- **Mejor caso:** la búsqueda termina con una sola comparación, así que es O(1).
- **Peor caso:** el tamaño del espacio de búsqueda es `N` al principio, luego `N/2`, `N/4` y así sucesivamente (`N/2^k` tras `k` pasos). La búsqueda termina cuando solo queda un elemento, es decir, `N / 2^k = 1`, de modo que `N = 2^k` y `k = log2(N)`.
- Por tanto, el número máximo de operaciones es `log2(N)` y la complejidad temporal en el peor caso es **O(log N)**: **tiempo logarítmico**.

**Referencia:** PDF parte 2, pág. 33

````python
# Cómo se reduce el espacio de búsqueda: N, N/2, N/4, ... hasta 1
n = 100
tamanos = [n]
while n > 1:
    n //= 2
    tamanos.append(n)
print(tamanos)           # [100, 50, 25, 12, 6, 3, 1]

# Cuántas veces se puede dividir entre 2 antes de llegar a 1
import math
for n in (8, 100, 1000, 1_000_000):
    print(f"n = {n:>9}: log2(n) = {math.log2(n):.2f}")
````

**Clave:** en tiempo logarítmico, duplicar el tamaño de los datos añade apenas **una** comparación más.

**Para el quiz:** peor caso O(log N); mejor caso O(1).

### Comparación con la búsqueda secuencial

**Ampliación (no está en el PDF):** se restablece `bin_search` sin el `print` y se cuentan las comparaciones en el peor caso de ambas búsquedas. Se usa `range(n)` como lista ordenada (admite índices sin ocupar memoria) y se busca un valor que no existe.

````python
def bin_search(nums, x):
    """Búsqueda binaria del PDF (versión sin print)."""
    low, high = 0, len(nums) - 1
    while low <= high:
        mid = (low + high) // 2
        if nums[mid] == x:
            return mid
        elif nums[mid] > x:
            high = mid - 1
        else:
            low = mid + 1
    return -1

def seq_search(nums, x):
    """Búsqueda secuencial de la unidad 24."""
    for i in range(len(nums)):
        if x == nums[i]:
            return i
    return -1

def bin_search_contando(nums, x):
    """Devuelve (posición, número de comparaciones de nums[mid] con x)."""
    low, high = 0, len(nums) - 1
    comparaciones = 0
    while low <= high:
        mid = (low + high) // 2
        comparaciones += 1
        if nums[mid] == x:
            return mid, comparaciones
        elif nums[mid] > x:
            high = mid - 1
        else:
            low = mid + 1
    return -1, comparaciones

print(f"{'n':>10} {'secuencial':>11} {'binaria':>8}")
for n in (8, 100, 1_000, 1_000_000):
    datos = range(n)                                   # lista ordenada 0..n-1
    peor_bin = bin_search_contando(datos, n)[1]        # n no existe: peor caso
    print(f"{n:>10} {n:>11} {peor_bin:>8}")
````

````python
# Medición orientativa de tiempos (los valores cambian según el equipo y la carga)
import timeit
datos = list(range(1_000_000))
t_sec = timeit.timeit(lambda: seq_search(datos, 999_999), number=3)
t_bin = timeit.timeit(lambda: bin_search(datos, 999_999), number=3)
print(f"Secuencial: {t_sec:.4f} s | Binaria: {t_bin:.6f} s (orientativo, 3 repeticiones, 1 000 000 de elementos)")
````

Los tiempos de `timeit` son **orientativos**: dependen del equipo. Lo que importa es el orden de magnitud: el número de comparaciones de la binaria crece con `log2(n)`, mientras que el de la secuencial crece con `n`.

**Ampliación (no está en el PDF):** el peor caso de la binaria coincide con `floor(log2(n)) + 1` comparaciones. Se comprueba para varios tamaños (para `n = 8` da 4, igual que en el PDF).

````python
import math
ok = all(bin_search_contando(range(n), n)[1] == math.floor(math.log2(n)) + 1 for n in range(1, 2000))
print("peor caso = floor(log2(n)) + 1 para n de 1 a 1999:", ok)
````

**Clave:** la búsqueda binaria es mucho más rápida que la secuencial en listas grandes, pero exige datos ordenados.

**Para el quiz:** en 8 elementos, mejor caso 1 comparación y peor caso 4.

## Cuestionarios del PDF

El PDF presenta (pág. 48) el código del juego de "adivinar el número" y pregunta: si `maximo = 100` y `numero = 51`, ¿cuántas "cuentas" se imprimirían? El cuestionario #2 (pág. 49) repite la pregunta con `numero = 25`.

El juego usa la misma idea de la búsqueda binaria: el programa adivina probando siempre el valor medio del rango. Se adapta el código del PDF a una función (en lugar de `input()`) para ejecutarlo con los dos casos.

````python
def adivinar_numero_pdf(maximum, number):
    count = 0
    low, high = 1, maximum
    while low < high:
        mid = (low + high) // 2
        count += 1
        if mid == number:
            print(f"Tu número es: {number}")
            break
        elif mid > number:
            high = mid - 1
        else:
            low = mid + 1
    print(f"Total de veces que se realizó la búsqueda: {count}")
    return count

print("Cuestionario 1: máximo = 100, número = 51")
adivinar_numero_pdf(100, 51)
print("Cuestionario 2: máximo = 100, número = 25")
adivinar_numero_pdf(100, 25)
````

**Respuesta:** el contador `count` cuenta las repeticiones del bucle y se imprime una sola vez al final. En el cuestionario #1 (`numero = 51`) el valor medio toma 50, 75, 62, 56, 53 y 51, por lo que `count` vale 6. En el #2 (`numero = 25`) el valor medio toma 50 y 25, por lo que `count` vale 2. Ambos casos imprimen `Tu número es: ...` y después el total de repeticiones.

**Nota sobre el PDF:** el código del cuestionario usa `while low < high`, y `from random import randint` no se usa. Con `<` hay un fallo: si el número buscado es justo el último candidato (por ejemplo, `numero = 100`), el bucle termina sin comprobarlo y no imprime `Tu número es`. Se muestra a continuación y se corrige con `<=`. También, "cuentas" en las preguntas es una traducción de `count` y significa "veces" o "repeticiones".

````python
print("Con el código del PDF y numero = 100 (no imprime 'Tu número es'):")
adivinar_numero_pdf(100, 100)

def adivinar_numero(maximum, number):
    """Versión corregida con low <= high: siempre encuentra el número."""
    count = 0
    low, high = 1, maximum
    while low <= high:
        mid = (low + high) // 2
        count += 1
        if mid == number:
            return count
        elif mid > number:
            high = mid - 1
        else:
            low = mid + 1
    return count

print("Versión corregida con numero = 100:", adivinar_numero(100, 100), "repeticiones")
````

## Proyecto de la unidad: caída de huevos con huevos infinitos

### Problema

Escribir un algoritmo que encuentre el **piso más alto** en el que un huevo no se rompe al dejarlo caer, dada la altura del edificio (pág. 37). Condiciones:

1. La **rotura** (`breaking`) es el piso más bajo donde se rompe el huevo; se selecciona aleatoriamente entre 1 y la altura (en esta unidad, el usuario la introduce).
2. El experimento se expresa como `do_experiment(floor, breaking)`: devuelve `True` si el huevo se rompe al caer desde ese piso y `False` si no.
3. `do_experiment()` se puede llamar sin límite porque hay un número infinito de huevos.

### Plan de programación (págs. 38–43)

1. Recibir del usuario la altura (`height`) y la rotura (`breaking`) con `input()` y convertirlas a `int`.
2. Implementar `do_experiment(floor, breaking)`, igual que en la unidad 24: devuelve `floor >= breaking`.
3. Implementar `find_highest_safe_floor2(height, breaking)` con búsqueda binaria: inicializar `low = 1` y `high = height`; mientras `low < high`, calcular `mid`; si el huevo se rompe en `mid`, `high = mid` (la solución está en ese piso o debajo); si no, `low = mid + 1` (la solución está por encima).
4. La búsqueda termina cuando `low` es igual a `high`, que es el **piso más bajo donde se rompe el huevo**. Por eso el piso más alto seguro es `low - 1`.
5. Imprimir el resultado con una f-string.

**Referencia:** PDF parte 2, págs. 17–18 y 37–43

````python
def do_experiment(floor, breaking):
    """True si el huevo se rompe al caer desde `floor`."""
    return floor >= breaking

def find_highest_safe_floor2(height, breaking):
    """Búsqueda binaria del piso más alto en el que el huevo no se rompe."""
    low, high = 1, height          # espacio de búsqueda inicial: todos los pisos
    while low < high:
        mid = (low + high) // 2    # piso intermedio
        if do_experiment(mid, breaking):
            high = mid             # se rompe: el primer piso de rotura es mid o está debajo
        else:
            low = mid + 1          # no se rompe: la rotura está por encima de mid
    return low - 1                 # low es el primer piso de rotura

# Ejemplo de las págs. 11-16: 10 pisos, el huevo se rompe desde el piso 7
print(find_highest_safe_floor2(10, 7))     # 6
# Ejemplos de las págs. 38-39 y 45-46
print(find_highest_safe_floor2(100, 93))   # 92
print(find_highest_safe_floor2(100, 50))   # 49
print(find_highest_safe_floor2(100, 51))   # 50
````

**Nota sobre el PDF:** la solución de la pág. 18 tiene un caso límite sin resolver. Si el huevo **no se rompe en ningún piso** (`breaking > height`), el bucle termina con `low == height` y devuelve `height - 1`, cuando la respuesta correcta es `height` (como en la versión secuencial de la unidad 24). El PDF supone que la rotura está entre 1 y la altura, pero la versión que pide la rotura al usuario (pág. 38) permite escribir un número mayor. La corrección es usar `high = height + 1`, que representa "no se rompe en ningún piso". Los ejemplos de la pág. 17 muestran dos peticiones (`input`) pero solo una línea de entrada y la salida `94`, que no corresponde a los valores mostrados. En la pág. 39 la salida dice `93` para una rotura en el piso 93, pero el propio texto de la diapositiva (y la pág. 38, que imprime `92`) indica que el resultado debe ser `rotura - 1`, es decir, 92.

````python
print("PDF, rotura mayor que la altura:", find_highest_safe_floor2(10, 11), "(debería ser 10)")

def find_highest_safe_floor2_corregida(height, breaking):
    """Versión corregida: high = height + 1 representa 'no se rompe en ningún piso'."""
    low, high = 1, height + 1
    while low < high:
        mid = (low + high) // 2
        if do_experiment(mid, breaking):
            high = mid
        else:
            low = mid + 1
    return low - 1

print("Corregida, rotura mayor que la altura:", find_highest_safe_floor2_corregida(10, 11))

# Comprobación exhaustiva contra la búsqueda secuencial de la unidad 24
def secuencial(height, breaking):
    for n in range(1, height + 1):
        if do_experiment(n, breaking):
            return n - 1
    return height

coinciden = all(find_highest_safe_floor2_corregida(h, b) == secuencial(h, b)
                for h in range(1, 80) for b in range(1, h + 3))
print("La versión corregida coincide con la secuencial en todos los casos:", coinciden)
````

### Evaluación de los resultados (págs. 44–46)

Para contar cuántos experimentos se hacen, el PDF añade `print(low, high, mid)` dentro del bucle. Si el primer piso de rotura es el 50, la búsqueda se hace 6 veces; si es el 51, 7 veces.

````python
def find_highest_safe_floor2_traza(height, breaking):
    low, high = 1, height
    while low < high:
        mid = (low + high) // 2
        print(low, high, mid)          # línea añadida para contar las repeticiones
        if do_experiment(mid, breaking):
            high = mid
        else:
            low = mid + 1
    return low - 1

for rotura in (50, 51):
    print(f"Altura 100, rotura en {rotura}:")
    piso = find_highest_safe_floor2_traza(100, rotura)
    print(f"El huevo estará a salvo hasta el piso número {piso}\n")
````

```python
def raiz_entera(n):
    low, high = 0, n
    resultado = 0
    while low <= high:
        mid = (low + high) // 2
        if mid * mid <= n:
            resultado = mid        # mid es candidato; se busca uno mayor
            low = mid + 1
        else:
            high = mid - 1
    return resultado
```

El espacio de búsqueda son los posibles valores de `r` (de 0 a `n`). Si `mid * mid <= n`, `mid` es candidato y se busca uno mayor; si no, se descarta la mitad superior. Se hacen unas `log2(n)` iteraciones en lugar de `n`.

## Apuntes para estudiar

**Clave**

- La búsqueda binaria requiere una lista **ordenada** y reduce el espacio de búsqueda a la mitad en cada comparación.
- Se mantienen dos límites, `low` y `high`; se compara `nums[mid]` con `x`, con `mid = (low + high) // 2`.
- Si `nums[mid] == x` se devuelve `mid`; si `nums[mid] > x`, `high = mid - 1`; si es menor, `low = mid + 1`. El bucle es `while low <= high` y, si termina, se devuelve `-1`.
- Mejor caso O(1); peor caso O(log N): tiempo logarítmico. En 8 elementos, entre 1 y 4 comparaciones.
- Caída de huevos con huevos infinitos: `find_highest_safe_floor2` con `low = 1`, `high = height`; `low` converge al primer piso de rotura y el resultado es `low - 1`.
- Con un solo huevo se usa búsqueda secuencial O(n); con infinitos huevos, búsqueda binaria O(log n).

**Errores comunes**

- Usar `low < high` en lugar de `low <= high` en la búsqueda de un elemento (no comprueba el último candidato).
- Olvidar el `+ 1` o el `- 1` al mover `low` o `high`, lo que puede causar bucles infinitos.
- Aplicar la búsqueda binaria a una lista sin ordenar.
- Devolver `low` en lugar de `low - 1` en el problema de los huevos.
- No tratar el caso en que el huevo no se rompe en ningún piso (`breaking > height`).

**Para el quiz**

- Peor caso de la búsqueda binaria en `N` elementos: `log2(N)`, O(log N).
- Para `S` de 8 elementos y `x = 77`, se hacen 4 comparaciones; para `x = 28`, una.
- Tras cada comparación, el espacio de búsqueda pasa de `N` a `N/2`.
- El juego de adivinar un número entre 1 y 100 con búsqueda binaria necesita como máximo 7 intentos.
