## 1. La sentencia `while`: qué es y su sintaxis

La sentencia `while` ejecuta repetidamente un bloque, igual que `for`, pero la repetición depende de una **condición**: el bloque se repite **mientras la condición sea verdadera**. `for` termina cuando agota la secuencia que sigue a `in`; `while` termina cuando la condición pasa a falsa.

```text
valor inicial
while expresión_condicional:
    bloque de código que se repite
```

Aunque es una sentencia de bucle, se parece mucho a `if`: si la condición es verdadera se ejecuta el bloque; la diferencia es que después vuelve a evaluar la condición. Si la condición sigue siendo verdadera siempre, el bucle no termina: hay que escribir el código con cuidado. `while True:` entra en un bucle infinito.

El ejemplo del PDF imprime "Bienvenido" cinco veces. Se inicializa `i` en 0; en cada vuelta se imprime y se incrementa `i` en 1 mientras `i < 5`. Como el valor inicial es 0, se hacen cinco iteraciones.

````python
i = 0                      # valor inicial
while i < 5:               # el bloque se ejecuta cuando la condición del bucle es verdadera
    print('Welcome to everyone!!')
    i += 1
````

**Clave:** un `while` necesita (1) un valor inicial, (2) una condición y (3) algo dentro del bloque que acerque la condición a falsa (aquí, `i += 1`).

**Error común:** olvidar actualizar la variable de la condición provoca un bucle infinito.

**Para el quiz:** `i = 0` con `while i < 5` ejecuta el bloque 5 veces (con `i` valiendo 0, 1, 2, 3 y 4).

## 2. `while` frente a `for`

Para sumar los enteros de 1 a `n` con `while` se necesitan más líneas que con `for`: hay que inicializar el contador, escribir la condición y aumentar el contador a mano. Por eso, **si el número de iteraciones está claro, es mejor `for`**. Entonces, ¿cuándo conviene `while`? Cuando la **condición** de repetición está clara pero **no se conoce de antemano el número de iteraciones**.

````python
def suma_hasta_n(n):
    s = 0
    i = 1
    while i <= n:
        s = s + i
        i += 1
    print('La suma de 1 a {} es {}'.format(n, s))
    return s

_ = suma_hasta_n(5)
````

````python
# Comparación: for frente a while para sumar de 1 a 10
s = 0
for i in range(1, 11):
    s += i
print('Suma de 1 a 10:', s)

n = 10
s = 0
i = 1
while i <= n:
    s = s + i
    i += 1
print('La suma de 1 a {} es {}'.format(n, s))
````

El PDF también muestra cómo usar `while` para un número de repeticiones conocido (aunque no sea lo más práctico): se inicializa un contador y se actualiza dentro del bloque.

````python
count = 1
s = 0                    # s guarda los valores acumulados; se reinicia en 0
while count <= 10:
    s = s + count        # añade el valor de count a s cada vez
    count = count + 1    # aumenta el valor de count en 1 cada vez
print("Suma es", s)
````

**Nota sobre el PDF:** en la comparación de la pág. 506 el código `for` aparece como `print('Suma de 1 a 10:'.s)` (con un punto en lugar de coma, error de transcripción). Se muestra corregido: `print('Suma de 1 a 10:', s)`.

**Clave:** `for` cuando se conoce cuántas veces se repite; `while` cuando solo se conoce la condición de salida.

**Para el quiz:** ¿cuándo se prefiere `while`? Cuando la condición es clara pero el número de repeticiones no.

## 3. `while` y condición de entrada (validación de datos)

El uso típico de `while` es **pedir un dato hasta que sea válido**. El PDF pide al usuario "tijeras", "piedra" o "papel" y vuelve a preguntar mientras la entrada sea otra cosa. Con `for` este problema es difícil de resolver, porque no se sabe cuántos intentos fallidos habrá.

El código del PDF usa `input()`. Para que se pueda ejecutar sin escribir nada, se encapsula en una función que recibe la función de lectura; abajo se prueba con entradas simuladas.

````python
def simular_entradas(*valores, eco=True):
    """Ampliación (no está en el PDF): crea una función que reemplaza a input() con valores de prueba."""
    iterador = iter(valores)
    def leer(mensaje=""):
        valor = next(iterador)
        if eco:
            print(f"{mensaje}{valor}")
        return valor
    return leer
````

````python
def pedir_opcion(leer=input):
    seleccionado = None                      # sin valor inicial el while fallaría (ver abajo)
    while seleccionado not in ['tijeras', 'piedra', 'papel']:
        seleccionado = leer('Elige entre tijeras, piedra, papel> ')
    print('Valor elegido:', seleccionado)
    return seleccionado

_ = pedir_opcion(simular_entradas("no", "¿por qué?", "papel"))
# Para usarlo de forma interactiva, quita el comentario de la siguiente línea
# pedir_opcion()
````

**Nota sobre el PDF:** el código de la pág. 507 usa `selected` al asignar y `seleccionado` al imprimir (nombres traducidos de forma inconsistente), lo que daría un `NameError`. Aquí se usa un solo nombre, `seleccionado`.

**`selected = None`.** La condición `while selected not in [...]` necesita que `selected` exista antes de evaluarse por primera vez. `None` significa "sin valor", pero cumple un papel importante en el flujo: hace que la condición sea verdadera en el primer intento. Si se omite esa línea, el código produce un error:

````python
def sin_valor_inicial():
    while seleccionado not in ['tijeras', 'piedra', 'papel']:   # 'seleccionado' no existe aún
        seleccionado = input()

try:
    sin_valor_inicial()
except UnboundLocalError as error:
    print("UnboundLocalError:", error)
````

Otro caso del PDF: sumar de 1 a `n` con `for` cuando el usuario escribe un número negativo da un resultado absurdo (error lógico).

````python
def suma_con_for(n):
    s = 0
    for i in range(1, n + 1):
        s = s + i
    print('La suma de 1 a {} es {}'.format(n, s))
    return s

_ = suma_con_for(-10)      # el rango queda vacío: la suma es 0
````

La solución es **limitar la entrada a enteros positivos** con un `while` que repita `input()` mientras el valor no sea válido (`n <= 0`). El valor inicial `n = -1` fuerza a entrar al bucle la primera vez.

````python
def suma_con_validacion(leer=input):
    n = -1
    while n <= 0:                       # se repite hasta que se introduce un valor positivo
        n = int(leer('Introduzca un número positivo para sumar :'))
    s = 0
    for i in range(1, n + 1):
        s = s + i
    print('La suma de 1 a {} es {}'.format(n, s))
    return s

_ = suma_con_validacion(simular_entradas("-1", "0", "7"))
````

Otro ejemplo del PDF: pedir la contraseña hasta que sea correcta. Se parte de una contraseña vacía y se repite la pregunta mientras sea distinta de la correcta.

````python
def pedir_contrasena(leer=input):
    password = ""
    while password != "pythonisfun":
        password = leer("Introduzca la contraseña: ")
    print("** Inicio de sesión correcto **")

pedir_contrasena(simular_entradas("aaaa", "pythonisfun"))
````

**Nota sobre el PDF:** la salida mostrada en la pág. 533 es `# Inicio de sesión exitoso #`, pero el código imprime `** Inicio de sesión correcto **`. Se usa el texto del código.

El mismo esquema aparece en la pregunta de la construcción: "¿Se ha terminado la construcción?" se repite hasta que la respuesta es "Sí".

````python
def esperar_construccion(leer=input):
    under_construction = True
    while under_construction:
        response = leer("¿Se ha terminado la construcción? ")
        if response == "Sí":
            under_construction = False
    print("Escapó con éxito.")

esperar_construccion(simular_entradas("No", "Sí"))
````

**Clave:** `while` + valor inicial que obliga a entrar + actualización con `input()` dentro del bloque = validación de entrada.

**Error común:** no inicializar la variable de la condición (`NameError` o `UnboundLocalError`), o escribir una condición que nunca se vuelve falsa.

**Para el quiz:** `selected = None` se escribe para que la variable exista y la condición sea verdadera la primera vez.

## 4. Sumar números que introduce el usuario (centinela)

Si no se conoce cuántos números va a introducir el usuario, se usa `while` con una variable de respuesta: el bucle continúa mientras el usuario conteste "sí" y termina cuando conteste "no".

````python
def sumar_numeros_usuario(leer=input):
    total = 0
    answer = 'sí'                                   # se inicializa para entrar al bucle
    while answer == 'sí':
        number = int(leer('Introduzca el número: '))
        total = total + number                      # acumula el valor
        answer = leer('¿Continuar?(sí/no): ')       # decide si el bucle se repite
    print('La suma es : ', total)
    return total

_ = sumar_numeros_usuario(simular_entradas("5", "sí", "6", "no"))
# Para usarlo de forma interactiva, quita el comentario de la siguiente línea
# sumar_numeros_usuario()
````

**Nota sobre el PDF:** el código del PDF inicializa `answer = 'yes'` y compara con `'yes'`, pero el diálogo de ejemplo responde "sí". Aquí se usa `'sí'` de forma coherente.

**Clave:** la variable de respuesta se inicializa con el valor que hace entrar al bucle y se actualiza al final de cada vuelta.

**Referencia:** PDF semana 0, págs. 535–536

## 5. Bucle infinito y la sentencia `break`

`while True:` repite para siempre porque la condición siempre es verdadera. Para salir se usa `break` dentro del bloque, normalmente dentro de un `if`. El PDF muestra la estructura: un bloque que se repite infinitamente, con una condición interna que ejecuta `break` cuando se cumple.

`break` **finaliza el bucle** inmediatamente: no se ejecuta el resto del bloque ni se vuelve a evaluar la condición.

````python
contador = 0
while True:                       # la condición siempre es True: bucle infinito
    contador += 1
    print('Repetir', contador)
    if contador == 3:             # expresión condicional que decide salir
        break
print('Fin del bucle')
````

````python
# Ejemplo del PDF con break: se detiene en la primera vocal
st = 'Programming'
for ch in st:
    if ch in ['a', 'e', 'i', 'o', 'u']:
        break                     # finaliza el bucle si es vocal
    print(ch)
print('The end')
````

**Clave:** `break` sale del bucle (`for` o `while`) más interno donde se encuentra.

**Error común:** un `while True` sin ningún `break` alcanzable es un bucle infinito de verdad.

**Referencia:** PDF semana 0, págs. 517–519, 531

## 6. La sentencia `continue` y los riesgos de `break` y `continue`

`continue` **omite el resto del bloque** en la vuelta actual y salta al principio del bucle para seguir con la siguiente iteración. A diferencia de `break`, **no finaliza el bucle**. En el diagrama del PDF, solo se omiten las sentencias que están debajo de `continue`; el bucle termina cuando la condición pasa a falsa.

````python
# Ejemplo del PDF con continue: solo imprime las consonantes
st = 'Programming'
for ch in st:
    if ch in ['a', 'e', 'i', 'o', 'u']:
        continue                  # omitir la ejecución siguiente si es vocal
    print(ch)
print('The end')
````

**Riesgo.** `break` y `continue` son útiles para controlar eficazmente los programas, pero si se usan demasiado el programa se vuelve difícil de entender por su flujo inconsistente. Se recomienda usarlos solo cuando sea necesario.

**Clave:** `break` termina el bucle; `continue` termina solo la vuelta actual.

**Error común:** confundir ambos: con `break` el texto "Programming" imprime `P r` y se detiene en la `o`; con `continue` imprime `P r g r m m n g`.

**Para el quiz:** `continue` no finaliza el bucle.

**Referencia:** PDF semana 0, págs. 517, 520–522

## 7. El módulo `random` (números aleatorios)

Python tiene muchos **módulos** creados por organizaciones y desarrolladores; los que vienen con la instalación de Python forman la **biblioteca estándar**. Un módulo es un archivo con funciones, variables o clases. Para usarlo se escribe `import nombre_del_módulo` y, al usarlo, se llama a sus componentes con un punto: `modulo.funcion()`. Por comodidad se le puede dar una abreviatura con `as`:

```text
import módulo1 [, módulo2, ...]
import nombre_del_módulo as abreviatura
```

Funciones principales del módulo `random`:

| Función | Tarea |
|---|---|
| `random()` | Número real entre 0 y 1 (sin incluir el 1). |
| `randrange()` | Entero dentro del rango especificado (con inicio, parada y paso, como `range`). |
| `randint(a, b)` | Entero aleatorio N tal que `a <= N <= b`. |
| `shuffle(seq)` | Mezcla aleatoriamente los elementos de una lista. |
| `choice(seq)` | Selecciona un elemento cualquiera de la secuencia. |
| `sample(seq, k)` | Selecciona aleatoriamente `k` elementos. |

**Nota sobre el PDF:** la tabla de la pág. 512 muestra los nombres traducidos (`aleatorio()`, `rango()`, `barajar`, `elección`, `muestra`); los nombres reales en Python son `random`, `randrange`, `shuffle`, `choice` y `sample`.

````python
import random as rd      # comando de importación: llama al módulo y le da la abreviatura rd

print(rd.random())             # real aleatorio en [0, 1)
print(rd.random())             # distinto cada vez
print(rd.randrange(1, 7))      # entero >= 1 y < 7
print(rd.randrange(0, 10, 2))  # múltiplo de 2 >= 0 y < 10
print(rd.randint(1, 10))       # entero >= 1 y <= 10 (incluye 1 y 10)
````

````python
# Se comprueban las propiedades (los valores concretos cambian en cada ejecución)
x = rd.random()
verificar("random() está en [0, 1)", 0 <= x < 1, True)
verificar("randrange(1, 7) está entre 1 y 6", 1 <= rd.randrange(1, 7) <= 6, True)
verificar("randrange(0, 10, 2) es par", rd.randrange(0, 10, 2) % 2, 0)
verificar("randint(1, 10) está entre 1 y 10", 1 <= rd.randint(1, 10) <= 10, True)
````

`shuffle` mezcla **en el sitio** una lista (la modifica y devuelve `None`); `choice` extrae un elemento; `sample` devuelve una **lista** de `k` elementos distintos.

````python
numlist = [10, 20, 30, 40, 50]
rd.shuffle(numlist)
print(numlist)                 # mismos elementos, orden distinto cada vez
print(rd.choice(numlist))      # un elemento cualquiera
print(rd.sample(numlist, 3))   # tres elementos distintos
````

````python
a = list(range(1, 11))     # enteros del 1 al 10
rd.shuffle(a)                  # mezclar aleatoriamente los elementos de la lista
print(a)
verificar("shuffle conserva los elementos", sorted(a), list(range(1, 11)))

b = rd.sample(a, 3)            # muestrea tres enteros arbitrarios
print('Número entero arbitrario entre 1 y 10 :', b)
verificar("sample devuelve 3 elementos", len(b), 3)
verificar("los elementos de sample vienen de la lista", all(v in a for v in b), True)
````

````python
# Ampliación (no está en el PDF): con la misma semilla se reproduce la misma secuencia
rd.seed(7)
primero = [rd.randint(1, 10) for _ in range(5)]
rd.seed(7)
segundo = [rd.randint(1, 10) for _ in range(5)]
verificar("misma semilla, misma secuencia", primero, segundo)
````

**Clave:** `random.random()` -> real en [0, 1); `randrange(a, b)` excluye `b`; `randint(a, b)` incluye `b`.

**Error común:** confundir `randint(1, 10)` (incluye el 10) con `randrange(1, 10)` (no lo incluye); olvidar que `shuffle` modifica la lista original y no devuelve nada (`print(rd.shuffle(a))` imprime `None`).

**Para el quiz:** `randint(1, 10)` incluye 1 y 10; `randrange(0, 10, 2)` devuelve un múltiplo de 2 menor que 10.

## 8. Doble bucle (bucles anidados)

Un **doble bucle** es un bucle contenido dentro de otro. El bucle de arriba se llama **bucle exterior** y el de dentro **bucle interior**. Por cada vuelta del exterior, el interior se ejecuta completo.

El ejemplo del PDF imprime la tabla de multiplicar de la 2 a la 9: `i` (exterior) toma valores de 2 a 9 y `j` (interior) de 1 a 9. Hay 8 x 9 = 72 cálculos. El `print()` vacío del final del bloque exterior cambia de línea después de cada tabla. `{}` es un marcador de posición de `format` (visto en la Unidad 8).

````python
for i in range(2, 10):                                  # bucle for externo
    for j in range(1, 10):                              # bucle for interno
        print('{}x{} = {:2d}'.format(i, j, i * j), end=' ')
    print()                                             # ejecuta el bucle interno y cambia de línea
````

````python
# Se cuenta cuántas veces se ejecuta el bloque interno: 8 valores de i x 9 valores de j
vueltas = 0
for i in range(2, 10):
    for j in range(1, 10):
        vueltas += 1
verificar("el doble bucle hace 72 cálculos", vueltas, 72)
````

El PDF muestra una tabla con la ejecución: con `i = 2`, `j` va de 1 a 9 (`2*1` ... `2*9`); luego `i = 3`, y así hasta `i = 9`. Los bucles triples ya son difíciles de entender, por eso los bucles anidados no suelen superar los bucles triples.

**Clave:** el bucle interior se reinicia y se ejecuta completo en cada vuelta del exterior; el total de vueltas es el producto de ambos.

**Error común:** poner el `print()` de salto de línea dentro del bucle interior (cada resultado quedaría en su propia línea) o fuera de ambos (todo en una sola línea).

**Referencia:** PDF semana 0, págs. 523–525

## Proyecto de la unidad: caracol en un pozo

**Problema.** Hay un pozo de 30 metros de profundidad y un caracol en el fondo. Durante el día el caracol sube 7 metros; por la noche, mientras duerme, resbala 5 metros. ¿Cuántos días tardará en salir? Pensar que avanza 2 metros netos por día (30 / 2 = 15 días) es un error, porque el último día sale sin resbalar; el PDF anticipa que la respuesta es 13 días y lo confirma programando. Se usa `while` porque no se conoce de antemano el número de repeticiones.

**Plan de programación (pseudocódigo del PDF).**

1. Inicio.
2. Inicializar la posición del caracol, la fecha, la posición objetivo, la distancia diaria de subida y de bajada.
3. Mientras la posición del caracol no haya alcanzado la posición objetivo:
4. El caracol se cae mientras duerme.
5. Pasa al día siguiente (el día aumenta).
6. Sube de nuevo.
7. Fin del mientras. Imprimir el día en que escapó.
8. Fin.

**Código del PDF (comentado).** Los identificadores son los del PDF: `current_pos`, `days`, `move_up`, `move_down` y `destination`.

````python
current_pos = 0      # Posición actual del caracol
days = 1             # Fecha
move_up = 7          # distancia que sube al día : en metros
move_down = 5        # distancia que baja mientras duerme : en metros
destination = 30     # Distancia objetivo : en metros
current_pos += move_up   # posición del caracol el primer día

print('día :', days, 'Posición del caracol :', current_pos, 'metros')
while current_pos <= destination:      # repetir si el caracol no ha alcanzado el objetivo
    current_pos -= move_down           # vuelve a caer
    days += 1                          # next day
    current_pos += move_up             # distancia que subió
    print('día :', days, 'Posición del caracol :', current_pos, 'metros')
print()
print('Se tardó {} días en salir del pozo.'.format(days))
````

**Resultado esperado.** Posiciones al final de cada día: 7, 9, 11, 13, ..., 29 y 31 metros; el caracol sale el día 13 (la salida del PDF dice "Tardó 13 días en salir del pozo").

**Nota sobre el PDF:** (1) el código imprime `'Se tardó {} días ...'` pero la salida mostrada dice "Tardó 13 días ..."; (2) el pseudocódigo dice "si la posición es **inferior** a la meta", pero el código usa `<=`. Con `<=`, si el caracol llega **exactamente** a la meta sigue repitiendo (resbala y sube otra vez) cuando en realidad ya salió. Con 30 metros no importa, pero con otra meta da un resultado incorrecto. La versión corregida usa `<`; se encapsula en una función y se compara con la de `<=`.

````python
def simular_caracol(subida, bajada, meta):
    """Devuelve los días que tarda el caracol y las posiciones al final de cada día."""
    posicion = subida          # el primer día sube sin resbalar
    dias = 1
    posiciones = [posicion]
    while posicion < meta:     # versión corregida: si ya llegó a la meta, sale
        posicion -= bajada
        dias += 1
        posicion += subida
        posiciones.append(posicion)
    return dias, posiciones

def dias_con_menor_o_igual(subida, bajada, meta):
    """Versión del PDF (con <=), solo para comparar."""
    posicion = subida
    dias = 1
    while posicion <= meta:
        posicion -= bajada
        dias += 1
        posicion += subida
    return dias

dias, posiciones = simular_caracol(7, 5, 30)
print(posiciones)
verificar("pozo de 30 m: 13 días", dias, 13)
verificar("posiciones 7, 9, ..., 31", posiciones, list(range(7, 32, 2)))
verificar("pozo de 30 m: la versión del PDF coincide", dias_con_menor_o_igual(7, 5, 30), 13)

# Meta de 29 m: el caracol llega exactamente a 29 el día 12
verificar("pozo de 29 m: versión corregida", simular_caracol(7, 5, 29)[0], 12)
verificar("pozo de 29 m: la versión con <= da 13 (incorrecto)", dias_con_menor_o_igual(7, 5, 29), 13)
````

```python
    lineas = []

    while i <= 9:
```

```python
def es_palindromo(n):
    original = n
    invertido = 0
    while n > 0:
        digito = n % 10                  # último dígito
        invertido = invertido * 10 + digito
        n = n // 10                      # quita el último dígito
    return invertido == original
```

En cada vuelta se toma el último dígito con `% 10`, se agrega al número invertido y se elimina con `// 10`. El bucle termina cuando no quedan dígitos.

```python
def adivinar(secreto, intentos):
    pistas = []
    total = 0

    while not adivinado:
```

## Apuntes para estudiar

**Clave**

- `while condición:` repite el bloque mientras la condición sea verdadera; necesita un valor inicial y una actualización dentro del bloque.
- `for` si el número de repeticiones está claro; `while` si solo se conoce la condición de salida.
- Validación de entrada: `while` + valor inicial que obliga a entrar + `input()` dentro del bloque.
- `while True:` es un bucle infinito; se sale con `break`.
- `break` termina el bucle; `continue` omite el resto de la vuelta actual y sigue con la siguiente.
- Usar `break` y `continue` solo cuando sea necesario: abusar de ellos vuelve el programa difícil de seguir.
- `import random as rd`: `random()` real en [0, 1), `randrange(a, b, paso)` excluye `b`, `randint(a, b)` incluye `b`, `shuffle(lista)` mezcla en el sitio, `choice(seq)` elige uno, `sample(seq, k)` devuelve `k` elementos.
- Doble bucle: el interior se ejecuta completo por cada vuelta del exterior (tablas de multiplicar: 8 x 9 = 72 cálculos).
- Proyecto del caracol: 13 días con 7 m de subida, 5 m de bajada y 30 m de pozo.

**Errores comunes**

- No actualizar la variable de la condición (bucle infinito).
- No inicializar la variable de la condición (`NameError` o `UnboundLocalError`).
- Usar `<=` en lugar de `<` cuando el objetivo se alcanza exactamente.
- Confundir `break` con `continue`.
- Confundir `randint` (incluye el límite superior) con `randrange` (no lo incluye); esperar que `shuffle` devuelva una lista nueva.
- Reiniciar mal los contadores en un doble bucle (el contador interior debe volver a su valor inicial en cada vuelta exterior).

**Para el quiz**

- Número de iteraciones de `i = 0; while i < 5: ...; i += 1`: 5.
- Elegir `while` cuando el número de repeticiones no se conoce.
- `break` finaliza el bucle; `continue` no lo finaliza.
- `randint(1, 10)` devuelve un entero entre 1 y 10, ambos incluidos.
- Un doble bucle `for i in range(2, 10)` / `for j in range(1, 10)` hace 72 vueltas internas.
- Con `Programming`: `break` en la primera vocal imprime `P r`; `continue` imprime `P r g r m m n g`.
