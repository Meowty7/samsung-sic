## Estructuras básicas de control

Un programa, por complejo que sea, se construye con solo tres estructuras (el PDF lo compara con los bloques de Lego):

- **Secuencia:** los comandos se ejecutan uno tras otro, en el orden en que están escritos.
- **Selección:** se elige y ejecuta una de varias instrucciones según una condición.
- **Iteración:** la misma orden se ejecuta repetidamente.

El PDF usa la analogía de una carretera: la secuencia es una vía recta, la selección es una bifurcación donde el coche elige un camino y la iteración es una rotonda donde se da vueltas.

Las sentencias que controlan el flujo del programa se llaman **sentencias de control**:

- Sentencias condicionales: `if`, `if-else`, `if-elif-else`.
- Sentencias de bucle: `for`, `while`.
- Comandos que cambian el flujo de un bucle: `break`, `continue`.

Sin estructura de selección, el programa repetiría siempre la misma acción y llegaría siempre a la misma conclusión.

**Referencia:** PDF semana 0, págs. 315–320

### Estructura secuencial

Ejemplo del PDF: cada línea se ejecuta en orden, de modo que `num` pasa de 100 a 200 y luego a 300.

````python
num = 100
print('num =', num)     # Se imprime 100
num = num + 100
print('num =', num)     # 100 se suma a num y se imprime 200
num = num + 100
print('num =', num)     # 100 se añade de nuevo a num y se imprime 300
````

**Clave:** en la estructura secuencial todas las líneas se ejecutan siempre y en el mismo orden. La selección rompe esa regla: algunas líneas se ejecutan solo si se cumple una condición.

**Referencia:** PDF semana 0, pág. 318

## Flujo de la sentencia condicional y expresión condicional

Para elegir ejecutar solo una de varias sentencias se usa una **rama**: un diagrama de flujo con una pregunta (la condición) y dos salidas, Verdadero y Falso.

- Si la condición es verdadera se ejecuta la <declaración ejecutable 1>.
- Si es falsa se ejecuta la <declaración ejecutable 2> (o ninguna, si no existe).

La condición debe ser una **expresión condicional**: una expresión que se evalúa como un valor booleano, `True` o `False`. Los operadores de comparación de la unidad anterior (`<`, `>`, `<=`, `>=`, `==`, `!=`) devuelven justamente esos valores, por eso se usan en las condiciones.

Situaciones del PDF:

- Situación 1: imprimir "descuento para jóvenes" si la edad es inferior a 20 años.
- Situación 2: imprimir "objetivo alcanzado" si el recuento de pasos supera los 1.000.

````python
edad = 18
cuenta_paseos = 1500

# Una expresión condicional se evalúa a un valor booleano
print(edad < 20)                 # True
print(cuenta_paseos >= 1000)     # True
print(type(edad < 20))           # <class 'bool'>
````

**Clave:** la condición de un `if` siempre se reduce a `True` o `False`.

**Referencia:** PDF semana 0, págs. 321–322

## Sintaxis de la sentencia `if`

```text
if <expresión condicional>:
    <bloque que se ejecuta cuando la condición es verdadera>
```

Partes: la palabra clave `if`, la expresión condicional, los **dos puntos** (`:`) y el bloque con **sangría** en las líneas siguientes.

Ejemplo del PDF con `edad = 18` y con `edad = 24`:

````python
# Ejemplo 1: la edad es 18, edad < 20 es True y se imprime el mensaje
edad = 18
if edad < 20:
    print("descuento para jóvenes")

# Ejemplo 2: la edad es 24, edad < 20 es False y no se imprime nada
edad = 24
if edad < 20:
    print("descuento para jóvenes")
print("Fin del ejemplo 2 (no se imprimió ningún descuento)")
````

````python
# Situación 2 del PDF: objetivo de pasos
cuenta_paseos = 1200
if cuenta_paseos >= 1000:
    print('objetivo alcanzado')
````

**Clave:** si la condición es falsa, el bloque simplemente se omite y el programa sigue con la siguiente línea sin sangría.

**Error común:** olvidar los dos puntos al final de la línea del `if` provoca un `SyntaxError`.

**Para el quiz:** con `edad = 24`, `if edad < 20:` no imprime nada.

**Referencia:** PDF semana 0, págs. 323–325

## Bloque y sangría

Un **bloque** es un trozo de código que se ejecuta cuando la condición es verdadera. El bloque de una sentencia condicional **debe tener sangría**; de lo contrario Python lanza un `IndentationError`.

En Python la sangría es importante: según la indentación, el mismo código produce resultados distintos.

````python
# Ejemplo didáctico del error (se captura para que el notebook siga):
codigo_sin_sangria = """
edad = 18
if edad < 20:
print("descuento para jóvenes")
"""
try:
    compile(codigo_sin_sangria, "<ejemplo>", "exec")
except IndentationError as error:
    print("IndentationError:", error.msg)
````

A continuación, los cuatro códigos de sangría del PDF. Fíjate en qué líneas pertenecen al bloque y cuáles no.

````python
# Código de sangría 1: se cumple la condición (edad < 20)
edad = 18
if edad < 20:
    print("descuento para jóvenes")
print("Bienvenido")          # sin sangría: se ejecuta siempre
````

````python
# Código de sangría 2: no se cumple la condición
edad = 24
if edad < 20:
    print("descuento para jóvenes")
print("Bienvenido")
````

````python
# Código de sangría 3: tres líneas dentro del bloque, edad = 18
edad = 18
if edad < 20:
    print('edad', edad)
    print("bienvenida a los jóvenes")
    print("descuento para jóvenes")
````

````python
# Código de sangría 4: las mismas tres líneas, edad = 24 (sin salida)
edad = 24
if edad < 20:
    print('edad', edad)
    print("bienvenida a los jóvenes")
    print("descuento para jóvenes")
print("(el bloque no se ejecutó, por eso no hay salida arriba)")
````

### Reglas de la sangría

- Un bloque son varias líneas con **el mismo número de espacios** delante. Si el número difiere dentro del mismo bloque, hay `IndentationError` ("unexpected indent" o "unindent does not match any outer indentation level").
- Tras la línea con los dos puntos (`if`, pero también `else`, `elif`, `for`, `while`, `def`, `class`) debe haber un bloque con sangría.
- La guía PEP-8 recomienda **4 espacios** por nivel. Los editores (IDE) de Python suelen poner la sangría automáticamente.
- Se recomienda usar la barra espaciadora en lugar del tabulador.
- Si el bloque es de una sola línea, puede escribirse a continuación de los dos puntos: `if edad < 20: print("descuento para jóvenes")`.

````python
# Dos formas equivalentes de escribir un bloque de una sola línea
edad = 18
if edad < 20: print("descuento para jóvenes")

if edad < 20:
    print("descuento para jóvenes")
````

````python
# Sangría incorrecta: líneas dentro del bloque con distinta cantidad de espacios
codigo_mal = """
edad = 18
if edad < 20:
    print('edad', edad)
      print("bienvenida a los jóvenes")
"""
try:
    compile(codigo_mal, "<ejemplo>", "exec")
except IndentationError as error:
    print("IndentationError:", error.msg)
````

Otro ejemplo del PDF, con una puntuación: todas las sentencias del bloque comparten la misma sangría.

````python
puntuacion = 95          # en el PDF: int(input("Introduzca la puntuación: "))
if puntuacion >= 90:     # si la condición es verdadera, se ejecuta todo el bloque
    print("Enhorabuena.")
    print("Estás aceptado.")
    print("Tienes derecho a una beca.")
````

**Clave:** lo que determina qué líneas pertenecen al `if` es la sangría, no el color ni los espacios entre líneas.

**Error común:** poner sangría distinta a líneas del mismo bloque, o no sangrar la línea posterior a los dos puntos.

**Para el quiz:** el tamaño de sangría recomendado por PEP-8 es de 4 espacios.

## Operador de asignación `=` frente a operador relacional `==`

El PDF insiste en esto porque los principiantes lo confunden con frecuencia:

- `=` es la **asignación**: pone el valor de la derecha en la variable de la izquierda.
- `==` es el operador **relacional**: devuelve `True` o `False` según si los valores de ambos lados son idénticos.

````python
x = 100          # asignación: x pasa a valer 100
print(x == 100)  # comparación: True
print(x == 50)   # comparación: False

if x > 1:
    print("x es mayor que 1.")
````

````python
# Ejemplo didáctico: usar = en lugar de == en una condición es un error de sintaxis
codigo = """
x = 100
if x = 100:
    print("x vale 100")
"""
try:
    compile(codigo, "<ejemplo>", "exec")
except SyntaxError as error:
    print("SyntaxError:", error.msg)
````

**Error común:** escribir `if x = 5:` en lugar de `if x == 5:`.

## La sentencia `pass`

`pass` no hace nada. Sirve como **marcador de posición**: deja un bloque vacío para completar más adelante, por ejemplo cuando todavía no se ha decidido qué hará cierta rama.

Un bloque no puede quedar vacío: si se omite el contenido, Python lanza `IndentationError: expected an indented block`.

````python
num = 2
if num % 2 == 0:
    print('Número par')
if num % 3 == 0:
    pass            # bloque vacío: aquí se programará algo más adelante
````

````python
# pass también sirve para definir una función sin cuerpo
def func():
    pass

func()   # no hace nada, pero no da error
print("func() se ejecutó sin errores")
````

````python
# Sin pass (ni otra sentencia) el bloque queda vacío y hay error. Se captura para mostrarlo:
codigo = """
def func():

func()
"""
try:
    compile(codigo, "<ejemplo>", "exec")
except IndentationError as error:
    print("IndentationError:", error.msg)
````

El PDF muestra además la forma general de una función (`def nombre_funcion(x1, x2, ...):` con su cuerpo y `return`) y la de una función sin cuerpo, que solo lleva `pass`. Las funciones se estudian más adelante.

**Nota sobre el PDF:** la traducción automática convirtió `pass` en "pase" y "paso" en varios textos (por ejemplo, "declaración de paso" o "declaración de conformidad"). La palabra clave de Python sigue siendo `pass`.

**Clave:** `pass` es un relleno sintáctico, no tiene efecto en la ejecución.

**Para el quiz:** `pass` cumple el papel de marcador de posición en un bloque vacío.

## Procesar entradas con `split()` e `int()`

`input()` solo devuelve cadenas. El método `split()` corta una cadena en partes y devuelve una **lista de cadenas**; el carácter que separa las partes se llama **separador**. Por defecto el separador es el espacio.

````python
words = 'Python is beautiful'
word_list = words.split()
print(word_list)
````

````python
# Simulamos lo que escribiría el usuario (en el PDF se usa input())
str1 = '10 20 30'
print('Entrada: ', str1)
print('Entrada separada por el método split(): ', str1.split())

str1 = '10,20,30'
print('Entrada: ', str1)
print('Entrada separada por el método split(): ', str1.split(','))
````

Los elementos de la lista siguen siendo cadenas; para operar con ellos hay que convertirlos a entero con `int()` (o a real con `float()`).

````python
entrada = '1,2,3'                          # en el PDF: input('Introduzca tres números enteros separados por ,: ')
num1, num2, num3 = entrada.split(',')
num1, num2, num3 = int(num1), int(num2), int(num3)
print(num1, num2, num3)
print("suma:", num1 + num2 + num3)
````

````python
# Con otro separador (el punto), como en el PDF con la entrada 5.6.7
entrada = '5.6.7'
num1, num2, num3 = entrada.split('.')
num1, num2, num3 = int(num1), int(num2), int(num3)
print(num1, num2, num3)
````

**Nota sobre el PDF:** en la pág. 353 el separador del segundo ejemplo aparece como `‘,.'` (comillas rotas por la traducción). Para la entrada `5.6.7` el separador correcto es `'.'`, como se muestra arriba.

**Clave:** `split()` devuelve una lista de cadenas; hay que convertir cada elemento con `int()` antes de calcular.

**Error común:** usar el separador equivocado. `'10 20 30'.split(',')` devuelve una lista con un solo elemento: la cadena completa.

## Proyecto de la unidad

### Problema: máquina de lotería automática

El programa genera tres números ganadores (en el PDF son fijos: 2, 3 y 9). El usuario introduce tres números de un dígito separados por espacios, en cualquier orden. Si acierta los tres, el programa imprime "¡Te ha tocado la lotería!".

### Plan de programación (pseudocódigo del PDF)

1. Inicio.
2. Se generan tres números: 2, 3, 9 (`n1, n2, n3`).
3. Se reinicia `n = 0` (contador de aciertos).
4. Se recibe la entrada de tres números `a, b, c`.
5. Si `a` coincide con uno de 2, 3, 9, se suma 1 a `n`.
6. Lo mismo con `b`.
7. Lo mismo con `c`.
8. Si `n` es 3, se imprime "¡Te ha tocado la lotería!".
9. Fin.

**Referencia:** PDF semana 0, págs. 310–312

### Código del PDF (con entrada simulada)

El PDF usa `input(...).split()` y comentarios como `# Si uno de los números coincide con el número ganador`. Aquí se encapsula en una función que recibe el texto que escribiría el usuario.

````python
def loteria(texto_usuario):
    """Máquina de lotería del PDF: devuelve el mensaje para la entrada indicada."""
    n1, n2, n3 = 2, 3, 9                    # números ganadores
    n = 0                                   # contador de aciertos

    a, b, c = texto_usuario.split()         # en el PDF: input('Introduzca un número de lotería de tres cifras : ').split()
    a = int(a)
    b = int(b)
    c = int(c)

    if a == n1 or a == n2 or a == n3:       # Si uno de los números coincide con el número ganador
        n += 1                              # Incrementar n
    if b == n1 or b == n2 or b == n3:
        n += 1
    if c == n1 or c == n2 or c == n3:
        n += 1

    if n == 3:                              # Si n es 3, tres números coinciden
        return '¡Te ha tocado la lotería!'
    return 'Sin premio'

# Entradas de prueba (no interactivo); el for repite el bloque para cada elemento (se estudia en la Unidad 8)
for prueba in ['1 2 9', '2 3 9', '9 3 2', '5 5 5']:
    print(f"Entrada: {prueba!r} -> {loteria(prueba)}")
````

Resultado esperado: con `1 2 9` hay solo dos aciertos (2 y 9), por eso no hay premio; con `2 3 9` y con `9 3 2` hay tres aciertos y el programa imprime "¡Te ha tocado la lotería!".

Para usarla de forma interactiva, quita el comentario de la última línea del siguiente bloque.

````python
# Para usarlo de forma interactiva, quita el comentario de la siguiente línea
# print(loteria(input('Introduzca un número de lotería de tres cifras : ')))
````

**Nota sobre el PDF:** en la pág. 311 la entrada de ejemplo se muestra como `1, 2, 9` (con comas), pero el código usa `.split()` sin argumento, que separa por espacios. Con comas daría `ValueError` al convertir `'1,'` en entero. La entrada correcta para ese código es `1 2 9` (o habría que usar `.split(',')`).

````python
# Ejemplo didáctico del problema descrito en la nota anterior
try:
    print(loteria('1, 2, 9'))
except ValueError as error:
    print("ValueError:", error)
````

**Nota sobre el PDF (limitación del programa):** el código cuenta cada número introducido que coincide con alguno de los ganadores, sin comprobar que no se repitan. Con la entrada `2 2 2` cuenta tres aciertos y daría premio aunque falten el 3 y el 9. Una versión más estricta compararía conjuntos.

````python
print(loteria('2 2 2'))   # premio indebido con la lógica del PDF

# Versión estricta (ampliación): comparar como conjuntos
def loteria_estricta(texto_usuario, ganadores=(2, 3, 9)):
    elegidos = {int(x) for x in texto_usuario.split()}
    return '¡Te ha tocado la lotería!' if elegidos == set(ganadores) else 'Sin premio'

print(loteria_estricta('2 2 2'))
print(loteria_estricta('9 3 2'))
````

**Ampliación (no está en el PDF):** el módulo `random` permite que los números ganadores sean realmente aleatorios. `random.seed(valor)` fija la semilla del generador, de modo que la "lotería" sea reproducible (los mismos números en cada ejecución); sin semilla cambiarían cada vez. `random.sample(range(1, 10), 3)` elige 3 números distintos entre 1 y 9. El módulo `random` se estudia en la Unidad 7.

````python
import random

def loteria_aleatoria(texto_usuario, semilla=42):
    """Genera 3 números ganadores distintos (1 a 9) y los compara con la entrada."""
    random.seed(semilla)                      # semilla fija: resultado reproducible
    ganadores = random.sample(range(1, 10), 3)
    elegidos = [int(x) for x in texto_usuario.split()]
    aciertos = 0
    for numero in elegidos:
        if numero in ganadores:
            aciertos += 1
    if aciertos == 3:
        mensaje = '¡Te ha tocado la lotería!'
    else:
        mensaje = f'Aciertos: {aciertos}'
    return ganadores, mensaje

ganadores, mensaje = loteria_aleatoria('1 2 3')
print("Números ganadores:", ganadores)
print(mensaje)
ganadores, mensaje = loteria_aleatoria(' '.join(str(g) for g in ganadores))
print(mensaje)
````

## Apuntes para estudiar

**Clave**

- Los programas se construyen con tres estructuras: secuencia, selección e iteración.
- Una sentencia `if` ejecuta su bloque solo si la expresión condicional es `True`.
- La condición se escribe con operadores relacionales (`<`, `>`, `<=`, `>=`, `==`, `!=`) y se evalúa como booleano.
- Sintaxis: `if condición:` y, debajo, el bloque con sangría (4 espacios según PEP-8).
- `=` asigna y `==` compara.
- `x % n == 0` comprueba si `x` es múltiplo de `n`.
- `pass` es un marcador de posición para bloques vacíos.
- `input()` devuelve cadenas: `split()` las divide en una lista y `int()` convierte cada elemento.

**Errores comunes**

- Olvidar los dos puntos al final de la línea del `if`.
- No sangrar el bloque o mezclar distintos números de espacios (`IndentationError`).
- Confundir `=` con `==`.
- Dejar un bloque vacío sin `pass`.
- Olvidar convertir la entrada con `int()` antes de operar.
- Usar el separador equivocado en `split()`.

**Para el quiz**

- Con `edad = 24`, `if edad < 20:` no imprime nada.
- Sangría recomendada por PEP-8: 4 espacios.
- Las tres estructuras de control: secuencia, selección, iteración (`if`, `for`/`while`).
- La condición de un `if` es de tipo `bool`.
- `10 % 3 == 0` es `False`; `15 % 3 == 0` es `True`.
- `pass` no hace nada.
