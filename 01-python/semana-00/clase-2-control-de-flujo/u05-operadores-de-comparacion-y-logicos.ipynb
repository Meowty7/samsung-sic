## Contexto

El problema del PDF es construir una **calculadora de años bisiestos**. Un año normal no dura exactamente 365 días: para que el calendario coincida con el año solar real se añade un día a febrero cada cierto tiempo (febrero tiene 28 días, o 29 en año bisiesto). El calendario gregoriano, usado en casi todo el mundo, añade un día cada cuatro años pero resta tres días cada 400 años para reducir la desviación respecto al calendario solar.

Reglas para decidir si un año es bisiesto:

2. Aunque sea divisible por 4, **no** debe ser divisible por 100.
3. Un año divisible por 400 es siempre bisiesto.

Ejemplos del PDF: bisiestos 1984, 1988, 2004, 2024, 1600, 2000, 2400; comunes (no bisiestos) 1700, 1800, 1900, 2100, 2200, 2300.

Para traducir esas reglas a código necesitamos **comparar** valores (`%`, `==`, `!=`) y **combinar** condiciones (`and`, `or`). Eso es lo que estudia esta unidad: el resultado de cada comparación o combinación es un valor booleano, `True` o `False`.

## 1. Operadores de comparación

Un **operador de comparación** examina la relación de tamaño entre dos operandos comparables (números, cadenas) con criterios como "mayor que" o "menor que" y devuelve un resultado `True` o `False`. El tipo de datos que solo tiene esos dos valores se llama **booleano** (`bool`).

| Operador | Descripción | Con `a = 100`, `b = 200` |
|---|---|---|
| `==` | `True` si ambos operandos tienen el mismo valor | `a == b` es `False` |
| `!=` | `True` si los valores son diferentes | `a != b` es `True` |
| `>` | `True` si el izquierdo es mayor que el derecho | `a > b` es `False` |
| `<` | `True` si el izquierdo es menor que el derecho | `a < b` es `True` |
| `>=` | `True` si el izquierdo es mayor o igual | `a >= b` es `False` |
| `<=` | `True` si el izquierdo es menor o igual | `a <= b` es `True` |

````python
a = 100
b = 200
print(a == b)
print(a != b)
print(a > b)
print(a < b)
print(a >= b)
print(a <= b)
````

Ejemplo del PDF con valores literales (en Python los resultados se muestran como `True` y `False`; el PDF los traduce como "Verdadero" y "Falso"):

````python
print('Resultado de 10 > 5: ', 10 > 5)
print('Resultado de 5 < 1: ', 5 < 1)
print('Resultado de 5 == 5: ', 5 == 5)
print('Resultado de 5 != 5: ', 5 != 5)
print("Resultado de 'a' > 'b': ", 'a' > 'b')
````

**Nota sobre el PDF:** en la pág. 264 la segunda línea está escrita `print('Resultado de 5 < 1: ', 5 > 1)`, pero la etiqueta dice `5 < 1` y la salida muestra `Falso`. Con `5 > 1` el resultado sería `True`; el valor correcto de `5 < 1` es `False`, que es lo que ejecuta la celda anterior.

El PDF usa los operadores de comparación para preguntas como "¿es par este número?": un entero es par si el resto de dividirlo entre 2 es 0.

````python
# Se simula la entrada del usuario con una variable (en un programa real: n = int(input(...)))
n = 60
print('¿Es par el número entero?', n % 2 == 0)
````

**Clave:** toda comparación devuelve `bool`. `n % 2 == 0` es la forma de decidir si un entero es par.

**Error común:** confundir `=` (asigna) con `==` (compara).

**Para el quiz:** `5 != 5` es `False`; `10 > 5` es `True`.

## 2. El tipo de datos booleano y la función `bool()`

El tipo **booleano** (o lógico) solo toma dos valores, `True` y `False`. La función `bool(x)` convierte cualquier valor a booleano. Reglas del PDF:

- El número `0` se convierte en `False`; **todos los demás números** (por ejemplo `9` o `-1`) son `True`.
- La **cadena vacía** `''` es `False`; cualquier cadena con contenido (`'hello'`) es `True`.
- `None` (palabra clave que significa "no hay valor") es `False`.
- Una lista vacía `[]` es `False`; una lista con elementos (`[10, 20]`) es `True`.

Equivalencias numéricas: `bool(1)` es `True` y `bool(0)` es `False`.

````python
print(bool(1))
print(bool(0))
print(bool(True))
print(bool(False))
````

````python
# Prueba de impresión del tipo booleano (pág. 266)
print(bool(-1))
print(bool(0))
print(bool(None))
print(bool(''))
print(bool('hello'))
print(bool([]))
print(bool([10, 20]))
````

Las comparaciones pueden mostrarse junto con `bool()` (pág. 286):

````python
print(10 > 20)       # 10 es menor que 20, así que 10 > 20 es False
print(10 < 20)       # 10 es menor que 20, entonces 10 < 20 es True
print(bool(9))       # 9 no es 0, así que es True
print(bool(-1))      # -1 tampoco es 0, así que es True
print(bool(0))       # 0 es el único valor numérico que es False
print(bool(None))    # None significa que no hay valor, por lo que es False
print(bool(''))      # la cadena vacía es False
print(bool('hello')) # contiene caracteres, por lo que es True
````

**Nota sobre el PDF:** el texto de las págs. 266 y 285 dice que "los espacios en blanco como `' '`" son `False`, y un comentario de la pág. 286 afirma "el espacio en blanco es una cadena, así que es Falso". Esto es inexacto: el código del PDF usa `bool('')` (cadena **vacía**, sin ningún carácter), que es `False`. Una cadena que contiene un espacio, `' '`, **sí** es `True` porque tiene un carácter. Se comprueba:

````python
print(repr(''), bool(''))      # cadena vacía
print(repr(' '), bool(' '))    # cadena con un espacio
````

**Clave:** valores "falsos": `0`, `''`, `None`, `[]` (y `False`). Todo lo demás es `True`.

**Error común:** creer que `bool(' ')` o `bool('0')` son `False`: ambas son cadenas con contenido, por lo tanto `True`.

**Para el quiz:** `bool(0)` es `False`, `bool(9)` y `bool(-1)` son `True`, `bool(None)` es `False`.

## 3. Comparación de cadenas, `in` y el orden alfabético

Con cadenas, `==` es `True` solo si ambas coinciden completamente, y `!=` es `True` si no coinciden.

````python
print('aaa' == 'aaa')
print('aaa' == 'bbb')
print('aaa' != 'aaa')
print('aaa' != 'bbb')
````

El operador `in` (Un paso más del PDF) comprueba si una cadena está **contenida parcialmente** dentro de otra.

````python
print('aaa' in 'aaa-bbb-ccc')
print('bbb' in 'aaa-bbb-ccc')
print('ddd' in 'aaa-bbb-ccc')
````

**Ampliación (no está en el PDF):** las cadenas también se comparan con `<` y `>` según su orden alfabético (en realidad, según el código de cada carácter). Por eso `'a' > 'b'` es `False` (el ejemplo del PDF) y las mayúsculas se consideran menores que las minúsculas.

````python
print('a' < 'b')
print('Z' < 'a')            # las mayúsculas van antes en el orden de caracteres
print('casa' == 'Casa')     # distingue mayúsculas de minúsculas
````

**Clave:** `==` y `!=` comparan cadenas completas; `in` busca una subcadena.

**Error común:** olvidar que `'casa'` y `'Casa'` son diferentes.

**Referencia:** PDF semana 0, págs. 270–271, 273

## 4. Diferencia entre `==` e `is`

- `==` compara los **valores**: ¿valen lo mismo?
- `is` comprueba si son **el mismo objeto**: ¿están guardados como el mismo objeto en memoria?

Ejemplo del PDF: `a = 1` (entero) y `b = 1.0` (real) valen lo mismo pero son objetos distintos.

````python
a = 1
b = 1.0
print(a == b)     # mismo valor
print(a is b)     # no son el mismo objeto (uno es int y el otro float)
````

**Ampliación (no está en el PDF):** con listas se ve claramente: dos listas con los mismos elementos son iguales (`==`) pero son dos objetos distintos (`is`). Una lista asignada a otra variable es el mismo objeto. `is` se usa sobre todo con `None` (`x is None`).

````python
x = [1, 2, 3]
y = [1, 2, 3]
z = x
print(x == y)         # True: mismo contenido
print(x is y)         # False: dos objetos distintos
print(x is z)         # True: z apunta al mismo objeto que x
print(id(x) == id(z)) # id() identifica el objeto

valor = None
print(valor is None)
````

**Nota sobre el PDF:** la explicación de la pág. 272 dice que `a` y `b` "están almacenados en memorias diferentes" por ser entero y real; es una simplificación válida para este caso. La regla general: `is` pregunta por identidad, no por valor, y no debe usarse para comparar números o cadenas.

**Clave:** `==` compara valores; `is` compara identidad.

**Error común:** usar `is` para comparar números o textos.

**Referencia:** PDF semana 0, pág. 272

## 5. Comparaciones encadenadas y precaución con el orden de los símbolos

Python permite usar **dos o más operadores de comparación** seguidos, como en matemáticas. `0 < n < 200` es `True` cuando `n` está entre 0 y 200 (el PDF lo usa dentro de un `if`, que se verá más adelante).

````python
n = 100
print('n =', n)
print(0 < n < 200)
print(0 < n < 50)
````

Precaución del PDF: el símbolo `<=` no puede escribirse como `=<`, y no debe haber espacios dentro de `!=` ni de `>=`. Se muestra el error de forma controlada:

````python
for codigo in ["num1 =< num2", "num1 ! = num2", "num1 > = num2"]:
    try:
        compile(codigo, "<prueba>", "eval")
        print("Válido:  ", codigo)
    except SyntaxError as error:
        print("Inválido:", codigo, "->", error.msg)
````

**Clave:** el signo `=` va siempre en segundo lugar en `<=`, `>=`, `!=` y `==`, sin espacios.

**Error común:** escribir `=<`, `=>` o `! =`.

**Referencia:** PDF semana 0, págs. 274–275

## 6. Operadores lógicos: `and`, `or`, `not`

Los **operadores lógicos** combinan valores booleanos y devuelven un nuevo valor booleano.

| Operador | Descripción |
|---|---|
| `x and y` | Es `True` solo si **ambos** son `True`; si alguno es `False`, el resultado es `False` |
| `x or y` | Es `True` si **al menos uno** es `True`; solo es `False` cuando ambos son `False` |
| `not x` | Invierte el valor: `True` pasa a `False` y viceversa |

El PDF ilustra `and` con dos interruptores en serie (la bombilla se enciende solo si ambos están en ON), `or` con dos interruptores en paralelo (se enciende si alguno está en ON) y muestra los símbolos de las puertas lógicas, muy usados en ingeniería eléctrica e informática.

````python
x = True
y = False
print(x and y)
print(x or y)
print(not x)
print(not y)
````

Ejemplo del PDF: ¿es el número un entero par entre 0 y 100? Se combinan tres condiciones con `and`:

````python
num = 99      # simula int(input('Introduzca un número entero: '))
result = (num >= 0 and num <= 100 and num % 2 == 0)
print('¿Es la entrada un número entero par entre 0 y 100?', result)
````

Con 99, las dos primeras condiciones son `True`, pero `99 % 2` vale 1, de modo que `1 == 0` es `False`; como `and` exige que **todas** sean `True`, el resultado es `False`.

**Ampliación (no está en el PDF):** `and` y `or` usan **evaluación en cortocircuito**: en cuanto el resultado queda decidido, no evalúan el resto. Esto evita errores, por ejemplo una división entre cero.

````python
divisor = 0
# Si divisor es 0, la primera condición ya es False y no se evalúa 10 / divisor
print(divisor != 0 and 10 / divisor > 1)
````

**Clave:** `and` exige todos verdaderos; `or` basta con uno; `not` invierte.

**Error común:** escribir `num >= 0 and <= 100`: cada condición de un `and` debe ser una expresión completa (`num >= 0 and num <= 100`).

**Para el quiz:** `True and False` es `False`; `True or False` es `True`; `not False` es `True`.

**Referencia:** PDF semana 0, págs. 276–280, 288

### Tabla de verdad

Una **tabla de verdad** muestra el resultado del operador para todas las combinaciones posibles de entradas. Para `and`: solo la fila `True, True` da `True`. Para `or`: solo la fila `False, False` da `False`. Para `not`: dos filas.

| x | y | x and y | x or y |
|---|---|---|---|
| False | False | False | False |
| False | True | False | True |
| True | False | False | True |
| True | True | True | True |

| x | not x |
|---|---|
| False | True |
| True | False |

El PDF muestra además la tabla en binario (0 = `False`, 1 = `True`) junto a los símbolos de las puertas lógicas AND y OR. Con Python podemos generarla:

````python
print("x       y       x and y   x or y")
for x in (False, True):
    for y in (False, True):
        print(f"{str(x):7} {str(y):7} {str(x and y):10} {x or y}")

print()
print("x       not x")
for x in (False, True):
    print(f"{str(x):7} {not x}")
````

**Nota sobre el PDF:** en la pág. 289 la tabla de `x or y` tiene errores: la fila (Falso, Verdadero) aparece con resultado "Falso" (debería ser "Verdadero"). La tabla correcta es la de arriba (y la que muestra la pág. 276 y la pág. 280): `or` da `True` en cuanto una entrada es `True`. Además, las columnas "x e y", "x o y" y "no x" de las tablas son traducciones de `x and y`, `x or y` y `not x`.

**Para el quiz:** `and` solo da `True` con dos `True`; `or` solo da `False` con dos `False`.

## 7. Precedencia de los operadores

Cuando una expresión combina varios operadores, la **precedencia** (prioridad) decide cuál se evalúa primero. Como en matemáticas, `*` y `/` se evalúan antes que `+` y `-`; los **paréntesis** tienen la mayor prioridad. Cuanto más arriba esté un operador en la tabla, mayor es su precedencia.

| Operador | Descripción |
|---|---|
| `**` | Exponenciación |
| `~`, `+`, `-` (unarios) | Operadores unarios |
| `*`, `/`, `%`, `//` | Multiplicación, división, resto, división entera |
| `+`, `-` | Suma, resta |
| `>>`, `<<` | Desplazamiento de bits |
| `&` | AND de bits |
| `^`, `\|` | XOR y OR de bits |
| `<=`, `<`, `>`, `>=` | Comparación |
| `==`, `!=` | Igualdad, diferencia |
| `=`, `%=`, `/=`, `//=`, `-=`, `+=`, `*=`, `**=` | Asignación y asignación compuesta |
| `is`, `is not` | Identidad |
| `in`, `not in` | Pertenencia |
| `not`, `or`, `and` | Lógicos |

**Nota sobre el PDF:** la tabla de la pág. 292 tiene tres imprecisiones respecto a Python real: (1) los lógicos se evalúan en el orden `not`, luego `and`, luego `or` (el PDF los lista "no, o, y"); (2) `is`, `is not`, `in` y `not in` están al mismo nivel que las comparaciones (`<`, `==`...), no debajo de la asignación; (3) la asignación `=` no es un operador dentro de una expresión: se ejecuta al final. Para este curso basta con la idea general y la regla práctica: **ante la duda, usa paréntesis**.

Ejemplo del PDF: el promedio de tres valores. Sin paréntesis, `x + y + z / 3` calcula primero `z / 3`.

````python
x, y, z = 10, 20, 30     # simula tres int(input(...))

avg = (x + y + z) / 3
print("Media =", avg)

print("Sin paréntesis:", x + y + z / 3)   # incorrecto para el promedio
````

**Ampliación (no está en el PDF):** efecto de la precedencia en operadores lógicos y de comparación.

````python
print(True or False and False)     # and se evalúa antes que or -> True or (False and False)
print((True or False) and False)   # con paréntesis cambia el resultado
print(not True == False)           # == antes que not -> not (True == False)
print(2 + 3 * 4 ** 2)              # ** primero, luego *, luego +
print(1 + 2 < 4 and 10 % 3 == 1)   # aritmética, comparación y por último and
````

**Clave:** paréntesis, `**`, `* / // %`, `+ -`, comparaciones, `not`, `and`, `or`.

**Error común:** olvidar los paréntesis en el promedio: `x + y + z / 3` no es `(x + y + z) / 3`.

**Para el quiz:** en `2 + 3 * 4` se evalúa primero `3 * 4`; los paréntesis siempre mandan.

**Referencia:** PDF semana 0, págs. 291–293

## Proyecto de la unidad: calculadora de años bisiestos

**Problema.** Escribir un programa que reciba un año e indique `True` si es bisiesto y `False` si no.

**Plan de programación (pseudocódigo del PDF):**

1. Inicio.
2. Introducir el año.
3. Usar la fórmula del año bisiesto para decidir si es verdadero o falso: `(año % 4 == 0 and año % 100 != 0) or año % 400 == 0`.
4. Si la condición no se cumple, el resultado impreso es `False`.
5. Fin.

La fórmula traduce las reglas: divisible por 4 **y no** por 100, **o** divisible por 400.

**Referencia:** PDF semana 0, págs. 257–261

El código del PDF es:

```python
year = int(input("Introduzca el año: "))
leap_year = (year % 4 == 0 and year % 100 != 0) or year % 400 == 0
print(leap_year)
```

Lo encapsulamos en una función con la lectura inyectable y lo probamos con años de prueba.

````python
def calculadora_bisiesto(leer=input):
    """Pide un año y muestra True si es bisiesto y False si no (pasos del plan del PDF)."""
    year = int(leer("Introduzca el año: "))                           # [2] entrada
    leap_year = (year % 4 == 0 and year % 100 != 0) or year % 400 == 0  # [3] fórmula
    print(leap_year)                                                  # [4] resultado
    return leap_year

# Versión no interactiva: lista de años de prueba; 2021 es el ejemplo del PDF
for anio_de_prueba in ["2021", "2024", "1900", "2000", "2100", "1600"]:
    resultado = calculadora_bisiesto(lambda mensaje: (print(mensaje + anio_de_prueba), anio_de_prueba)[1])

# Para usarlo de forma interactiva, quita el comentario de la siguiente línea
# calculadora_bisiesto()
````

**Resultado esperado:** 2021 -> `False`; 2024 -> `True`; 1900 -> `False` (divisible por 100 pero no por 400); 2000 -> `True` (divisible por 400); 2100 -> `False`; 1600 -> `True`.

## Apuntes para estudiar

**Clave**

- Toda comparación (`== != > < >= <=`) devuelve un `bool`: `True` o `False`.
- `bool(x)` es `False` para `0`, `''` (cadena vacía), `None` y `[]`; el resto es `True`.
- `==` compara valores; `is` compara identidad (si es el mismo objeto).
- `in` comprueba si una cadena está contenida en otra.
- `and` exige todos verdaderos; `or` basta con uno; `not` invierte.
- Comparaciones encadenadas: `0 < n < 200`.
- Precedencia: paréntesis, aritmética, comparaciones, `not`, `and`, `or`. Ante la duda, paréntesis.
- Año bisiesto: `(año % 4 == 0 and año % 100 != 0) or año % 400 == 0`.

**Errores comunes**

- Confundir `=` (asignación) con `==` (comparación).
- Escribir `=<`, `=>` o `! =` en lugar de `<=`, `>=` y `!=`.
- Creer que `bool(' ')` es `False`: solo la cadena vacía lo es.
- Usar `is` para comparar números o cadenas en lugar de `==`.
- Olvidar los paréntesis en `(x + y + z) / 3`.
- Mezclar `and` y `or` sin paréntesis y obtener un resultado distinto del que se esperaba.
- Comparar texto sin considerar mayúsculas y minúsculas (`'casa' != 'Casa'`).

**Para el quiz**

- `10 > 5` es `True`; `5 != 5` es `False`; `'a' > 'b'` es `False`.
- `bool(0)`, `bool(None)`, `bool('')` y `bool([])` son `False`; `bool(9)` y `bool(-1)` son `True`.
- `a = 1`, `b = 1.0`: `a == b` es `True` y `a is b` es `False`.
- `True and False` es `False`; `True or False` es `True`; `not True` es `False`.
- Un número `n` es par si `n % 2 == 0`.
- `2021` no es bisiesto; `2000` sí lo es; `1900` no.
