## La sentencia `else`: qué hacer cuando la condición es falsa

Con `if` solo se ejecuta algo cuando la condición es verdadera. Con `if-else` se añade un bloque alternativo que se ejecuta cuando es falsa. Ejemplo del PDF: una beca se concede con nota igual o superior a 90; si no, no se acepta.

```text
if <condición>:
    <bloque si es verdadera>
else:
    <bloque si es falsa>
```

Se ejecuta exactamente uno de los dos bloques.

````python
score = 95
if score >= 90:
    print("Aceptado")
    print("Reciba también una beca")
else:
    print("No se acepta")

score = 80
if score >= 90:
    print("Aceptado")
    print("Reciba también una beca")
else:
    print("No se acepta")
````

**Nota sobre el PDF:** en la pág. 371 el código usa `if score > 90:`, pero el texto dice "nota igual o superior a 90". Con `>` una nota de exactamente 90 sería rechazada. Para que el código coincida con el enunciado se usa `>=`, como arriba.

**Clave:** `else` no lleva condición; se ejecuta cuando la del `if` es falsa. Siempre va al mismo nivel de sangría que su `if`.

**Referencia:** PDF semana 0, pág. 371

## `if` sin `else` frente a `if-else`

Ejemplo del PDF: imprimir "Es por la mañana" antes de las 12 y "Es por la tarde" desde las 12. Con solo `if` hacen falta **dos** sentencias `if`, y el programa evalúa las dos condiciones siempre.

````python
hour = 10
if hour < 12:
    print('Es por la mañana.')
if hour >= 12:
    print('Es por la tarde.')
````

Las dos condiciones son **mutuamente excluyentes** (no pueden ser verdaderas a la vez; el PDF habla de "relación exclusiva"). Cuando es así, `if-else` es más claro y simplifica el diagrama de flujo: una sola pregunta y dos caminos.

````python
hour = 10
if hour < 12:
    print('Es por la mañana.')
else:
    print('Es por la tarde.')

hour = 15
if hour < 12:
    print('Es por la mañana.')
else:
    print('Es por la tarde.')
````

**Nota sobre el PDF:** en los diagramas de las págs. 372 y 374 la traducción automática puso "Ya es de día." en lugar de "Es por la mañana." y "Es por la tarde" sin punto; el significado es el mismo. Aquí se usan los textos del código.

**Clave:** si las dos condiciones son excluyentes, usa `if-else` en lugar de dos `if`.

**Para el quiz:** una relación exclusiva es la de dos condiciones que no pueden cumplirse a la vez.

## `if-else` anidados

Dentro de un bloque se puede escribir otra sentencia `if` o `if-else`. Se llama sentencia **anidada** (o doble). La interna solo se evalúa si se ejecuta el bloque que la contiene.

Ejemplo del PDF: recibir un entero y decir si es negativo, cero o positivo. Si `num >= 0`, hace falta una segunda comprobación para distinguir 0 de positivo.

````python
num = -20           # en el PDF: int(input("Introduzca un número entero: "))
if num >= 0:        # asegúrese de sangrar el bloque
    if num == 0:    # comprobar condiciones específicas dentro del bloque
        print("Es 0.")
    else:
        print("Es positivo.")
else:
    print("Es negativo.")
````

Otro ejemplo del PDF: dentro del `else` de `num < 0` se determina si el número es par o impar, y solo se imprime para los números no negativos.

````python
num = 100
if num < 0:
    print(num, 'es negativo')
else:
    print(num, 'no es negativo')
    if num % 2 == 0:
        print(num, 'es un número par')
    else:
        print(num, 'es un número impar')
````

Con `num = -100` solo se ejecuta el primer bloque: el `else` externo (y el `if-else` interno) no se evalúa.

````python
num = -100
if num < 0:
    print(num, 'es negativo')
else:
    print(num, 'no es negativo')
    if num % 2 == 0:
        print(num, 'es un número par')
    else:
        print(num, 'es un número impar')
````

El PDF nombra así las partes: la sentencia `if-else` exterior y la interior; las líneas del bloque interior tienen una sangría adicional.

**Nota sobre el PDF:** en las págs. 378–381 los `print` aparecen como `print(num, es negativo)` sin comillas, por la traducción. En Python válido el texto va entre comillas: `print(num, 'es negativo')`, como arriba.

**Clave:** cada nivel de anidamiento añade un nivel de sangría.

**Error común:** perder la cuenta de a qué `if` pertenece cada `else`; la sangría lo indica.

**Referencia:** PDF semana 0, págs. 376–381

## `elif`: examinar más condiciones cuando la anterior es falsa

El anidamiento complica el código cuando hay muchos casos. `elif` (abreviatura de "else if") permite encadenar condiciones: se evalúa la siguiente solo si las anteriores fueron falsas. Forma general:

```text
if <condición 1>:
    <bloque 1>
elif <condición 2>:
    <bloque 2>
else:
    <bloque final>
```

Solo se ejecuta el bloque de la **primera** condición verdadera; si ninguna lo es, se ejecuta el `else`. Ejemplo del PDF: positivo, cero o negativo.

````python
num = 6
if num > 0:
    print("Es positivo.")
elif num == 0:
    print("Es 0.")
else:
    print("Es negativo.")

num = -5
if num > 0:
    print("Es positivo.")
elif num == 0:
    print("Es 0.")
else:
    print("Es negativo.")

num = 0
if num > 0:
    print("Es positivo.")
elif num == 0:
    print("Es 0.")
else:
    print("Es negativo.")
````

**Clave:** en una cadena `if-elif-else` se ejecuta como máximo un bloque. El `elif` solo se examina si todas las condiciones anteriores fueron falsas.

**Para el quiz:** `elif` es la abreviatura de "else if".

**Referencia:** PDF semana 0, págs. 382–383

## Generador de calificaciones

Problema del PDF: clasificar una puntuación en una calificación según la tabla.

| Puntuación | Calificación |
|---|---|
| 90 a 100 | A |
| De 80 a menos de 90 | B |
| De 70 a menos de 80 | C |
| De 60 a menos de 70 | D |
| Por debajo de 60 | F |

El PDF resuelve el problema de tres maneras para comparar.

### Ejemplo A: varias sentencias `if`

Cada rango se escribe con una condición compuesta (`and`). Funciona, pero es largo y propenso a errores: hay que acertar en los límites de cada condición.

````python
def calificacion_a(score):
    grade = ''
    if score >= 90:
        grade = 'A'
    if score < 90 and score >= 80:
        grade = 'B'
    if score < 80 and score >= 70:
        grade = 'C'
    if score < 70 and score >= 60:
        grade = 'D'
    if score < 60:
        grade = 'F'
    return grade

score = 88                       # en el PDF: int(input('Introduzca la puntuación : '))
print('Tu nota es: ', calificacion_a(score))
````

**Nota sobre el PDF:** en las págs. 385–390 el último `print` es `print('Tu nota es: ', nota)`, pero la variable se llama `grade` (el texto aparece como "grado" en los diagramas). Con ese código real Python lanzaría `NameError: name 'nota' is not defined`. Aquí se usa `grade` de forma coherente.

````python
# Ejemplo didáctico del error de nombre descrito en la nota
try:
    grade = 'B'
    print('Tu nota es: ', nota)
except NameError as error:
    print("NameError:", error)
````

### Ejemplo B: `if-else` anidados

Cada `else` contiene otro `if-else`. Se reduce la probabilidad de errores en las condiciones, pero aparecen demasiadas sangrías y la legibilidad sigue siendo deficiente (el `if-else` solo representa dos condiciones).

````python
def calificacion_b(score):
    if score >= 90:       # Si 90 o más, 'A'
        grade = 'A'
    else:
        if score >= 80:   # Si menos de 90, 80 o más, 'B'
            grade = 'B'
        else:
            if score >= 70:
                grade = 'C'
            else:
                if score >= 60:
                    grade = 'D'
                else:
                    grade = 'F'
    return grade

print('Tu nota es: ', calificacion_b(88))
````

### Comparación entre `if-else` anidado e `if-elif`

Ambos códigos hacen la misma tarea. La versión con `elif` tiene menos sangrías y menos líneas: es más fácil de leer.

```text
if score >= 90:            if score >= 90:
    ...                        ...
else:                      elif score >= 80:
    if score >= 80:            ...
        ...                else:
    else:                      ...
        ...
```

### Ejemplo C: `if-elif-else`

El código es más fácil de entender porque cada condición es clara: si la nota no es 'A', se mira 'B', y así sucesivamente. Además la probabilidad de error es menor.

````python
def calificacion_c(score):
    if score >= 90:
        grade = 'A'
    elif score >= 80:     # Si no es 'A', 'B' si es 80 o más
        grade = 'B'
    elif score >= 70:     # Si no es 'B', 'C' si es 70 o más
        grade = 'C'
    elif score >= 60:     # Si no es 'C', 'D' si es 60 o más
        grade = 'D'
    else:                 # De lo contrario, 'F'
        grade = 'F'
    return grade

print('Tu nota es: ', calificacion_c(88))
````

### Traza para `score = 88`

1. Se evalúa `score >= 90`: es `False`, se pasa a la siguiente condición.
2. Se evalúa `score >= 80`: es `True`, se ejecuta `grade = 'B'`.
3. Se salta el resto de las ramas `elif` y el `else`.
4. Se imprime "Tu nota es: B".

Comprobamos que los tres ejemplos coinciden:

````python
for puntos in (100, 90, 89, 80, 79, 70, 69, 60, 59, 0):      # el for se estudia en la Unidad 8
    a, b, c = calificacion_a(puntos), calificacion_b(puntos), calificacion_c(puntos)
    print(puntos, a, b, c, "coinciden" if a == b == c else "difieren")
````

**Clave:** al encadenar con `elif`, el orden de las condiciones importa: se evalúan de arriba hacia abajo y se ejecuta la primera verdadera. Por eso basta con `score >= 80` en la segunda rama: ya se sabe que no es 90 o más.

**Error común:** escribir las condiciones en orden incorrecto, por ejemplo `score >= 60` primero: toda nota de 60 o más recibiría 'D'.

**Para el quiz:** en el ejemplo C con `score = 88` se imprime B.

## Condiciones complejas con operadores de comparación y lógicos

Para condiciones más sofisticadas se combinan operadores de comparación y operadores lógicos. Todos devuelven valores booleanos.

| Operador de comparación | Explicación |
|---|---|
| `==` | `True` si los dos operandos tienen idéntico valor |
| `!=` | `True` si los valores de los operandos son diferentes |
| `>` | `True` si el operando izquierdo es mayor que el derecho |
| `<` | `True` si el operando izquierdo es menor que el derecho |
| `>=` | `True` si el izquierdo es mayor o igual que el derecho |
| `<=` | `True` si el izquierdo es menor o igual que el derecho |

| Operador lógico | Significado |
|---|---|
| `x and y` | Si uno de `x` o `y` es `False`, es `False`; solo es `True` si ambos son `True` |
| `x or y` | Si `x` e `y` son `False`, es `False`; es `True` si alguno es `True` |
| `not x` | Si `x` es verdadero es `False`, y viceversa |

Ejemplo del PDF con dos números `a` y `b`.

````python
a = 10
b = 14
if (a % 2 == 0) and (b % 2 == 0):
    print('Ambos son pares.')
if (a % 2 == 0) or (b % 2 == 0):
    print('Al menos uno es par.')
````

````python
# Si b se cambia a 13, no se cumple la primera condición
a = 10
b = 13
if (a % 2 == 0) and (b % 2 == 0):
    print('Ambos son pares.')
if (a % 2 == 0) or (b % 2 == 0):
    print('Al menos uno es par.')
````

Con `b = 13`: `(b % 2 == 0)` es `False`, así que el `and` completo es `False`; pero `(a % 2 == 0)` es `True`, por lo que el `or` es `True`.

**Nota sobre el PDF:** en las págs. 394 y 395 los mensajes dicen "Ambos son positivos" y "Al menos uno es positivo" aunque la condición comprueba si los números son pares (`% 2 == 0`). Es un error de traducción de "even" (par) y se muestra corregido.

**Clave:** `and` exige que todas las condiciones sean verdaderas; `or` basta con una.

## Aplicación: la frutería de David

Enunciado del PDF: en la frutería de David hay cuatro frutas con precio fijo. El programa muestra el menú, recibe el número de artículo (1 a 4) y la cantidad (1 a 10), imprime el precio total y pide un depósito (dinero). Si el dinero es insuficiente imprime "No hay suficiente dinero."; si no, imprime el cambio.

| Artículo | Fruta | Precio |
|---|---|---|
| 1 | Manzana | 5000 won |
| 2 | Uva | 6000 won |
| 3 | Melón | 8000 won |
| 4 | Naranja | 2000 won |

El núcleo del programa es una cadena `if-elif-else` que asigna la fruta y el precio según el número del pedido. Se encapsula en una función con entradas simuladas.

````python
def fruteria(order, count, money):
    """Frutería de David con los datos ya recibidos (en el PDF se piden con input)."""
    if order == 1:
        fruit = 'Manzana'
        price = 5000
    elif order == 2:
        fruit = 'Uva'
        price = 6000
    elif order == 3:
        fruit = 'Melón'
        price = 8000
    else:
        fruit = 'Naranja'
        price = 2000

    print('Fruta seleccionada :', fruit)
    print('Precio :', price)
    print('Cantidad :', count)
    print('El precio total es de', price * count, 'won.')

    if money < price * count:
        print('No hay suficiente dinero.')
    else:
        change = money - price * count
        print(money, 'won recibidos. Su cambio es de', change, 'won.')

# Ejemplo del PDF: Uva (2), cantidad 2, se inserta 20000
fruteria(2, 2, 20000)
print()
fruteria(4, 5, 8000)       # dinero insuficiente
````

## El módulo `random`: números aleatorios

Para generar números aleatorios hay que importar el módulo `random` con `import random`.

- `random.randint(a, b)` devuelve un entero aleatorio entre `a` y `b`, ambos incluidos.
- `random.randrange(n)` devuelve un entero aleatorio entre 0 y `n - 1`. Con `randrange(2)` se obtiene 0 o 1.

Ejemplo del PDF: un número entre 1 y 100.

````python
import random      # llama al módulo random

number = random.randint(1, 100)
print(number)           # cambia en cada ejecución
print(1 <= number <= 100)
````

### Lanzar una moneda con `randrange(2)`

`random.randrange(2)` devuelve 0 o 1. Con un `if-else` se asigna una cara a cada valor. El PDF imprime 'front' y 'back'; en el código del PDF aparecen como "Frente"/"Atrás" por la traducción ("Delante"/"Atrás" en la salida).

````python
import random

print("Comienza el juego de lanzar la moneda.")
coin = random.randrange(2)
if coin == 0:
    print("Frente.")
else:
    print("Atrás")
print("Juego terminado.")
````

**Ampliación (no está en el PDF):** `random.seed(valor)` fija la semilla del generador. Con la misma semilla se obtiene siempre la misma secuencia de números, lo que sirve para probar un programa de forma reproducible. Cada ejecución sin semilla da resultados distintos.

````python
import random

random.seed(7)
primera = [random.randrange(2) for _ in range(5)]
random.seed(7)
segunda = [random.randrange(2) for _ in range(5)]
print(primera, segunda, primera == segunda)
````

**Nota sobre el PDF:** en la pág. 415 el código tiene comillas rotas (`print("Comienza el juego de lanzar la moneda".)`, `print("Frente".)`). Aquí se muestra corregido.

**Clave:** `randint(1, 3)` incluye ambos extremos; `randrange(2)` no incluye el 2.

## Proyecto de la unidad

### Problema: tanda de penaltis

El usuario lanza un penalti contra el ordenador. Elige una de tres zonas (izquierda, centro, derecha). El ordenador, como portero, defiende una de las tres zonas al azar. Si coinciden, el portero para el balón; si no, es gol.

### Plan de programación (pseudocódigo del PDF)

1. Inicio.
2. Recibir una entrada aleatoria (la zona del portero).
3. El usuario introduce una de las tres opciones.
4. Se comprueba si la dirección del portero y la del balón son idénticas.
5. Si son idénticas, imprimir 'Ataque fallido'.
6. Si son diferentes, imprimir 'Ofensa exitosa'.
7. Fin.

**Referencia:** PDF semana 0, págs. 367–368

### Código del PDF

El PDF genera `n = random.randint(1, 3)` y asigna `computer_choice` con `if-elif-else`. Aquí se pasa `n` como parámetro para poder probarlo sin azar; la versión con azar viene después.

````python
import random

def penalti_pdf(user_choice, n):
    """Código del PDF, con n recibido para que el resultado sea predecible en la prueba."""
    if n == 1:
        computer_choice = "Izquierda"
    elif n == 2:
        computer_choice = "Medio"
    else:
        computer_choice = "Derecha"

    if computer_choice == user_choice:
        print("Ofensa fallida.")
    else:
        print("¡Felicidades! La ofensiva tuvo éxito.")
    print('posición de defensa del ordenador :', computer_choice)

penalti_pdf("derecha", 1)      # ejemplo del PDF: el portero va a la izquierda
````

**Nota sobre el PDF:** el código tiene tres problemas:

1. El ordenador guarda `"Izquierda"`, `"Medio"` y `"Derecha"` con mayúscula inicial, pero la pregunta pide escribir `izquierda`, `centro` o `derecha` en minúsculas. Como `"Izquierda" == "izquierda"` es `False`, el usuario nunca consigue que coincida con "Izquierda", y aunque lance a la misma zona se anunciaría gol. Además la zona central se llama `Medio` en el código y `centro` en la pregunta.
2. El mensaje de la salida de ejemplo ("¡Felicidades! Lo has conseguido.") no coincide con el del código ("¡Felicidades! La ofensiva tuvo éxito.").
3. Entre las comillas aparece una comilla tipográfica (`’`) en una línea por la traducción.

````python
# Demostración del problema 1: lanza a la izquierda y el portero también (n = 1), pero no coinciden
penalti_pdf("izquierda", 1)
````

### Versión corregida

Se usan las mismas palabras en las dos partes (minúsculas, `centro`), se normaliza la entrada con `strip()` y `lower()` y se valida que sea una opción permitida. Devuelve el resultado en lugar de imprimirlo para poder comprobarlo.

````python
import random

def penalti(user_choice, n=None):
    """Tanda de penaltis. Si n es None, la zona del portero es aleatoria (1, 2 o 3)."""
    if n is None:
        n = random.randint(1, 3)        # genera aleatoriamente números entre 1, 2, 3
    if n == 1:
        computer_choice = "izquierda"
    elif n == 2:
        computer_choice = "centro"
    else:
        computer_choice = "derecha"

    user_choice = user_choice.strip().lower()
    if user_choice not in ("izquierda", "centro", "derecha"):
        return "Opción no válida", computer_choice
    if computer_choice == user_choice:
        return "Ofensa fallida.", computer_choice
    return "¡Felicidades! La ofensiva tuvo éxito.", computer_choice

# Pruebas deterministas (n fijo)
print(penalti("derecha", 1))
print(penalti("Izquierda", 1))
print(penalti("centro", 2))
print(penalti("arriba", 3))
````

**Ampliación (no está en el PDF):** con `random.seed` el portero "al azar" es reproducible; sirve para probar el programa con el azar activo.

````python
import random

random.seed(2024)                  # semilla fija: la secuencia de porteros se repite siempre
tandas = [penalti("izquierda") for _ in range(5)]
for resultado, portero in tandas:
    print(f"Portero: {portero:10} -> {resultado}")
````

## Apuntes para estudiar

**Clave**

- `if-else` ejecuta uno de dos bloques: el del `if` si la condición es verdadera, el del `else` si es falsa.
- Dos condiciones excluyentes se expresan mejor con `if-else` que con dos `if`.
- Se pueden anidar sentencias `if-else`; cada nivel añade sangría.
- `elif` ("else if") encadena condiciones: se evalúa solo si las anteriores fueron falsas, y se ejecuta únicamente la primera rama verdadera.
- El orden de las condiciones en `if-elif-else` importa.
- Las condiciones complejas combinan comparaciones con `and`, `or` y `not`.
- Año bisiesto: `(year % 4 == 0 and year % 100 != 0) or (year % 400 == 0)`.
- `random.randint(a, b)` incluye ambos extremos; `random.randrange(n)` devuelve de 0 a `n - 1`.

**Errores comunes**

- Dejar sin comillas el texto de un `print` o de una asignación de cadena.
- Comparar con `>` cuando el enunciado dice "igual o superior" (necesita `>=`).
- Usar `if` independientes cuando las ramas son excluyentes y debería usarse `elif`/`else`.
- Poner las condiciones del `elif` en el orden equivocado.
- Comparar con el nombre de otra variable o con una función incorporada (`id`) en lugar de la variable correcta.
- Olvidar `import random` antes de llamar a `random.randint`.

**Para el quiz**

- Con `score = 88`, el generador de calificaciones imprime B.
- `elif` es la abreviatura de "else if".
- `random.randrange(2)` devuelve 0 o 1.
- `x or y` es `True` si alguno es `True`; `x and y` solo si ambos lo son.
- 1900 no es bisiesto; 2000 sí lo es.
- Una "relación exclusiva" es la de dos condiciones que no pueden cumplirse a la vez.
