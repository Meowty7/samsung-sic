## 1. Qué es una variable

Una **variable** es un nombre que apunta a un valor guardado en la memoria del computador. El PDF la compara con una caja con una etiqueta: la etiqueta es el nombre (`peso`) y dentro va el valor, que puede cambiar en cualquier momento. Ese valor son los **datos**.

Para qué sirven: si un valor se usa en varios lugares, lo guardamos una vez con un nombre y lo reutilizamos. Así un cambio se hace en un solo sitio y se reducen los errores. Una variable puede guardar un único valor (`peso = 78.7`) o varios a la vez (`persona = ('David', 78.7)`, una *tupla*).

A diferencia de C o Java, **Python no exige declarar el tipo** de la variable: se crea al asignarle un valor por primera vez.

````python
peso = 78.7                  # una variable con un solo valor
persona = ('David', 78.7)    # una variable con varios valores (tupla)
print(peso)
print(persona)

peso = 80.2                  # el valor de la "caja" puede cambiar
print(peso)
````

**Clave:** la variable es el nombre; el valor es lo que contiene, y puede reemplazarse.

**Nota sobre el PDF:** el PDF escribe los decimales con coma (`78,7`) por la traducción automática. En Python el separador decimal es el punto: `78.7`. Con coma se crea una tupla, no un número.

**Referencia:** PDF semana 0, págs. 201–202

````python
x = 78,7          # con coma se crea una tupla, no un decimal
print(x, type(x))
````

## 2. El operador de asignación `=` y cómo se guarda un valor

El `=` **no significa "igual"**. Significa: *calcula lo de la derecha y guárdalo con el nombre de la izquierda*. Por eso `resultado = 100 + 200` primero evalúa `100 + 200` (300) y luego guarda 300 en `resultado`. Este `=` se llama **operador de asignación**.

El PDF también explica qué pasa en memoria (pág. 210): Python reserva un espacio en la memoria principal, le pone un nombre (por ejemplo `radio`) y guarda allí el valor (`5.0`). Es más cómodo usar el nombre `radio` que la dirección de memoria (algo como `0x12FF06`).

Asignar y recuperar valores con variables minimiza los errores. Ejemplo del PDF con el círculo:

````python
resultado = 100 + 200
print(resultado)

radius = 4.0
print('Radio del círculo', radius)
print('Área del círculo', 3.14 * radius * radius)
print('Circunferencia del círculo', 2.0 * 3.14 * radius)
````

Si cambiamos `radius`, los demás resultados se actualizan sin tocar las fórmulas.

**Nota sobre el PDF:** en las págs. 203 y 242 el texto traducido usa `print('Radio del círculo', radio)` pero la variable se llamó `radius`; ese código lanzaría `NameError`. Aquí se usa un solo nombre (`radius`), como en la imagen original. También se corrigen los decimales (`3.14`, no `3,14`).

**Clave:** primero se evalúa la derecha, después se guarda en la izquierda.

**Error común:** leer `x = 5` como "x es igual a 5" y esperar que `5 = x` también funcione. No funciona (ver la sección sobre el signo igual).

## 3. Nombres de variables (identificadores) y palabras reservadas

El nombre de una variable se llama **identificador**. Sirve para distinguir unos datos de otros, igual que usamos nombres para distinguir a las personas. Reglas del PDF:

1. Puede combinar letras, dígitos y el guion bajo `_`, pero no puede ser solo un símbolo especial.
2. **No puede empezar por un dígito.**
3. No puede contener espacios ni tabuladores.
4. **Distingue mayúsculas de minúsculas:** `indice` e `INDICE` son variables distintas.
5. No puede ser una **palabra clave (reservada)** de Python.

Ejemplo del PDF: el área de un rectángulo con los identificadores `width` y `height`.

````python
width = 10     # identificador para la anchura del rectángulo
height = 5     # identificador para la altura del rectángulo
rectangle_area = width * height
print('rectangle_area =', rectangle_area)

indice = 1
INDICE = 2     # distinta de "indice": Python distingue mayúsculas y minúsculas
print(indice, INDICE)
````

Verificamos las reglas con ejemplos que fallan. Los errores se capturan para que el notebook siga ejecutándose:

````python
def probar(codigo):
    """Ejecuta una línea de código y muestra el error de sintaxis si lo hay."""
    try:
        exec(codigo, {})
        print(f"Válido:   {codigo}")
    except SyntaxError as error:
        print(f"Inválido: {codigo}   ->   SyntaxError: {error.msg}")

probar("mi_peso = 70")
probar("peso2 = 70")
probar("_peso = 70")
probar("2peso = 70")        # empieza por un dígito
probar("mi peso = 70")      # contiene un espacio
probar("peso-kg = 70")      # el guion medio es un operador (resta), no parte del nombre
probar("global = 500")      # palabra reservada (ejemplo del PDF)
````

### Palabras clave (reservadas) de Python

Python reserva palabras con significado especial (`if`, `for`, `class`, `global`, `True`...). No se pueden usar como nombre de variable: provoca un `SyntaxError`. El PDF muestra cómo consultar la lista con el módulo `keyword` (el `import` se estudia más adelante en el curso):

````python
import keyword

# Búsqueda de palabras clave Python
print("Python keyword list...")
print(keyword.kwlist)
print("Cantidad:", len(keyword.kwlist))
````

**Nota sobre el PDF:** la tabla de la pág. 206 omite algunas palabras (por ejemplo `async` y `await`); la lista completa y siempre actual es la que imprime `keyword.kwlist`. La cantidad depende de la versión de Python.

**Ampliación (no está en el PDF):** `str.isidentifier()` indica si un texto cumple las reglas de forma de un identificador y `keyword.iskeyword()` si es palabra reservada. Un nombre es utilizable si cumple lo primero y no lo segundo.

````python
def es_nombre_utilizable(nombre):
    return nombre.isidentifier() and not keyword.iskeyword(nombre)

for candidato in ["peso", "2peso", "mi peso", "global", "Global", "_x"]:
    print(f"{candidato!r:10} -> {es_nombre_utilizable(candidato)}")
````

## 4. Cadenas de texto

Una variable también puede guardar texto. Los datos de texto se llaman **cadenas** (*strings*) y se escriben entre comillas simples `'...'` o dobles `"..."` (son equivalentes). La variable `s1` es la "etiqueta" que apunta al objeto cadena, formado por una secuencia de caracteres. Las cadenas se pueden unir con `+`.

````python
s1 = 'Hello World'
print(s1)

s1 = "Hello"
s2 = "World!"
print(s1 + s2)               # el + une las cadenas sin añadir espacios
print(s1 + " " + s2)         # ampliación: añadimos el espacio nosotros
````

**Clave:** comillas simples y dobles son equivalentes; `+` entre cadenas las concatena.

**Error común:** esperar un espacio automático al concatenar: `"Hello" + "World!"` da `HelloWorld!`.

## 5. Tipos de datos y `type()`

Un **tipo de dato** es la clase de dato que un lenguaje puede procesar. Los básicos: booleanos, números (enteros, reales y complejos), cadenas, listas, tuplas, conjuntos y diccionarios. La función `type()` devuelve el tipo de un objeto. Cada variable necesita espacio de memoria acorde a su valor, pero en Python **no se especifica el tipo**: Python elige el más adecuado.

````python
num = 85
print(type(num))

pi = 3.14159
print(type(pi))

message = "Good morning"
print(type(message))

name = 'David'
age = 27
print(type(name))
print(type(age))
````

### Tipado dinámico frente a estático (Un paso más)

| Tipificación estática | Tipificación dinámica |
|---|---|
| Examina los tipos al compilar | No hace falta especificar cada tipo: código más sencillo |
| Es necesario especificar cada tipo | Puede dar errores de tipo durante la ejecución |
| C, C++, C#, Java, Objective-C, Pascal | Python, Basic, Ruby, PHP, JavaScript |

Python usa **tipado dinámico**: la misma variable puede apuntar a valores de tipos distintos en momentos distintos.

````python
dato = 85
print(type(dato))
dato = "ahora soy texto"     # misma variable, otro tipo
print(type(dato))
````

**Clave:** `type(x)` dice qué tipo tiene el valor; el tipo lo determina Python, no el programador.

## 6. Asignación simultánea, múltiple y el signo igual

- **Asignación simultánea:** varias variables en una sola línea, `x, y = 100, 200`.
- **Asignación múltiple:** el mismo valor a varias variables, `num1 = num2 = num3 = 200`. Según el PDF, el orden de asignación empieza por la derecha: primero `num3`, luego `num2`, luego `num1`.
- El `=` **no** compara: `300 = 300` es un error porque el valor 300 no puede recibir otro valor. Para comparar se usa `==` (se estudia en la Unidad 5).

````python
x, y = 100, 200
result = x + y
print(result)

num1 = num2 = num3 = 200
print(num1, num2, num3)
````

````python
# El signo igual no significa que dos valores sean iguales
try:
    exec("300 = 300")
except SyntaxError as error:
    print("SyntaxError:", error.msg)
````

**Ampliación (no está en el PDF):** la asignación simultánea permite intercambiar dos valores sin variable auxiliar.

````python
a, b = 1, 2
a, b = b, a
print(a, b)
````

## 7. Operadores de asignación compuestos

Combinan una operación aritmética con la asignación: `i += 10` equivale a `i = i + 10`. Son habituales en el código de otras personas, así que hay que saber leerlos.

| Operador | Descripción | Ejemplo |
|---|---|---|
| `+=` | suma y asigna | `i += 10` |
| `-=` | resta y asigna | `i -= 10` |
| `*=` | multiplica y asigna | `i *= 10` |
| `/=` | divide y asigna | `i /= 10` |
| `%=` | módulo y asigna | `i %= 10` |
| `^=` | XOR a nivel de bits y asigna | `i ^= 10` |

Ejemplo del PDF (suma acumulativa):

````python
num = 200
num += 100
print(num)
num -= 100
print(num)
num *= 20
print(num)
num /= 2
print(num)
````

Observa que tras `/=` el resultado es `2000.0` (decimal): la división con `/` siempre devuelve un número real.

**Ampliación (no está en el PDF):** también existen `//=` y `**=` (aparecen en la tabla de precedencias de la Unidad 5).

````python
i = 17
i //= 5
print(i)      # 3
i **= 3
print(i)      # 27
i %= 10
print(i)      # 7
````

**Clave:** `a op= b` equivale a `a = a op b`.

**Error común:** usar `+=` sobre una variable que todavía no existe (`NameError`).

**Referencia:** PDF semana 0, págs. 221–222

## 8. Operadores aritméticos (incluye potencia y resto)

En `a + b`, el `+` es el operador y `a`, `b` son los **operandos**. En `a = 10 + 1` primero se evalúa `10 + 1` y el resultado (11) se asigna a `a`.

| Operador | Significado | Ejemplo | Resultado |
|---|---|---|---|
| `+` | Adición | `7 + 4` | 11 |
| `-` | Resta | `7 - 4` | 3 |
| `*` | Multiplicación | `7 * 4` | 28 |
| `/` | División real (siempre devuelve decimal) | `7 / 4` | 1.75 |
| `//` | División entera (cociente, descarta decimales) | `7 // 4` | 1 |
| `%` | Resto o módulo (no tiene relación con porcentajes) | `7 % 4` | 3 |
| `**` | Potenciación (eleva el izquierdo al derecho) | `7 ** 2` | 49 |

Pista del PDF: la raíz cuadrada de 4 se obtiene con `4 ** 0.5`.

````python
print("7 + 4 =", 7 + 4)
print("7 - 4 =", 7 - 4)
print("7 * 4 =", 7 * 4)
print("7 / 4 =", 7 / 4)
print("7 // 4 =", 7 // 4)
print("7 % 4 =", 7 % 4)
print("7 ** 2 =", 7 ** 2)
print("4 ** 0.5 =", 4 ** 0.5)
````

**Nota sobre el PDF:** en la tabla de la pág. 224 la descripción de `**` ("multiplica el operando izquierdo por la potencia del operando derecho") es imprecisa: `a ** b` **eleva** `a` a la potencia `b`. Además la columna "Firme" de la pág. 225 es "Signo" (el símbolo del operador) y "De vuelve" es "Devuelve".

### Resto y cociente juntos

`//` y `%` se complementan: `a == (a // b) * b + a % b`. Con ellos se puede, por ejemplo, saber si un número es par (`n % 2 == 0`) o separar dígitos.

**Ampliación (no está en el PDF):** `divmod(a, b)` devuelve cociente y resto a la vez, y la prioridad entre operadores es la habitual de la matemática (se detalla en la Unidad 5).

````python
a, b = 47, 5
print(a // b, a % b)
print(divmod(a, b))
print((a // b) * b + a % b == a)
print(2 + 3 * 4, (2 + 3) * 4)     # la multiplicación se evalúa antes que la suma
````

### Precaución: división entre cero

Dividir un valor entre 0 provoca `ZeroDivisionError`. El PDF indica leer atentamente el mensaje de error.

````python
num1 = 3
num2 = 0
try:
    print(num1 / num2)
except ZeroDivisionError as error:
    print("ZeroDivisionError:", error)
````

**Clave:** `/` siempre da decimal; `//` da el cociente entero; `%` el resto; `**` la potencia.

**Error común:** creer que `%` calcula porcentajes; y dividir (o hacer `%`) entre cero.

**Para el quiz:** `7 // 4 = 1`, `7 % 4 = 3`, `7 / 4 = 1.75`, `7 ** 2 = 49`.

## 9. Entrada del usuario con `input()`

`input("mensaje")` muestra un mensaje, espera a que el usuario escriba y **devuelve siempre una cadena** (`str`), aunque escriba dígitos. Para operar con números hay que convertir: `int(...)` para enteros y `float(...)` para decimales.

Ejemplos del PDF (se muestran con la entrada simulada; en un notebook interactivo `input()` abriría un cuadro de texto). Para poder ejecutar el notebook sin interrupciones, definimos una pequeña función que *simula* a un usuario con una lista de respuestas. En tus propios programas usarás `input()` directamente.

````python
respuestas = iter(["David", "300", "400"])

def input_simulado(mensaje=""):
    """Imita input(): muestra el mensaje y devuelve la siguiente respuesta de prueba."""
    respuesta = next(respuestas)
    print(mensaje + respuesta)
    return respuesta

# Ejemplo de la pág. 235-236
name = input_simulado("Introduzca el nombre: ")
print("Hola,", name + ".")
print("Bienvenido al mundo de la programación en Python.")

# Ejemplo de la pág. 237: suma de dos enteros introducidos por el usuario
x = int(input_simulado("Introduzca el primer número entero: "))
y = int(input_simulado("Introduzca el segundo número entero: "))
s = x + y
print("La suma de", x, "y", y, "es", s)
````

**Nota sobre el PDF:** en la pág. 237 la versión traducida aparece como `print("La suma de", x, "y" y, "es", s)`, que no es válida (falta una coma tras `"y"`). Correcto: `print("La suma de", x, "y", y, "es", s)`.

### `input()` devuelve una cadena

Esto explica el `ValueError` y el `TypeError` que muestra el PDF.

````python
x = "100"                       # lo que devolvería input() al escribir 100
print("Tipo de datos de entrada:", type(x))

# Convertimos solo si la cadena contiene solo dígitos (idea del PDF, pág. 239)
if x.isdigit() == True:
    x = int(x)
    print("Tipo de datos de entrada:", type(x))
else:
    print("x no es un tipo numérico.")
````

### Cuidado con el `ValueError` y el `TypeError`

Si el usuario escribe `cien`, `int("cien")` falla con `ValueError` (en el PDF aparece traducido como "ErrorValor"). Y sumar una cadena con un entero produce `TypeError`.

````python
try:
    int("cien")
except ValueError as error:
    print("ValueError:", error)

my_age = 22
my_height = '177'
my_age = my_age + 1
try:
    my_height = my_height + 1
except TypeError as error:
    print("TypeError:", error)
print(my_age, my_height)
````

**Nota sobre el PDF:** el texto de la pág. 244 menciona la variable `my_heigth` (con errata); el código usa `my_height`.

**Ampliación (no está en el PDF):** para sumar texto y número hay dos caminos: convertir el texto a número (`int(my_height) + 1`) o el número a texto (`my_height + str(1)`, que concatena y da `'1771'`).

````python
my_height = '177'
print(int(my_height) + 1)       # 178: aritmética
print(my_height + str(1))       # '1771': concatenación
````

**Clave:** `input()` → siempre `str`; convertir con `int()` / `float()` antes de calcular.

**Error común:** usar el resultado de `input()` directamente en una suma o un `**`; o escribir un decimal con coma (`1,75`), que `float()` no acepta.

**Para el quiz:** el tipo que devuelve `input()` es `str`.


Para poder ejecutarlo sin interrupciones lo encapsulamos en una función que recibe cómo leer los datos (`leer`), y la probamos con una lista de respuestas de prueba.

````python
def calculadora_imc(leer=input):
    """Pide nombre, altura (cm) y peso (kg) y muestra el IMC (pasos del plan del PDF)."""
    name = leer("Introduzca el nombre: ")                    # [2] nombre
    height_cm = int(leer("Introduzca la altura (cm): "))     # [3] altura en cm
    height_m = height_cm / 100                               # [4] cm -> m
    weight = int(leer("Introduzca el peso (kg): "))          # [5] peso
    bmi = weight / (height_m ** 2)                           # [6] IMC
    print("El IMC de", name, "es", bmi)                      # [7] resultado
    return bmi

# Versión no interactiva con respuestas de prueba (las del ejemplo del PDF)
respuestas = iter(["David", "170", "70"])
imc = calculadora_imc(input_simulado)

# Para usarlo de forma interactiva, quita el comentario de la siguiente línea
# calculadora_imc()
````

## Apuntes para estudiar

**Clave**

- Variable = nombre que apunta a un valor; `=` asigna (evalúa la derecha y guarda a la izquierda), no compara.
- Identificadores: letras, dígitos y `_`; no empiezan por dígito; sin espacios; distinguen mayúsculas; no pueden ser palabras reservadas.
- `type(x)` indica el tipo; Python tiene tipado dinámico (no se declara el tipo).
- Asignación simultánea `x, y = 1, 2`; múltiple `a = b = c = 0`.
- Compuestos: `+=`, `-=`, `*=`, `/=`, `%=`, `//=`, `**=`; `a op= b` equivale a `a = a op b`.
- Aritméticos: `+ - * / // % **`; `/` siempre devuelve decimal.
- `input()` devuelve siempre `str`; convertir con `int()` o `float()`.
- IMC = peso (kg) / altura (m) ** 2; la altura en cm se divide entre 100.

**Errores comunes**

- Usar el resultado de `input()` en una operación sin convertirlo (`TypeError` o concatenación inesperada).
- Escribir decimales con coma (`1,75`): `float()` falla y `78,7` crea una tupla.
- Nombrar variables con palabras reservadas (`global`, `class`) o empezar por un número.
- Confundir `=` (asignar) con `==` (comparar); `300 = 300` es `SyntaxError`.
- Creer que `%` calcula porcentajes: es el resto de la división.
- Dividir o calcular el resto entre cero (`ZeroDivisionError`).
- Usar un nombre distinto del definido (`radio` frente a `radius`): `NameError`.

**Para el quiz**

- `7 // 4 = 1`, `7 % 4 = 3`, `7 / 4 = 1.75`, `7 ** 2 = 49`.
- `type(input())` es `str`; para enteros se usa `int(input(...))`.
- `global = 500` produce `SyntaxError: invalid syntax`.
- `indice` e `INDICE` son variables distintas.
- Tras `num = 200; num += 100; num -= 100; num *= 20; num /= 2` el valor es `2000.0`.
- IMC normal según la OMS: 18.5 a 24.9; sobrepeso 25.0 a 29.9; obesidad desde 30.0.
