## 2. Pensar como un ordenador y las 3 estructuras de la programación

El PDF recomienda la charla TED **"3 maneras de tomar mejores decisiones pensando como un ordenador"** de Tom Griffiths (enlace de la diapositiva: https://youtu.be/1mLQFm3wEfw).

Para pensar y resolver problemas como un informático hay que comprender los fundamentos del funcionamiento de la programación. A lo largo del curso se mejora la eficiencia minimizando las tareas repetitivas, operando en **secuencia** y tomando **decisiones** basadas en condiciones claras. La primera tarea consiste en usar `print()` para aprender la **estructura secuencial** y hacer que un programa funcione en orden.

Las tres estructuras que muestra el PDF mediante diagramas de flujo son:

- **Secuencia:** las instrucciones se ejecutan una detrás de otra, en orden.
- **Bucle:** un bloque se repite mientras se cumpla una condición.
- **Selección:** según una condición (verdadera o falsa) se elige uno de dos caminos.

**Referencia:** PDF semana 0, págs. 21–22

**Ampliación (no está en el PDF):** el PDF solo muestra los diagramas de flujo de las tres estructuras. Este es un ejemplo mínimo de cada una en Python; `for` e `if` se estudian más adelante, no hace falta dominarlos todavía.

````python
# Secuencia: cada línea se ejecuta una vez, de arriba hacia abajo
print("Paso 1")
print("Paso 2")

# Bucle: un bloque que se repite
for vuelta in range(3):
    print("Repetición número", vuelta + 1)

# Selección: se elige un camino según una condición
temperatura = 30
if temperatura > 25:
    print("Hace calor")
else:
    print("La temperatura es agradable")
````

**Clave:** toda la programación se construye combinando estas tres estructuras; esta unidad trabaja la primera, la secuencia.

## 3. Secuencia

La **secuencia** es el orden en que un ordenador ejecuta sus códigos. Como el programa es muy rápido, puede parecer que las instrucciones se ejecutan a la vez, pero en realidad el ordenador las procesa en orden. Incluso con los mismos comandos, si cambia el orden se produce un resultado completamente distinto. Por eso, al escribir código hay que tener en cuenta la prioridad y el propósito de los comandos antes de decidir su orden.

El PDF lo ilustra con la vida cotidiana: poner pasta de dientes en el cepillo, usar el cepillo para limpiar los dientes y enjuagar el cepillo. Si se cambia el orden (enjuagar primero), el resultado ya no es el esperado.

**Referencia:** PDF semana 0, pág. 26

````python
print("Orden correcto:")
print("1. poner pasta de dientes en el cepillo")
print("2. utilizar el cepillo de dientes para limpiar los dientes")
print("3. enjuagar el cepillo de dientes")

print()  # línea en blanco para separar
print("Orden cambiado (el resultado ya no es el esperado):")
print("1. enjuagar el cepillo de dientes")
print("2. poner pasta de dientes en el cepillo")
print("3. utilizar el cepillo de dientes para limpiar los dientes")
````

**Clave:** Python ejecuta las instrucciones de arriba hacia abajo; el orden forma parte del significado del programa.

**Error común:** pensar que el orden de las líneas no importa porque el ordenador "lo hace todo a la vez".

**Para el quiz:** la secuencia es el orden en que el ordenador ejecuta sus códigos.

### 4.2 Expresión, declaración y programa

- Una **expresión** es un código simple que produce un valor: números, fórmulas y cadenas (por ejemplo `352`, `10 + 20`, `"¡Hola Mundo!"`, `17.2`).
- Una **declaración** (o sentencia) se compone de una o más expresiones (por ejemplo `print('¡Hola Mundo!')`).
- Un **programa** se compone de una o más declaraciones.

**Nota sobre el PDF:** en la diapositiva, la traducción automática convirtió el código real en "tiempo de importación" y "timesleep(5)". El código original es `import time` y `time.sleep(5)`: importar el módulo `time` y esperar 5 segundos.

**Referencia:** PDF semana 0, págs. 27–28

````python
import time

# Expresiones: producen un valor
print(352)
print(10 + 20)
print("¡Hola Mundo!")
print(17.2)

# Declaración: una instrucción que se ejecuta
time.sleep(1)  # el PDF espera 5 segundos; aquí 1 para no hacer lento el cuaderno

# Programa: varias declaraciones juntas
import time
time.sleep(1)
print("¡Hola Mundo!")
````

**Clave:** expresión (produce un valor) < declaración (una o más expresiones) < programa (una o más declaraciones).

**Error común:** copiar `timesleep(5)` tal como aparece en el PDF; el código correcto es `time.sleep(5)` con el punto.

**Para el quiz:** un programa se compone de declaraciones, y una declaración se compone de expresiones.

### 4.3 Anotación (comentarios)

Una **anotación** es una frase que describe la función del código dentro del programa. El intérprete omite su ejecución. Las anotaciones ayudan a otros programadores a entender tu código y son útiles como referencia futura en códigos complejos.

Hay dos formas de añadir anotaciones:

1. **De una sola línea:** se añade el signo `#` al principio de la frase; toda la línea se convierte en anotación. También puede ir al final de una línea de código.
2. **De varias líneas:** se escriben tres comillas simples (`'''`) o tres comillas dobles (`"""`) en la línea anterior y en la posterior al texto.

**Referencia:** PDF semana 0, págs. 32–34

**Ampliación (no está en el PDF):** este pequeño simulador ejecuta los comandos de Q1 en orden y muestra la cuadrícula resultante. Usa un bucle `for` y una lista de listas, temas que se verán más adelante; sirve para comprobar tu resultado en papel.

````python
def ejecutar_comandos(comandos, filas=3, columnas=3):
    """Ejecuta una secuencia de comandos sobre una cuadrícula; empieza arriba a la izquierda."""
    cuadricula = [["."] * columnas for _ in range(filas)]   # "." = blanco
    fila, columna = 0, 0
    for comando in comandos:
        if comando == "derecha":
            columna += 1
        elif comando == "izquierda":
            columna -= 1
        elif comando == "abajo":
            fila += 1
        elif comando == "arriba":
            fila -= 1
        elif comando == "rellenar":
            cuadricula[fila][columna] = "#"                 # "#" = negro
    return cuadricula


comandos_q1 = ["derecha", "rellenar", "derecha", "abajo", "rellenar", "izquierda", "izquierda",
               "rellenar", "abajo", "derecha", "rellenar", "derecha"]

for linea in ejecutar_comandos(comandos_q1):
    print(" ".join(linea))
````

**Clave:** un algoritmo en secuencia es una lista ordenada de comandos; cambiar el orden cambia el resultado final.

### 6.3 Entorno virtual: definición y necesidad

Al programar con Python es probable que se utilicen varios paquetes externos. Si hay varios proyectos con paquetes distintos, o se necesita actualizar a la última versión de Python o usar una versión diferente, la gestión de módulos y el control de versiones se vuelven difíciles y pueden surgir problemas de compatibilidad entre versiones.

Por eso se construye un **entorno virtual**: un espacio independiente para gestionar proyectos, módulos y versiones; es decir, crear varios espacios de trabajo en un solo ordenador (como tener varios servidores). Ejemplo del PDF: el entorno `data_science` usa Python 3.7; el entorno `Python_work` usa Python 3.8 con pandas instalado; ambos conviven en el mismo ordenador.

## 8. La función `print()`

### 8.1 Los fundamentos

`print()` es un comando que muestra en pantalla los valores que se escriben dentro de los paréntesis. El valor entre comillas abiertas y cerradas se llama **cadena** (string). Se pueden usar comillas simples o dobles, pero el símbolo de inicio y el de fin tienen que coincidir.

Consejo del PDF: a diferencia de otros lenguajes, Python empieza automáticamente una nueva línea después de imprimir una cadena con `print()`.

**Referencia:** PDF semana 0, pág. 56

### 8.2 Imprimir cadena y valor

Para imprimir una cadena se encierra entre comillas. Para imprimir un valor numérico se escribe el número **sin comillas**; si se escribe entre comillas, el tipo de dato será una cadena y no un número. También se pueden hacer operaciones aritméticas dentro de `print()` (las operaciones con tipos numéricos se tratarán en la Unidad 3).

**Referencia:** PDF semana 0, pág. 57

````python
print('Hello World!')
print("Hello World!")

print(10)
print(7.5)

print(10 + 7.5)
print(10 - 7.5)

# Una cadena que parece número sigue siendo una cadena
print("10")
print(type(10), type("10"))
````

**Clave:** sin comillas se imprime un valor (número); con comillas, una cadena. Las comillas de apertura y cierre deben ser del mismo tipo.

**Error común:** abrir con comilla simple y cerrar con doble (o dejar la cadena sin cerrar), lo que produce un error de sintaxis.

**Referencia:** PDF semana 0, págs. 56–57

### 8.3 Añadir una nueva línea

`print()` pasa a una nueva línea después de cada impresión. Para insertar saltos de línea **dentro** de una misma cadena se usa la **secuencia de escape** `\n`: un carácter especial que controla la salida de una cadena. Otra forma es encerrar la cadena entre tres comillas dobles (`"""`) o simples (`'''`), con lo que el salto de línea del código se muestra tal cual. Si se pone una barra invertida `\` al final de una línea dentro de esas comillas triples, la línea no se rompe: la línea siguiente se conecta a la misma línea.

**Referencia:** PDF semana 0, págs. 58–59

````python
print('Hello\nWorld!')

print('''Hello
World!''')

# La barra invertida al final de la línea une las dos líneas
print('''Hello\
World!''')
````

**Clave:** `\n` es el salto de línea; las comillas triples conservan los saltos del código; la barra invertida al final los anula.

**Para el quiz:** `print('Hello\nWorld!')` imprime `Hello` y `World!` en líneas separadas.

**Referencia:** PDF semana 0, págs. 58–59

### 8.4 Los operadores coma (,), suma (+) y multiplicación (*)

- **Coma:** entre dos valores, imprime un espacio entre ellos y luego el salto de línea. Permite imprimir dos o más cadenas o valores, e incluso mezclar una cadena y un número.
- **Suma `+`:** entre cadenas las une (concatena) sin espacio; entre números los suma. **No se pueden combinar una cadena y un número con `+`:** se produce un `TypeError`.
- **Multiplicación `*`:** una cadena por un entero positivo repite la cadena esa cantidad de veces, sin espacios. Entre números multiplica. **Una cadena por otra cadena produce un `TypeError`.**

**Nota sobre el PDF:** la introducción de la pág. 65 escribe "coma (+)" cuando se refiere a la coma `,`.

**Referencia:** PDF semana 0, págs. 60–62

````python
print('Hello', 'World!')
print(10, 20)
print('Hello', 10)        # la coma sí permite mezclar cadena y número

print('Hello' + 'World!')
print(10 + 20)

print('Hello' * 3)
print(10 * 20)
````

````python
# Los dos errores que muestra el PDF (se capturan para no detener el cuaderno)
try:
    print('Hello' + 10)
except TypeError as error:
    print("TypeError:", error)

try:
    print('Hello' * 'World!')
except TypeError as error:
    print("TypeError:", error)
````

**Clave:** la coma separa con espacio y admite tipos mezclados; `+` y `*` exigen tipos compatibles.

**Error común:** `print('Edad: ' + 23)` produce `TypeError`; usa `print('Edad:', 23)`.

**Para el quiz:** `'Hello' * 3` da `HelloHelloHello`; `'Hello' * 'World!'` da `TypeError`.

**Referencia:** PDF semana 0, págs. 60–62

### 8.5 Parámetros `end` y `sep`

- **`end`:** por defecto `print()` termina con un salto de línea. Con `end=''` (o cualquier cadena) se imprime esa cadena en lugar del salto de línea. Ejemplo: `end=''` imprime "Hello" e inmediatamente después "World!" sin romper la línea; `end=','` imprime una coma en lugar del salto.
- **`sep`:** por defecto la coma entre valores imprime un espacio. Con `sep=''` o `sep=','` se imprime esa cadena en lugar del espacio.

**Nota sobre el PDF:** los títulos dicen "Parámetros End y Split" (y la pág. 65 habla de "división (sep)"). El segundo parámetro se llama `sep` (de *separator*, separador); `split` es otra cosa (un método de las cadenas que se verá en la sección de errores). Además, el texto describe `sep` como si dividiera, pero lo que hace es separar los valores.

**Referencia:** PDF semana 0, págs. 63–64

````python
print('Hello', end='')
print('World!')

print('Hello', end=',')
print('World!')

print('Hello', 'World!', sep='')
print('Hello', 'World!', sep=',')
````

**Clave:** `end` controla lo que se imprime al final; `sep` controla lo que se imprime entre valores.

**Error común:** confundir `sep` con `split`, o escribir `end` con mayúscula (`End`): los parámetros van en minúscula.

**Para el quiz:** `print('Hello', 'World!', sep=',')` imprime `Hello,World!`.

**Referencia:** PDF semana 0, págs. 63–64

## 9. Método Python: estilo de codificación y convención de nomenclatura

### 9.1 Por qué importa el estilo

Es importante formar un buen estilo y hábito de codificación desde el principio. El objetivo es escribir códigos claros y concisos. Usar las convenciones de codificación que usan las comunidades de desarrolladores también ayuda a colaborar con otros. Existen dos guías de estilo principales para Python: **PEP 8** (de la comunidad oficial de Python, https://www.python.org/dev/peps/pep-0008/) y la guía de estilo de Google. Esta unidad (y el curso) usa PEP 8.

### 9.2 Disposición del código

- Python permite escribir varias sentencias en una línea con punto y coma, pero tiene mala legibilidad: se recomienda **una sentencia por línea**.
- Si una sentencia es demasiado larga, se puede romper la línea con la **barra invertida `\`** al final.
- Los operadores y operandos llevan un espacio entre ellos, y al enumerar elementos se pone un espacio después de la coma.

**Referencia:** PDF semana 0, págs. 66–68

````python
# Mala legibilidad (varias sentencias en una línea)
a = 10; a += 10; print(a)     # Bad case

# Buena práctica: una sentencia por línea
a = 10                        # Good case
a += 10
print(a)

# Sentencia larga partida con barra invertida
width, height, color, emphasis, highlight = 1, 1, 'red', 'strong', 50
if width == 0 and height == 0 and color == 'red' and \
   emphasis == 'strong' or highlight > 100:
    print('Sorry, you lose.')
else:
    print('Todavía no pierdes')

lista_mala = [1,2,3]          # Bad
lista_buena = [1, 2, 3]       # Good
print(lista_buena)
````

### 9.3 Convención de nomenclatura

Los nombres usados en programación (variables, funciones, clases, etc.) se llaman **identificadores**. En Python, los nombres de variables y funciones siguen generalmente el estilo serpiente y los nombres de clases el estilo Pascal.

| Estilo | Descripción | Ejemplo |
|---|---|---|
| Serpiente (snake) | Primera letra en minúscula y `_` en lugar de espacios entre palabras | `hello_world` |
| Camello (camel) | Primera letra en minúscula; cada palabra siguiente empieza en mayúscula | `holaMundo` |
| Pascal | Como camello, pero la primera letra también en mayúscula | `HolaMundo` |

Reglas de los identificadores (según el PDF):

1. Generalmente se componen de caracteres ingleses, números y guiones bajos (`_`).
2. No debe haber espacios en medio.
3. La primera palabra debe empezar por una letra inglesa o un guion bajo (`_`).
4. Se distinguen mayúsculas de minúsculas (`Count` y `count` son distintos).
5. No hay límite para la longitud.
6. Las palabras clave/reservadas no pueden utilizarse como identificadores.
7. Los identificadores que comienzan por `__` (dos guiones bajos) se refieren a atributos especiales o nombres de métodos específicos; no se usan como nombre de función o variable general.

Identificadores **correctos** del PDF: `number4`, `my_list`, `__code__`, `for_loop` (una palabra clave puede formar parte de un nombre más largo). **Incorrectos:** `1st_variable` (empieza por número), `my list` (tiene espacio), `global` (palabra reservada), `ver2.9` y `num&co` (caracteres especiales distintos del guion bajo).

**Referencia:** PDF semana 0, págs. 69–72

````python
import keyword

correctos = ["number4", "my_list", "__code__", "for_loop"]
incorrectos = ["1st_variable", "my list", "global", "ver2.9", "num&co"]

for nombre in correctos + incorrectos:
    valido = nombre.isidentifier() and not keyword.iskeyword(nombre)
    print(f"{nombre!r:15} -> {'válido' if valido else 'inválido'}")

# Mayúsculas y minúsculas distinguen identificadores
Count = 1
count = 2
print(Count, count)
````

**Ampliación (no está en el PDF):** la regla 1 dice "caracteres ingleses", pero Python 3 también admite letras de otros alfabetos en los identificadores (por ejemplo `año`). Aun así, en este curso se recomienda usar solo letras inglesas, números y guion bajo, como pide el PDF.

**Clave:** variables y funciones en estilo serpiente (`mi_variable`); sin espacios, sin empezar por número, sin palabras reservadas.

**Error común:** nombrar una variable `1st_variable` o `my list`; usar `Count` y `count` creyendo que son la misma variable.

**Para el quiz:** `count` y `Count` son identificadores diferentes; el estilo `holaMundo` se llama camello.

**Referencia:** PDF semana 0, págs. 66–72

## 10. Errores: error de sintaxis y error de tiempo de ejecución

### 10.1 Error de sintaxis

Los lenguajes de programación se rigen por normas y convenciones estrictas. Escribir un código que se aleje de las reglas gramaticales da lugar a un **error de sintaxis** (`SyntaxError`). Hay que aprender y practicar la gramática de Python y codificar según las reglas. Ejemplo del PDF: para imprimir la cadena `'Hola'` hay que poner comillas a ambos lados; el código `print('Hello)` deja la cadena sin cerrar y produce `SyntaxError: EOL while scanning string literal`. El código **no llega a ejecutarse**: el error se detecta antes.

**Referencia:** PDF semana 0, pág. 73

### 10.2 Error de tiempo de ejecución

A diferencia de los errores de sintaxis, un código gramaticalmente correcto puede dar lugar a errores **durante la ejecución**: son los **errores de tiempo de ejecución**. El desarrollador debe tener en cuenta que el usuario puede introducir datos incorrectos y sugerir una solicitud fácil de usar para fomentar la entrada de datos correctos; además, conviene probar el código con diversas condiciones. En el ejemplo del PDF el usuario introduce la cadena `two` en lugar de un entero y se produce `ValueError: invalid literal for int() with base 10: 'two'`.

**Referencia:** PDF semana 0, pág. 74

````python
# Error de sintaxis: se detecta ANTES de ejecutar; lo comprobamos compilando el texto del código
try:
    compile("print('Hello)", "<celda>", "exec")
except SyntaxError as error:
    print("Error de sintaxis:", error.msg)

# Error de tiempo de ejecución: el código es válido, falla al ejecutarse
try:
    numero = int("two")        # el PDF usa int(input('Input an integer: ')) y el usuario escribe 'two'
except ValueError as error:
    print("Error de tiempo de ejecución (ValueError):", error)
````

**Nota sobre el PDF:** el mensaje `EOL while scanning string literal` es el de versiones antiguas de Python; las versiones recientes dicen `unterminated string literal (detected at line 1)` (la celda anterior muestra el mensaje de la versión que usa este cuaderno).

### 10.3 Un paso más: resolver errores con `try ... except`

El PDF añade un ejemplo de error de ejecución común, `ZeroDivisionError` (dividir entre cero), y cómo resolverlo con las sentencias `try ~ except`. Si el usuario introduce correctamente dos enteros (por ejemplo `10 2`), se imprime `10/2 = 5.0`; si introduce una cadena en lugar de un entero (`10 dos`) o divide entre cero (`10 0`), la sentencia `except` se encarga del caso de excepción y se imprime el mensaje "Compruebe si los números enteros se han introducido correctamente."

El código original usa `input()` y `.split()`. Aquí se encapsula en una función que recibe el texto como argumento, para poder probarlo sin escribir por teclado.

**Referencia:** PDF semana 0, págs. 75–76

````python
def dividir_dos_enteros(texto):
    """Divide dos enteros escritos en un texto separados por un espacio, como en la pág. 75 del PDF."""
    try:
        a, b = texto.split()
        resultado = int(a) / int(b)
        print('{}/{} = {}'.format(a, b, resultado))
    except:
        print('Compruebe si los números enteros se han introducido correctamente.')


# Casos del PDF: "10 2", "10 dos" y "10 0"
for entrada in ["10 2", "10 dos", "10 0"]:
    print("Entrada:", entrada)
    dividir_dos_enteros(entrada)

# Para usarlo de forma interactiva, quita el comentario de la siguiente línea
# dividir_dos_enteros(input('Introduzca dos enteros. '))
````

**Ampliación (no está en el PDF):** un `except` sin tipo atrapa cualquier error y puede ocultar fallos inesperados. Lo recomendable es indicar qué errores se esperan, como en esta variante.

````python
def dividir_dos_enteros_seguro(texto):
    try:
        a, b = texto.split()
        print('{}/{} = {}'.format(a, b, int(a) / int(b)))
    except (ValueError, ZeroDivisionError) as error:
        print('Compruebe si los números enteros se han introducido correctamente. Detalle:', error)


dividir_dos_enteros_seguro("10 0")
dividir_dos_enteros_seguro("10 dos")
````

**Clave:** error de sintaxis = el código está mal escrito y ni siquiera arranca; error de tiempo de ejecución = el código está bien escrito pero falla mientras corre (por ejemplo `int('two')` o dividir entre cero).

**Error común:** creer que un programa sin errores de sintaxis ya está libre de errores; los datos del usuario pueden provocar errores de ejecución.

**Para el quiz:** `ZeroDivisionError` y `ValueError` son errores de tiempo de ejecución; `SyntaxError` es de sintaxis. Se tratan con `try ... except`.

````python
import io
from contextlib import redirect_stdout


def salida_de(funcion, *argumentos):
    """Ejecuta funcion(*argumentos) y devuelve como texto todo lo que imprimió con print()."""
    buffer = io.StringIO()
    with redirect_stdout(buffer):
        funcion(*argumentos)
    return buffer.getvalue()
````

## Apuntes para estudiar

**Clave**

- El pensamiento computacional es formular problemas y soluciones de modo que un agente procesador pueda ejecutarlos eficazmente.
- Las tres estructuras de la programación son secuencia, bucle y selección; esta unidad trabaja la secuencia.
- Secuencia: el ordenador ejecuta el código en orden, de arriba hacia abajo; cambiar el orden cambia el resultado.
- Expresión (valor) < declaración (una o más expresiones) < programa (una o más declaraciones).
- `#` comenta una línea; tres comillas encierran una anotación de varias líneas.
- Las palabras reservadas (`if`, `for`, `global`, `import`...) no pueden usarse como identificadores.
- Un entorno virtual aísla versiones de Python y paquetes por proyecto (conda en el PDF; `uv` + `.venv` en este cuaderno).
- `print()` muestra cadenas (con comillas) y valores (sin comillas); `end` cambia el final y `sep` cambia el separador.
- `\n` inserta un salto de línea dentro de una cadena.
- Identificadores: estilo serpiente (`mi_variable`) para variables y funciones; no empiezan por número, sin espacios ni símbolos.

**Errores comunes**

- Abrir y cerrar una cadena con comillas distintas, o dejarla sin cerrar: error de sintaxis.
- Sumar una cadena y un número con `+` (`'Hello' + 10`), o multiplicar dos cadenas: `TypeError`.
- Usar una palabra reservada como nombre de variable, o nombres con espacios o que empiezan por número.
- Confundir `Count` con `count`: Python distingue mayúsculas de minúsculas.
- Creer que si no hay error de sintaxis tampoco habrá errores de ejecución.
- Copiar del PDF código con la traducción rota (`timesleep(5)`, `concrete-n`) en lugar del original.
