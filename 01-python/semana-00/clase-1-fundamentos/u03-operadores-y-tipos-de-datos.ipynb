## 1. Operadores aritméticos con datos numéricos

En programación se hacen operaciones con información numérica (la altura y el peso de una persona, 177 cm y 58 kg). Los datos que intervienen en una operación se llaman **operandos** y el símbolo que indica la operación es el **operador**. En `100 + 20`, `100` y `20` son operandos y `+` es el operador.

| Operador | Significado | Acción |
|---|---|---|
| `+` | Suma | Suma el operando izquierdo y el derecho |
| `-` | Resta | Resta el operando derecho del izquierdo |
| `*` | Multiplicación | Multiplica los dos operandos |
| `/` | División | Divide; siempre devuelve un número real (`float`) |
| `//` | División entera (cociente) | Descarta los decimales y deja solo la parte entera |
| `%` | Resto (módulo) | Obtiene el resto de la división; no tiene relación con el porcentaje |
| `**` | Potencia | Eleva el operando izquierdo al exponente derecho; con `**0.5` se calcula la raíz cuadrada |

````python
print(100 + 20)    # suma de 100 y 20
print(100 * 20)    # multiplicación de 100 y 20
print(100 - 20)    # resta de 100 y 20
print(100 / 20)    # división: devuelve un número real, 5.0
print(100 // 20)   # división entera: el cociente
print(100 % 20)    # resto de la división
````

**Nota sobre el PDF:** en la pág. 143, el comentario de la última línea (`100 % 20`) dice "encuentra el cociente", igual que el de `//`. El operador `%` devuelve el **resto**, no el cociente; aquí el comentario está corregido.

Otros ejemplos del PDF (págs. 161–165):

````python
print(1 + 5)       # suma: 6
print(1 - 8)       # resta: -7
print(4 * 5)       # multiplicación: 20
print(5 // 3)      # división entera: 1 (el cociente de 5 entre 3)
print(2 ** 10)     # potencia: 2 elevado a 10 = 1024
print(9 % 5)       # resto de 9 entre 5 = 4
````

**Ampliación (no está en el PDF):** `//` y `%` trabajan juntos. Se cumple siempre que `dividendo = divisor * cociente + resto`, y `** 0.5` calcula la raíz cuadrada.

````python
dividendo, divisor = 17, 5
cociente = dividendo // divisor
resto = dividendo % divisor
print(f"{dividendo} = {divisor} * {cociente} + {resto}")
print(divisor * cociente + resto == dividendo)   # True

print(25 ** 0.5)   # raíz cuadrada de 25 -> 5.0
print(-7 // 2)     # la división entera redondea hacia abajo: -4
````

### Operaciones con números complejos

Python también opera con números complejos, que se escriben `x + yj` (la `j` indica la parte imaginaria).

````python
print((3 + 4j) + (5 + 2j))   # suma de números complejos
print((3 + 4j) - (5 + 2j))   # resta de números complejos
print((3 + 4j) * (5 + 2j))   # multiplicación de números complejos
print((3 + 4j) / (5 + 2j))   # división de números complejos
````

**Clave:** `/` siempre da `float`; `//` da el cociente; `%` da el resto; `**` es la potencia.
**Error común:** leer `%` como porcentaje. En Python es el resto de la división.
**Para el quiz:** `5 // 3` es 1, `9 % 5` es 4 y `2 ** 10` es 1024.

## 2. Tipos de datos

En Python, distinguir el tipo de datos es muy importante: los caracteres y los números pueden verse iguales, pero sus propiedades son muy diferentes. El número `10` se puede calcular; el texto `"10"` no. Según el tipo de datos, los operadores disponibles son diferentes: dividir la cadena `"10"` entre 2 produce un `TypeError`.

````python
print(10 / 2)          # el número 10 sí se puede dividir

try:
    print("10" / 2)    # la cadena "10" no se puede dividir
except TypeError as error:
    print("TypeError:", error)
````

Además, un mismo operador se comporta distinto según el tipo. Con números, `*` multiplica; con cadenas, `*` **repite**:

````python
print(100 * 2)         # multiplicación numérica: 200
print("Hello" * 2)     # repetición de la cadena: HelloHello
print("100" * 2)       # repetición de la cadena: 100100
````

### Cómo distinguir los tipos de datos: `type()`

## 3. Precedencia de los operadores

Si una expresión mezcla suma y multiplicación, la multiplicación se calcula primero. Lo que está entre **paréntesis** se calcula antes que todo lo demás. El resultado depende del orden en que se procesan los operadores.

````python
print(10 + 20 * 30)     # la multiplicación se calcula primero: 610
print((10 + 20) * 30)   # el paréntesis se calcula primero: 900

print(1 + 2 * 5)        # 11
print((1 + 2) * 5)      # 15
````

Tabla de precedencia del PDF (cuanto más arriba, mayor prioridad):

| Operador | Explicación |
|---|---|
| `()` | Paréntesis |
| `**` | Potencia |
| `~`, `+`, `-` | Operadores monádicos (unarios) |
| `*`, `/`, `%`, `//` | Multiplicación, división, resto, división entera |
| `+`, `-` | Suma y resta |
| `>>`, `<<` | Desplazamiento de bits |
| `&` | AND de bits |
| `^`, `\|` | XOR de bits, OR de bits |
| `<=`, `<`, `>`, `>=` | Comparación |
| `==`, `!=` | Igual, diferente |
| `=`, `%=`, `/=`, `//=`, `-=`, `+=`, `*=`, `**=` | Asignación y asignación compleja |
| `is`, `is not` | Identidad |
| `in`, `not in` | Pertenencia |
| `not`, `or`, `and` | Operadores lógicos |

Para esta unidad basta con recordar el orden: paréntesis, potencia, multiplicación/división/resto, suma/resta.

**Nota sobre el PDF:** la tabla del PDF pone `not`, `or`, `and` en una sola fila; en realidad no tienen la misma prioridad (`not` va antes que `and`, y `and` antes que `or`). **Ampliación (no está en el PDF):** esto se verá en la unidad de operadores lógicos.

````python
# La potencia se calcula antes que la multiplicación
print(2 * 3 ** 2)       # 2 * 9 = 18
print((2 * 3) ** 2)     # 6 ** 2 = 36

# Misma prioridad: se calcula de izquierda a derecha
print(100 / 5 * 2)      # 20.0 * 2 = 40.0
````

**Clave:** ante la duda, usa paréntesis; hacen explícito el orden que quieres.
**Error común:** olvidar los paréntesis al calcular un promedio: `a + b / 2` no es `(a + b) / 2`.
**Para el quiz:** `10 + 20 * 30` da 610 y `(10 + 20) * 30` da 900.

## 4. Conversión de tipos de datos

Los datos se pueden calcular convirtiendo su tipo:

- `int("10")` convierte la cadena `"10"` en el entero `10`, así que `int("10") / 2` se ejecuta sin error.
- `str(10)` convierte el número `10` en la cadena `"10"`, que se puede unir con otra cadena con `+`.
- `float("10.5")` convierte la cadena en un número real.

````python
print(int("50"))                # 50: la cadena "50" pasa a entero
print(int("50") + 20)           # 70: ya se puede sumar
print("10" + str(10))           # "1010": str() convierte el número en cadena
print(float("10.5"))            # 10.5: la cadena pasa a número real

print(int("10") / 2)            # convierte la cadena '10' en entero y divide: 5.0
print("I like number " + str(10))   # une una cadena con el número convertido
````

Sin conversión, mezclar una cadena con un número da un `TypeError`. El PDF lo muestra en el apartado "Un paso más": `"A" + 1` y `"10" + 10` fallan, porque `"10"` parece un número pero, al estar entre comillas, es una cadena.

````python
for expresion in ['"A" + 1', '"10" + 10']:
    try:
        print(eval(expresion))
    except TypeError as error:
        print(f"{expresion} -> TypeError: {error}")
````

**Clave:** convierte con `int()`, `float()` y `str()` antes de operar.
**Error común:** sumar `"10" + 10`; hay que decidir si quieres `int("10") + 10` (20) o `"10" + str(10)` ("1010").
**Para el quiz:** `int("50") + 20` da 70 y `"10" + str(10)` da `"1010"`.

## 5. Cadenas y operadores `+` y `*`

El tipo `str` es muy adecuado para mostrar información en el software: el nombre, el ID o la contraseña de un usuario se expresan como cadena. Una cadena es una serie de caracteres conectados entre sí. Como son texto, no se pueden hacer cálculos aritméticos con ellas, pero sí **suma y multiplicación**:

- `+` **concatena** (conecta) dos cadenas.
- `*` **repite** una cadena un número **entero** de veces.
- La resta y la división no funcionan con cadenas.

````python
print("Hello" + "World")          # concatenación: HelloWorld
print("I" + " Love" + " Python!")  # se pueden unir varias cadenas
print("text" + "repeat" * 10)      # repetición: "repeat" diez veces
print("Python " * 10)              # 'Python ' repetido 10 veces
print("*" * 50)                    # una línea de 50 asteriscos
print(str(100) * 10)               # la cadena '100' repetida 10 veces
````

Si el multiplicador no es un entero, o si se suma un número a una cadena, ocurre un `TypeError`:

````python
try:
    "*" * 3.1
except TypeError as error:
    print("TypeError:", error)

try:
    "A" + 1
except TypeError as error:
    print("TypeError:", error)
````

### Indexación de cadenas

La **indexación** permite obtener solo algunos caracteres de una cadena larga, usando corchetes `[]` y un número. **La indexación comienza en 0.** Python admite también **índices negativos**: `-1` es el último carácter y, en una cadena de 5 caracteres, `-5` es el primero.

````python
palabra = "hello"
print(palabra[0])    # 'h': el índice 0 es la primera letra
print(palabra[2])    # 'l': el índice 2 es la tercera letra
print(palabra[-1])   # 'o': el índice -1 es el último carácter
print(palabra[-5])   # 'h': el índice -5 es el primer carácter (la cadena tiene 5 caracteres)
````

Un índice fuera de rango produce un `IndexError`. La cadena `"hello"` tiene 5 caracteres, así que sus índices van de 0 a 4 (de -5 a -1 en negativo).

````python
try:
    print("hello"[5])
except IndexError as error:
    print("IndexError:", error)

print(len("hello"))   # cantidad de caracteres
````

**Nota sobre el PDF:** el texto de las págs. 180–183 habla de la cadena "hola" (traducción de "hello"), pero el código usa `"hello"`. Con "hola" (4 letras) el índice `-5` también fallaría; la explicación "5 caracteres, rango de 0 a 4" corresponde a `"hello"`.

**Clave:** `+` une cadenas, `*` repite con un entero; el primer carácter es el índice 0 y el último el -1.
**Error común:** `"10" + 10` (TypeError), `"*" * 3.1` (el multiplicador debe ser entero) y usar un índice igual al largo de la cadena (IndexError).
**Para el quiz:** `"hello"[0]` es `'h'`, `"hello"[-1]` es `'o'` y `"hello"[5]` es un error.

Como además los `print` etiquetan `circum` como "Área" y `area` como "Circunferencia", la salida impresa es correcta, pero los nombres confunden. La salida del PDF para radio 4,0 es "Área 50,24" y "Circunferencia 25,12" (con coma decimal por la traducción; Python usa punto). Abajo se muestra la versión con nombres corregidos.

**Ampliación (no está en el PDF):** se encapsula en funciones. `calcular_circulo` hace el cálculo y `mostrar_circulo` imprime. El `round(..., 2)` evita salidas como `31.400000000000002`, producidas por la representación binaria de los números reales. La versión interactiva con `input()` queda comentada; abajo se ejecuta con los radios sugeridos por el PDF.

````python
PI = 3.14   # [2] definir el valor de PI


def calcular_circulo(radio):
    """Devuelve (circunferencia, area) de un círculo con el radio dado."""
    circunferencia = 2.0 * PI * radio    # [4] circunferencia = 2 * PI * r
    area = PI * radio * radio            # [5] área = PI * r al cuadrado
    return round(circunferencia, 2), round(area, 2)


def mostrar_circulo(radio):
    """Imprime el radio, el área y la circunferencia (paso [6])."""
    circunferencia, area = calcular_circulo(radio)
    print("Radio del círculo", radio)
    print("Área del círculo", area)
    print("Circunferencia del círculo", circunferencia)


def pedir_radio():
    """Versión interactiva: pide el radio con input() y muestra el resultado (paso [3])."""
    radio = float(input("Introduzca el radio :"))
    mostrar_circulo(radio)


# Para usarlo de forma interactiva, quita el comentario de la siguiente línea
# pedir_radio()

# Versión no interactiva: radios de prueba del PDF
for radio in [4.0, 5, 6, 10, 20, 33]:
    mostrar_circulo(radio)
    print("-" * 30)
````

**Resultado esperado:** para el radio 4,0 la salida es `Área del círculo 50.24` y `Circunferencia del círculo 25.12`, igual que la salida de la pág. 136 (con punto decimal). Para los demás radios, la circunferencia y el área crecen con r y r², respectivamente.

## Apuntes para estudiar

**Clave**

- Operandos son los datos; el operador es el símbolo: `+ - * / // % **`.
- `/` siempre da `float`; `//` es el cociente; `%` es el resto; `**` es la potencia (`** 0.5` es la raíz cuadrada).
- Tipos básicos: `int`, `float`, `str`, `bool`; también `complex` (`4 + 3j`) y, más adelante, lista, tupla y diccionario.
- `type(valor)` devuelve el tipo del objeto; `100` es `int` y `100.0` es `float`.
- Precedencia: paréntesis, potencia, `* / % //`, `+ -`. Los paréntesis ordenan el cálculo.
- Conversión: `int()`, `float()`, `str()`.
- Cadenas: `+` concatena y `*` repite con un entero; `,` en `print` añade un espacio; `\n`, `\t`, `\"` y `\'` son secuencias de escape.
- Indexación: empieza en 0; `[-1]` es el último carácter.
- Circunferencia = 2πr; área = πr².

**Errores comunes**

- Leer `%` como porcentaje.
- Sumar `"10" + 10` sin convertir (`TypeError`).
- Multiplicar una cadena por un `float` (`"*" * 3.1`).
- Usar un índice fuera de rango (`"hello"[5]` es `IndexError`).
- Olvidar paréntesis: `a + b / 2` no es `(a + b) / 2`.
- Mezclar los nombres de variables: en el código del PDF, `circum` guarda el área y `area` la circunferencia.

**Para el quiz**

- `5 // 3` es 1; `9 % 5` es 4; `2 ** 10` es 1024; `100 / 20` es 5.0.
- `10 + 20 * 30` es 610 y `(10 + 20) * 30` es 900.
- `"Hello" * 2` es `HelloHello`; `"100" * 2` es `100100`; `100 * 2` es 200.
- `int("50") + 20` es 70; `"10" + str(10)` es `"1010"`.
- `type(100.0)` es `float`; `type("100")` es `str`; `type(10 + 3j)` es `complex`.
- `"hello"[0]` es `h`; `"hello"[-1]` es `o`.

## Autoevaluación

**1. ¿Qué diferencia hay entre `/`, `//` y `%`?**

<details><summary>Ver respuesta</summary>

`/` divide y devuelve siempre un número real (`float`); `//` es la división entera y devuelve solo el cociente, sin decimales; `%` devuelve el resto de la división. Por ejemplo, `5 / 3` es 1.666..., `5 // 3` es 1 y `5 % 3` es 2.
</details>

**2. ¿Por qué `"10" / 2` produce un error pero `int("10") / 2` no?**

<details><summary>Ver respuesta</summary>

Porque `"10"` es una cadena (`str`) y la división no está definida para cadenas (`TypeError`). `int("10")` la convierte en el entero 10, y entonces `10 / 2` es 5.0.
</details>

**3. ¿Qué resultado dan `100 * 2` y `"100" * 2`, y por qué son distintos?**

<details><summary>Ver respuesta</summary>

`100 * 2` da 200 (multiplicación numérica) y `"100" * 2` da `"100100"` (repetición de la cadena). El operador `*` se comporta según el tipo de datos de los operandos.
</details>

**4. ¿En qué orden se calcula `10 + 20 * 30` y cómo se cambia?**

<details><summary>Ver respuesta</summary>

Primero se multiplica (20 * 30 = 600) y luego se suma (10 + 600 = 610). Para forzar la suma primero se usan paréntesis: `(10 + 20) * 30` da 900.
</details>

**5. ¿Qué devuelve `"hello"[-1]` y por qué `"hello"[5]` falla?**

<details><summary>Ver respuesta</summary>

`"hello"[-1]` devuelve `'o'`, el último carácter, porque los índices negativos cuentan desde el final. `"hello"[5]` falla con `IndexError` porque la cadena tiene 5 caracteres y sus índices válidos son del 0 al 4 (o del -5 al -1).
</details>
