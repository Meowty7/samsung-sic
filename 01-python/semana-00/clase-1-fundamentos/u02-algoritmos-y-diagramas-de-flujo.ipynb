## 1. De la invención al proceso de diseño técnico (EDP)

El PDF describe el proceso que siguió William para pasar del problema a la invención:

1. Encontrar problemas.
2. Aportar ideas para resolver los problemas.
3. Planificación de una invención.
4. Realización de un prototipo.
5. Experimento y corrección.
6. Mejorar sus ideas.

Los pasos 5 y 6 se repiten: el desarrollo es **continuo mediante la repetición**. Este proceso lo usan también muchos científicos e inventores y se llama **proceso de diseño técnico** (*engineering design process*, EDP). Lo más importante de todo es **planificar concretamente las ideas** y seguir mejorándolas con experimentación, corrección y retroalimentación.

La programación es lo mismo: quien quiere hacer algo debe empezar por planificar pensamientos e ideas en detalle, **en lugar de trabajar directamente en el ordenador**.

### El EDP en detalle

El EDP es una serie de etapas que siguen ingenieros y científicos para encontrar una solución. Cada etapa consiste en fijar objetivos y limitaciones, crear un prototipo y hacer pruebas y evaluaciones continuas. Hay versiones de 5, 7 o más etapas; el PDF muestra una de 12 etapas en forma de ciclo:

```text
Definir el problema
 -> Brainstorm: pensar posibles soluciones
 -> Investigar / explorar posibilidades
 -> Establecer criterios y restricciones
 -> Considerar soluciones alternativas
 -> Seleccionar un enfoque
 -> Desarrollar una propuesta de diseño
 -> Hacer un modelo o prototipo
 -> Probar y evaluar
 -> Refinar el diseño
 -> Crear la solución
 -> Comunicar el resultado
 (y se vuelve a empezar si hace falta mejorar)
```

En programación, la etapa **"Desarrollar una propuesta de diseño"** (planificación) se expresa en forma de **pseudocódigo** y de **diagrama de flujo**.

**Nota sobre el PDF:** el texto en español dice "PDE" en varios lugares (por ejemplo "proceso de diseño técnico (engineering design process, PDE)"). ## 3. Algoritmo

Un **algoritmo** es un conjunto de instrucciones para hacer algo. No basta con un montón de órdenes: debe tener una **secuencia razonable** de cálculos y tareas para resolver un problema. Es un proceso computacional bien definido que recibe una **entrada** (un valor o un conjunto de valores), la procesa con una serie de **comandos** y produce una **salida**.

El PDF lo ilustra con un algoritmo de ordenación: entrada `[31, 41, 59, 26, 41, 58]`, salida `[26, 31, 41, 41, 58, 59]`.

La planificación del algoritmo se hace con **pseudocódigo** y con **diagrama de flujo**. La diferencia con un programa: el algoritmo es la idea (el procedimiento), el programa es esa idea escrita en un lenguaje de programación concreto.

````python
# Un algoritmo recibe una entrada, ejecuta comandos y entrega una salida.
entrada = [31, 41, 59, 26, 41, 58]

salida = sorted(entrada)   # sorted() aplica un algoritmo de ordenación
print("Entrada:", entrada)
print("Salida: ", salida)
````

**Nota sobre el PDF:** en la pág. 93 la salida aparece como `[26, 31, 42, 41, 58, 59]`. Es una errata: el 42 no existe en la entrada y la lista no está ordenada. La salida correcta es `[26, 31, 41, 41, 58, 59]`, como se ve arriba.

**Ampliación (no está en el PDF):** `sorted()` es una función incorporada que oculta el algoritmo. Más adelante en el curso verás cómo se escribe uno a mano (por ejemplo, ordenación por selección, que el PDF menciona en la pág. 105).

**Clave:** algoritmo = entrada + comandos en orden + salida.
**Para el quiz:** un algoritmo no es solo un conjunto de órdenes; necesita una secuencia razonable que resuelva el problema.

## 4. Pseudocódigo

El **pseudocódigo** es una representación de un algoritmo usando lenguaje cotidiano. Es un buen comienzo cuando aún no tienes suficientes conocimientos de programación: sirve de **puente entre el diagrama de flujo y el código real**. Lo esencial es que explique y planifique claramente lo que debe hacer cada línea.

Reglas que da el PDF:

- Escribe al principio una frase que describa el propósito del algoritmo.
- Escribe de forma concisa.
- Crea y usa reglas de nombres simples y distinguibles.
- La sangría y los espacios en blanco son puntos clave.
- Expresa las estructuras de secuencia, selección y repetición usando `if`, `for`, `while`.

Ejemplo del PDF (pseudocódigo para decir si un número es par o impar):

```text
Este programa permitirá al usuario comprobar si un número es par o impar.

    imprimir resultado "Este número es par"

```

**Nota sobre el PDF:** la diapositiva escribe `if "4"` e `if "3"` (casos de ejemplo) en lugar de una condición general. Aquí se muestra la condición escrita en lenguaje cotidiano, que es lo que pide la regla "explica claramente cada línea".

Ahora el mismo algoritmo como código Python real (el pseudocódigo se traduce casi línea a línea):

````python
def comprobar_par_impar(numero):
    """Este programa permite comprobar si un número es par o impar."""
    if numero % 2 == 0:                       # si el resto de dividir entre 2 es 0
        print(f"{numero}: Este número es par")
    else:
        print(f"{numero}: Este número es impar")


comprobar_par_impar(4)
comprobar_par_impar(3)
````

**Clave:** el pseudocódigo no se ejecuta; es una guía. Usa los comandos de entrada (read, get), salida (print, show), cálculo (compute), iteración (for, while) y decisión (if-then-else).

## 5. Diagrama de flujo

El **diagrama de flujo** expresa los comandos con formas normalizadas y el flujo de trabajo con flechas. Las formas son las establecidas por la ISO (Organización Internacional de Normalización). Se dibuja **de arriba abajo y de izquierda a derecha, sin superponer** elementos.

| Forma | Uso según el PDF |
|---|---|
| Óvalo (terminador) | Inicio y fin del diagrama |
| Hexágono (preparación) | Inicializar datos |
| Rectángulo (acción o proceso) | Operaciones aritméticas o procesamiento de un solo paso |
| Rombo (decisión) | Bifurcarse según una condición (sí / no) |
| Paralelogramo (entrada, salida) | Datos de entrada y de salida |
| Documento | Representar un documento o informe impreso |
| Círculo (conector en la página) | Conectar con otro organigrama (empalme) |
| Flecha de flujo | Representar el flujo de procesamiento |

**Ampliación (no está en el PDF):** en los cuadernos de Jupyter puedes dibujar diagramas con **Mermaid** dentro de una celda markdown (JupyterLab 4.1 o superior lo muestra como dibujo; si no, verás el texto del diagrama). Ejemplo mínimo con las formas principales:

```mermaid
flowchart TD
    A([Inicio]) --> B{{"Preparación: definir datos"}}
    B --> C[/"Entrada: leer un número"/]
    C --> D["Proceso: calcular"]
    D --> E{"Decisión: condición"}
    E -- Sí --> F[/"Salida: mostrar resultado"/]
    E -- No --> D
    F --> G([Fin])
```

**Clave:** óvalo = inicio/fin, rectángulo = proceso, rombo = decisión, paralelogramo = entrada/salida.
**Para el quiz:** el rombo es el que se bifurca (sí/no); las flechas indican el orden.

## 6. Las tres estructuras de ejecución en la vida cotidiana

Las tres estructuras de la Unidad 1 (**secuencia**, **selección** e **iteración**) sirven no solo para programar, también en la vida real. El PDF las aplica a las indicaciones del transeúnte:

- **Secuencia:** una orden tras otra ("entra en el cruce", "sigue recto").
- **Selección:** elegir según una condición ("si es una oficina de correos, gira a la derecha; si no, sigue recto").
- **Iteración:** repetir ("repite tres veces: sigue recto y pasa el edificio").

````python
# Las tres estructuras en tres líneas de Python
print("Secuencia: paso 1, paso 2, paso 3")           # una orden tras otra

edificio = "oficina de correos"
if edificio == "oficina de correos":                  # selección: decide según una condición
    print("Selección: giro a la derecha")
else:
    print("Selección: sigo recto")

for paso in range(1, 4):                              # iteración: repite 3 veces
    print(f"Iteración: paso {paso} de 3, sigo recto y paso el edificio")
````

## 7. Eficacia del algoritmo: búsqueda lineal y binaria

Un algoritmo que da el resultado correcto **no necesariamente es eficiente**. El PDF compara dos algoritmos que buscan el valor 6 en siete datos.

**Algoritmo A: búsqueda lineal.** Comprueba los datos uno a uno desde el principio hasta encontrar el buscado. En `[41, 3, 6, 5, 9, 36, 23]` se hacen **3 comparaciones** (41, 3 y 6).

**Algoritmo B: búsqueda binaria.** Primero hay que **ordenar** los datos: `[3, 5, 6, 9, 23, 36, 41]`. Se compara con el valor del centro y se descarta la mitad donde no puede estar. Para buscar 6: centro 9 (6 es menor, va a la izquierda), centro 5 (6 es mayor, a la derecha), y se encuentra 6 en la tercera comparación.

Con 100 datos, en el peor caso la lineal compara **100 veces** (n pasos); la binaria reduce el rango a la mitad cada vez: 100 → 50 → 25 → 13 → 7 → 4 → 2 → resultado, es decir **log₂ 100 ≈ 7 pasos**.

````python
def busqueda_lineal(datos, objetivo):
    """Algoritmo A: revisa los datos uno a uno. Devuelve (indice, comparaciones)."""
    comparaciones = 0
    for indice, dato in enumerate(datos):
        comparaciones += 1                    # cada dato revisado es una comparación
        if dato == objetivo:
            return indice, comparaciones
    return -1, comparaciones                  # -1 significa "no encontrado"


datos = [41, 3, 6, 5, 9, 36, 23]
indice, comparaciones = busqueda_lineal(datos, 6)
print(f"Lineal: 6 está en el índice {indice} tras {comparaciones} comparaciones")
````

````python
def busqueda_binaria(datos_ordenados, objetivo):
    """Algoritmo B: compara con el centro y descarta la mitad. Devuelve (indice, comparaciones)."""
    inicio, fin = 0, len(datos_ordenados) - 1
    comparaciones = 0
    while inicio <= fin:
        centro = (inicio + fin) // 2          # posición del valor mediano
        comparaciones += 1
        valor = datos_ordenados[centro]
        print(f"  Paso {comparaciones}: rango {datos_ordenados[inicio:fin + 1]}, centro = {valor}")
        if valor == objetivo:
            return centro, comparaciones
        if valor < objetivo:
            inicio = centro + 1               # el objetivo está a la derecha
        else:
            fin = centro - 1                  # el objetivo está a la izquierda
    return -1, comparaciones


ordenados = sorted(datos)                     # la binaria exige datos ordenados
print("Datos ordenados:", ordenados)
indice, comparaciones = busqueda_binaria(ordenados, 6)
print(f"Binaria: 6 está en el índice {indice} tras {comparaciones} comparaciones")
````

````python
# Caso del PDF: 100 datos y reducción a la mitad en cada paso
n = 100
secuencia = [n]
while n > 1:
    n = (n + 1) // 2          # redondeo hacia arriba: 100 -> 50 -> 25 -> 13 -> 7 -> 4 -> 2 -> 1
    secuencia.append(n)
print(" -> ".join(str(x) for x in secuencia))
print("Pasos (flechas):", len(secuencia) - 1)
````

**Nota sobre el PDF:** la diapositiva pide que "agonices" para encontrar el algoritmo más eficiente; es una traducción automática literal (la idea original es "esfuérzate/piensa a fondo" por encontrar el más eficiente).

**Clave:** lineal = n pasos en el peor caso; binaria = log₂ n pasos, pero **solo funciona con datos ordenados**.
**Error común:** aplicar búsqueda binaria a una lista sin ordenar; el resultado es incorrecto.
**Para el quiz:** con 100 datos, lineal necesita hasta 100 comparaciones y binaria unas 7.

## 8. Rendimiento del algoritmo: espacio y tiempo

Hay dos formas de medir el rendimiento de un algoritmo:

- **Complejidad espacial:** cuánta memoria ocupa.
- **Complejidad temporal:** con qué rapidez resuelve el problema.

### Complejidad espacial

La memoria del ordenador es limitada, así que el algoritmo que ocupa menos espacio es más eficiente en términos de complejidad espacial. El PDF recuerda las áreas de memoria de un programa:

| Área | Qué guarda |
|---|---|
| Área de código | Los códigos del programa; la CPU toma los comandos uno a uno |
| Área de datos | Variables globales y estáticas; se asigna al iniciar y se destruye al terminar |
| Área del montón (heap) | Asignación dinámica; la gestiona el programador y su tamaño se decide en tiempo de ejecución |
| Área de pila (stack) | Variables locales y parámetros de llamadas a funciones; se destruye al terminar la llamada |

Comparando la cantidad total de cada zona entre el algoritmo A y el B, el que ocupa menos memoria (en el ejemplo, el B) es más eficaz en espacio.

### Complejidad temporal y notación Big O

La **notación Big O** es una representación matemática del rendimiento del algoritmo. Como la velocidad real depende del ordenador, en lugar de medir segundos se expresan los **pasos de trabajo** con notación matemática. La complejidad temporal es más fácil de medir que la espacial, por eso es la más usada.

`O(n)` se lee "O de n": la **O** viene de "orden de" y entre paréntesis va el **recuento de operaciones** del algoritmo en función del número de datos n.

- Búsqueda lineal: **O(n)**.
- Búsqueda binaria: **O(log n)**.

El PDF supone 1 milisegundo por cada dato comprobado y compara los tiempos:

````python
import math


def formatear_ms(ms):
    """Convierte milisegundos en una unidad legible (ms, s o días)."""
    if ms < 1000:
        return f"{ms:.0f} milisegundos"
    if ms < 86_400_000:
        return f"{ms / 1000:.0f} segundos"
    return f"{ms / 86_400_000:.1f} días"


print(f"{'Datos':>14} | {'Búsqueda lineal':>20} | {'Búsqueda binaria':>20}")
for n in [100, 10_000, 10**9]:
    lineal = n                                  # O(n): n comprobaciones de 1 ms
    binaria = math.ceil(math.log2(n))           # O(log n): log2(n) comprobaciones de 1 ms
    print(f"{n:>14,} | {formatear_ms(lineal):>20} | {formatear_ms(binaria):>20}")
````

**Nota sobre el PDF:** en la tabla del PDF, para 10⁹ datos la búsqueda binaria aparece con 32 milisegundos. Calculado, log₂(10⁹) ≈ 29,9, es decir, unos 30 pasos. La conclusión no cambia (11 días frente a unas decenas de milisegundos), pero el 32 es aproximado. Los demás valores (7 y 14 ms; 100 ms, 10 s y 11 días) sí coinciden.

### Comparar otras notaciones Big O

El PDF muestra un gráfico de complejidad Big O con zonas de calidad (de mejor a peor: asombroso, buena, feria [regular], mala, terrible) y menciona estos ejemplos:

- **O(log n)**, tiempo logarítmico: búsqueda binaria.
- **O(n)**, tiempo lineal: búsqueda lineal.
- **O(n log n)**: ordenación rápida.
- **O(n²)**: ordenación por selección.

**Ampliación (no está en el PDF):** las columnas O(1), O(2ⁿ) y O(n!) del gráfico se pueden comparar contando operaciones para un mismo n.

````python
import math

n = 20
operaciones = {
    "O(1)": 1,
    "O(log n)": math.ceil(math.log2(n)),
    "O(n)": n,
    "O(n log n)": math.ceil(n * math.log2(n)),
    "O(n^2)": n ** 2,
    "O(2^n)": 2 ** n,
    "O(n!)": math.factorial(n),
}
print(f"Operaciones para n = {n}")
for notacion, cantidad in operaciones.items():
    print(f"{notacion:>11}: {cantidad:,}")
````

**Clave:** Big O compara el crecimiento de los pasos al aumentar los datos, no el tiempo real. Menor crecimiento = más eficiente.
**Error común:** pensar que Big O da segundos; da cómo crece el número de operaciones con n.
**Para el quiz:** búsqueda lineal O(n), binaria O(log n), ordenación por selección O(n²), ordenación rápida O(n log n).

## Proyecto de la unidad

### Búsqueda de rutas: encontrar el hotel

**Problema.** Encontrar el hotel reservado siguiendo las indicaciones: "Siga recto hacia el oeste en este cruce. Verás cuatro edificios. Cuando vea la oficina de correos, gire a la derecha cruzándola. Siga recto, y cuando vea un restaurante italiano, el tercer edificio es el hotel que busca."

**Plan de programación.** El PDF construye el diagrama de flujo en cuatro bloques, y en cada uno señala qué estructura se usa:

1. **Salida y dirección oeste** (secuencia + selección): entrar en la encrucijada, elegir un camino y comprobar si va hacia el oeste; si no, se elige el siguiente camino; si sí, se camina recto.
2. **Encontrar la oficina de correos** (secuencia + selección): llegar al primer edificio y comprobar si es una oficina de correos; si lo es, girar a la derecha; si no, seguir recto y comprobar el siguiente.
3. **Encontrar el restaurante italiano** (secuencia + selección): tras girar, seguir recto hasta el primer edificio y comprobar si es un restaurante italiano; si no, seguir y comprobar el siguiente.
4. **Encontrar el hotel** (iteración + secuencia): repetir 3 veces "seguir recto y pasar el edificio", comprobar el nombre del edificio y terminar.

#### Pseudocódigo

```text
Este algoritmo encuentra el hotel a partir de las indicaciones del transeúnte.

Entrar en la encrucijada
repeat
    elegir un camino de la encrucijada
until el camino elegido va hacia el oeste
caminar recto

repeat
    llegar al edificio
    if el edificio es una oficina de correos then
        girar a la derecha
    else
        caminar recto
until el edificio es una oficina de correos

repeat
    caminar recto
    llegar al edificio
until el edificio es un restaurante italiano

for 3 veces
    seguir recto y pasar el edificio
comprobar el nombre del edificio
fin: se encontró el hotel
```

#### Diagrama de flujo (Mermaid, equivalente al de la pág. 91)

```mermaid
flowchart TD
    A([Inicio]) --> B{{"Entre en una encrucijada"}}
    B --> C["Elige un camino de la encrucijada"]
    C --> D{"Juzgue si el camino elegido es hacia el oeste"}
    D -- No --> C
    D -- Sí --> E["Camina recto"]
    E --> F["Llegue al edificio"]
    F --> G{"¿Es una oficina de correos?"}
    G -- No --> H["Camina recto"]
    H --> F
    G -- Sí --> I["Gire a la derecha"]
    I --> J["Camina recto"]

    L -- No --> J
    L -- Sí --> M["Repetir 3 veces: siga recto y pase el edificio"]
    M --> N["Compruebe el nombre del edificio"]
    N --> O([Encontrar el hotel ha terminado])
```

**Ampliación (no está en el PDF):** el PDF pide solo el diagrama. Para comprobar que el algoritmo funciona, lo convertimos en código Python y lo ejecutamos sobre una calle de prueba. Cada bloque del diagrama corresponde a un bloque del código.

````python
def encontrar_hotel(caminos, calle_oeste, calle_tras_correos):
    """Simula el algoritmo de búsqueda de rutas del PDF.

    caminos: direcciones disponibles en la encrucijada, por ejemplo ["norte", "este", "oeste", "sur"].
    calle_oeste: edificios que se ven al ir hacia el oeste.
    calle_tras_correos: edificios que se ven después de girar a la derecha en la oficina de correos.
    Devuelve el nombre del hotel.
    """
    print("Inicio: entro en la encrucijada")

    # Bloque 1: elegir caminos hasta encontrar el que va hacia el oeste (selección)
    for camino in caminos:
        print(f"  Elijo el camino hacia el {camino}")
        if camino == "oeste":
            print("  Es hacia el oeste: camino recto")
            break
        print("  No es hacia el oeste: pruebo el siguiente")

    # Bloque 2: recorrer los edificios hasta la oficina de correos (selección)
    for edificio in calle_oeste:
        print(f"  Llego al edificio: {edificio}")
        if edificio == "oficina de correos":
            print("  Es la oficina de correos: giro a la derecha")
            break
        print("  No es la oficina de correos: sigo recto")

    # Bloque 3: recorrer los edificios hasta el restaurante italiano (selección)
    posicion_restaurante = None
    for posicion, edificio in enumerate(calle_tras_correos):
        print(f"  Llego al edificio: {edificio}")
        if edificio == "restaurante italiano":
            print("  Es el restaurante italiano: empiezo a buscar el hotel")
            posicion_restaurante = posicion
            break
        print("  No es el restaurante italiano: sigo recto")

    # Bloque 4: repetir 3 veces seguir recto y pasar el edificio (iteración)
    posicion = posicion_restaurante
    for _ in range(3):
        posicion += 1
        print(f"  Sigo recto y paso el edificio (ahora en la posición {posicion})")

    hotel = calle_tras_correos[posicion]
    print(f"Fin: compruebo el nombre del edificio: {hotel}")
    return hotel


encrucijada = ["norte", "este", "oeste", "sur"]
calle_oeste = ["banco", "farmacia", "oficina de correos", "librería"]
tras_correos = ["panadería", "restaurante italiano", "zapatería", "floristería", "Hotel Aurora"]

hotel = encontrar_hotel(encrucijada, calle_oeste, tras_correos)
print("Resultado:", hotel)
````

**Resultado esperado:** el algoritmo recorre la encrucijada hasta el camino oeste, llega a la oficina de correos, gira, encuentra el restaurante italiano, pasa tres edificios y termina en `Hotel Aurora`.

**Clave:** el diagrama usa las tres estructuras: secuencia (pasos en orden), selección (los rombos "¿es...?") e iteración (repetir 3 veces).

def binaria(datos_ordenados, objetivo):
    inicio, fin = 0, len(datos_ordenados) - 1
    comparaciones = 0
    while inicio <= fin:
        centro = (inicio + fin) // 2
        comparaciones += 1
        if datos_ordenados[centro] == objetivo:
            return centro, comparaciones
        if datos_ordenados[centro] < objetivo:
            inicio = centro + 1
        else:

def factores_primos(n):
    factores = []
    divisor = 2
    while n > 1:
        if n % divisor == 0:        # decisión: divisor es factor

            n //= divisor
        else:
            divisor += 1            # siguiente divisor
    return factores

## Apuntes para estudiar

**Clave**

- Antes de programar se planifica: EDP (diseño técnico) y, en programación, pseudocódigo y diagrama de flujo.
- Algoritmo = entrada, comandos en secuencia razonable y salida. Programa = algoritmo escrito en un lenguaje.
- Pseudocódigo: lenguaje cotidiano, conciso, con sangría y con `if`, `for`, `while`; sirve de puente al código.
- Diagrama de flujo: formas ISO (óvalo, rectángulo, rombo, paralelogramo...), de arriba abajo y de izquierda a derecha.
- Tres estructuras: secuencia, selección e iteración.
- Eficiencia: lineal es O(n); binaria es O(log n) y requiere datos ordenados.
- Complejidad espacial (memoria) y temporal (pasos); Big O mide pasos, no segundos.

**Errores comunes**

- Programar directamente sin planificar.
- Aplicar búsqueda binaria a datos sin ordenar.
- Confundir la complejidad (número de pasos) con el tiempo real en segundos.
- Dibujar un rombo de decisión con una sola salida (debe tener sí y no).
- Escribir el pseudocódigo como código Python exacto en lugar de explicar cada paso.

**Para el quiz**

- La sigla correcta del proceso de diseño técnico es EDP.
- Con 100 datos: lineal hasta 100 pasos, binaria 7.
- O(log n) es mejor que O(n); O(n²) es peor que O(n log n).
- Los comandos de pseudocódigo del PDF: read/obtain/get, print/show/display, compute/calculate/determine, for/while, if-then-else, repeat-until, true/false.
- En el diagrama de la misión, la iteración es "repetir 3 veces: seguir recto y pasar el edificio".
