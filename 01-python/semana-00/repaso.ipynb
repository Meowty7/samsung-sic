## Unidad 3 — Conceptos básicos de tipos de datos numéricos y operaciones aritméticas

[Ir a la unidad](clase-1-fundamentos/u03-operadores-y-tipos-de-datos.ipynb)

```python
5 // 3  # división entera: devuelve el cociente (1)
9 % 5  # resto de la división (4)
2 ** 10  # potencia (1024)
25 ** 0.5  # raíz cuadrada (5.0)
100 / 20  # división, siempre devuelve float (5.0)
(3 + 4j) * (5 + 2j)  # operación con números complejos
type(100)  # devuelve el tipo del objeto (int)
int("50") + 20  # convierte la cadena a entero y suma (70)
"10" + str(10)  # convierte a cadena y concatena ("1010")
float("10.5")  # convierte la cadena a número real
"Hello" + "World"  # concatenación de cadenas
"*" * 50  # repite la cadena 50 veces
print("I am", 23)  # la coma añade un espacio entre los elementos
print("line\nbreak")  # \n cambia de línea; \t inserta tabulación
"hello"[-1]  # último carácter ('o'); el primer índice es 0
```

**Errores comunes**

- Leer `%` como porcentaje: en Python es el resto de la división.
- Sumar una cadena con un número sin convertir (`"10" + 10`) produce TypeError.
- Multiplicar una cadena por un número real (`"*" * 3.1`): el multiplicador debe ser entero.
- Usar un índice fuera de rango (`"hello"[5]`) produce IndexError; en una cadena de 5 caracteres el último índice es 4.
- Olvidar los paréntesis: `a + b / 2` no es `(a + b) / 2`.
- Asumir que `100` y `100.0` son del mismo tipo: el primero es int y el segundo float.

## Unidad 5 — Operaciones lógicas y de comparación

[Ir a la unidad](clase-2-control-de-flujo/u05-operadores-de-comparacion-y-logicos.ipynb)

```python
10 > 5  # True: comparación mayor que
5 != 5  # False: distinto de
a == b  # True si ambos valores son iguales
a is b  # True solo si son el mismo objeto
0 < n < 200  # comparación encadenada: n entre 0 y 200
'aaa' in 'aaa-bbb'  # True: la cadena está contenida
bool(0)  # False: el 0 es falso
bool('hello')  # True: cadena con contenido
x and y  # True solo si ambos son True
x or y  # True si al menos uno es True
not x  # invierte el valor booleano
n % 2 == 0  # True si n es par
n // 100 == 3  # True si la centena de n es 3
(x + y + z) / 3  # promedio: los paréntesis mandan
(y % 4 == 0 and y % 100 != 0) or y % 400 == 0  # año bisiesto
```

**Errores comunes**

- Confundir `=` (asignación) con `==` (comparación).
- Escribir `=<` o `! =` en lugar de `<=` y `!=`: es `SyntaxError`.
- Creer que `bool(' ')` es False: solo la cadena vacía `''` lo es.
- Usar `is` para comparar números o textos en lugar de `==`.
- Olvidar paréntesis: `x + y + z / 3` no calcula el promedio.
- Mezclar `and` y `or` sin paréntesis: `and` se evalúa antes que `or`.

## Unidad 7 — Declaración condicional-2: toma de decisiones en dos direcciones y aplicación de enunciados condicionales

[Ir a la unidad](clase-2-control-de-flujo/u07-if-else-elif.ipynb)

```python
if x > 0:  # primera condición
elif x == 0:  # se evalúa solo si la anterior fue falsa
else:  # se ejecuta si ninguna condición fue verdadera
if n % 2 == 0: print("par")  # decide par o impar
if a >= 0:  # if exterior; dentro puede haber otro if-else sangrado
a and b  # True solo si ambos son True
a or b  # True si alguno es True
not a  # invierte el valor booleano
(y % 4 == 0 and y % 100 != 0) or (y % 400 == 0)  # año bisiesto
import random  # carga el módulo de números aleatorios
random.randint(1, 3)  # entero entre 1 y 3, ambos incluidos
random.randrange(2)  # devuelve 0 o 1
random.seed(7)  # fija la semilla (ampliación): resultados reproducibles
x in ('a', 'e', 'i', 'o', 'u')  # comprueba pertenencia a una tupla
```

## Unidad 8 — Bucle-1: la sentencia for y la función range

[Ir a la unidad](clase-2-control-de-flujo/u08-bucle-for.ipynb)

```python
for i in range(5):  # repite 5 veces con i = 0, 1, 2, 3, 4
for _ in range(10):  # repite 10 veces sin usar la variable
range(inicio, parada, paso)  # enteros desde inicio hasta parada-1 de paso en paso
list(range(0, 10, 2))  # [0, 2, 4, 6, 8]
range(1, 11)  # enteros del 1 al 10
range(10, 0, -1)  # cuenta regresiva del 10 al 1
for n in [11, 22, 33]:  # recorre una lista
for ch in 'Hello':  # recorre una cadena carácter por carácter
s = 0; for i in range(1, 11): s += i  # suma acumulada (55)
sum(numbers)  # suma los elementos de una lista sin bucle
fact = 1; for i in range(1, n + 1): fact *= i  # factorial de n
print(i, end=' ')  # imprime sin salto de línea
print('s', 'e', 'p', sep='-')  # imprime s-e-p
'{:5d}'.format(100)  # entero con 5 espacios de ancho
for b in panes: for m in carnes: print(b, m)  # bucles anidados: todas las combinaciones
```

**Errores comunes**

- Creer que `range(10)` genera del 1 al 10: genera de 0 a 9 (para 1 a 10 se usa `range(1, 11)`).
- Olvidar los dos puntos al final del `for` o no sangrar el bloque (`SyntaxError` o `IndentationError`).
- Pasar números reales a `range` (`TypeError`); para reales existe `numpy.arange` (más adelante).
- Llamar `sum` a una variable y después intentar usar la función `sum()`.
- Inicializar el acumulador dentro del bucle (se reinicia en cada vuelta) o en 0 cuando se multiplica.
- Usar un paso negativo con inicio menor que la parada: el rango queda vacío.

## Unidad 9 — Bucle-2: la sentencia while, break y continue, random y bucle doble

[Ir a la unidad](clase-2-control-de-flujo/u09-bucle-while.ipynb)

```python
while i < 5:  # repite mientras la condición sea verdadera
i += 1  # actualiza la variable de la condición dentro del bloque
while True:  # bucle infinito (se sale con break)
break  # finaliza el bucle inmediatamente
continue  # salta al principio del bucle sin ejecutar el resto de la vuelta
n = -1; while n <= 0: n = int(input('n> '))  # repite hasta recibir un entero positivo
while password != 'pythonisfun':  # repite hasta que la contraseña sea correcta
import random as rd  # importa el módulo y le da la abreviatura rd
rd.random()  # real aleatorio en [0, 1)
rd.randint(1, 10)  # entero aleatorio de 1 a 10 (ambos incluidos)
rd.randrange(0, 10, 2)  # múltiplo de 2 de 0 a 8 (el 10 se excluye)
rd.shuffle(lista)  # mezcla la lista en el sitio
rd.choice(lista)  # elige un elemento al azar
rd.sample(lista, 3)  # lista con 3 elementos distintos al azar
for i in range(2, 10): for j in range(1, 10):  # doble bucle: 72 vueltas internas
```

**Errores comunes**

- No actualizar la variable de la condición dentro del `while`: queda un bucle infinito.
- No inicializar la variable de la condición antes del bucle (`NameError` o `UnboundLocalError`).
- Usar `<=` cuando el objetivo se alcanza exactamente (caso del caracol): conviene `<`.
- Confundir `break` (termina el bucle) con `continue` (solo termina la vuelta actual).
- Confundir `randint(1, 10)` (incluye el 10) con `randrange(1, 10)` (no lo incluye), o esperar que `shuffle` devuelva una lista nueva.
- Abusar de `break` y `continue`: el programa se vuelve difícil de seguir.
