## Mapa rápido de la semana

| Algoritmo | Idea | Peor caso | Caso típico | Memoria extra |
|---|---|---|---|---|
| Burbuja | Intercambiar vecinos desordenados; el mayor "sube" al final | O(N²) | O(N²) | O(1) |
| Selección | Buscar el mínimo y ponerlo al principio | O(N²) | O(N²) | O(1) |
| Inserción | Insertar cada elemento en su lugar dentro de la parte ya ordenada | O(N²) | O(N) si casi ordenada | O(1) |
| Fusión (merge sort) | Dividir a la mitad, ordenar cada mitad y fusionar | O(N log N) | O(N log N) | O(N) |
| Rápido (quick sort) | Partir alrededor de un pivote y ordenar cada lado | O(N²) con mal pivote | O(N log N) | O(log N) in situ |

| Técnica | Cuándo sirve | Ejemplo |
|---|---|---|
| Greedy | Cuando la mejor elección local lleva a la óptima global | Cambio de monedas, selección de actividades |
| Divide y vencerás | Cuando el problema se parte en subproblemas del mismo tipo | Búsqueda binaria, merge sort, quick sort, ruta del triángulo |

**Clave:** greedy no siempre da el óptimo (con monedas `[4, 3, 1]` y 6, greedy usa 3 monedas y lo óptimo son 2). Un pivote aleatorio evita el peor caso de quicksort con listas ya ordenadas.

## Unidad 28 — Ordenamiento por fusión (merge sort)

[Ir a la unidad](clase-1-algoritmos-de-ordenamiento/u28-ordenamiento-por-fusion.ipynb)

```python
mid = len(S) // 2  # índice medio para dividir
L, R = S[:mid], S[mid:]  # copias de las dos mitades (mergesort1)
if len(S) > 1:  # caso recursivo; si no, ya está ordenada
mergesort1(L); mergesort1(R)  # ordenar cada mitad recursivamente
merge1(S, L, R)  # fusionar L y R sobre S
L[0] <= R[0]  # compara los primeros elementos de dos listas ordenadas
S[k] = L.pop(0)  # coloca el menor y lo descarta de su lista
mergesort2(S, 0, len(S) - 1)  # llamada inicial por rangos de índices
if low < high:  # solo recurre si el rango tiene más de un elemento
mid = (low + high) // 2  # punto medio del rango
merge2(S, low, mid, high)  # fusiona S[low..mid] y S[mid+1..high]
R.append(S[i]); i += 1  # añade el menor a la lista auxiliar y avanza
S[low + k] = R[k]  # vuelca la lista auxiliar en el rango original
```

**Errores comunes**

- Olvidar copiar los elementos sobrantes cuando una de las listas se vacía antes que la otra.
- No definir el caso base y caer en recursión infinita.
- Llamar a `mergesort2(S)` sin los argumentos `low` y `high`; la llamada inicial es `mergesort2(S, 0, len(S) - 1)`.
- Creer que dividir ordena: el orden se obtiene al fusionar, no al dividir.
- Usar `<` en lugar de `<=` al fusionar y perder la estabilidad (el `merge2` del PDF usa `<`).
- Olvidar que los cortes `S[:mid]` y `S[mid:]` son copias: ordenar `L` no cambia `S` hasta fusionar.
