## La estrategia divide y vencerás

El método resuelve un problema en tres pasos:

1. **Dividir:** partir la instancia de entrada en dos o más instancias más pequeñas.
2. **Conquistar:** resolver cada instancia pequeña. Si todavía no es lo bastante pequeña para resolverse de inmediato, se aplica recursividad.
3. **Combinar:** si hace falta, unir las respuestas de las instancias pequeñas para obtener la respuesta de la instancia original.

La recursividad encaja de forma natural: cada llamada resuelve una instancia más pequeña y necesita un caso base (condición de terminación) para detenerse.

**Ampliación (no está en el PDF):** ejemplo mínimo con los tres pasos visibles, sumar los elementos de una lista.

````python
def suma_dc(lista):
    """Suma los elementos de una lista con divide y vencerás."""
    # Caso base: una lista vacía o de un elemento se resuelve directamente
    if len(lista) == 0:
        return 0
    if len(lista) == 1:
        return lista[0]
    # Dividir
    mitad = len(lista) // 2
    izquierda, derecha = lista[:mitad], lista[mitad:]
    # Conquistar (recursivamente)
    suma_izq = suma_dc(izquierda)
    suma_der = suma_dc(derecha)
    # Combinar
    return suma_izq + suma_der

print(suma_dc([3, 1, 4, 1, 5, 9, 2, 6]))
````

**Clave:** divide y vencerás = dividir, conquistar (recursivamente) y combinar. Algunos algoritmos tienen un paso de combinar costoso (merge sort) y otros casi no lo tienen (quick sort, búsqueda binaria).

**Error común:** olvidar el caso base y provocar recursión infinita (`RecursionError`).

**Para el quiz:** los tres pasos del método son dividir, conquistar y combinar; la recursividad se usa cuando la instancia dividida aún no es suficientemente pequeña.

## Búsqueda binaria como divide y vencerás

La búsqueda binaria (Unidad 25) busca un elemento en una lista **ordenada**. Compara `x` con el elemento central: si es igual, termina; si `x` es menor, solo puede estar en la mitad izquierda; si es mayor, en la mitad derecha. Si el espacio de búsqueda queda vacío, `x` no existe en la lista.

El PDF usa la lista `S = [11, 17, 26, 28, 37, 45, 53, 59, 77]` y busca `x = 26`: compara con 37 (central), reduce a `[11, 17, 26, 28]`, compara con 17, reduce a `[26, 28]` y encuentra 26.

Detalle importante del PDF: la búsqueda binaria **resuelve solo uno** de los dos subproblemas en que divide la entrada, mientras que el merge sort resuelve los dos.

### Versión iterativa (con índices bajo y alto)

````python
def bin_search(nums, x):
    low, high = 0, len(nums) - 1
    while low <= high:
        mid = (low + high) // 2
        if nums[mid] == x:
            return mid
        elif nums[mid] > x:
            high = mid - 1
        else:
            low = mid + 1
    return -1

S = [11, 17, 26, 28, 37, 45, 53, 59, 77]
print(bin_search(S, 26))
print(bin_search(S, 30))
````

### Versión recursiva (divide y vencerás)

Caso base: si `low > high`, el rango está vacío y se devuelve `-1`. Si no, se calcula `mid`; si `nums[mid] == x` se devuelve `mid`; si `x` es menor se busca en `[low, mid - 1]` y si es mayor, en `[mid + 1, high]`.

````python
def bin_search2(nums, x, low, high):
    if low > high:
        return -1
    else:
        mid = (low + high) // 2
        if nums[mid] == x:
            return mid
        elif nums[mid] > x:
            return bin_search2(nums, x, low, mid - 1)
        else:
            return bin_search2(nums, x, mid + 1, high)

S = [11, 17, 26, 28, 37, 45, 53, 59, 77]
print(bin_search2(S, 26, 0, len(S) - 1))
print(bin_search2(S, 30, 0, len(S) - 1))
````

**Ampliación (no está en el PDF):** traza de la búsqueda de `x = 26` para ver cómo se reduce el rango en cada llamada.

````python
def bin_search_traza(nums, x, low, high, nivel=0):
    sangria = "  " * nivel
    if low > high:
        print(f"{sangria}rango vacio (low={low}, high={high}): no existe")
        return -1
    mid = (low + high) // 2
    print(f"{sangria}low={low}, high={high}, mid={mid}, nums[mid]={nums[mid]}")
    if nums[mid] == x:
        print(f"{sangria}encontrado en el indice {mid}")
        return mid
    elif nums[mid] > x:
        return bin_search_traza(nums, x, low, mid - 1, nivel + 1)
    else:
        return bin_search_traza(nums, x, mid + 1, high, nivel + 1)

S = [11, 17, 26, 28, 37, 45, 53, 59, 77]
bin_search_traza(S, 26, 0, len(S) - 1)
print()
bin_search_traza(S, 30, 0, len(S) - 1)
````

**Clave:** dividir = elegir la mitad donde puede estar `x`; conquistar = buscar solo en esa mitad; combinar = no hace falta (se devuelve el resultado tal cual).

**Error común:** usar la búsqueda binaria en una lista sin ordenar; el resultado deja de ser confiable. Otro error frecuente es escribir `low < high` en lugar de `low <= high` y no encontrar el último elemento del rango.

**Para el quiz:** la condición de terminación de `bin_search2` es `low > high` y devuelve `-1`; resuelve un único subproblema.

**Referencia:** PDF parte 2, págs. 64–68 y 73–75

## Merge sort y divide y vencerás

El merge sort (Unidad 28) divide la lista en dos mitades del mismo tamaño, ordena **cada** mitad recursivamente y luego **combina** las dos listas ordenadas con un procedimiento aparte (`mezclar`). El PDF lo ilustra con `[27, 10, 12, 20, 25, 13, 15, 22]`: primero divide hasta listas de un elemento y luego mezcla de a pares hasta obtener `[10, 12, 13, 15, 20, 22, 25, 27]`.

**Ampliación (no está en el PDF):** el PDF solo muestra el diagrama (págs. 69–70); esta implementación es propia.

````python
def mezclar(a, b):
    """Combina dos listas ya ordenadas en una sola lista ordenada."""
    resultado = []
    i = j = 0
    while i < len(a) and j < len(b):
        if a[i] <= b[j]:
            resultado.append(a[i])
            i += 1
        else:
            resultado.append(b[j])
            j += 1
    # Una de las dos listas se agoto: se agrega lo que queda de la otra
    resultado.extend(a[i:])
    resultado.extend(b[j:])
    return resultado

def merge_sort(lista):
    if len(lista) <= 1:          # caso base
        return lista
    mitad = len(lista) // 2
    izquierda = merge_sort(lista[:mitad])    # conquistar la mitad izquierda
    derecha = merge_sort(lista[mitad:])      # conquistar la mitad derecha
    return mezclar(izquierda, derecha)       # combinar

datos = [27, 10, 12, 20, 25, 13, 15, 22]
print(merge_sort(datos))
print(datos)  # la lista original no se modifica
````

````python
def merge_sort_traza(lista, nivel=0):
    sangria = "  " * nivel
    print(f"{sangria}dividir {lista}")
    if len(lista) <= 1:
        return lista
    mitad = len(lista) // 2
    izq = merge_sort_traza(lista[:mitad], nivel + 1)
    der = merge_sort_traza(lista[mitad:], nivel + 1)
    res = mezclar(izq, der)
    print(f"{sangria}mezclar {izq} + {der} -> {res}")
    return res

merge_sort_traza([27, 10, 12, 20, 25, 13, 15, 22])
````

**Clave:** el trabajo está en el paso de **combinar** (`mezclar`); dividir es solo partir por la mitad. Resuelve los dos subproblemas.

**Error común:** olvidar agregar el resto de la lista que no se agotó al final de `mezclar`, perdiendo elementos.

**Para el quiz:** en merge sort los dos subproblemas son del mismo tamaño y hace falta un procedimiento separado para fusionar las listas ordenadas.

**Referencia:** PDF parte 2, págs. 69–70

## Quick sort y divide y vencerás

El quick sort (Unidad 29) elige un **pivote** y divide la lista en dos partes: elementos menores que el pivote y elementos mayores o iguales. Ordena cada parte recursivamente. A diferencia del merge sort:

- los dos subproblemas **no** tienen necesariamente el mismo tamaño;
- **no existe** un paso de combinar listas ordenadas: el pivote ya queda en su posición final y basta concatenar.

El PDF ilustra con `[15, 10, 12, 20, 25, 13, 22]` y pivote 15, que genera `[13, 10, 12]` y `[25, 20, 22]`.

**Ampliación (no está en el PDF):** implementación propia con listas por comprensión.

````python
def quick_sort(lista):
    if len(lista) <= 1:      # caso base
        return lista
    pivote = lista[0]
    menores = [x for x in lista[1:] if x < pivote]
    mayores = [x for x in lista[1:] if x >= pivote]
    return quick_sort(menores) + [pivote] + quick_sort(mayores)

print(quick_sort([15, 10, 12, 20, 25, 13, 22]))
````

````python
def quick_sort_traza(lista, nivel=0):
    sangria = "  " * nivel
    if len(lista) <= 1:
        print(f"{sangria}{lista} (caso base)")
        return lista
    pivote = lista[0]
    menores = [x for x in lista[1:] if x < pivote]
    mayores = [x for x in lista[1:] if x >= pivote]
    print(f"{sangria}{lista} pivote={pivote} -> menores={menores} mayores={mayores}")
    return quick_sort_traza(menores, nivel + 1) + [pivote] + quick_sort_traza(mayores, nivel + 1)

quick_sort_traza([15, 10, 12, 20, 25, 13, 22])
````

**Nota sobre el PDF:** el diagrama muestra `[13, 10, 12]` a la izquierda del pivote, es decir, el orden que resulta de una partición *in situ* (intercambiando elementos). Nuestra versión conserva el orden relativo y obtiene `[10, 12, 13]`; el resultado final ordenado es el mismo. Además, el PDF usa una lista de 8 elementos para merge sort y una de 7 para quick sort.

**Clave:** el pivote determina el reparto; con un pivote malo (lista ya ordenada) las particiones quedan desbalanceadas.

**Error común:** creer que quick sort necesita mezclar sublistas como merge sort. No lo necesita.

**Para el quiz:** quick sort divide en dos instancias de tamaño no necesariamente igual y no tiene proceso de integración.

**Referencia:** PDF parte 2, pág. 71

## Ruta mínima del triángulo con divide y vencerás

Un triángulo se representa como una lista de listas `T`: `T[fila][columna]`. Desde la posición `(fila, columna)` solo se puede bajar a `(fila + 1, columna)` (abajo) o a `(fila + 1, columna + 1)` (abajo a la derecha).

Sea `C(fila, columna)` la suma mínima de un camino que parte de esa posición. Con `n = len(T)`:

```
C(fila, columna) = T[fila][columna] + min(C(fila + 1, columna), C(fila + 1, columna + 1))
C(n, columna)    = 0
```

El caso base `C(n, columna) = 0` ocurre al pasar la última fila (se llegó a la base del triángulo). El problema se divide en dos subproblemas (bajar o bajar a la derecha), se resuelven recursivamente y se combina con `min` y la suma del costo actual.

**Referencia:** PDF parte 2, págs. 76–78

````python
def find_minimum(row, col, triangle):
    if row == len(triangle):
        return 0
    else:
        minimum = min(find_minimum(row + 1, col, triangle),
                      find_minimum(row + 1, col + 1, triangle))
    return triangle[row][col] + minimum

triangle = [
    [2],
    [3, 4],
    [6, 5, 7],
    [4, 1, 8, 3]
]
minimum = find_minimum(0, 0, triangle)
print("The minimum cost is ", minimum)
````

El PDF explica el código así: el caso base `row == len(triangle)` devuelve 0 (líneas 1–3); si no se llegó al fondo, se calculan recursivamente los dos caminos y se elige el menor (líneas 4–6); finalmente se suma el costo de la posición actual (línea 7). La llamada inicial es `find_minimum(0, 0, triangle)` porque la cima es `triangle[0][0]`.

**Nota sobre el PDF:** el texto traducido habla de "valores de fila y largo" (es `row` y `len(triangle)`) y de "triángulo de costo [fila][columna]" (es `triangle[row][col]`). La salida del PDF, `The minimum cost is  11`, tiene dos espacios porque `print` con coma añade un espacio tras la cadena que ya termina en espacio; el código es correcto.

**Referencia:** PDF parte 2, págs. 79–82

**Ampliación (no está en el PDF):** esta recursión repite subproblemas. Cada llamada genera dos más, así que para `n` filas se hacen `2^(n+1) - 1` llamadas. Se puede comprobar contando llamadas y comparando con una versión con memoria (memoización), que guarda cada `C(fila, columna)` ya calculado.

````python
llamadas = 0

def find_minimum_contando(row, col, triangle):
    global llamadas
    llamadas += 1
    if row == len(triangle):
        return 0
    minimum = min(find_minimum_contando(row + 1, col, triangle),
                  find_minimum_contando(row + 1, col + 1, triangle))
    return triangle[row][col] + minimum

def triangulo_de_unos(n):
    return [[1] * (fila + 1) for fila in range(n)]

for n in (4, 10, 15):
    llamadas = 0
    find_minimum_contando(0, 0, triangulo_de_unos(n))
    print(f"n={n}: {llamadas} llamadas (2^(n+1) - 1 = {2 ** (n + 1) - 1})")
````

````python
def find_minimum_memo(triangle):
    memo = {}
    def C(row, col):
        if row == len(triangle):
            return 0
        if (row, col) not in memo:
            memo[(row, col)] = triangle[row][col] + min(C(row + 1, col), C(row + 1, col + 1))
        return memo[(row, col)]
    return C(0, 0)

print(find_minimum_memo(triangle))
print(find_minimum_memo(triangulo_de_unos(300)))  # con la version simple seria imposible
````

**Clave:** la relación recursiva `C(fila, col) = T[fila][col] + min(abajo, abajo-derecha)` y el caso base `fila == n -> 0` resumen todo el algoritmo.

**Error común:** usar `len(triangle) - 1` como caso base devolviendo 0; se perdería el costo de la última fila. También llamar a la función con `(1, 0)` en lugar de `(0, 0)`.

**Para el quiz:** la suma mínima del triángulo del ejemplo es 11 (camino 2 + 3 + 5 + 1); la llamada inicial es `find_minimum(0, 0, triangle)`.

**Referencia:** PDF parte 2, págs. 76–82

## Examen sorpresa: la moneda más pesada

El PDF plantea: hay ocho monedas idénticas numeradas del 1 al 8 y solo una es más pesada. Con una balanza de dos platillos, diseñar un algoritmo para hallarla y decir cuántas pesadas se necesitan como mínimo. En la pregunta P2 se repite con nueve monedas.

**Nota sobre el PDF:** el texto traducido dice "pesar los iones" y "básculas de brazo"; significa "pesar las monedas" en una balanza de dos brazos.

La idea divide y vencerás: en lugar de dividir en 2 grupos, se divide en **3** grupos (casi iguales). Se comparan dos grupos del mismo tamaño: si uno pesa más, ahí está la moneda; si empatan, está en el tercer grupo. Cada pesada reduce el problema a un tercio, por lo que la respuesta es 2 pesadas tanto para 8 como para 9 monedas (con 3 pesadas se resuelven hasta 27).

**Ampliación (no está en el PDF):** la solución es propia; el PDF no la incluye.

````python
def pesar(pesos, grupo_a, grupo_b):
    """Simula una pesada: devuelve 'izquierda', 'derecha' o 'igual'."""
    a = sum(pesos[i] for i in grupo_a)
    b = sum(pesos[i] for i in grupo_b)
    if a > b:
        return "izquierda"
    if b > a:
        return "derecha"
    return "igual"

def buscar_pesada(pesos, candidatas=None):
    """Devuelve (indice de la moneda pesada, numero de pesadas usadas)."""
    if candidatas is None:
        candidatas = list(range(len(pesos)))
    if len(candidatas) == 1:
        return candidatas[0], 0
    tercio = -(-len(candidatas) // 3)            # techo de len/3
    g1 = candidatas[:tercio]
    g2 = candidatas[tercio:2 * tercio]
    g3 = candidatas[2 * tercio:]
    resultado = pesar(pesos, g1, g2)
    if resultado == "izquierda":
        moneda, usadas = buscar_pesada(pesos, g1)
    elif resultado == "derecha":
        moneda, usadas = buscar_pesada(pesos, g2)
    else:
        moneda, usadas = buscar_pesada(pesos, g3)
    return moneda, usadas + 1

for n in (8, 9):
    peor = 0
    for pesada in range(n):
        pesos = [1] * n
        pesos[pesada] = 2
        moneda, usadas = buscar_pesada(pesos)
        assert moneda == pesada
        peor = max(peor, usadas)
    print(f"{n} monedas: peor caso = {peor} pesadas")
````

**Clave:** dividir en tres grupos reduce el problema a un tercio por pesada: el mínimo es el menor `k` tal que `3^k >= n`.

**Para el quiz:** 8 monedas y 9 monedas se resuelven ambas con 2 pesadas.

**Referencia:** PDF parte 2, págs. 84–85

**Problema (págs. 89–94):** diseñar un algoritmo de divide y vencerás para el problema del tromino. Un tromino son tres cuadrados unidos en forma de L. Se tiene un tablero de `m x m` cuadrados (con `m` potencia de 2) donde una casilla está marcada con una X. Hay que cubrir todo el tablero con trominós de modo que: no se cubra la X, no se superpongan y no sobresalgan del tablero. La pista del PDF sugiere dividir el tablero en cuatro.

**Ampliación (no está en el PDF):** el PDF solo da el enunciado y pistas con figuras (pág. 94 muestra un tablero de 4x4 resuelto con la X en la segunda columna de la fila superior). La implementación siguiente es propia.

Idea: se divide el tablero en 4 cuadrantes. La X está en uno de ellos. Se coloca un tromino en el centro cubriendo una casilla de cada uno de los otros tres cuadrantes; así cada cuadrante tiene exactamente una casilla "ocupada" y se convierte en un subproblema idéntico, de la mitad de tamaño. El caso base es el tablero de 1x1.

````python
def tromino(m, fila_x, col_x):
    """Cubre un tablero m x m (m potencia de 2) con trominos, dejando libre la casilla X."""
    tablero = [[0] * m for _ in range(m)]
    tablero[fila_x][col_x] = -1          # -1 marca la X
    contador = [0]                       # numero del siguiente tromino

    def cubrir(arriba, izq, tam, fx, cx):
        if tam == 1:
            return
        contador[0] += 1
        numero = contador[0]
        mitad = tam // 2
        centro_f, centro_c = arriba + mitad, izq + mitad
        # Los cuatro cuadrantes: (fila inicial, columna inicial, casilla del centro que le toca)
        cuadrantes = [
            (arriba, izq, centro_f - 1, centro_c - 1),
            (arriba, centro_c, centro_f - 1, centro_c),
            (centro_f, izq, centro_f, centro_c - 1),
            (centro_f, centro_c, centro_f, centro_c),
        ]
        for f0, c0, cf, cc in cuadrantes:
            if f0 <= fx < f0 + mitad and c0 <= cx < c0 + mitad:
                cubrir(f0, c0, mitad, fx, cx)      # aqui esta la X (o la casilla ya ocupada)
            else:
                tablero[cf][cc] = numero           # el tromino ocupa esta casilla central
                cubrir(f0, c0, mitad, cf, cc)

    cubrir(0, 0, m, fila_x, col_x)
    return tablero

def mostrar(tablero):
    for fila in tablero:
        print(" ".join(" X" if v == -1 else f"{v:2d}" for v in fila))

t4 = tromino(4, 0, 1)
mostrar(t4)
````

````python
def es_cobertura_valida(tablero):
    """Cada numero debe aparecer exactamente 3 veces y no debe haber casillas vacias salvo la X."""
    cuentas = {}
    for fila in tablero:
        for v in fila:
            cuentas[v] = cuentas.get(v, 0) + 1
    return cuentas.get(-1) == 1 and cuentas.get(0, 0) == 0 and all(
        n == 3 for k, n in cuentas.items() if k > 0)

print(es_cobertura_valida(t4))
t8 = tromino(8, 5, 2)
mostrar(t8)
print(es_cobertura_valida(t8))
````

**Clave:** el truco es colocar un tromino central para que los cuatro cuadrantes tengan una casilla ocupada; cada uno se vuelve una copia más pequeña del problema original.

**Para el quiz:** el problema del tromino se resuelve dividiendo el tablero en cuatro cuadrantes; `m` debe ser potencia de 2.

**Referencia:** PDF parte 2, págs. 87–94

## Proyecto de la unidad: Triangle Path Finder

**Problema:** dada una lista de listas que representa un triángulo de enteros positivos, calcular la suma del camino de menor costo desde la cima hasta la base, avanzando abajo o abajo a la derecha.

**Plan (pasos del PDF):**
1. Representar el triángulo como lista de listas `T`.
2. Definir `C(fila, col) = T[fila][col] + min(C(fila + 1, col), C(fila + 1, col + 1))`.
3. Definir el caso base: `fila == len(T)` devuelve 0.
4. Implementar `find_minimum(row, col, triangle)` recursivamente.
5. Llamar `find_minimum(0, 0, triangle)` y mostrar el resultado.

**Resultado esperado:** `The minimum cost is  11`.

**Referencia:** PDF parte 2, págs. 59–61 y 82

````python
def find_minimum(row, col, triangle):
    # Caso base: se paso de la ultima fila, no queda costo por pagar
    if row == len(triangle):
        return 0
    # Dividir: bajar (row + 1, col) o bajar a la derecha (row + 1, col + 1)
    # Conquistar: cada opcion se resuelve recursivamente
    minimum = min(find_minimum(row + 1, col, triangle),
                  find_minimum(row + 1, col + 1, triangle))
    # Combinar: costo de la casilla actual mas el mejor camino restante
    return triangle[row][col] + minimum

triangle = [
    [2],
    [3, 4],
    [6, 5, 7],
    [4, 1, 8, 3]
]
minimum = find_minimum(0, 0, triangle)
print("The minimum cost is ", minimum)
````

def mezclar(a, b):
    resultado = []
    i = j = 0
    while i < len(a) and j < len(b):
        if a[i] <= b[j]:
            resultado.append(a[i])
            i += 1
        else:
            resultado.append(b[j])

    resultado.extend(a[i:])

## Apuntes para estudiar

**Clave**
- Divide y vencerás: dividir la instancia, conquistar cada parte (con recursividad si aún es grande) y combinar las respuestas.
- Búsqueda binaria: divide la lista ordenada y resuelve solo una de las dos mitades; condición de terminación `low > high` devuelve `-1`.
- Merge sort: divide en dos mitades iguales, resuelve ambas y necesita un procedimiento para mezclar las dos listas ordenadas.
- Quick sort: divide con un pivote en dos partes de tamaño no necesariamente igual y no necesita combinar.
- Ruta del triángulo: `C(fila, col) = T[fila][col] + min(C(fila + 1, col), C(fila + 1, col + 1))` y `C(n, col) = 0`; el ejemplo da 11.
- La recursión simple del triángulo repite subproblemas; la memoización lo evita (ampliación).

**Errores comunes**
- Omitir el caso base y obtener `RecursionError`.
- Aplicar búsqueda binaria a una lista sin ordenar.
- Olvidar agregar el resto de la lista al final de `mezclar`.
- Confundir quick sort con merge sort: el primero no tiene paso de integración.
- Usar un caso base equivocado en el triángulo (`len(triangle) - 1`) y perder la última fila.
- Pensar que la programación en pareja es una estrategia de dividir y vencer el trabajo; el PDF dice que no lo es.

**Para el quiz**
- Los tres pasos: dividir, conquistar, combinar.
- Búsqueda binaria, merge sort y quick sort aplican divide y vencerás.
- Búsqueda binaria resuelve un solo subproblema; merge sort resuelve los dos.
- Merge sort necesita fusionar; quick sort no.
- Suma mínima del triángulo del PDF: 11 (2 + 3 + 5 + 1); llamada inicial `find_minimum(0, 0, triangle)`.
- Ocho y nueve monedas: 2 pesadas.
