## 1. Qué es el enfoque Greedy

El enfoque Greedy (voraz) es una técnica de diseño de algoritmos que llega a una solución **eligiendo, en cada paso de una secuencia de decisiones, la mejor opción del momento** según un criterio definido de antemano. Es una técnica sencilla y muy eficiente para problemas de optimización como el cambio de monedas y la selección de actividades.

El enfoque resuelve el problema en tres pasos:

| Paso | Nombre en el PDF | Pregunta que responde |
|---|---|---|
| 1 | Proceso de selección | ¿Qué elemento elijo ahora para añadirlo al conjunto? |
| 2 | Comprobación de la viabilidad | ¿El nuevo conjunto es apropiado para ser la respuesta (sigue siendo válido)? |
| 3 | Comprobación de la solución | ¿El nuevo conjunto ya es la respuesta al problema? |

**Clave:** una decisión greedy nunca se reconsidera: lo elegido queda elegido. Por eso es rápido, pero también por eso necesita un criterio de selección bien escogido.

**Para el quiz:** los tres pasos son selección, comprobación de viabilidad y comprobación de solución.

## 2. Problema del cambio de monedas

En Corea hay cuatro monedas relevantes en el ejemplo: 500, 100, 50 y 10 won. Al devolver un cambio se busca usar **el menor número de monedas** (se supone que hay monedas suficientes de cada tipo).

Para devolver 870 won, el PDF aplica el criterio greedy "tomar siempre la moneda más grande posible":

| Paso | Cambio antes | Moneda elegida | Cambio después |
|---|---|---|---|
| 1 | 870 | 500 | 370 |
| 2 | 370 | 100 | 270 |
| 3 | 270 | 100 | 170 |
| 4 | 170 | 100 | 70 |
| 5 | 70 | 50 | 20 |
| 6 | 20 | 10 | 10 |
| 7 | 10 | 10 | 0 |

Cuando una moneda ya no cabe en el cambio restante (por ejemplo, 500 con 370 pendientes), deja de poder elegirse y se pasa a la siguiente más grande. Resultado: 7 monedas, `[500, 100, 100, 100, 50, 10, 10]`.

El siguiente código reproduce la tabla con una simulación paso a paso (es un ejemplo propio para visualizar el proceso; el algoritmo del PDF está más abajo).

**Ampliación (no está en el PDF):** la simulación imprime cada paso.

````python
# Simulación paso a paso del cambio de 870 won (ejemplo propio para visualizar)
coins = [500, 100, 50, 10]
amount = 870
paso = 0
largest = 0
while amount > 0:
    if amount < coins[largest]:
        largest += 1          # esta moneda ya no cabe: pasamos a la siguiente más grande
    else:
        paso += 1
        antes = amount
        amount -= coins[largest]
        print(f"Paso {paso}: cambio {antes:>3} -> se elige {coins[largest]:>3} -> queda {amount:>3}")
````

**Referencia:** PDF parte 2, págs. 15–23

## 3. Diseño del algoritmo Greedy para el cambio de monedas

El PDF expresa el proceso en pseudocódigo (en inglés en el original):

```text
while (there are more coins and the problem is not solved) {
    grab the largest remaining coin;                                    <- proceso de selección
    if (adding the coin doesn't make the change exceed the amount owed) <- comprobación de la viabilidad
        add the coin to the change;
    if (the total value of the change equals the amount owed)           <- comprobación de la solución
        the problem is solved;
}
```

Traducido: mientras queden monedas y el problema no esté resuelto, (1) tomar la moneda más grande que queda, (2) añadirla solo si no hace que el cambio supere el importe a devolver, (3) si el total del cambio iguala el importe, el problema está resuelto.

Correspondencia con los tres pasos del enfoque:

- **Selección:** se elige la moneda de mayor valor nominal que queda.
- **Viabilidad:** se comprueba que la moneda no supere el importe restante.
- **Solución:** se comprueba si el total de las monedas elegidas ya es igual al importe del cambio.

**Clave:** en el código del PDF la comprobación de viabilidad es la condición `amount < coins[largest]` (si la moneda no cabe, se descarta) y la comprobación de solución es la condición del `while` (`amount > 0`: cuando el resto llega a 0, el cambio está completo).

**Referencia:** PDF parte 2, págs. 24–28

## 4. Implementación del cambio de monedas

La entrada es la lista de tipos de moneda **ordenada de mayor a menor** y el importe; la salida es la lista de monedas elegidas. Código del PDF (los identificadores se mantienen como en el original):

- Líneas 2–3: `changes` es la "billetera" del cambio y empieza vacía; `largest` es la posición de la moneda más grande que aún puede usarse, y empieza en 0 (la primera, la mayor).
- Líneas 4–9: el `while` se repite mientras quede cambio por devolver. Si `amount < coins[largest]`, la moneda no cabe y se avanza a la siguiente; si no, se agrega la moneda al cambio y se resta su valor.
- Línea 10: al terminar el bucle, `changes` suma exactamente el importe.

````python
def coin_change(coins, amount):
    changes = []
    largest = 0
    while amount > 0:
        if amount < coins[largest]:
            largest += 1
        else:
            changes.append(coins[largest])
            amount -= coins[largest]
    return changes

coins = [500, 100, 50, 10]
changes = coin_change(coins, 870)
print(changes, len(changes))
````

**Nota sobre el PDF:** en la pág. 32 el ejemplo usa `input("Input the amount: ")` y el usuario escribe 870; el resultado del PDF es `[500, 100, 100, 100, 50, 10, 10] 7`, igual al que se obtiene arriba. Aquí se llama a la función directamente con 870 para que la celda se ejecute sola. Más abajo, en el proyecto, la versión interactiva se encapsula en una función.

El PDF supone que el importe de entrada **siempre puede formarse** con las monedas dadas. Si no es así, el código falla. Veamos qué ocurre:

**Error común:** ejecutar `coin_change` con un importe que no se puede formar (por ejemplo, 875 con monedas de 500, 100, 50 y 10) agota la lista de monedas y lanza `IndexError`. El PDF no lo contempla; lo mostramos capturado con `try/except`.

````python
try:
    print(coin_change([500, 100, 50, 10], 875))
except IndexError as error:
    print("IndexError:", error)
    print("No se puede formar 875 con esas monedas: 'largest' se salió de la lista.")
````

## 5. Cuándo el enfoque Greedy NO da la mejor solución

El PDF propone un "examen sorpresa" que pone a prueba esta idea: se agrega una moneda de 400 won y se pide el resultado de `coin_change` para 800 won con `coins = [500, 400, 100, 50, 10]`.

Aplicando el algoritmo: 500 cabe (quedan 300); 400 ya no cabe (300 < 400), se descarta; 100 cabe tres veces. Resultado: `[500, 100, 100, 100]`, 4 monedas. Pero existe una solución mejor: `[400, 400]`, solo 2 monedas.

````python
coins = [500, 400, 100, 50, 10]
changes = coin_change(coins, 800)
print(changes, len(changes))
print("Mejor solución posible: [400, 400] ->", len([400, 400]), "monedas")
````

**Nota sobre el PDF:** el enunciado dice "supongamos que hay una moneda de 400 wones" y pide "el resultado de cómo el algoritmo `coin_change()` determinará el cambio de la moneda de 800 wones". Es una traducción confusa: lo que se pide es el cambio **de 800 won**. La respuesta del algoritmo es `[500, 100, 100, 100]`.

**Ampliación (no está en el PDF):** esto muestra que el criterio greedy "la moneda más grande" no siempre produce el mínimo de monedas. Funciona con el sistema 500, 100, 50, 10 (cada moneda es múltiplo de la siguiente), pero no con cualquier conjunto. Otro ejemplo clásico es `[4, 3, 1]` con importe 6: greedy da `[4, 1, 1]` (3 monedas) y el óptimo es `[3, 3]` (2 monedas). Cuando el enfoque greedy no garantiza el óptimo se usan otras técnicas, como la programación dinámica (Unidad 32).

````python
# Ampliación: otro contraejemplo (sistema de monedas 4, 3, 1)
print(coin_change([4, 3, 1], 6), "<- greedy (3 monedas)")
print([3, 3], "<- óptimo (2 monedas)")
````

**Clave:** greedy es rápido y simple, pero su corrección depende del problema y del criterio de selección. En el cambio de monedas con 500/100/50/10 y en la selección de actividades (con el criterio correcto) sí da la respuesta óptima.

## 6. Problema de selección de actividades (reuniones en una sala)

Planteamiento: se dan 7 solicitudes de uso de una sala de conferencias, cada una con hora de inicio y de fin. Se quiere el **máximo número de reuniones que no se solapan**.

El PDF usa estos datos (pág. 10, con la versión de las págs. 11, 34–45; véase la nota abajo):

| id | 0 | 1 | 2 | 3 | 4 | 5 | 6 |
|---|---|---|---|---|---|---|---|
| Inicio | 1 | 3 | 2 | 0 | 5 | 7 | 5 |
| Fin | 2 | 4 | 5 | 6 | 6 | 9 | 9 |

**Nota sobre el PDF:** la tabla de las págs. 10 y 33 muestra los inicios `1, 3, 2, 1, 5, 8, 5`, mientras que el código y las tablas de las págs. 11 y 34–45 usan `1, 3, 2, 0, 5, 7, 5`. Son datos distintos entre diapositivas. Con ambos juegos de datos el resultado es el mismo, `[0, 1, 4, 5]`, como se comprueba más abajo. En este notebook se usan los datos del código (págs. 11 y 45).

**Para dibujar el calendario** usamos una función auxiliar propia. Cada fila es una reunión; `#` marca las horas ocupadas y las filas elegidas se indican con `*`.

**Ampliación (no está en el PDF):** la función `dibujar_calendario` es propia, para visualizar el problema como lo hacen las diapositivas.

````python
def dibujar_calendario(start, finish, elegidas=()):
    """Dibuja cada reunión como una fila de '#' sobre el eje de horas (ejemplo propio)."""
    ancho = max(finish) + 1
    print("id  inicio fin | " + "".join(str(h % 10) for h in range(ancho)))
    for i in range(len(start)):
        fila = "".join("#" if start[i] <= h < finish[i] else "." for h in range(ancho))
        marca = "*" if i in elegidas else " "
        print(f"{i:>2}{marca} {start[i]:>6} {finish[i]:>3} | {fila}")

start = [1, 3, 2, 0, 5, 7, 5]
finish = [2, 4, 5, 6, 6, 9, 9]
dibujar_calendario(start, finish)
````

### Criterio greedy: elegir siempre la reunión que termina antes

El PDF sigue estas decisiones, con las reuniones **ordenadas por hora de finalización** (la reunión 0 es la que termina antes):

1. Al principio todas son posibles, así que se elige la reunión 0 (inicio 1, fin 2).
2. La reunión 1 empieza en 3, que es mayor que el fin de la reunión 0 (2): se puede elegir.
3. La reunión 2 empieza en 2, antes del fin de la reunión 1 (4): no se puede elegir.
4. La reunión 3 empieza en 0, antes del fin de la reunión 1 (4): no se puede elegir.
5. La reunión 4 empieza en 5, después del fin de la reunión 1 (4): se puede elegir.
6. La reunión 5 empieza en 7, después del fin de la reunión 4 (6): se puede elegir.
7. La reunión 6 empieza en 5, antes del fin de la reunión 5 (9): no se puede elegir.

Resultado: las reuniones `[0, 1, 4, 5]`, un máximo de 4.

**Nota sobre el PDF:** en la pág. 41 el texto dice "la hora de inicio de la reunión 6, que es la 5, es menor que la hora de inicio de la reunión 5". La traducción es errónea: lo correcto es comparar con la hora de **fin** de la última reunión elegida (la reunión 5 termina en 9, y 5 < 9). Asimismo, en las págs. 37 y 38 el texto habla de "primera" y "segunda reunión" cuando se refiere a las reuniones 1 y 2; hay que guiarse por las tablas.

**Relación con los tres pasos:**

- **Selección:** se toma la siguiente reunión en el orden de fin.
- **Viabilidad:** su inicio debe ser mayor o igual que el fin de la última reunión elegida.
- **Solución:** se acaba al revisar todas las reuniones.

**Referencia:** PDF parte 2, págs. 33–42

### Implementación

Código del PDF (nombre `activity_selection1`). Recibe las listas `start` y `finish`, ya ordenadas por hora de fin:

- Líneas 1–4: `result` guardará los números de reunión; `i` es la última reunión elegida; se elige la reunión 0 desde el inicio.
- Líneas 5–9: se prueban las reuniones desde 1 hasta `len(start) - 1`. Si `finish[i] <= start[j]`, la reunión `j` no se solapa con la última elegida: se añade al resultado y `i` pasa a ser `j`.

````python
def activity_selection1(start, finish):
    result = []
    i = 0
    result.append(i)
    for j in range(1, len(start)):
        if finish[i] <= start[j]:
            result.append(j)
            i = j
    return result

start = [1, 3, 2, 0, 5, 7, 5]
finish = [2, 4, 5, 6, 6, 9, 9]
meetings = activity_selection1(start, finish)
maximum = len(meetings)
print(meetings, maximum)
dibujar_calendario(start, finish, meetings)
````

````python
# Con los datos de la tabla de la pág. 10 (inicios 1, 3, 2, 1, 5, 8, 5) el resultado es el mismo
print(activity_selection1([1, 3, 2, 1, 5, 8, 5], [2, 4, 5, 6, 6, 9, 9]))
````

**Nota sobre el PDF:** el algoritmo solo es correcto si las reuniones llegan **ordenadas por hora de fin**. El PDF lo usa con datos que ya cumplen esa condición, pero no lo dice explícitamente. Además, el código solo funciona si hay al menos una reunión (siempre elige la 0).

**Error común:** aplicar `activity_selection1` a reuniones sin ordenar por hora de fin. Veamos un ejemplo con tres reuniones donde la primera de la lista termina tarde.

**Ampliación (no está en el PDF):** para datos sin ordenar hay que ordenarlos primero por hora de fin. Se muestra un ejemplo con `sorted` y una clave.

````python
# Reuniones como pares (id, inicio, fin) en desorden
reuniones = [(0, 0, 10), (1, 1, 2), (2, 3, 4)]

# Sin ordenar: se toma la reunión 0 (0-10) y todo lo demás se solapa
print("Sin ordenar:", activity_selection1([r[1] for r in reuniones], [r[2] for r in reuniones]))

# Ordenando por hora de fin, el algoritmo funciona
ordenadas = sorted(reuniones, key=lambda r: r[2])
indices = activity_selection1([r[1] for r in ordenadas], [r[2] for r in ordenadas])
print("Ordenadas :", [ordenadas[k][0] for k in indices], "(ids originales)")
````

**Ampliación (no está en el PDF):** el criterio "elegir la que termina antes" es el que garantiza el óptimo. Otros criterios que parecen razonables fallan: elegir la que empieza antes lo hace con el ejemplo anterior (elige la reunión 0-10 y se pierde las otras dos).

**Clave:** en selección de actividades el criterio greedy es **ordenar por hora de fin** y tomar cada reunión cuyo inicio sea mayor o igual que el fin de la última elegida. El costo dominante es la ordenación.

**Para el quiz:** el resultado del ejemplo del PDF es `[0, 1, 4, 5]` y el máximo es 4.

## 7. Diseño e implementación recursiva de algoritmos Greedy

Uno de los objetivos de la unidad es **diseñar e implementar algoritmos Greedy de forma recursiva**. En las páginas de este PDF las implementaciones mostradas usan bucles (`while` y `for`); la versión recursiva no se desarrolla en el código del PDF.

**Nota sobre el PDF:** el objetivo declara la implementación recursiva, pero las diapositivas solo muestran las versiones iterativas. Lo que sigue es una ampliación para cubrir ese objetivo.

**Ampliación (no está en el PDF):** cómo convertir un bucle greedy en recursión.

Un algoritmo greedy tiene la forma "elegir, comprobar, repetir con el problema restante". Eso es exactamente una recursión:

- **Caso base:** el problema está resuelto (cambio 0) o ya no hay más candidatos.
- **Paso recursivo:** tomar la mejor opción, comprobar su viabilidad y resolver el subproblema que queda.

Cambio de monedas recursivo: el estado es `(amount, largest)`.

````python
def coin_change_rec(coins, amount, largest=0):
    """Cambio de monedas con el criterio greedy, de forma recursiva.

    Devuelve la lista de monedas o None si no se puede formar el importe.
    """
    if amount == 0:                       # caso base: problema resuelto
        return []
    if largest >= len(coins):             # caso base: no quedan monedas que probar
        return None
    if amount < coins[largest]:           # no viable: probar con la siguiente moneda
        return coin_change_rec(coins, amount, largest + 1)
    resto = coin_change_rec(coins, amount - coins[largest], largest)   # elegir y resolver el resto
    if resto is None:
        return None
    return [coins[largest]] + resto

print(coin_change_rec([500, 100, 50, 10], 870))
print(coin_change_rec([500, 100, 50, 10], 875))   # no se puede formar: None en lugar de IndexError
````

Selección de actividades recursiva: el estado es la última reunión elegida `i` y la siguiente candidata `j`.

````python
def activity_selection_rec(start, finish, i, j, result):
    """Selección de actividades greedy, recursiva. i = última elegida, j = candidata."""
    if j >= len(start):                   # caso base: no quedan candidatas
        return result
    if finish[i] <= start[j]:             # viable: no se solapa con la última elegida
        result.append(j)
        return activity_selection_rec(start, finish, j, j + 1, result)
    return activity_selection_rec(start, finish, i, j + 1, result)   # descartar j

def activity_selection2(start, finish):
    """Envoltorio: elige la reunión 0 y llama a la función recursiva."""
    if not start:
        return []
    return activity_selection_rec(start, finish, 0, 1, [0])

print(activity_selection2([1, 3, 2, 0, 5, 7, 5], [2, 4, 5, 6, 6, 9, 9]))
````

**Ampliación (no está en el PDF):** la versión recursiva hace lo mismo que la iterativa, pero cada llamada ocupa espacio en la pila. Python limita la profundidad (por defecto, 1000 llamadas), así que con muchas monedas o reuniones la versión iterativa es más segura. La recursión es útil para entender cómo cada decisión reduce el problema a uno más pequeño.

**Clave:** convertir greedy a recursión = caso base (resuelto o sin candidatos) + elegir lo mejor + llamar sobre lo que queda.

## Proyecto de la unidad

### Problema: ¿cuántas reuniones son posibles?

Una empresa tiene una sola sala de conferencias. Varios equipos quieren celebrar reuniones allí, pero dos equipos no pueden usar la sala al mismo tiempo. El gestor de la sala quiere asignarla de forma que se celebre el máximo número de reuniones sin solapamientos. Conociendo la hora de inicio y de fin de N solicitudes, ¿cuál es ese máximo? (Problema de selección de actividades.)

Datos de ejemplo del PDF (N = 7):

| id | 0 | 1 | 2 | 3 | 4 | 5 | 6 |
|---|---|---|---|---|---|---|---|
| Inicio | 1 | 3 | 2 | 0 | 5 | 7 | 5 |
| Fin | 2 | 4 | 5 | 6 | 6 | 9 | 9 |

**Referencia:** PDF parte 2, págs. 9–11

### Plan de programación

1. Representar las solicitudes con dos listas, `start` y `finish`, ordenadas por hora de fin.
2. Elegir la primera reunión (la que termina antes) y recordarla como la última elegida.
3. Recorrer las demás: si su inicio es mayor o igual que el fin de la última elegida, es viable; elegirla y actualizarla.
4. Al terminar, el número de reuniones elegidas es el máximo.

**Referencia:** PDF parte 2, págs. 33–45 (selección de actividades)

### Código final

Es el código final del PDF (`activity_selection1`, pág. 12). Se repite aquí con comentarios para que el proyecto se lea de una vez.

````python
def activity_selection1(start, finish):
    result = []                          # números de las reuniones elegidas
    i = 0                                # última reunión elegida
    result.append(i)                     # la reunión 0 termina antes: siempre se elige
    for j in range(1, len(start)):       # probar las demás reuniones
        if finish[i] <= start[j]:        # viabilidad: empieza cuando la última ya terminó
            result.append(j)
            i = j                        # j pasa a ser la última elegida
    return result

start = [1, 3, 2, 0, 5, 7, 5]
finish = [2, 4, 5, 6, 6, 9, 9]
meetings = activity_selection1(start, finish)
maximum = len(meetings)
print(meetings, maximum)
````

Resultado esperado (pág. 11): `[0, 1, 4, 5] 4`. Comprobación automática:

````python
verificar("reuniones elegidas", meetings, [0, 1, 4, 5])
verificar("máximo de reuniones", maximum, 4)
````

````python
def leer_numeros(texto):
    """Convierte '500 100 50' en [500, 100, 50]. Devuelve None si el texto no es válido."""
    try:
        return [int(parte) for parte in texto.split()]
    except ValueError:
        return None

def sesion_reuniones(leer=input):
    """Pregunta inicios y fines de las reuniones y muestra cuántas caben en la sala.

    Termina si el usuario cancela (entrada vacía o None).
    """
    while True:
        texto = leer("Horas de inicio separadas por espacios (vacío para salir): ")
        if texto is None or texto.strip() == "":
            print("Sesión finalizada.")
            return
        inicios = leer_numeros(texto)

        texto = leer("Horas de fin separadas por espacios (vacío para salir): ")
        if texto is None or texto.strip() == "":
            print("Sesión finalizada.")
            return
        fines = leer_numeros(texto)

        if not inicios or not fines or len(inicios) != len(fines):
            print("Datos no válidos: deben ser números enteros y la misma cantidad de inicios y fines.")
            continue

        # Se ordenan las reuniones por hora de fin (el algoritmo lo exige)
        orden = sorted(range(len(inicios)), key=lambda k: fines[k])
        s = [inicios[k] for k in orden]
        f = [fines[k] for k in orden]
        elegidas = activity_selection1(s, f)
        ids = [orden[k] for k in elegidas]
        print(f"Reuniones elegidas (posiciones en la lista original): {sorted(ids)}")
        print(f"Máximo de reuniones posibles: {len(ids)}")

def lector_de_prueba(respuestas):
    """Devuelve una función que simula input() con una lista de textos y luego cancela."""
    pendientes = list(respuestas)
    def leer(pregunta):
        texto = pendientes.pop(0) if pendientes else ""
        print(f"{pregunta}{texto}")
        return texto
    return leer

# Versión no interactiva: se simulan las respuestas del usuario
sesion_reuniones(lector_de_prueba([
    "1 3 2 0 5 7 5", "2 4 5 6 6 9 9",     # ejemplo del PDF
    "5 1 3", "6 2 4",                      # reuniones sin ordenar
    "1 2", "3",                            # datos no válidos
]))

# Para usarlo de forma interactiva, quita el comentario de la siguiente línea
# sesion_reuniones()
````

**Clave:** el máximo para el ejemplo del PDF es 4, y la solución `[0, 1, 4, 5]` es la que produce el criterio "terminar antes".

El PDF muestra el programa que lee las monedas con `input`, usa una función `coin_change2` (que **no se muestra**) y da este resultado: monedas `500 50 50 100 50 10 10` e importe 710, salida `[500, 100, 50, 50, 10] 5`.

**Nota sobre el PDF:** la pista aparece traducida (`monedas.ordenar(reversa = Verdadero)`); en Python es `coins.sort(reverse=True)`. La función `coin_change2` no se muestra en las diapositivas: la escribimos aquí. Con monedas limitadas no se puede reutilizar una moneda, así que se recorre la lista ordenada **una sola vez** y se toma cada moneda si cabe.

````python
def coin_change2(coins, amount):
    """Cambio con cantidad limitada de monedas: cada moneda de la lista se usa como máximo una vez."""
    changes = []
    for coin in coins:                 # coins ya viene ordenada de mayor a menor
        if coin <= amount:             # viabilidad: la moneda cabe en el importe restante
            changes.append(coin)
            amount -= coin
        if amount == 0:                # solución: cambio completo
            break
    return changes

# Versión no interactiva del programa de la pág. 52 (el PDF lee el texto con input)
texto_monedas = "500 50 50 100 50 10 10"
coins = list(map(int, texto_monedas.split()))
coins.sort(reverse=True)
print(coins)
changes = coin_change2(coins, 710)
print(changes, len(changes))

# Para usarlo de forma interactiva, quita el comentario de las siguientes líneas
# coins = list(map(int, input("Input the coins: ").split()))
# coins.sort(reverse=True)
# amount = int(input("Input the amount: "))
# changes = coin_change2(coins, amount)
# print(changes, len(changes))
````

## Apuntes para estudiar

**Clave**

- Greedy elige en cada paso la mejor opción local según un criterio fijado de antemano y no la reconsidera.
- Tres pasos: proceso de selección, comprobación de la viabilidad y comprobación de la solución.
- Cambio de monedas: tomar siempre la moneda más grande que quepa en el importe restante. Con 500/100/50/10 y 870 won se obtienen 7 monedas: `[500, 100, 100, 100, 50, 10, 10]`.
- En el código del PDF, `amount < coins[largest]` es la viabilidad y `amount > 0` (condición del `while`) la solución.
- Selección de actividades: ordenar por hora de fin y elegir cada reunión cuyo inicio sea mayor o igual que el fin de la última elegida. Ejemplo del PDF: `[0, 1, 4, 5]`, máximo 4.
- Un algoritmo greedy puede escribirse con bucle (versión del PDF) o de forma recursiva: caso base + elegir + resolver lo que queda.

**Errores comunes**

- Dar las monedas sin ordenar de mayor a menor a `coin_change`: el algoritmo supone orden descendente.
- Importe imposible de formar: `coin_change` termina en `IndexError`.
- Aplicar `activity_selection1` a reuniones que no están ordenadas por hora de fin.
- Comparar el inicio de la candidata con el inicio (en lugar del fin) de la última reunión elegida.
- Creer que greedy siempre da el óptimo: con `[500, 400, 100, 50, 10]` y 800 won da 4 monedas cuando existen 2 (`400 + 400`).
- Con monedas limitadas, no comprobar que la suma del cambio coincide con el importe.
