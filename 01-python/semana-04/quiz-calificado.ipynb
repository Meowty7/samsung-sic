## Q 27-01 (30 puntos)

### Consigna

¿Cuántas comparaciones se ejecutaron en el siguiente proceso de ordenamiento de burbuja?

```python
def bubbleSort(S):
    n = len(S)
    for i in range(n):
        print(S)
        for j in range(n-1):
            if S[j] > S[j+1]:
               S[j] , S[j+1] = S[j+1] , S[j]

S = [50, 30, 40, 10, 20]
bubbleSort(S)
print(S)
```

**Dónde se ve:** [Unidad 27](clase-1-algoritmos-de-ordenamiento/u27-burbuja-seleccion-insercion.ipynb)

### Respuesta

````
def bubbleSort(S):
    n = len(S)
    comparaciones = 0
    for i in range(n):
        print(S)
        for j in range(n - 1):
            comparaciones += 1            # se cuenta cada S[j] > S[j+1]
            if S[j] > S[j + 1]:
                S[j], S[j + 1] = S[j + 1], S[j]
    return comparaciones


S = [50, 30, 40, 10, 20]
total = bubbleSort(S)
print(S)
print('Comparaciones:', total)
````

**Explicación:** **Respuesta: 20 comparaciones.** El bucle externo da n = 5 vueltas y el interno compara n − 1 = 4 pares en cada una: 5 × 4 = 20. Esta versión no se detiene antes aunque la lista ya esté ordenada (desde la tercera vuelta) y siempre compara n − 1 pares; por eso hace n(n − 1) comparaciones, no n(n − 1)/2. Se agregó un contador para comprobarlo; la salida impresa es la misma del enunciado.

## Q 27-02 (30 puntos)

### Consigna

¿Cuántas comparaciones se ejecutaron en el siguiente proceso de ordenamiento de inserción?

```python
def insertionSort2(S):

    for i in range(1,n):
        print(S)
        x = S[i]
        j = i - 1
        while j >= 0 and S[j] > x:
            S[j+1] = S[j]
            j -= 1

S = [50, 30, 40, 10, 20]

```

**Dónde se ve:** [Unidad 27](clase-1-algoritmos-de-ordenamiento/u27-burbuja-seleccion-insercion.ipynb)

### Respuesta

````
def insertionSort2(S):
    n = len(S)
    comparaciones = 0
    for i in range(1, n):
        print(S)
        x = S[i]
        j = i - 1
        while j >= 0:
            comparaciones += 1            # se cuenta cada S[j] > x
            if S[j] > x:
                S[j + 1] = S[j]
                j -= 1
            else:
                break
        S[j + 1] = x
    return comparaciones


S = [50, 30, 40, 10, 20]
total = insertionSort2(S)
print(S)
print('Comparaciones:', total)
````

**Explicación:** **Respuesta: 10 comparaciones** entre elementos (`S[j] > x`):

| i | x | Comparaciones | Lista después |
|---|---|---|---|
| 1 | 30 | 1 (con 50) | [30, 50, 40, 10, 20] |
| 2 | 40 | 2 (con 50 y 30) | [30, 40, 50, 10, 20] |
| 3 | 10 | 3 (con 50, 40 y 30) | [10, 30, 40, 50, 20] |
| 4 | 20 | 4 (con 50, 40, 30 y 10) | [10, 20, 30, 40, 50] |

Total: 1 + 2 + 3 + 4 = 10. Cuando `j` llega a −1, `j >= 0` es falso y ya no se compara ningún elemento. El contador se agregó separando la condición del `while`, sin cambiar lo que hace el algoritmo.

**Nota sobre el enunciado:** el "Output" que muestra es el de la Q 27-01 (burbuja), copiado por error. La salida real de este código es la de arriba: `[50, 30, 40, 10, 20]`, `[30, 50, 40, 10, 20]`, `[30, 40, 50, 10, 20]`, `[10, 30, 40, 50, 20]` y `[10, 20, 30, 40, 50]`.

## Q 29-01 (40 puntos)

### Consigna

Dada la lista a continuación, escriba la salida después de ejecutar la función `partition1()`.

```python
S = [15,10,12,20,25,13,22]
partition1(S, 0, len(S)-1)
print(S)
```

Output

```text
[15,10,12,20,25,13,22] 0 6 pivot = 15
...
```

**Dónde se ve:** [Unidad 29](clase-1-algoritmos-de-ordenamiento/u29-ordenamiento-rapido.ipynb)

### Respuesta

````
def partition1(S, low, high):
    pivot, left, right = S[low], low, high     # pivote: el primer elemento
    while left < right:
        print(S, left, right, "pivot = ", pivot)
        while left < high and S[left] <= pivot:   # avanza mientras sea menor o igual al pivote
            left += 1
        while right > low and pivot <= S[right]:  # retrocede mientras sea mayor o igual al pivote
            right -= 1
        if left < right:
            S[left], S[right] = S[right], S[left]
    S[low], S[right] = S[right], S[low]          # el pivote va a su posición final
    return right


S = [15, 10, 12, 20, 25, 13, 22]
pivotpoint = partition1(S, 0, len(S) - 1)
print(S)
print('Posición final del pivote:', pivotpoint)
````

**Explicación:** **Respuesta (salida):**

```text
[15, 10, 12, 20, 25, 13, 22] 0 6 pivot =  15
[15, 10, 12, 13, 25, 20, 22] 3 5 pivot =  15
[13, 10, 12, 15, 25, 20, 22]
```

Paso a paso con pivote 15: `left` avanza hasta el 20 (índice 3) y `right` retrocede hasta el 13 (índice 5); como `left < right`, se intercambian 20 y 13. En la segunda vuelta `left` llega al 25 (índice 4) y `right` al 13 (índice 3); ya se cruzaron, así que termina el bucle y el pivote se intercambia con `S[3]`. Resultado: **`[13, 10, 12, 15, 25, 20, 22]`**, con el 15 en su lugar definitivo (índice 3): a la izquierda los menores y a la derecha los mayores.

**Nota sobre el enunciado:** el formato de salida que muestra (`lista left right pivot = valor`) es el de la partición de la unidad 29 con pivote fijo en el primer elemento; con la `partition1` de las diapositivas, que imprime solo la lista, el resultado final es el mismo `[13, 10, 12, 15, 25, 20, 22]`.
