```python
import random


def hacer_quiz(unidad=None, cantidad=10):
    """Hace preguntas al azar. unidad=None mezcla todas las unidades."""
    banco = [p for p in PREGUNTAS if unidad is None or p["unidad"] == unidad]
    seleccion = random.sample(banco, min(cantidad, len(banco)))
    falladas = []
    for i, p in enumerate(seleccion, 1):
        print(f"\nPregunta {i} de {len(seleccion)} (Unidad {p['unidad']})")
        print(p["pregunta"])
        for letra, texto in p["opciones"].items():
            print(f"  {letra}) {texto}")
        respuesta = input("Tu respuesta (A-D): ").strip().upper()[:1]
        if respuesta == p["correcta"]:
            print("Correcto.")
        else:
            print(f"Incorrecto. La respuesta es {p['correcta']}: {p['porque']}")
            falladas.append(p)
    aciertos = len(seleccion) - len(falladas)
    print(f"\nResultado: {aciertos} de {len(seleccion)}")
    for p in falladas:
        print(f"- Repasar: {p['pregunta']} ({p.get('referencia', '')})")


print(f"{len(PREGUNTAS)} preguntas cargadas.")
```
