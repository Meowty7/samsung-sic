## Semana 0 — Fundamentos de programación con Python

### [Unidad 1 — Programación secuencial](01-python/semana-00/clase-1-fundamentos/u01-pensamiento-computacional-y-entorno.ipynb)

- **pensamiento computacional:** procesos de pensamiento para formular problemas y sus soluciones de modo que un agente procesador de información pueda ejecutarlas eficazmente.
- **secuencia:** orden en que el ordenador ejecuta las instrucciones, de arriba hacia abajo; cambiar el orden cambia el resultado.
- **expresión:** código simple que produce un valor, como `10 + 20` o `"Hola"`; varias expresiones forman una declaración.
- **declaración:** instrucción formada por una o más expresiones, como `print('Hola')`; varias declaraciones forman un programa.
- **anotación (comentario):** texto que el intérprete ignora y que explica el código; se escribe con `#` o entre tres comillas.
- **palabra reservada:** palabra con función predeterminada en Python (`if`, `for`, `global`) que no puede usarse como identificador.
- **entorno virtual:** espacio independiente con su propia versión de Python y sus paquetes, aislado de otros proyectos.
- **error de sintaxis:** error por romper las reglas del lenguaje; el código no llega a ejecutarse.
- **error de tiempo de ejecución:** error en código correcto que aparece mientras se ejecuta, como `ValueError` o `ZeroDivisionError`.

### [Unidad 2 — Planificación de la programación](01-python/semana-00/clase-1-fundamentos/u02-algoritmos-y-diagramas-de-flujo.ipynb)

- **algoritmo:** conjunto de instrucciones con una secuencia razonable que, a partir de una entrada, produce una salida para resolver un problema.
- **pseudocódigo:** representación de un algoritmo en lenguaje cotidiano; sirve de puente entre el diagrama de flujo y el código real.
- **diagrama de flujo:** representación gráfica de un algoritmo con formas normalizadas (ISO) y flechas que indican el flujo.
- **EDP (proceso de diseño técnico):** serie de etapas (definir, idear, prototipar, probar, mejorar) que siguen ingenieros y científicos para resolver un problema.
- **búsqueda lineal:** algoritmo que revisa los datos uno a uno desde el principio; necesita n pasos en el peor caso.
- **búsqueda binaria:** algoritmo que, sobre datos ordenados, compara con el centro y descarta la mitad; necesita log2 n pasos.
- **notación Big O:** representación matemática de cómo crece el número de operaciones de un algoritmo según el número de datos.
- **complejidad espacial y temporal:** medidas de rendimiento de un algoritmo; la primera mide la memoria usada y la segunda el número de pasos.

### [Unidad 3 — Conceptos básicos de tipos de datos numéricos y operaciones aritméticas](01-python/semana-00/clase-1-fundamentos/u03-operadores-y-tipos-de-datos.ipynb)

- **operando:** dato o variable sobre el que actúa un operador; en `100 + 20`, 100 y 20 son operandos.
- **operador:** símbolo que indica la operación a realizar (`+`, `-`, `*`, `/`, `//`, `%`, `**`).
- **tipo de datos:** clase de un valor (`int`, `float`, `str`, `bool`, `complex`) que determina qué operaciones admite.
- **type():** función incorporada que devuelve el tipo de un objeto.
- **precedencia:** orden en que se calculan los operadores de una expresión; los paréntesis tienen la mayor.
- **conversión de tipos:** cambio del tipo de un valor con `int()`, `float()` o `str()`.
- **concatenación:** unión de dos cadenas con el operador `+`.
- **indexación:** acceso a un carácter de una cadena con corchetes; empieza en 0 y admite índices negativos.

### [Unidad 4 — Variables y entradas](01-python/semana-00/clase-1-fundamentos/u04-variables.ipynb)

- **variable:** nombre que apunta a un valor guardado en la memoria; el valor puede cambiar.
- **operador de asignación (`=`):** guarda el valor calculado a la derecha con el nombre de la izquierda; no significa "igual".
- **identificador:** nombre de una variable; usa letras, dígitos y `_`, no empieza por dígito y no puede ser palabra reservada.
- **palabra clave (reservada):** palabra con significado especial en Python (`if`, `global`, `class`...) que no puede usarse como nombre de variable.
- **operador de asignación compuesto:** combina una operación con la asignación (`+=`, `-=`, `*=`, `/=`, `%=`).
- **operador aritmético:** realiza cálculos: `+ - * / // % **`.
- **tipado dinámico:** Python decide el tipo del valor al ejecutar; no se declara el tipo de la variable.
- **input():** función que lee texto del usuario y siempre devuelve una cadena (`str`).

### [Unidad 5 — Operaciones lógicas y de comparación](01-python/semana-00/clase-2-control-de-flujo/u05-operadores-de-comparacion-y-logicos.ipynb)

- **booleano:** tipo de datos con solo dos valores, `True` y `False`.
- **operador de comparación:** compara dos valores y devuelve un booleano: `== != > < >= <=`.
- **operador lógico:** combina booleanos: `and`, `or`, `not`.
- **tabla de verdad:** lista el resultado de un operador lógico para todas las combinaciones de entradas.
- **is:** operador de identidad; indica si dos nombres apuntan al mismo objeto.
- **bool():** función que convierte cualquier valor a booleano (0, `''`, `None` y `[]` dan `False`).
- **precedencia:** orden en que se evalúan los operadores de una expresión; los paréntesis tienen la mayor.
- **año bisiesto:** año de 366 días: divisible por 4 y no por 100, o divisible por 400.

### [Unidad 6 — Declaración condicional-1: condiciones y toma de decisiones](01-python/semana-00/clase-2-control-de-flujo/u06-sentencia-if.ipynb)

- **estructura secuencial:** forma de ejecución en la que los comandos se ejecutan uno tras otro, en el orden en que están escritos.
- **estructura de selección:** estructura en la que se elige y ejecuta una de varias instrucciones según una condición.
- **estructura de iteración:** estructura en la que se ejecuta repetidamente la misma orden.
- **expresión condicional:** expresión que se evalúa como un valor booleano (`True` o `False`) y decide qué rama se ejecuta.
- **bloque:** conjunto de líneas con sangría que se ejecutan cuando se cumple la condición de la sentencia que las encabeza.
- **sangría:** espacios al inicio de una línea que indican a qué bloque pertenece; PEP-8 recomienda 4 espacios.
- **pass:** sentencia que no hace nada; sirve como marcador de posición en un bloque vacío.
- **separador:** carácter que `split()` usa para cortar una cadena en partes (por defecto, el espacio).

### [Unidad 7 — Declaración condicional-2: toma de decisiones en dos direcciones y aplicación de enunciados condicionales](01-python/semana-00/clase-2-control-de-flujo/u07-if-else-elif.ipynb)

- **if-else:** sentencia condicional doble: ejecuta el bloque del `if` si la condición es verdadera y el del `else` si es falsa.
- **elif:** abreviatura de "else if"; examina otra condición solo cuando las anteriores fueron falsas.
- **if anidado:** sentencia `if` o `if-else` escrita dentro del bloque de otra; cada nivel añade sangría.
- **relación exclusiva:** relación entre dos condiciones que no pueden cumplirse a la vez.
- **operador lógico:** `and`, `or` y `not`, que combinan valores booleanos en condiciones más complejas.
- **módulo random:** módulo de Python para generar números aleatorios; se importa con `import random`.
- **randint:** función `random.randint(a, b)` que devuelve un entero entre `a` y `b`, ambos incluidos.
- **randrange:** función `random.randrange(n)` que devuelve un entero de 0 a `n - 1`.

### [Unidad 8 — Bucle-1: la sentencia for y la función range](01-python/semana-00/clase-2-control-de-flujo/u08-bucle-for.ipynb)

- **iteración:** repetición de un bloque de código varias veces; también se llama bucle.
- **bucle for:** sentencia que repite un bloque una vez por cada elemento de una secuencia (`for variable in secuencia:`).
- **range:** función que genera enteros consecutivos con la forma `range(inicio=0, parada, paso=1)`; la parada se excluye.
- **variable de control del bucle:** variable (`i`, `j`, `k`) que toma un valor distinto en cada vuelta; se puede reemplazar por `_` si no se usa.
- **objeto secuencial:** objeto con varios elementos ordenados, como una lista o una cadena, que `for` puede recorrer.
- **acumulador:** variable inicializada antes del bucle y actualizada en cada vuelta (por ejemplo `s += i`).
- **sangría (indentación):** espacios al inicio de las líneas que forman el bloque repetido del `for`.
- **bucle infinito:** iteración que nunca sale del círculo; no ocurre con `for` sobre una secuencia finita.

### [Unidad 9 — Bucle-2: la sentencia while, break y continue, random y bucle doble](01-python/semana-00/clase-2-control-de-flujo/u09-bucle-while.ipynb)

- **sentencia while:** bucle que repite un bloque mientras una condición sea verdadera.
- **bucle infinito:** bucle cuya condición nunca se vuelve falsa (por ejemplo `while True:`) y no se detiene sin `break`.
- **break:** palabra clave que finaliza el bucle de inmediato.
- **continue:** palabra clave que omite el resto de la vuelta actual y pasa a la siguiente iteración sin finalizar el bucle.
- **módulo:** archivo con funciones, variables o clases que se trae con `import`; los incluidos con Python forman la biblioteca estándar.
- **random:** módulo de la biblioteca estándar para generar números aleatorios y mezclar o elegir elementos.
- **doble bucle (bucle anidado):** bucle dentro de otro; el interior se ejecuta completo por cada vuelta del exterior.
- **validación de entrada:** uso de `while` con `input()` para repetir la pregunta hasta recibir un valor válido.

## Semana 1 — Secuencias de datos en Python

### [Unidad 10 — Tipos de datos de lista y tupla](01-python/semana-01/clase-1-listas-tuplas-diccionarios/u10-listas-y-tuplas.ipynb)

- **lista:** estructura de datos mutable que guarda varios elementos, de cualquier tipo, bajo un solo nombre, entre corchetes y separados por comas.
- **índice:** número que indica la posición de un elemento; en una lista de n elementos va de 0 a n-1 (o de -1 a -n con índices negativos).
- **indexación:** acceso al valor de un elemento mediante su índice de posición, por ejemplo `lista[0]`.
- **rebanado (slicing):** operación `lista[inicio:fin:paso]` que crea una lista nueva con una sección de otra; el índice final no se incluye.
- **tupla:** secuencia inmutable, parecida a la lista, que se crea con paréntesis y no admite cambios una vez creada.
- **mutable / inmutable:** un tipo mutable (list, set, dict) puede modificarse tras crearse; uno inmutable (int, float, str, tuple) no.
- **operadores in / not in:** devuelven True o False según un elemento esté o no en una secuencia (cadena, lista o tupla).
- **método:** función propia de un objeto que se invoca con punto, como `lista.append(x)`.

### [Unidad 11 — Tipos de datos del diccionario](01-python/semana-01/clase-1-listas-tuplas-diccionarios/u11-diccionarios-y-json.ipynb)

- **diccionario (dict):** tipo de datos formado por pares clave-valor entre llaves, que permite obtener un valor a partir de su clave.
- **clave:** elemento que identifica un valor dentro del diccionario; debe ser única y no se puede obtener a partir del valor.
- **valor:** dato asociado a una clave; puede repetirse y ser de cualquier tipo.
- **enfoque clave y asignación:** uso de la clave para leer, insertar o modificar valores, con `d[clave]` y `d[clave] = valor`.
- **KeyError:** error que se produce al leer o borrar una clave que no existe en el diccionario.
- **JSON:** formato de texto, independiente del lenguaje, para intercambiar datos formados por pares clave-valor.
- **json.loads:** función del módulo json que convierte una cadena con formato JSON en un diccionario de Python.
- **format:** método de cadenas que reemplaza marcadores `{}` por valores y controla ancho y decimales.

### [Unidad 12 — Tipos de datos de secuencia](01-python/semana-01/clase-1-listas-tuplas-diccionarios/u12-iteracion-y-busqueda.ipynb)

- **objeto secuencia:** objeto creado a partir de un tipo de dato de secuencia (lista, tupla, rango o cadena); cada valor que contiene se llama elemento.
- **operador de pertenencia:** `in` y `not in`; devuelven `True` o `False` según el elemento esté o no en la secuencia.
- **vinculación (enlace):** unir dos secuencias del mismo tipo con el operador `+`, que crea un objeto nuevo; no funciona con `range`.
- **iteración con `*`:** repetir una secuencia con `secuencia * entero`; el multiplicador debe ser un entero y `range` no lo admite.
- **índice:** número que indica la posición de un elemento; va de 0 a n-1, o de -1 a -n si es negativo.
- **corte (slicing):** `secuencia[inicio:fin]`, que extrae una porción incluyendo `inicio` y excluyendo `fin`.
- **tupla:** secuencia inmutable; con un solo elemento necesita coma, como en `(100,)`.
- **empaquetado y desempaquetado:** juntar valores en una variable (`a = (1, 2)`) y repartirlos en varias (`x, y = c`).

### [Unidad 13 — Listas bidimensionales](01-python/semana-01/clase-1-listas-tuplas-diccionarios/u13-listas-bidimensionales.ipynb)

- **lista bidimensional:** lista cuyos elementos son otras listas; se accede a un elemento con `m[fila][columna]`.
- **indexación bidimensional:** uso de dos índices, el primero para la fila y el segundo para la columna, ambos desde 0.
- **lista irregular (dentada):** lista bidimensional cuyas filas tienen distinto número de elementos.
- **doble bucle:** un `for` dentro de otro; el externo recorre las filas y el interno los elementos de cada fila.
- **referencia:** asignar `list1 = list_array` no copia la lista; ambas variables apuntan al mismo objeto.
- **deepcopy:** función de `copy` que crea una copia independiente, con las filas interiores en memoria separada.
- **comprensión de listas:** sintaxis `[expresión for x in secuencia if condición]` para crear listas en una sola línea.
- **modo de apertura:** letra de `open` que define el uso del archivo: `r` lee, `w` sobrescribe, `a` añade, `x` solo crea.

### [Unidad 14 — Diccionario, método 1: añadir, editar, borrar y guardar](01-python/semana-01/clase-2-diccionarios-y-conjuntos/u14-diccionarios-crud.ipynb)

- **diccionario:** tipo de dato que almacena pares clave-valor y se accede por clave, no por índice.
- **clave-valor:** par formado por una clave única y el valor asociado a ella.
- **setdefault:** método que añade una clave con un valor por defecto solo si la clave no existe y devuelve el valor de la clave.
- **update:** método que añade pares nuevos o sobrescribe el valor de claves existentes.
- **pop:** método que borra una clave y devuelve su valor; con un segundo argumento evita el KeyError.
- **popitem:** método que borra y devuelve como tupla el último par insertado (Python 3.6 o superior).
- **get:** método que devuelve el valor de una clave sin borrarla; devuelve None o un valor por defecto si no existe.
- **split / join:** split divide una cadena en una lista de subcadenas y join une una lista de cadenas con un separador.

### [Unidad 15 — Diccionario, método 2: fromkeys, defaultdict, diccionario doble y copia](01-python/semana-01/clase-2-diccionarios-y-conjuntos/u15-diccionarios-avanzados.ipynb)

- **fromkeys:** método `dict.fromkeys(claves, valor)` que crea un diccionario con las claves dadas y un mismo valor inicial (None por defecto).
- **defaultdict:** diccionario del módulo `collections` que crea automáticamente un valor por defecto para las claves ausentes.
- **función generadora de valor por defecto:** función sin argumentos (int, list...) que se pasa a `defaultdict` para producir el valor inicial.
- **diccionario doble (anidado):** diccionario cuyos valores son otros diccionarios, accesible con `d[clave1][clave2]`.
- **copia superficial:** `x.copy()` duplica solo el primer nivel; los objetos interiores se siguen compartiendo.
- **copia profunda:** `copy.deepcopy(x)` duplica todos los niveles en memoria independiente.
- **módulo:** archivo de Python con funciones, variables o clases que se carga con `import`.
- **epoch:** instante de referencia del sistema Unix (1 de enero de 1970, 0:0:0 UTC) desde el que `time.time()` cuenta segundos.

### [Unidad 16 — Establecer tipos de datos: conjuntos (set)](01-python/semana-01/clase-2-diccionarios-y-conjuntos/u16-conjuntos.ipynb)

- **set (conjunto):** tipo de datos desordenado que no admite elementos duplicados y no tiene índices.
- **unión:** conjunto con todos los elementos de dos conjuntos (`s1 | s2` o `s1.union(s2)`).
- **intersección:** elementos comunes a dos conjuntos (`s1 & s2` o `s1.intersection(s2)`).
- **diferencia simétrica:** elementos que están en uno de los conjuntos pero no en ambos (`s1 ^ s2`).
- **subconjunto propio:** conjunto contenido en otro y distinto de él (`B < A`).
- **conjuntos disjuntos:** conjuntos sin elementos en común (`isdisjoint`).
- **compresión de conjunto:** sintaxis `{expr for x in iterable if cond}` que crea un conjunto filtrando elementos.
- **zip:** función que agrupa varios iterables en tuplas elemento a elemento y se detiene en el más corto.

## Semana 2 — Funciones, clausuras y clases

### [Unidad 17 — Función](01-python/semana-02/clase-1-funciones-recursividad-lambda/u17-funciones.ipynb)

- **función:** conjunto de código con nombre que realiza una operación específica y se ejecuta cuando es llamado; se define con `def`.
- **parámetro:** variable definida en la cabecera de la función que recibe el valor real cuando se la llama.
- **argumento:** valor real que se pasa a la función en la llamada.
- **argumento arbitrario (`*args`):** parámetro precedido por `*` que recoge una cantidad variable de argumentos en una tupla.
- **parámetro por defecto:** parámetro con un valor asignado en la definición (`b=2`) que se usa cuando no se pasa el argumento.
- **argumento de palabra clave:** argumento pasado indicando el nombre del parámetro (`get_root(a=1, c=-8, b=2)`), de modo que el orden no importa.
- **variable global / local:** la global se declara fuera de las funciones y es visible en todo el código; la local se crea dentro de una función y solo existe allí.
- **`global`:** palabra clave que declara dentro de una función que un nombre se refiere a la variable global, para poder modificarla.

### [Unidad 18 — Llamada a función de recursión](01-python/semana-02/clase-1-funciones-recursividad-lambda/u18-recursividad.ipynb)

- **recursión (llamada recursiva):** técnica en la que una función se llama a sí misma dentro de su propio cuerpo para resolver un problema dividiéndolo en versiones más pequeñas.
- **condición de terminación (caso base):** situación en la que la función recursiva devuelve un valor sin volver a llamarse; es obligatoria para que la recursión se detenga.
- **factorial:** producto de todos los números naturales de 1 a n, escrito `n!`; se define de forma recursiva como `n! = n * (n-1)!` con `1! = 1`.
- **sucesión de Fibonacci:** secuencia `0, 1, 1, 2, 3, 5, 8...` en la que cada término es la suma de los dos anteriores, `F(n) = F(n-1) + F(n-2)`.
- **memorización (memoization):** método que guarda los resultados ya calculados de subproblemas repetidos (por ejemplo en un diccionario) para no volver a calcularlos.
- **divide y vencerás:** estrategia que divide un problema en problemas más pequeños, los resuelve y combina los resultados; el factorial no lo es porque su recursión tiene una sola rama.
- **caja negra:** forma de ver una función: se entrega la entrada correcta y se usa la salida, sin conocer el proceso interno (por ejemplo `len('ABCD')` devuelve 4).
- **Torre de Hanoi:** juego de discos y tres varillas que se resuelve con recursión; con n discos requiere un mínimo de `2^n - 1` movimientos.

### [Unidad 19 — Lambda](01-python/semana-02/clase-1-funciones-recursividad-lambda/u19-lambda-filter-map-reduce.ipynb)

- **función lambda (expresión lambda):** función anónima, sin nombre y de una sola expresión, escrita como `lambda parámetros: expresión`; devuelve el valor de la expresión.
- **filter:** función incorporada que recibe una función y un iterable y devuelve solo los elementos para los que la función devuelve `True`.
- **map:** función incorporada que aplica una función a cada elemento de un iterable y devuelve los resultados.
- **reduce:** función del módulo `functools` que combina los elementos de un iterable, de izquierda a derecha, en un único valor.
- **iterable:** objeto que se puede recorrer elemento a elemento, como una lista, una tupla, una cadena o un rango.
- **iterador:** objeto que entrega un elemento cada vez con `next()` y lanza `StopIteration` cuando se agota; se obtiene con `iter()`.
- **comprensión de listas:** sintaxis `[expresión for x in iterable if condición]` que crea una lista en una sola línea.
- **excepción:** error que se produce al ejecutar el código y que se atiende con `try ... except`, con bloques opcionales `else` y `finally`.

### [Unidad 20 — Ámbito de variables y clausuras (closure)](01-python/semana-02/clase-2-ambito-clausuras-y-clases/u20-ambito-y-clausuras.ipynb)

- **regla de ámbito (LEGB):** orden en que Python busca un nombre: Local, Enclosing (función envolvente), Global y Built-in.
- **variable local:** variable creada dentro de una función; existe solo durante la llamada y desaparece al terminar.
- **variable global:** variable definida a nivel de módulo, fuera de las funciones; una función solo puede asignarla si declara `global`.
- **función anidada:** función definida dentro de otra función; puede leer las variables de la función externa.
- **función de primera clase:** función tratada como un objeto: se pasa como argumento, se asigna a una variable, se devuelve o se guarda en una lista.
- **nonlocal:** palabra clave que enlaza un nombre con la variable del espacio de nombres envolvente más cercano (vinculación).
- **clausura (closure):** función anidada devuelta por la función externa que conserva las variables de esta aunque ya haya terminado.
- **variable libre:** variable usada en un bloque que no es global ni está definida en ese bloque (por ejemplo, `a` dentro de `mult`).

### [Unidad 21 — Clases y objetos: programación orientada a objetos](01-python/semana-02/clase-2-ambito-clausuras-y-clases/u21-clases-y-objetos.ipynb)

- **objeto:** elemento de un programa que reúne atributos (datos) y métodos (funciones); en Python todo valor es un objeto.
- **clase:** plantilla abstracta que define los atributos y acciones de un tipo de objeto (`class Cat:`).
- **instancia:** objeto concreto creado a partir de una clase, con sus propios valores de atributo.
- **constructor (`__init__`):** método especial que se ejecuta al crear una instancia y asigna sus valores iniciales.
- **`self`:** primer parámetro de un método de instancia; hace referencia a la propia instancia que lo llama.
- **variable de clase:** variable definida en el cuerpo de la clase, fuera de los métodos, compartida por todas las instancias.
- **herencia:** mecanismo por el que una clase hija (subclase) reutiliza los atributos y métodos de una clase padre (superclase).
- **encapsulación:** protección de los atributos de una clase (privados con `__`) mediante métodos de acceso, evitando modificaciones externas arbitrarias.

## Semana 3 — Pilas, colas, búsqueda y tablas hash

### [Unidad 22 — Pila](01-python/semana-03/clase-1-pilas-colas-busqueda-secuencial/u22-pilas.ipynb)

- **pila (stack):** tipo de datos abstracto en el que solo se añaden o eliminan elementos por un extremo, la cima.
- **LIFO (Last-In-First-Out):** regla por la que el último elemento en entrar es el primero en salir.
- **push:** operación que añade un elemento en la cima de la pila (el PDF traducido la escribe "pulse").
- **pop:** operación que quita y devuelve el elemento de la cima de la pila.
- **cima (top) y fondo:** extremo por el que se opera la pila y extremo opuesto, respectivamente.
- **tipo de datos abstracto:** tipo de datos definido por el usuario, en Python una clase, descrito por sus operaciones.
- **is_empty():** comprobación que devuelve `True` si la pila no tiene elementos; evita el `IndexError` al sacar.
- **paréntesis equilibrados:** cadena en la que cada paréntesis de cierre empareja con su apertura del mismo tipo y en orden correcto.

### [Unidad 23 — Cola](01-python/semana-03/clase-1-pilas-colas-busqueda-secuencial/u23-colas.ipynb)

- **cola (queue):** tipo de dato abstracto en el que se añade por la parte trasera y se elimina por la parte delantera.
- **FIFO:** First-In-First-Out; el primer elemento que entra es el primero que sale.
- **enqueue:** operación que añade un elemento al final (parte trasera) de la cola.
- **dequeue:** operación que elimina y devuelve el elemento del frente de la cola; devuelve None si está vacía.
- **tipo de dato abstracto (ADT):** descripción de una estructura por sus operaciones, independiente de cómo se implemente.
- **problema de Josefo (Josephus):** N personas en círculo; se elimina cada K-ésima hasta que queda una; se pide el orden de eliminación.
- **deque:** cola de doble extremo; permite añadir y eliminar por ambos lados.

### [Unidad 24 — Búsqueda secuencial](01-python/semana-03/clase-1-pilas-colas-busqueda-secuencial/u24-busqueda-secuencial.ipynb)

- **búsqueda secuencial:** algoritmo que revisa los elementos de una lista uno por uno, en orden, hasta encontrar el valor buscado o agotar la lista.
- **problema de búsqueda:** tarea de encontrar la posición de un elemento específico dentro de unos datos dados, o determinar que no existe.
- **tiempo lineal:** complejidad O(n); el número de operaciones crece en proporción al tamaño de los datos.
- **caso base (mejor caso):** situación en que el elemento buscado es el primero y basta una sola comparación, O(1).
- **peor caso:** situación en que el elemento no existe y se comparan los n elementos, O(n).
- **valor centinela:** valor especial, como -1, que indica "no encontrado" porque no es un índice válido.
- **find_largest:** función que encuentra el índice del mayor elemento recorriendo la lista una sola vez y guardando el índice del mayor visto hasta el momento.
- **rotura (breaking):** en el problema de la caída de huevos, el piso más bajo desde el cual el huevo se rompe.

### [Unidad 25 — Búsqueda binaria](01-python/semana-03/clase-2-busqueda-binaria-tablas-hash/u25-busqueda-binaria.ipynb)

- **búsqueda binaria:** algoritmo que busca un valor en una lista ordenada comparándolo con el elemento medio y descartando la mitad donde no puede estar.
- **espacio de búsqueda:** rango de índices, entre `low` y `high`, donde el valor buscado todavía podría encontrarse.
- **low / high / mid:** índice inferior, índice superior e índice medio, calculado como `(low + high) // 2`.
- **tiempo logarítmico:** complejidad O(log N); el número de operaciones crece muy despacio, una más cada vez que se duplica el tamaño de los datos.
- **lista ordenada:** requisito de la búsqueda binaria; los elementos están en orden ascendente.
- **huevos infinitos:** variante del problema de la caída de huevos en que se puede seguir experimentando aunque se rompan huevos, lo que permite usar búsqueda binaria.
- **operador //:** división entera; devuelve el cociente sin decimales, usado para calcular el elemento medio.

### [Unidad 26 — Tabla hash](01-python/semana-03/clase-2-busqueda-binaria-tablas-hash/u26-tablas-hash.ipynb)

- **tabla hash:** tipo de dato abstracto que almacena y extrae datos asignando claves a valores mediante hashing.
- **hashing:** designar la ubicación donde se guarda o se busca un dato a partir del valor de su clave.
- **función hash:** función que convierte una clave de longitud arbitraria en una clave hash de longitud fija; en el PDF, `f(x) = x % 8`.
- **hashkey (clave hash):** posición de la tabla (por ejemplo 0 a 7) calculada por la función hash para una clave.
- **cubo (bucket):** ranura de la tabla que, en la clase del PDF, es una lista capaz de guardar varios valores.
- **colisión:** situación en que dos claves distintas producen la misma clave hash y caen en el mismo cubo.
- **diccionario:** estructura de Python de pares clave-valor que funciona como una tabla hash con búsqueda en tiempo constante (promedio).
- **tiempo constante:** coste O(1) de buscar o guardar, independiente de la cantidad de datos (en promedio, con buena función hash).

## Semana 4 — Ordenamiento, greedy y divide y vencerás

### [Unidad 27 — Ordenamiento por burbujas, selección e inserción](01-python/semana-04/clase-1-algoritmos-de-ordenamiento/u27-burbuja-seleccion-insercion.ipynb)

- **problema de ordenamiento:** tarea de reubicar los elementos de una colección para que queden en orden (por ejemplo, de menor a mayor peso).
- **ordenamiento de burbujas:** algoritmo que compara pares de elementos adyacentes e intercambia los que están invertidos; el mayor sube al final en cada pasada.
- **ordenamiento por selección:** algoritmo que encuentra el elemento más pequeño de la parte no ordenada y lo coloca en su posición, a la izquierda.
- **ordenamiento por inserción:** algoritmo que toma cada elemento de la parte no ordenada y lo inserta en su lugar dentro de la parte ordenada, desplazando los mayores.
- **intercambio (swap):** operación que permuta los valores de dos posiciones; en Python, `S[x], S[y] = S[y], S[x]`.
- **tiempo cuadrático:** complejidad `O(N²)`, en la que el trabajo crece con el cuadrado del tamaño de la entrada; es la de los tres algoritmos de la unidad en el peor caso.
- **algoritmo in situ:** algoritmo que ordena en la misma lista, sin usar una lista adicional.
- **anagrama:** palabra formada reordenando todas las letras de otra, usando cada letra exactamente una vez.

### [Unidad 28 — Ordenamiento por fusión (merge sort)](01-python/semana-04/clase-1-algoritmos-de-ordenamiento/u28-ordenamiento-por-fusion.ipynb)

- **ordenamiento por fusión (merge sort):** algoritmo que divide la lista en dos mitades, ordena cada una recursivamente y las fusiona en una lista ordenada.
- **fusionar (merge):** construir una lista ordenada a partir de dos listas ya ordenadas, comparando sus primeros elementos.
- **divide y vencerás:** técnica que divide un problema en subproblemas más pequeños, los resuelve (aquí de forma recursiva) y combina los resultados.
- **caso base:** situación que detiene la recursión; en merge sort, una lista de 0 o 1 elementos, que ya está ordenada.
- **memoria adicional:** espacio extra que usa el algoritmo además de la lista de entrada; `mergesort1` usa 2N según el PDF.
- **O(N log N):** complejidad temporal del ordenamiento por fusión: unos log N niveles de recursión con N trabajo de fusión por nivel.
- **estabilidad:** propiedad de conservar el orden relativo de los elementos con igual clave; depende de usar `<=` al fusionar.

### [Unidad 29 — Ordenamiento rápido (quicksort)](01-python/semana-04/clase-1-algoritmos-de-ordenamiento/u29-ordenamiento-rapido.ipynb)

- **ordenamiento rápido (quicksort):** algoritmo de ordenamiento de divide y vencerás que divide la lista según un pivote y ordena recursivamente cada parte.
- **pivote:** elemento de referencia con el que se comparan los demás para separar menores y mayores.
- **partición:** operación que reordena un tramo de la lista de modo que los menores queden a la izquierda del pivote y los mayores a la derecha.
- **pivotpoint (punto de pivote):** posición final del pivote al terminar la partición; ya no cambia.
- **in situ:** ordenamiento que reorganiza los elementos dentro de la propia lista, sin lista auxiliar del tamaño de la entrada.
- **peor caso:** entrada que produce particiones de n - 1 elementos en cada llamada; con primer pivote, la lista ordenada o la inversa.
- **pivote aleatorio:** estrategia que elige una posición al azar (randint) y la intercambia con la primera antes de particionar.
- **selección rápida (quickselect):** uso de la partición para hallar el k-ésimo elemento siguiendo solo el lado donde está el índice buscado.

### [Unidad 30 — Enfoque Greedy](01-python/semana-04/clase-2-greedy-divide-y-venceras/u30-enfoque-greedy.ipynb)

- **algoritmo Greedy (voraz):** técnica de diseño que construye la solución eligiendo en cada paso la mejor opción del momento según un criterio predefinido, sin reconsiderarla.
- **proceso de selección:** primer paso del enfoque Greedy; elegir el siguiente elemento que se añade al conjunto solución.
- **comprobación de la viabilidad:** segundo paso; verificar que el conjunto con el nuevo elemento sigue siendo apropiado para ser la respuesta.
- **comprobación de la solución:** tercer paso; determinar si el conjunto actual ya es la respuesta al problema.
- **problema del cambio de monedas:** devolver un importe con el menor número de monedas posible; el criterio greedy toma siempre la moneda más grande que quepa.
- **selección de actividades:** dado un conjunto de reuniones con hora de inicio y de fin, elegir el máximo número de ellas que no se solapen.
- **caso base:** condición que detiene una función recursiva (por ejemplo, importe igual a 0 o no quedar candidatas).

### [Unidad 31 — Divide y vencerás (divide y conquista)](01-python/semana-04/clase-2-greedy-divide-y-venceras/u31-divide-y-venceras.ipynb)

- **divide y vencerás:** estrategia que divide un problema en instancias más pequeñas, las resuelve (recursivamente si hace falta) y combina sus respuestas.
- **caso base:** condición de terminación de una función recursiva, donde la respuesta se obtiene sin más llamadas.
- **búsqueda binaria:** búsqueda en una lista ordenada que descarta la mitad del rango en cada paso; resuelve solo uno de los dos subproblemas.
- **merge sort:** ordenamiento que divide la lista en dos mitades iguales, ordena ambas y las fusiona con un procedimiento de mezcla.
- **quick sort:** ordenamiento que divide la lista según un pivote, sin paso de combinación y con partes de tamaño no necesariamente igual.
- **pivote:** elemento de referencia que separa los menores de los mayores en quick sort.
- **problema de la ruta mínima del triángulo:** hallar la menor suma al descender por un triángulo de enteros pasando a uno de los dos números adyacentes de la fila inferior.
- **tromino:** pieza en forma de L formada por tres cuadrados unidos.

## Semana 5 — Programación dinámica, backtracking y primeros pasos con pandas

### [Unidad 32 — Programación dinámica](01-python/semana-05/clase-1-programacion-dinamica-y-backtracking/u32-programacion-dinamica.ipynb)

- **programación dinámica:** técnica de diseño que resuelve un problema a partir de una relación recursiva, guardando las respuestas de los subproblemas en una tabla para no recalcularlas.
- **subproblemas superpuestos:** llamadas recursivas duplicadas que resuelven varias veces el mismo subproblema, como `fib1(3)` dentro de `fib1(5)`.
- **memorización:** técnica de almacenar en una tabla (diccionario o lista) las respuestas ya calculadas para devolverlas directamente cuando se repite el subproblema.
- **bottom-up (ascendente):** enfoque que resuelve primero los casos pequeños y construye con ellos la respuesta del caso completo, normalmente con bucles.
- **top-down (descendente):** enfoque de dividir y vencer que parte del problema completo y lo descompone mediante llamadas recursivas.
- **coeficiente binomial:** número C(n, k) que cumple C(n, k) = C(n-1, k-1) + C(n-1, k), con valor 1 cuando k = 0 o k = n.
- **triángulo de Pascal:** matriz triangular de coeficientes binomiales donde cada valor es la suma de los dos que tiene encima.
- **subestructura óptima:** propiedad por la cual la solución óptima del problema se construye con soluciones óptimas de sus subproblemas.

### [Unidad 33 — Retroceso (backtracking)](01-python/semana-05/clase-1-programacion-dinamica-y-backtracking/u33-backtracking.ipynb)

- **retroceso (backtracking):** técnica que construye una solución de forma incremental y vuelve atrás cuando el candidato no puede conducir a una solución válida.
- **búsqueda en profundidad (DFS):** recorrido que explora cada rama hasta el fondo antes de volver; se implementa con recursión usando la pila de llamadas.
- **árbol del espacio de estados:** árbol que organiza todos los candidatos parciales; cada nodo es un estado y sus hijos son las extensiones posibles.
- **prometedor (promising):** estado en el que aún puede existir una solución en el subárbol del nodo visitado.
- **poda (pruning):** volver sin visitar el subárbol de un nodo no prometedor.
- **problema de las N reinas:** colocar N reinas en un tablero N*N sin que dos compartan fila, columna o diagonal.
- **suma de subconjuntos:** encontrar los subconjuntos de una lista cuya suma es un valor dado; se resuelve con retroceso podando las sumas que ya superan el objetivo.

### [Unidad 34 — Uso de módulos de Python](01-python/semana-05/clase-2-modulos-y-pandas/u34-modulos-de-python.ipynb)

- **módulo:** archivo `.py` que agrupa funciones, clases y variables (y puede incluir código de ejecución); puede ser estándar o externo.
- **módulo estándar:** módulo incorporado en Python que no requiere instalación, como `os`, `math`, `random` o `datetime`.
- **módulo externo:** módulo creado por terceros que hay que instalar antes de usarlo, como `pandas`, `numpy` o `matplotlib`.
- **pip:** herramienta que instala módulos externos desde PyPI con el comando `pip install nombre`.
- **import / from / as:** sentencias para traer un módulo completo, solo algunos de sus nombres, o asignarles un alias.
- **marca de tiempo Unix:** segundos transcurridos desde las 00:00:00 UTC del 1 de enero de 1970; la devuelve `time.time()` como `float`.
- **timedelta:** objeto de `datetime` que representa una duración o la diferencia entre dos fechas; se suma y se resta con fechas.
- **semilla (seed):** valor inicial del generador aleatorio; con `random.seed(n)` la secuencia de números se repite igual.

### [Unidad 35 — Serie Pandas para el tratamiento de datos](01-python/semana-05/clase-2-modulos-y-pandas/u35-series-de-pandas.ipynb)

- **Pandas:** biblioteca de código abierto de Python para operar, limpiar y analizar datos; se importa como `import pandas as pd`.
- **Serie (Series):** estructura unidimensional de pandas en la que cada dato tiene un índice asociado (relación 1:1).
- **DataFrame:** estructura bidimensional de pandas formada por un índice y columnas; varias series con el mismo índice forman un DataFrame.
- **Índice (index):** etiquetas de una serie; por defecto enteros desde 0, o los nombres que se indiquen con el parámetro `index`.
- **NaN:** valor que indica un dato faltante; aparece al operar series cuyos índices no coinciden.
- **Gráfico lineal:** gráfico de matplotlib (`plt.plot`) que une puntos (x, y) con una línea.
- **Gráfico de barras:** gráfico (`plt.bar` vertical, `plt.barh` horizontal) donde la altura o longitud de cada barra es proporcional al valor.
- **Densidad de población:** número de habitantes por unidad de superficie, normalmente por km².

### [Unidad 36 — Pandas DataFrame para procesamiento de datos](01-python/semana-05/clase-2-modulos-y-pandas/u36-dataframes-de-pandas.ipynb)

- **DataFrame:** estructura de pandas bidimensional con filas y columnas etiquetadas; cada columna es una Serie y todas comparten el índice.
- **Índice:** etiquetas de las filas de un DataFrame; puede ser numérico, de texto o jerárquico (multiíndice).
- **Característica:** columna de la tabla que describe una propiedad de cada registro (también atributo, predictor o variable).
- **Registro:** fila de la tabla que representa un caso u observación.
- **E/S de pandas:** conjunto de funciones de lectura (`read_csv`, `read_excel`, `read_json`) y escritura (`to_csv`, `to_excel`, `to_json`) de datos externos.
- **API:** servicio remoto que entrega datos; `pandas-datareader` los convierte directamente en un DataFrame (FRED, Banco Mundial, etc.), a veces con clave API.
- **Correlación:** medida de la relación entre dos variables; el coeficiente de Pearson va de -1 a 1, y un valor cercano a 0 indica que no hay relación lineal.
- **Eje (axis):** dirección de una operación; `axis=0` recorre las filas (índice) y `axis=1` las columnas.

## Semana 6 — Datos ordenados, series temporales y mini proyectos

### [Unidad 37 — Ordenación de datos](01-python/semana-06/clase-1-ordenacion-y-series-temporales/u37-ordenacion-de-datos.ipynb)

- **NaN:** valor que representa un dato que falta en pandas (*Not a Number*); se detecta con `isnull()`/`isna()`.
- **datos ordenados (Tidy Data):** forma estándar de estructurar un conjunto de datos, término de Hadley Wickham: cada variable en una columna, cada observación en una fila y cada valor en una celda.
- **ordenación de datos:** proceso de organizar datos anómalos (con datos que faltan, unidades distintas o duplicados) para poder analizarlos.
- **estadísticas descriptivas:** medidas que resumen un conjunto de datos, como la media, la mediana, la moda, la varianza y la desviación típica.
- **coeficiente de variación (CV):** desviación típica dividida por la media; permite comparar la dispersión de variables con escalas distintas.
- **cuartil / IQR:** los cuartiles dividen los datos en cuatro partes (Q1 = 25 %, Q2 = mediana, Q3 = 75 %); el intervalo intercuartílico (IQR) es Q3 - Q1 y forma la caja del box plot.
- **coeficiente de correlación:** medida entre -1 y 1 del grado en que dos variables se mueven juntas; no depende de la escala, a diferencia de la covarianza.
- **diagrama de caja (boxplot):** gráfico que muestra mediana, cuartiles, bigotes y datos atípicos de una variable.

### [Unidad 38 — Datos de series temporales](01-python/semana-06/clase-1-ordenacion-y-series-temporales/u38-series-temporales.ipynb)

- **Serie temporal:** conjunto de valores de una o más variables medidos a lo largo de un periodo, a intervalos específicos y ordenados en el tiempo.
- **Timestamp (marca de tiempo):** un punto específico en el tiempo, representado en pandas con el tipo `datetime64`.
- **Period (periodo):** intervalo constante entre dos marcas de tiempo, definido por una frecuencia (por ejemplo, un mes); expone `start_time` y `end_time`.
- **DatetimeIndex:** índice formado por varias marcas de tiempo; ordena los datos por fecha y permite trocearlos con fechas.
- **Timedelta:** duración o diferencia entre dos fechas u horas, usada para sumar o restar intervalos.
- **Media móvil:** promedio de las últimas N observaciones de una serie (`rolling(window=N).mean()`); suaviza la volatilidad a corto plazo y muestra la tendencia.
- **Modelo aditivo:** descomposición de una serie como suma de tendencia, ciclo, estacionalidad y componente aleatorio.
- **Gráfico de áreas:** gráfico de líneas con el área inferior coloreada (`kind='area'`); permite comparar dos o más series superpuestas.

### [Unidad 39 — Mini proyecto de análisis de datos financieros](01-python/semana-06/clase-2-mini-proyectos/u39-analisis-datos-financieros.ipynb)

- **ticker:** código corto que identifica un valor o fondo en el mercado (por ejemplo, `SPY` para el S&P 500).
- **serie temporal:** conjunto de datos ordenados por fecha; en pandas, un `DataFrame` o `Series` con `DatetimeIndex`.
- **cambio porcentual diario:** variación entre el cierre de hoy y el de ayer, `(hoy - ayer) / ayer * 100`.
- **rendimiento acumulado:** efecto total de los cambios diarios a lo largo del periodo; el PDF lo aproxima con `.cumsum()` y el cálculo exacto es compuesto, `(1 + r).cumprod() - 1`.
- **coeficiente de correlación:** medida entre -1 y 1 de la asociación entre dos series; cerca de 1 se mueven juntas, cerca de 0 no tienen relación y negativo indica sentidos opuestos (`.corr()`).
- **volatilidad:** magnitud de la variación del precio en un periodo; se mide con la desviación estándar de los cambios diarios en una ventana móvil (`rolling().std()`).
- **ventana móvil:** subconjunto deslizante de las últimas `n` observaciones sobre el que se calcula una medida (media, desviación estándar).
- **mapa de calor:** representación de una matriz con colores según el valor de cada celda; se usa para la matriz de correlación.

### [Unidad 40 — Miniproyecto de análisis de la pandemia de Coronavirus en el mundo](01-python/semana-06/clase-2-mini-proyectos/u40-pandemia-coronavirus.ipynb)

- **Mapa coroplético:** mapa en el que cada área (país, distrito) se colorea según el valor de una variable; a mayor valor, color más oscuro.
- **Folium:** biblioteca de Python que genera mapas interactivos basados en Leaflet; se instala con `pip install folium`.
- **GeoJSON:** formato de archivo que describe los límites geográficos (polígonos) de áreas como países o distritos.
- **Latitud y longitud:** coordenadas de un punto del mapa; Folium las recibe en el orden latitud, longitud.
- **Marcador (`Marker`):** icono que se coloca en una coordenada del mapa y puede mostrar un mensaje emergente (`popup`).
- **Acumulado:** valor que incluye todo lo ocurrido hasta una fecha; para conocer el nivel actual se toma el último valor, no la suma de las filas.
- **Estadísticas descriptivas:** medidas que resumen un conjunto de datos (media, mediana, mínimo, máximo, cuartiles).
- **`key_on`:** parámetro de `folium.Choropleth` que indica el campo del GeoJSON que debe coincidir con la columna identificadora del marco de datos.
